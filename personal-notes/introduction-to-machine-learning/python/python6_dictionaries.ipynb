# 8. Data Structures: Dictionaries

The second data structure we will explore is **dictionaries**. These are an unordered set of elements written within curly braces, `{}`, consisting of keys and values.

Each pair `key:value` is separated by commas. Keys act as identifiers and precede `:`. Following the keys are the values, which can be elements (numeric, boolean, strings, lists, dictionaries, etc.) associated with that key.

Dictionaries, like lists, have the following properties:
- Heterogeneous: elements can be of different types within the same dictionary.
- Mutable: elements can be modified.

An example of a dictionary would be

```python
dicc = {"Bob": 32, "Alice": 21}
print(dicc)
```

In `Python`, dictionary keys must be unique, meaning there cannot be two keys that are exactly the same. If there happen to be two identical keys, Python retains the last value associated with that key.

```python
dicc = {"Bob": 32,
        "Alice": 21,
        "Bob": 23}
print(dicc)
```

## 8.1. Elements and Size of a Dictionary

To access the elements of a dictionary, we cannot do it by position, but rather through their keys.

The syntax is `dictionary[key]`.

```python
dicc = {
    "names": ["Bob", "Alice", "John"],
    "ages" : [31, 25, 16]
    }
dicc["names"]
```

```python
dicc["ages"]
```

Since it is a mutable structure, we can modify the values of dictionaries.

```python
dicc = {"names": ["Bob", "Alice", "John"],
        "ages": [31, 25, 16]}
dicc["names"] = ["David", "Emily", "Sarah"]
dicc["ages"][2] = 36
print(dicc)
```

To find out how many elements a dictionary contains, we can use the `len()` function like this:

```python
dicc = {"fruit": ["Apple", "Pear", "Orange"],
        "price": [2, 1.5, 1],
        "color": ["red", "green", "orange"]}
        
print(len(dicc))
```

## 8.4. Dictionary Methods

| Method         | Description                                                     |
|----------------|-----------------------------------------------------------------|
| `.keys()`      | Returns a view object that displays a list of all the keys       |
| `.values()`    | Returns a view object that displays a list of all the values     |
| `.items()`     | Returns a view object that displays a list of key-value tuples   |
| `.clear()`     | Removes all items from the dictionary                            |
| `.copy()`      | Returns a shallow copy of the dictionary                         |
| `.fromkeys()`  | Creates a new dictionary with keys from a sequence and values set to a default value |
| `.get()`       | Returns the value of the specified key                           |
| `.pop()`       | Removes the specified key and returns the corresponding value   |
| `.setdefault()`| Returns the value of the specified key. If the key does not exist, inserts the key with the specified value|
| `.update()`    | Updates the dictionary with the key-value pairs from another dictionary or iterable |

```python
# Define a dictionary
my_dict = {"name": "Alice", "age": 30, "city": "New York"}

# .keys(): Returns a view object that displays a list of all the keys
print(my_dict.keys())

# .values(): Returns a view object that displays a list of all the values
print(my_dict.values())

# .items(): Returns a view object that displays a list of key-value tuples
print(my_dict.items())

# .clear(): Removes all items from the dictionary
my_dict.clear()
print(my_dict)

# .copy(): Returns a shallow copy of the dictionary
my_dict = {"name": "Alice", "age": 30, "city": "New York"}
copied_dict = my_dict.copy()
print(copied_dict)

# .fromkeys(): Creates a new dictionary with keys from a sequence and values set to a default value
keys = ['a', 'b', 'c']
default_value = 0
new_dict = dict.fromkeys(keys, default_value)
print(new_dict)

# .get(): Returns the value of the specified key
print(my_dict.get("name"))

# .pop(): Removes the specified key and returns the corresponding value
removed_value = my_dict.pop("age")
print(removed_value)  # Output: 30
print(my_dict)

# .setdefault(): Returns the value of the specified key. If the key does not exist, inserts the key with the specified value
my_dict.setdefault("age", 25)  # 'age' key already exists, so no change
my_dict.setdefault("gender", "female")  # 'gender' key doesn't exist, so it's added
print(my_dict)

# .update(): Updates the dictionary with the key-value pairs from another dictionary or iterable
update_dict = {"age": 35, "country": "USA"}
my_dict.update(update_dict)
print(my_dict)
```

## 8.5. Loops and dictonaries

To iterate over the entire dictionary, we can make use of a `for` loop, as the dictionary is an iterable structure:

```python
dicc = {"username": "bob",
        "name": "Bob",
        "age": 22,
        "city": "Michigan"}

for key in dicc:
    print(key, ":", dicc[key])
```

Another way to iterate through the dictionary would be to obtain a list of tuples of the form `(key, value)` for each element of a dictionary, which we construct using the `.items()` method. Since it's a list, we know it's iterable, and we can display all its entries using a `for` loop.

```python
dicc.items()
```

```python
for item in dicc.items():
    print(item)
```

```python
for key, value in dicc.items():
  print(key, ":", value)
```

### 8.6.2. Problems

We have a dictionary with 5 keys: `Math`, `English`, `History`, `Science`, `IT`. Each key contains a list of size 3, where each of those entries corresponds to a grade from 0 to 10. The user will input these grades through the keyboard. For each key, display the average of the 3 grades. HINT: Make use of the `sum()` function.

```python
subjects = {"Math": [],
            "English": [],
            "History": [],
            "Science": [],
            "IT": []}

for key in subjects:
    print("===", key, "===")
    for i in range(1, 4):
        subjects[key].append(int(input("Grade for quarter {} ".format(i))))

print("\n Average grades per subject:")
for key, val in subjects.items():
    print(key, ":", sum(val)/len(val))
```

### 8.6.4. Problems

Prompt the user for 8 integer numbers from 0 to 9. These are assumed to be the numbers of their ID card, which we will store each one in a list entry. Then, with those numbers, calculate the corresponding letter according to the Spanish ID card rules and save it in a variable. Finally, create a dictionary with two keys, each one respectively storing the numbers and the letter of the ID card. Display the resulting dictionary.

```python
print("Enter the DNI numbers one by one:")
nums = []
for _ in range(8):
    nums.append(int(input()))

dni = 0
for i in range(len(nums)):
    dni += 10 ** (len(nums) - i - 1) * nums[i]

letters = {0: "T", 1: "R", 2: "W", 3: "A", 4: "G", 5: "M", 6: "Y",
           7: "F", 8: "P", 9: "D", 10: "X", 11: "B", 12: "N", 13: "J",
           14: "Z", 15: "S", 16: "Q", 17: "V", 18: "H", 19: "L", 20: "C",
           21: "K", 22: "E"}

letter = letters[dni % 23]

DNI = {"numbers": nums, "letter": letter}
print(DNI)
```

### 8.6.5. Problems

Read a string from the keyboard and return a dictionary with the count of each character's appearances in the string provided by the user.

```python
s = input("Enter a sentence: ")
counted = []
letters = {}

s = s.lower()
for c in s:
    if c not in counted and c != " ":
        counted.append(c)
        letters[c.upper()] = s.count(c)

print(letters)
```

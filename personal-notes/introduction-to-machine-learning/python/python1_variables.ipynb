# 3. Python variables

A variable comprises a designated space in the storage system, typically the main memory of a computer, and a symbolic name (an identifier) linked to that space. Essentially, a variable represents the connection between a name and an object situated within the computer's memory. In Python, values are assigned to variables using the format: `variable_name = value`.

```python
x = 1
x
```

```python
y = "hello"
y
```

## 3.1. Constraints on Variable Names

- Cannot start with or contain special characters.
- Cannot begin with numbers.
- Cannot be named the same as reserved keywords in Python.
- Cannot include spaces.

It is advisable to give variables meaningful names that reflect the data they hold, facilitating comprehension for anyone reading the code.

Currently, for variable names composed of multiple words, there are four accepted conventions:

- camelCase: `variableName`
- PascalCase: `VariableName`
- snake_case: `variable_name`
- kebab-case: `variable-name`

## 3.4. Performing Operations with a Numeric Variable

Once we have stored a numerical value in a variable, we can perform operations on it:

```python
x = 3
x + 1
```

We can even save that value in a new variable.

```python
y = x + 1
y
```

Alternatively, we can overwrite the original variable with that value.

```python
x = x + 1
x
```

In `Python`, if we want to overwrite a numeric variable by adding a certain amount to it, we can do so in the following way:

```python
x = 7
x
```

```python
x += 2
x
```

Just like `+=`, we also have `-=`, `*=`, `/=`, `//=`, `%=`, and `**=`, which are equivalent to `+=` for the other arithmetic operations in `Python`, and we will cover them in the next topic.

## 3.5. Comments

Given a code block, it can be helpful at times to explain what it does or its purpose, or to make sure a line doesn't execute for some reason but remains in the code. This is where comments come into play—they are part of the code but are not executed.

If we want to make a single-line comment, we use `#`.

```python
# Let's define the variable x and assign it the value 1
x = 1  # Here is our variable x
# This is another comment
# And none of the comments will be executed
x
```

Comments, can be very helpful in understanding our code. However, it is advisable not to overuse them, as demonstrated in the previous example, as the code transitions from being clear to becoming excessively lengthy and confusing.

# 14. Dataframes with the `pandas` module

**Dataframe.** It is a mutable two-dimensional data structure with the axes labeled where

- each row represents a different observation
- each column represents a different variable

In `Python`, to define a dataframe, we first need to import the `pandas` module.

```python
import pandas as pd
```

We can create dataframes from a dictionary, a list of lists, a list of dictionaries, etc.

```python
# Dataframes from a dictionary

data = {"x": [1, 2, 3, 4, 5], "y": [2, 4, 6, 8, 10]}
df1 = pd.DataFrame(data = data)
print(df1)
```

```python
# Dataframe from a list of lists

df2 = pd.DataFrame(data = [[1, 2], [2, 4], [3, 6], [4, 8], [5, 10]],
                   columns = ["x", "y"])
print(df2)
```

```python
# We can modify the name of the rows with the `index` parameter.

df3 = pd.DataFrame(data = data, index = ["obs1", "obs2", "obs3", "obs4", "obs5"])
print(df3)
```

```python
# Dataframe from a list of dictionaries

data = [{"x": 1, "y": 2},
        {"x": 2, "y": 4},
        {"x": 3, "y": 6},
        {"x": 4, "y": 8},
        {"x": 5, "y": 10}]
df4 = pd.DataFrame(data = data)
print(df4)
```

```python
# We can create a dataframe using the `zip()` function.

x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

data = list(zip(x, y))
print("Data:\n")
print(data)

print("-----------------------------------------")

print("Dataframe:\n")
df5 = pd.DataFrame(data, columns = ["x", "y"])
print(df5)
```

### 14.1.2. `csv` files

```python
import csv
```

```python
# To read a `csv` we use the `.reader()` method of the `csv` module.

# With the `open()` function we have accessed the file, which we have identified by `f`.
with open("csv_example.csv", "r") as f:
    # Next, we have used the `.reader()` method to read the file `f`. 
    # This method has returned us the iterable `reader`, of which we 
    # have shown all its rows iterating with a `for` loop.
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

```python
# The `.reader()` method allows us to configure said separator with the `delimiter` parameter

with open("csv_delimiter_example.csv", "r") as f:
    reader = csv.reader(f, delimiter = "|")
    for row in reader:
        print(row)
```

```python
# To remove extra whitespace, the `.reader()` method brings the `skipinitialspace` parameter. 
# If we set it to `True`, the extra spaces will disappear.

# skipinitialspace = False (default)
with open("csv_spaces_example.csv", "r") as f:
    reader = csv.reader(f, skipinitialspace = False)
    for row in reader:
        print(row)
```

```python
# If instead we want quotes, we have the `quoting` parameter, which admits different values:

with open("csv_quotation_example.csv", "r") as f:
    reader = csv.reader(f, quoting = csv.QUOTE_NONNUMERIC)
    for row in reader:
        print(row)
```

```python
with open("csv_quotation_example.csv", "r") as f:
    reader = csv.reader(f, quoting = csv.QUOTE_ALL)
    for row in reader:
        print(row)
```

```python
# Dialects

# We give the `.register_dialect()` method a name in `string` format and then set the `delimiter`,
# `skipinitialspace` and `quoting` parameters.

# Next, to the `.reader()` method we give the name of the dialect, `my_dialect`, to the `dialect` 
# parameter and thus we read the file correctly.

# Once the custom dialect is created, we can use it as many times as we want to open and read `csv` 
# files with the same format, in this case, as `csv_dialect_example.csv`. 


csv.register_dialect("my_dialect",
                     delimiter = "|",
                     skipinitialspace = True,
                     quoting = csv.QUOTE_NONNUMERIC)

with open("csv_dialect_example.csv", "r") as f:
    reader = csv.reader(f, dialect = "my_dialect")
    for row in reader:
        print(row)
```

```python
# Dictionaries and `csv`

# With the `.DictReader()` method of the `csv` module, we return an `OrderedDict` object, which is iterable.

with open("csv_example.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row)
```

```python
# # We could use the `dict()` function inside `print()` to display `OrderedDict` objects as dictionaries.

with open("csv_example.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(dict(row))
```

```python
# To create and write a `csv` we use the `open()` function together with the `csv.writer()` method:

data = [["id", "Name", "City", "Age"], 
        [1234, "Arturo", "Madrid", 22], 
        [2345, "Beatriz", "Barcelona", 25],
        [3456, "Carlos", "Sevilla", 18], 
        [4567, "Dolores", "Cuenca", 34]]

with open("csv_write.csv", "w") as f:
    writer = csv.writer(f)
    for row in data:
        writer.writerow(row)
```

```python
# We can do the same as the previous chunk of code without using a loop, using the .writerows() method.

with open("csv_write_writerows.csv", "w") as f:
    writer = csv.writer(f)
    writer.writerows(data)
```

```python
# We can write a `csv` from a dictionary. To do this we will have to use the `.DictWriter()` method.

data = [{"id": 1234, "Name": "Arturo", "City": "Madrid", "Age": 22},
        {"id": 2345, "Name": "Beatriz", "City": "Barcelona", "Age": 25},
        {"id": 3456, "Name": "Carlos", "City": "Sevilla", "Age": 18},
        {"id": 4567, "Name": "Dolores", "City": "Cuenca", "Age": 34}]

# The header is the list of the keys of any of the data entries
header = list(data[0].keys())

with open("csv_write_dict.csv", "w") as f:
    writer = csv.DictWriter(f, fieldnames = header)

    writer.writeheader()
    for d in data:
        writer.writerow(d)
```

```python
# We could have obtained the same result without using any loops, but using the `.writerows()` method

with open("csv_write_dict_writerows.csv", "w") as f:
    writer = csv.DictWriter(f, fieldnames = header)

    writer.writeheader()
    writer.writerows(data)
```

```python
# Create a dictionary with data
data = {'A': [1, 2, 3],
        'B': [4, 5, 6],
        'C': [7, 8, 9]}

# Create a DataFrame from the dictionary with 'index' orientation
df = pd.DataFrame.from_dict(data, orient='index', columns=['X', 'Y', 'Z'])

print(df)
```

```python
# Create a sample DataFrame
data = {'A': [1, 2, 3],
        'B': [4, 5, 6],
        'C': [7, 8, 9]}
df = pd.DataFrame(data)

# .shape - Returns the dimensions of the DataFrame
print("Shape of the DataFrame:", df.shape)

# .info() - Provides a concise summary of the DataFrame's info
print("\nInfo of the DataFrame:")
print(df.info())

# .describe() - Generates descriptive statistics of the DataFrame
print("\nDescriptive statistics of the DataFrame:")
print(df.describe())

# .size - Returns the number of elements in the DataFrame
print("\nSize of the DataFrame:", df.size)

# .ndim - Returns the number of dimensions (axes) of the DataFrame
print("\nNumber of dimensions of the DataFrame:", df.ndim)
```

```python
# Create a sample DataFrame
data = {'A': [1, 2, 3],
        'B': [4, 5, 6],
        'C': [7, 8, 9]}
df = pd.DataFrame(data)

# [] - Selects all rows of the DataFrame
print("Selecting all rows of the DataFrame:")
print(df[:])

# .columns[] - Accesses the column labels of the DataFrame
print("\nAccessing column labels of the DataFrame:")
print(df.columns)

# .loc[] - Accesses a group of rows and columns by labels
print("\nAccessing rows and columns by labels:")
print(df.loc[0:1, ['A', 'B']])

# .iloc[] - Accesses group of rows and columns by integer position
print("\nAccessing rows and columns by integer position:")
print(df.iloc[0:2, 0:2])

# .head() - Returns the first n rows of the DataFrame
print("\nFirst 2 rows of the DataFrame:")
print(df.head(2))

# .tail() - Returns the last n rows of the DataFrame
print("\nLast 2 rows of the DataFrame:")
print(df.tail(2))

# .copy() - Creates a deep copy of the DataFrame
print("\nDeep copy of the DataFrame:")
df_copy = df.copy()
print(df_copy)
```

```python
# Create a sample DataFrame
data = {'A': [1, 2, 3],
        'B': [4, 5, 6],
        'C': [7, 8, 9]}
df = pd.DataFrame(data)

# .rename() - Renames index/columns of the DataFrame (with inplace argument)
print("Renaming columns of the DataFrame:")
df.rename(columns={'A': 'New_A', 'B': 'New_B', 'C': 'New_C'}, inplace=True)
print(df)

# .insert() - Inserts new columns into the DataFrame
print("\nInserting a new column 'D' into the DataFrame:")
df.insert(loc=3, column='D', value=[10, 11, 12])
print(df)

# .drop() - Drops specified labels from rows or columns
print("\nDropping column 'D' from the DataFrame:")
df.drop(columns=['D'], inplace=True)
print(df)

# .pop() - Removes and returns a column from the DataFrame
print("\nPop column 'New_C' from the DataFrame:")
column_c = df.pop('New_C')
print("Column C:", column_c)
print("DataFrame after popping column C:")
print(df)

# .rank() - Computes numerical data ranks (with ascending argument)
print("\nRanking values of column 'New_A' in ascending order:")
print(df['New_A'].rank(ascending=True))

# .nunique() - Counts distinct observations over requested axis
print("\nNumber of unique values in column 'New_B':", df['New_B'].nunique())

# .duplicated() - Identifies duplicated rows in the DataFrame
print("\nIdentifying duplicated rows:")
print(df.duplicated())

# .drop_duplicates() - Removes duplicate rows from the DataFrame
print("\nRemoving duplicate rows:")
print(df.drop_duplicates())

# .nsmallest() - Returns the first n smallest elements from the DataFrame
print("\nGetting the two smallest values from column 'New_A':")
print(df['New_A'].nsmallest(2))

# .nlargest() - Returns the first n largest elements from the DataFrame
print("\nGetting the two largest values from column 'New_A':")
print(df['New_A'].nlargest(2))

# .dtypes - Returns the data types of the DataFrame columns
print("\nData types of the DataFrame columns:")
print(df.dtypes)
```

```python
import numpy as np

# Create a sample DataFrame with missing values
data = {'A': [1, np.nan, 3, 4, 5],
        'B': [np.nan, 2, np.nan, 4, 5],
        'C': [1, 2, 3, 4, 5]}
df = pd.DataFrame(data)

# .isnull() - Detects missing values in the DataFrame
print("Checking for missing values in the DataFrame:")
print(df.isnull())

# .notnull() - Detects non-missing values in the DataFrame
print("\nChecking for non-missing values in the DataFrame:")
print(df.notnull())

# .fillna() - Fills missing values with specified value
print("\nFilling missing values with 0:")
print(df.fillna(0))

# .replace() - Replaces specified values in the DataFrame
print("\nReplacing NaN values with -1:")
print(df.replace(np.nan, -1))

# .interpolate() - Interpolates missing values in the DataFrame
print("\nInterpolating missing values:")
print(df.interpolate())

# .dropna() - Drops missing values from the DataFrame
print("\nDropping rows with missing values:")
print(df.dropna())
```

```python
# Create a sample DataFrame
data = {'A': [1, 2, 3, 4, 5],
        'B': ['a', 'b', 'c', 'd', 'e']}
df = pd.DataFrame(data)

# Using .query() to filter rows based on a condition
filtered_df = df.query('A > 2')
print("Filtered DataFrame where column 'A' is greater than 2:")
print(filtered_df)
```

## 14.4. Filtering dataframes

```python
# Create a sample DataFrame
data = {'A': [1, 2, 3, 4, 5],
        'B': ['a', 'b', 'c', 'd', 'e']}
df = pd.DataFrame(data)
```

```python
# Filtering rows based on a single condition:
# Filter rows where column 'A' is greater than 2

filtered_df = df[df['A'] > 2]
print(filtered_df)
```

```python
# Filtering rows based on multiple conditions:
# Filter rows where column 'A' is greater than 2 and column 'B' is 'c'
filtered_df = df[(df['A'] > 2) & (df['B'] == 'c')]
print(filtered_df)
```

```python
# Filtering rows based on a condition using `.loc`:
# Filter rows where column 'A' is greater than 2
filtered_df = df.loc[df['A'] > 2]
print(filtered_df)
```

```python
# Filtering rows based on a condition and selecting specific columns:
# Filter rows where column 'A' is greater than 2 and select only columns 'A' and 'B'
filtered_df = df.loc[df['A'] > 2, ['A', 'B']]
print(filtered_df)
```

## 14.5. `pandas` series

**Series.** A `pandas` Series is like a column of a dataframe.

We can construct Series of `pandas` from a one-dimensional list

```python
a = [1, 2, 3, 4, 5]
my_series = pd.Series(a)
print(my_series)
```

**Note.** If we do not specify anything, by default the labels of the Series entries correspond to the index they occupy. Let's remember that in `Python` the indices start with 0.

These tags can be used to access a specific value

```python
print(my_series[1])
```

When we create a Series, we can modify its labels with the `index` parameter:

```python
my_series = pd.Series(a, index = ["a", "b", "c", "d", "e"])
print(my_series)
```

Now, to access a particular entry, we will do so with the new tags:

```python
print(my_series["c"])
```

We can also create Series from dictionaries. In this case, the keys will correspond to the series labels, and the dictionary values will correspond to the values taken by the Series entries.

```python
videos = {"day1": 5, "day2": 9, "day3": 7, "day4": 6, "day5": 8}
my_series = pd.Series(videos)
print(my_series)
```

## 14.6. DataFrame Combining Methods

| Method              | Description                                                                                   |
|---------------------|-----------------------------------------------------------------------------------------------|
| `join()`            | Combines DataFrames using columns as join keys.                                               |
| `merge()`           | Combines DataFrames using one or more columns as join keys.                                   |
| `concat()`          | Concatenates DataFrames along a specific axis (default is along rows).                        |
| `merge_ordered()`   | Merges DataFrames with ordered indices.                                                       |
| `merge_asof()`      | Merges DataFrames using an asof merge of the keys.                                            |
| `combine_first()`   | Combines DataFrames with priority given to one of the DataFrames.                             |
| `update()`          | Updates one DataFrame with values from another DataFrame, using non-NA values from the other. |

```python
# Sample DataFrames
df1 = pd.DataFrame({'A': [1, 2, 3],
                    'B': ['a', 'b', 'c']})

df2 = pd.DataFrame({'A': [4, 5, 6],
                    'B': ['d', 'e', 'f']})

# join()
result_join = df1.join(df2, lsuffix='_left', rsuffix='_right')
print("Join result:")
print(result_join)

# merge()
result_merge = pd.merge(df1, df2, on='A')
print("\nMerge result:")
print(result_merge)

# concat()
result_concat = pd.concat([df1, df2], axis=0)
print("\nConcatenation result:")
print(result_concat)

# merge_ordered()
df3 = pd.DataFrame({'A': [1, 3, 5],
                    'B': ['a', 'c', 'e']})
df4 = pd.DataFrame({'A': [1, 2, 4],
                    'C': ['x', 'y', 'z']})

result_merge_ordered = pd.merge_ordered(df3, df4, on='A')
print("\nMerge ordered result:")
print(result_merge_ordered)

# merge_asof()
df_left = pd.DataFrame({'time': [1, 2, 3, 4, 5],
                        'value': [10, 20, 30, 40, 50]})
df_right = pd.DataFrame({'time': [1, 2, 3, 6],
                         'value': [15, 25, 35, 45]})

result_merge_asof = pd.merge_asof(df_left, df_right, on='time')
print("\nMerge asof result:")
print(result_merge_asof)

# combine_first()
df5 = pd.DataFrame({'A': [1, 2, 3],
                    'B': [None, 'b', 'c']})
df6 = pd.DataFrame({'A': [1, 2, 3],
                    'B': ['x', 'y', 'z']})

result_combine_first = df5.combine_first(df6)
print("\nCombine first result:")
print(result_combine_first)

# update()
df7 = pd.DataFrame({'A': [1, 2, 3],
                    'B': [4, 5, 6]})
df8 = pd.DataFrame({'A': [1, 2],
                    'B': [7, 8]})

df7.update(df8)
print("\nUpdate result:")
print(df7)
```

### 14.7.6. Problem 6

Create a dataframe with 10 rows and 5 columns with the following characteristics:

- The first column will be `word` and will contain 10 words.
- The second column will be `length` and will contain the length of each word.
- The third column will be `start` and will contain the first letter of each word.
- The fourth column will be 'end' and will contain the last letter of each word.
- The fifth column will be `isPalindrome} and will contain boolean values indicating whether or not each word is a palindrome.

Next, transform the `word` column into the row names using the `.set_index()` method. Finally, calculate the dimensions of the dataframe.

```python
def is_palindrome(word):
     """
     Returns whether the word word is palindrome.
     Args:
         word: Word in string format
     Returns:
         isPalindrome: Boolean
     """
     word = word.lower()
     l = []
     isPalindrome = True
     for c in word:
         l.append(c)
     n = len(l)
     for i in range(int(n / 2)):
         if l[i] != l[n - (i + 1)]:
             isPalindrome = False
     return isPalindrome
```

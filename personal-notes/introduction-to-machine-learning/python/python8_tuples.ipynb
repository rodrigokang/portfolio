# 10. Data Structures: Tuples

The last structure we will see is tuples. These are a set of elements, which can be of different types, separated by commas and written within parentheses, `()`.

Tuples are:
- heterogeneous: elements can be of different types within the same tuple
- immutable: elements cannot be modified once the tuple has been created

An example of a tuple would be

```python
t = ("Bob", 32, "professor", True)
print(t)
```

We can declare a tuple without specifying its elements within parentheses.

```python
t = "Bob", 32, "professor", True
type(t)
```

We can declare tuples using the `tuple()` function.

```python
t = tuple(("Bob", 32, "professor", True))
type(t)
```

```python
tuple([1,2,3])
```

We can access the elements of a tuple by their index using square bracket syntax, `[]`.

```python
t = 1, "a", 2, "e", 3, "i", 4, "o", 5, "u"
print(t[0])
print(t[5])
```

Just like with lists, we can access tuple elements using negative indices.

```python
print(t[-1])
print(t[-4])
```

To access multiple entries of a tuple at once, we can use the colon `:` to indicate a range of indices.

```python
print(t[2:6])
print(t[:5])
print(t[5:])
```

```python
print(t[-5:-1])
```

To check if an element belongs to a tuple, we can use the `in` keyword.

```python
print(6 in t)
print("i" in t)
```

We have mentioned that tuples are immutable. That is, once a tuple is created, its elements cannot be modified.

```python
t = "Cherry", "Apple", "Pear"
t[1] = "Kiwi"
```

An alternative would be to convert to a list, make the modification, and then convert back to a tuple.

```python
t = "Cherry", "Apple", "Pear"
t = list(t)
t[1] = "Kiwi"
t = tuple(t)

print(t)
print(type(t))
```

## 10.1. Unpacking 

We can extract the values from a tuple into variables. This process is known as **unpacking**.

```python
fruits = "Cherry", "Apple", "Pear", "Orange"
print(type(fruits))

# Con paréntesis
(fruit1, fruit2, fruit3, fruit4) = fruits

print(fruit1)
print(fruit2)
print(fruit3)
print(fruit4)
```

It works the same if we don't declare the variables within parentheses.

```python
fruits = "Cherry", "Apple", "Pear", "Orange"
print(type(fruits))

# Sin paréntesis
fruit1, fruit2, fruit3, fruit4 = fruits

print(fruit1)
print(fruit2)
print(fruit3)
print(fruit4)
```

The number of variables must match the number of elements in the tuple. Otherwise, an asterisk should be used to store the remaining elements in a list.

```python
fruits = "Cherry", "Apple", "Pear", "Orange"

(fruit1, fruit2, *restFruits) = fruits

print(fruit1)
print(fruit2)
print(restFruits)
print(type(restFruits))
```

If the asterisk is added to any variable other than the last one, `Python` will store as many elements in the list as necessary to match the number of remaining elements with the number of remaining variables.

```python
fruits = "Cherry", "Kiwi", "Pear", "Orange", "Peach", "Watermelon", "Melon"

(fruit1, *restFruits, fruit2, fruit3) = fruits

print(fruit1)
print(restFruits)
print(fruit2)
print(fruit3)
```

```python
dot = (1, 2, 3)
x, _, z = dot
print(x + z)
```

```python
fruits = "Cherry", "Kiwi", "Pear", "Orange", "Peach", "Watermelon", "Melon"

(fruit1, *_, fruit2, fruit3) = fruits

print(fruit1)
print(fruit2)
print(fruit3)
```

## 10.2. Tuple Concatenation

We can concatenate tuples using the `+` operator, although the result will be a new tuple, as tuples, remember, cannot be modified.

```python
t1 = 1, 3
t2 = 2, 4

t1 + t2
```

## 10.3. Tuple repetition

We can repeat tuples a number $n$ of times using the `*` operator.

```python
t = ("a", "b", "c")
t * 5
```

## 10.4. Tuple size

We can calculate the number of elements in a tuple using the `len()` function.

```python
t = "John", 32, "professor", True
len(t)
```

If we wanted to create a tuple with a single element, we would have to do the following:

```python
t1 = ("apple", )
print(type(t1))

# Lo siguiente no es una tupla
t2 = ("apple")
print(type(t2))
```

## 10.5. Loops and tuples

We can iterate over a tuple using a `for` loop.

```python
fruits = "Cherry", "Kiwi", "Pear", "Orange", "Peach", "Watermelon", "Melon"

for fruit in fruits:
  print(fruit)
```

We can also use the unpacking technique in loops.

```python
t = ("cherry", "red"), ("kiwi", "yellow"), ("pear", "green"), ("orange", "orange")

for fruit, color in t:
  if fruit == "kiwi":
    print("El color del", fruit, "es", color)
  else:
    print("La {} es {}".format(fruit, color))
```

## 10.6. Tuples and the Rest of Data Structures

A tuple can contain lists, dictionaries, sets, and tuples.

```python
t = [4, 5, 6], {"vowels": ("a", "e", "i", "o", "u")}, {1, 2, 3}, ("x", "y")
type(t)
```

Similarly,

- Lists can contain dictionaries, sets, tuples, and other lists.
- Dictionaries can contain lists, sets, tuples, and other dictionaries.
- Sets cannot contain lists, dictionaries, tuples, or even other sets.

```python
l = [{"vowels": ("a", "e", "i", "o", "u")}, {1, 2, 3}, ("x", "y"), [4, 5, 6]]
type(l)
```

```python
dicc = {"list": [4, 5, 6], "set": {1, 2, 3}, "tuple": ("x", "y"), "dict": {"vowels": ("a", "e", "i", "o", "u")}}
type(dicc)
```

```python
set1 = {[4, 5, 6], {"vowels": ("a", "e", "i", "o", "u")}, ("x", "y"), {1, 2, 3}}
```

Given any iterable object in Python, we can convert it to a tuple using the `tuple()` function.

```python
print(tuple(l)) # From a list
print(type(tuple(l)))

print(tuple(dicc)) # From a dictionary only the keys are stored in the tuple
print(type(tuple(dicc)))

print(tuple({1, 2, 3, 4, 5})) # From a set
print(type(tuple({1, 2, 3, 4, 5})))
```

## 10.7. `zip()` function

The `zip()` function is used to combine lists into tuples.

```python
objects = ["notebook", "pen", "propelling pencil", "mine_pack"]
price = [5.00, 3.30, 1.29, 0.50]
items = zip(objects, price)
print(items)
```

We can convert the result of a `zip()` function to a list.

```python
items = zip(objects, price)
list(items)
```

We can convert the result of a `zip()` function to a dictionary.

```python
items = zip(objects, price)
dict(items)
```

We need to recreate the `zip()` object because the result of this function is an iterator, and once it has been converted to a list, dictionary, or tuple, it is considered a complete iteration and will not be able to generate more values.

We can convert the result of a `zip()` function to a tuple:

```python
items = zip(objects, price)
tuple(items)
```

```python
for obj, pr in zip(objects, price):
    print("The object {} costs {} €.".format(obj, pr))
```

### 10.8.1. Problem 1

Ask the user for the number of integers they will input. For each of these integers, create a tuple where the first entry is the integer and the second entry is the word "positive," "negative," or "zero" depending on the sign of the integer. Store all tuples in a list and display them.

```python
n = int(input("How many integers will you input? "))
nums = []

for _ in range(n):
    sign = ""
    num = int(input())
    if num > 0:
        sign = "positive"
    elif num == 0:
        sign = "zero"
    else:
        sign = "negative"
    nums.append((num, sign))

print(nums)
```

### 10.8.2. Problem 2

Ask the user for integers from 1 to 10 until they input 0. For each of these integers, create a tuple where the first entry is the integer and the second entry is the word "fail," "pass," "notable," or "excellent" according to the interval to which the integer belongs. Display the newly created tuple to the user.

The different categories are:

- fail if the grade belongs to [1, 5)
    - pass if the grade belongs to [5, 7)
    - notable if the grade belongs to [7, 9)
    - excellent if the grade belongs to [9, 10]

```python
num = int(input("Enter an integer from 1 to 10 "))

while num != 0:
    grade = ""
    if num < 1:
        print("You must enter an integer between 1 and 10")
    elif num < 5:
        grade = "Fail"
        print((num, grade))
    elif num < 7:
        grade = "Pass"
        print((num, grade))
    elif num < 9:
        grade = "Notable"
        print((num, grade))
    elif num <= 10:
        grade = "Excellent"
        print((num, grade))
    else:
        print("You must enter an integer between 1 and 10")

    num = int(input("Enter an integer from 1 to 10 "))
```

### 10.8.3. Problem 3

Given a phrase entered by the user, create a list with 3 tuples of 2 entries. The first tuple will contain the number of vowels; the second, the number of consonants; and the last, the number of spaces. For each tuple, the first entry will be an explanatory string and the second will be the corresponding value.

```python
s = input()
s = s.lower()
vowels = ["a", "e", "i", "o", "u"]

vowels_count = 0
consonants_count = 0
blanks_count = 0

for c in s:
    if c in vowels:
        vowels_count += 1
    elif c.isalpha():
        consonants_count += 1
    elif c == " ":
        blanks_count += 1

info = [("Vowels", vowels_count), ("Consonants", consonants_count), ("Spaces", blanks_count)]
print(info)
```

### 10.8.4. Problem 4

Ask the user for a subject ("Math", "Language", "History", "Computer Science", or "Music") and the grade in that subject until they input a different subject from the ones indicated. The user can repeat a subject as many times as they want. The grade must be from 1 to 10. We will store the information (subject, grade) in a tuple. The tuples will be stored in a list. Finally, for each subject, we will display the average grade.

```python
subjects = ["Math", "Language", "History", "Computer Science", "Music"]
grades = []

print(subjects)
s = input("Enter the subject: ")
while s in subjects:
    grade = int(input("Enter the grade between 1 and 10: "))
    grades.append((s, grade))
    s = input("Enter the subject: ")

means = {"Math": [],
         "Language": [],
         "History": [],
         "Computer Science": [],
         "Music": []}

for item in grades:
    means[item[0]].append(item[1])

print("\n=== AVERAGE GRADES ===")
for key, val in means.items():
    print("The average grade for {} is {}".format(key, "unknown" if len(val) == 0 else sum(val) / len(val)))
```

### 10.8.5. Problem 5

Ask the user for the number of points on a plane they want to input. For each point, we will ask for the x and y coordinates. Save the coordinates (x, y) in tuples of size 3, where the last entry corresponds to the quadrant to which that point belongs. All tuples of size 3 will be stored in a list. Finally, we will display all the tuples of size 3 created, with the format `The point (\{x\}, \{y\}) belongs to quadrant \{quadrant\}`.

```python
n = int(input("How many points will you input? "))
points = []

for i in range(n):
    x = float(input("Enter the x coordinate = "))
    y = float(input("Enter the y coordinate = "))

    if x >= 0 and y >= 0:
        quadrant = "I"
    if x <= 0 and y >= 0:
        quadrant = "II"
    if x <= 0 and y <= 0:
        quadrant = "III"
    if x >= 0 and y <= 0:
        quadrant = "IV"
    if x == 0 and y == 0:
        quadrant = "center"

    points.append((x, y, quadrant))

for point in points:
    x, y, quadrant = point
    print("The point ({}, {}) belongs to quadrant {}".format(x, y, quadrant))
```

### 10.8.6. Problem 6

Given a list of words, create another list of the same size that stores the first letter of each word in the corresponding position. Finally, using the `zip()` function, create a tuple of tuples with the word in the first entry and the starting letter in the second.

```python
words = ["wave", "snail", "pineapple", "beach", "barbecue", "window", "fly"]
first_letters = []

for w in words:
    first_letters.append(w[0])

print(tuple(zip(words, first_letters)))
```

## Problem 1 — Creating and Inspecting Arrays

### Problem

Create `values = [2, 4, 6, 8, 10]`, convert it into an `ndarray`, and inspect its type, `ndim`, `shape`, `size`, and `dtype`. Explain the difference between `ndim`, `shape`, and `size`.

### Solution

The array is one-dimensional. `ndim` counts axes, `shape` gives the length of each axis, and `size` counts all stored elements. For this array they are `1`, `(5,)`, and `5`, respectively.

#### Verification

```python
import numpy as np
values = np.array([2, 4, 6, 8, 10])
print(type(values))
print(values.ndim)
print(values.shape)
print(values.size)
print(values.dtype)
```

### Key Takeaway

`ndim`, `shape`, and `size` describe different aspects of array structure: number of axes, axis lengths, and total element count.

## Problem 2 — One-Dimensional and Two-Dimensional Arrays

### Problem

Inspect `a = np.array([1, 2, 3, 4, 5, 6])` and `A = np.array([[1, 2, 3], [4, 5, 6]])`. Determine `ndim`, `shape`, and `size`, and explain why both contain six elements but represent different structures.

### Solution

`a` has shape `(6,)` and one axis. `A` has shape `(2, 3)` and two axes. Both have `size == 6`, but shape determines how the same number of values is organized and therefore how indexing and algebraic operations behave.

#### Verification

```python
a = np.array([1, 2, 3, 4, 5, 6])
A = np.array([[1, 2, 3], [4, 5, 6]])
for name, arr in [('a', a), ('A', A)]:
    print(name, 'ndim=', arr.ndim, 'shape=', arr.shape, 'size=', arr.size)
```

### Key Takeaway

Equal size does not imply equal structure; shape is part of an array's meaning.

## Problem 3 — Array Creation Functions

### Problem

Create zeros, ones, a constant-filled array, integer ranges, evenly spaced values, and a `4 × 4` identity matrix using appropriate NumPy creation functions. Explain when `linspace()` is preferable to `arange()` for floating-point intervals.

### Solution

Use `zeros`, `ones`, `full`, `arange`, `linspace`, and `eye`. `arange()` is naturally step-based, whereas `linspace()` is point-count based and is usually clearer when an exact number of floating-point samples over an interval is required.

#### Verification

```python
a = np.zeros(5)
b = np.ones((3, 4))
c = np.full((2, 3), 7)
d = np.arange(10)
e = np.arange(2, 21, 2)
f = np.linspace(0, 1, 6)
g = np.eye(4)
for arr in [a, b, c, d, e, f, g]:
    print(arr, '\n')
```

### Key Takeaway

Use `zeros`, `ones`, `full`, `arange`, `linspace`, and `eye`. `arange()` is naturally step-based, whereas `linspace()` is point-count based and is usually clearer when an exact number of floating-point samples over an interval is required.

## Problem 7 — Boolean Masks

### Problem

Given `values = np.array([3,8,12,5,17,21,4,15])`, create masks for values greater than 10, between 5 and 15 inclusive, even values, and values below 5 or above 15.

### Solution

Array comparisons create element-wise boolean masks. Combine conditions using `&` and `|`, with parentheses around each comparison.

#### Verification

```python
values = np.array([3,8,12,5,17,21,4,15])
masks = [
    values > 10,
    (values >= 5) & (values <= 15),
    values % 2 == 0,
    (values < 5) | (values > 15),
]
for m in masks:
    print(m, '->', values[m])
```

### Key Takeaway

Array comparisons create element-wise boolean masks. Combine conditions using `&` and `|`, with parentheses around each comparison.

## Problem 10 — Inferring Dimensions

### Problem

For `a = np.arange(30)`, predict the shapes from `reshape(5,-1)`, `reshape(-1,3)`, and `reshape(2,-1,5)`. Explain `-1`.

### Solution

The shapes are `(5,6)`, `(10,3)`, and `(2,3,5)`. NumPy infers the missing dimension from total size divided by the product of the specified dimensions. Only one unknown is allowed because otherwise the shape would be underdetermined.

#### Verification

```python
a = np.arange(30)
print(a.reshape(5,-1).shape)
print(a.reshape(-1,3).shape)
print(a.reshape(2,-1,5).shape)
```

### Key Takeaway

The shapes are `(5,6)`, `(10,3)`, and `(2,3,5)`. NumPy infers the missing dimension from total size divided by the product of the specified dimensions. Only one unknown is allowed because otherwise the shape would be underdetermined.

### Key Takeaway

## Problem 12 — Assignment, View, or Copy?

### Problem

Given `b = a`, `c = a[1:4]`, and `d = a[1:4].copy()`, predict the effects of modifying each object.

### Solution

`b` is another name for the same array object; `c` is a distinct array object sharing underlying data; `d` owns independent copied data. Changes through `b` and `c` affect `a`, while changes through `d` do not.

#### Verification

```python
a = np.array([10,20,30,40,50])
b = a
c = a[1:4]
d = a[1:4].copy()
b[0] = 100
c[0] = 200
d[0] = 300
print('a', a)
print('b', b)
print('c', c)
print('d', d)
```

### Key Takeaway

`b` is another name for the same array object; `c` is a distinct array object sharing underlying data; `d` owns independent copied data. Changes through `b` and `c` affect `a`, while changes through `d` do not.

## Problem 13 — Why Views Matter

### Problem

Explain why slicing a multi-gigabyte array may return a view rather than an automatic copy, and discuss the trade-off.

### Solution

Views avoid allocating and copying large blocks of memory, reducing both memory use and data movement. The cost is aliasing: modifying the view may change the source. Copies provide independence but require additional memory and time.

### Key Takeaway

Views are a performance feature with semantic consequences: efficiency is gained by sharing storage.

## Problem 25 — Population and Sample Standard Deviation

### Solution

## Problem 30 — Reproducible Random Numbers

### Problem

Create two generators with seed 42, generate five values from each, and compare with a generator using a different seed.

## Problem 41 — Designing an Array-Based Preprocessing Function

### Problem

Write `standardize_features(X: np.ndarray) -> np.ndarray` that standardizes features, avoids mutation and explicit loops, and raises an error for zero-variance features.

### Solution

The feature statistics are computed along axis 0. Because arithmetic creates a new result array, the input is not mutated. A zero standard deviation must be rejected because division would be undefined.

#### Verification

```python
def standardize_features(X: np.ndarray) -> np.ndarray:
    """Return a feature-wise standardized copy of a two-dimensional array."""
    means = X.mean(axis=0)
    stds = X.std(axis=0)
    if np.any(stds == 0):
        raise ValueError("Cannot standardize a feature with zero standard deviation.")
    return (X - means) / stds

X=np.array([[1.,10.],[2.,20.],[3.,30.]])
Z=standardize_features(X)
print(Z)
print('original:', X)
print(Z.mean(axis=0), Z.std(axis=0))
```

### Key Takeaway

Array interfaces are often understood as much through expected shapes as through Python types.

## Problem 42 — Reasoning About an Unfamiliar Numerical Pipeline

### Problem

Analyse a function that filters rows with positive first feature, computes per-feature means with `keepdims=True`, centers values, computes root-mean-square scale per feature, and divides by that scale. Explain shapes, broadcasting, mutation, and the zero-scale risk.

### Solution

`X[:,0] > 0` creates one boolean per observation. Boolean indexing selects rows and, unlike basic slicing, produces a new array. `means` has `(1,n_features)` because axis 0 is reduced but retained. Subtraction broadcasts that row over all selected observations. Squaring is element-wise; `mean(..., axis=0)` reduces observations and yields `(n_features,)`; the square root gives RMS scale. Final division broadcasts scale across rows. The original `X` is not modified. A zero scale causes division by zero. Mathematically, the function filters observations, centers each feature, and scales each centered feature by its root-mean-square magnitude.

#### Verification

```python
def transform(X):
    selected = X[X[:, 0] > 0]
    means = selected.mean(axis=0, keepdims=True)
    centered = selected - means
    scale = np.sqrt(np.mean(centered ** 2, axis=0))
    return centered / scale

X=np.array([[-1.,10.],[1.,20.],[2.,30.],[3.,50.]])
result=transform(X)
print(result)
print(result.shape)
```

### Key Takeaway

Complex NumPy pipelines become tractable when each step is reconstructed from selection semantics, shape changes, axis reduction, and broadcasting.

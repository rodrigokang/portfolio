## Problem 1 — Truth Values

### Problem

Without running the code, determine which messages are printed:

```python
values = []

if values:
    print("A")
else:
    print("B")

if 0:
    print("C")

if "0":
    print("D")
```

Explain each result in terms of Python's truth-value rules.

### Solution

An empty list is false in a Boolean context, so the first `if` condition fails and the corresponding `else` block prints `"B"`.

The integer `0` is also false, so `"C"` is not printed.

The string `"0"` is **not** an empty string. Its contents happen to represent the character zero, but the object itself contains one character and is therefore true in a Boolean context. Consequently, `"D"` is printed.

The exact output is therefore:

```text
B
D
```

This illustrates an important distinction: Python evaluates the truth value of the **object**, not the apparent meaning of its textual contents.

#### Verification

````python
values = []

if values:
    print("A")
else:
    print("B")

if 0:
    print("C")

if "0":
    print("D")
````

### Key Takeaway

Empty containers and numeric zero values are false, while non-empty strings are true even when their contents look like a false or zero value.

## Problem 2 — Independent Conditions

### Problem

Predict the output:

```python
x = 12

if x > 0:
    print("positive")

if x % 2 == 0:
    print("even")

if x > 20:
    print("large")
```

Then explain how the behaviour would change if the second and third `if` statements were replaced by `elif`.

### Solution

The three `if` statements are independent. Python evaluates each condition regardless of whether a previous condition succeeded.

For `x = 12`:

- `x > 0` is true, so `"positive"` is printed;
- `x % 2 == 0` is true, so `"even"` is printed;
- `x > 20` is false, so `"large"` is not printed.

The output is therefore:

```text
positive
even
```

If the second and third statements were changed to `elif`, the structure would become one mutually exclusive conditional chain. Since `x > 0` is already true, Python would execute the first branch and skip all later `elif` branches. Only `"positive"` would be printed.

#### Verification

````python
x = 12

if x > 0:
    print("positive")

if x % 2 == 0:
    print("even")

if x > 20:
    print("large")

print("--- elif version ---")

if x > 0:
    print("positive")
elif x % 2 == 0:
    print("even")
elif x > 20:
    print("large")
````

### Key Takeaway

Consecutive `if` statements represent independent tests; an `if`-`elif` chain represents mutually exclusive alternatives.

## Problem 3 — Ordering Conditions

### Problem

Consider:

```python
score = 0.72

if score >= 0.5:
    category = "acceptable"
elif score >= 0.7:
    category = "good"
elif score >= 0.9:
    category = "excellent"
else:
    category = "poor"
```

What value is assigned to `category`?

Explain why the program does not implement the likely intended classification and rewrite the ordering of the conditions appropriately.

### Solution

`category` becomes `"acceptable"`.

The first condition, `score >= 0.5`, is true for `0.72`. Once an `if` or `elif` condition succeeds, Python executes that branch and skips the remainder of the chain. The more restrictive condition `score >= 0.7` is therefore never reached.

For threshold categories of this form, the conditions should normally be ordered from the most restrictive threshold to the least restrictive:

```python
if score >= 0.9:
    category = "excellent"
elif score >= 0.7:
    category = "good"
elif score >= 0.5:
    category = "acceptable"
else:
    category = "poor"
```

For `score = 0.72`, the corrected structure assigns `"good"`.

#### Verification

````python
score = 0.72

if score >= 0.9:
    category = "excellent"
elif score >= 0.7:
    category = "good"
elif score >= 0.5:
    category = "acceptable"
else:
    category = "poor"

category
````

### Key Takeaway

In an `if`-`elif` chain, the first true branch wins. Overlapping threshold conditions must therefore be ordered deliberately.

## Problem 5 — Point inside a Rectangle

### Problem

Given a point $(x,y)$, determine whether it belongs to the closed rectangle

$$
[a,b]\times[c,d].
$$

Write the condition using chained comparisons.

Then determine how the condition should change if the boundary of the rectangle is excluded.

### Solution

For the closed rectangle, both coordinates may lie on the boundary. The condition is

```python
a <= x <= b and c <= y <= d
```

which corresponds to

$$
a \leq x \leq b
\quad\text{and}\quad
c \leq y \leq d.
$$

If the boundary is excluded, strict inequalities are required:

```python
a < x < b and c < y < d
```

The second condition describes the interior of the rectangle rather than the closed rectangle.

#### Verification

````python
a, b = 0, 4
c, d = -1, 3
x, y = 4, 2

inside_closed = a <= x <= b and c <= y <= d
inside_open = a < x < b and c < y < d

inside_closed, inside_open
````

### Key Takeaway

Chained comparisons provide a direct translation of interval membership from mathematics into Python.

## Problem 6 — Leap Year

### Problem

A year is a leap year if it is divisible by 4, except for years divisible by 100, unless they are also divisible by 400.

Write a conditional structure that determines whether a given year is a leap year.

Then attempt to express the same rule using a single Boolean expression.

### Solution

A direct conditional implementation is:

```python
if year % 400 == 0:
    is_leap = True
elif year % 100 == 0:
    is_leap = False
elif year % 4 == 0:
    is_leap = True
else:
    is_leap = False
```

The order is important because divisibility by 400 is an exception to the century rule.

The same logic can be expressed as one Boolean expression:

```python
is_leap = year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)
```

That expression states that a year is leap if it is divisible by 400, or if it is divisible by 4 but not by 100.

#### Verification

````python
years = [1900, 2000, 2024, 2025]

for year in years:
    is_leap = year % 400 == 0 or (year % 4 == 0 and year % 100 != 0)
    print(year, is_leap)
````

### Key Takeaway

A compact Boolean expression is useful when it remains readable, but the underlying hierarchy of exceptions must still be understood.

## Problem 7 — Predicting a `for` Loop

### Problem

Without executing the code, determine the final values of `total` and `x`:

```python
total = 0

for x in [2, 5, 3]:
    total += x
```

Explain the state of both variables after every iteration.

### Solution

The loop processes the values `2`, `5`, and `3` in that order.

The state evolves as follows:

| Iteration | `x` | `total` after update |
|:---:|:---:|:---:|
| initial | — | `0` |
| 1 | `2` | `2` |
| 2 | `5` | `7` |
| 3 | `3` | `10` |

After the loop:

```python
total == 10
x == 3
```

The iteration variable remains bound after the loop and retains the value assigned in the final iteration.

#### Verification

````python
total = 0

for x in [2, 5, 3]:
    total += x
    print("x =", x, "total =", total)

print("final:", total, x)
````

### Key Takeaway

A loop can be reasoned about as a sequence of state transitions; the iteration variable normally remains bound to its final value afterwards.

## Problem 8 — `range()`

### Problem

Determine the values generated by each expression without executing them:

```python
range(5)
range(2, 7)
range(1, 10, 3)
range(10, 1, -2)
range(5, 5)
```

For each case, state how many iterations a corresponding `for` loop would perform.

### Solution

The generated values are:

```text
range(5)         -> 0, 1, 2, 3, 4          -> 5 iterations
range(2, 7)      -> 2, 3, 4, 5, 6          -> 5 iterations
range(1, 10, 3)  -> 1, 4, 7                -> 3 iterations
range(10, 1, -2) -> 10, 8, 6, 4, 2         -> 5 iterations

```

#### Verification

````python
ranges = [
    range(5),
    range(2, 7),
    range(1, 10, 3),
    range(10, 1, -2),
    range(5, 5),
]

for r in ranges:
    print(list(r), "iterations:", len(r))
````

### Key Takeaway

`range()` represents a half-open arithmetic sequence: `start` is included, `stop` is excluded, and `step` controls direction and spacing.

### Solution

```python
for i in range(1, 11):
```

$$

$$

````python
for i in range(1, 11):
    print(i)
````

### Key Takeaway

When the final integer must be included, the `stop` argument must lie one step beyond it.

## Problem 11 — Arithmetic Progression

### Problem

An arithmetic progression is defined by

$$
a_k = a_0 + kd.
$$

Given `a0`, `d`, and an upper bound `b`, use `range()` and a `for` loop to calculate the sum of all progression terms not exceeding `b`.

Assume that all values are integers and that `d > 0`.

### Solution

Because `range()` excludes its stopping boundary, `b + 1` can be used to include `b` when it is itself a term of the progression:

```python
total = 0

for term in range(a0, b + 1, d):
    total += term
```

For example, with

```python
a0 = 1
d = 4
b = 24
```

the terms are

$$
1,5,9,13,17,21,
$$

and their sum is

$$
66.
$$

The upper bound need not itself be a progression term; `range()` simply stops before producing a value greater than it.

#### Verification

````python
a0 = 1
d = 4
b = 24

total = 0

for term in range(a0, b + 1, d):
    total += term

total
````

### Key Takeaway

`range()` naturally represents integer arithmetic progressions when the common difference is an integer.

## Problem 14 — Filtering with `continue`

### Problem

Given

```python
values = [4, -2, 7, -1, 0, 9]
```

use a `for` loop and `continue` to print only the non-negative values.

Then rewrite the solution without `continue`.

Compare the control flow of the two versions.

### Solution

Using `continue`, negative values are handled as cases that should be skipped:

```python
for value in values:
    if value < 0:
        continue

    print(value)
```

Without `continue`, the desired cases can be selected positively:

```python
for value in values:
    if value >= 0:
        print(value)
```

Both versions print

```text
4
7
0
9
```

The first form exits the current iteration early for invalid values. The second executes the processing block only when the positive condition is satisfied. For this short example, the second is arguably simpler; `continue` becomes more useful when substantial processing follows the validation step.

#### Verification

````python
values = [4, -2, 7, -1, 0, 9]

print("with continue")
for value in values:
    if value < 0:
        continue
    print(value)

print("without continue")
for value in values:
    if value >= 0:
        print(value)
````

### Key Takeaway

`continue` is most valuable when early rejection keeps the normal processing path simple; it should not be used merely because it is available.

## Problem 15 — First Matching Value

### Problem

Given two integers `a` and `b`, find the first integer in the closed interval

$$
[a,b]
$$

that is divisible by 2, 3, and 5.

Terminate the search immediately after finding the first valid number.

If no such number exists, report that no match was found.

### Solution

A number divisible by 2, 3, and 5 is divisible by their least common multiple, 30. The search can still be expressed directly using the three divisibility tests:

```python
found = False

for n in range(a, b + 1):
    if n % 2 == 0 and n % 3 == 0 and n % 5 == 0:
        print(n)
        found = True
        break

if not found:
    print("No match found")
```

`break` is important here because the requirement is to find the **first** matching value. Once it is found, further iterations are unnecessary.

#### Verification

````python
a, b = 7, 37
found = False

for n in range(a, b + 1):
    if n % 2 == 0 and n % 3 == 0 and n % 5 == 0:
        print("First match:", n)
        found = True
        break

if not found:
    print("No match found")
````

### Key Takeaway

A search loop should stop as soon as the required result has been obtained when later iterations cannot improve the answer.

## Problem 16 — `for`-`else`

### Problem

Rewrite Problem 15 using a `for`-`else` structure so that no additional Boolean variable is required to determine whether a valid number was found.

Explain precisely when the `else` clause executes.

### Solution

The search can be written as

```python
for n in range(a, b + 1):
    if n % 2 == 0 and n % 3 == 0 and n % 5 == 0:
        print(n)
        break
else:
    print("No match found")
```

The `else` clause executes only if the `for` loop completes **without executing `break`**.

If a valid number is found, `break` terminates the loop and the `else` block is skipped. If the iterable is exhausted with no match, the `else` block runs.

#### Verification

````python
def first_match(a, b):
    for n in range(a, b + 1):
        if n % 2 == 0 and n % 3 == 0 and n % 5 == 0:
            print("First match:", n)
            break
    else:
        print("No match found")

first_match(7, 37)
first_match(1, 20)
````

### Key Takeaway

Loop `else` is best understood as the branch for normal completion without `break`, which makes it particularly useful for searches.

## Problem 17 — Infinite Loop Diagnosis

### Problem

Consider:

```python
x = 1

while x < 100:
    x *= 0.5
```

Does this loop terminate?

Explain the evolution of `x` and identify why the condition never becomes false.

Modify the loop so that it terminates naturally.

### Solution

The loop does not terminate under ordinary mathematical reasoning because `x` begins below 100 and is repeatedly halved:

$$
1,\quad
0.5,\quad
0.25,\quad
0.125,\ldots
$$

The values move toward zero, so the condition `x < 100` remains true.

With finite floating-point arithmetic, repeated halving eventually underflows to `0.0`; however, `0.0 < 100` is still true, so the loop remains infinite.

A naturally terminating version must update `x` in a direction that eventually makes the condition false, for example:

```python
x = 1

while x < 100:
    x *= 2
```

The sequence then grows until `x` reaches or exceeds 100.

#### Verification

````python
x = 1
history = []

while x < 100:
    history.append(x)
    x *= 2

history, x
````

### Key Takeaway

For a terminating `while` loop, the state changes must move execution toward making the continuation condition false.

## Problem 18 — Iteration until Convergence

### Problem

Consider the recurrence

$$
x_{k+1}
=
\frac{1}{2}
\left(
x_k+\frac{a}{x_k}
\right),
$$

with $a>0$ and $x_0>0$.

This is an iterative method for approximating $\sqrt{a}$.

Starting from

```python
a = 2
x = 1.0
```

repeatedly compute the next approximation until

$$
|x_{k+1}-x_k| < 10^{-10}.
$$

Use a `while` loop and count the number of iterations required.

Add a maximum number of iterations as a safety condition.

### Solution

A convenient implementation computes the proposed next state, measures the change, and then decides whether the process has converged:

```python
a = 2
x = 1.0
tolerance = 1e-10
max_iterations = 100
iteration = 0

while iteration < max_iterations:
    x_next = 0.5 * (x + a / x)
    iteration += 1

    if abs(x_next - x) < tolerance:
        x = x_next
        break

    x = x_next
```

For these values, the method converges rapidly to $\sqrt{2}$.

The maximum iteration count is a safety condition: if numerical or logical problems prevent convergence, execution still terminates.

#### Verification

````python
a = 2
x = 1.0
tolerance = 1e-10
max_iterations = 100
iteration = 0

while iteration < max_iterations:
    x_next = 0.5 * (x + a / x)
    iteration += 1

    if abs(x_next - x) < tolerance:
        x = x_next
        break

    x = x_next

print("iterations:", iteration)
print("approximation:", x)
print("squared error:", abs(x * x - a))
````

### Key Takeaway

Iterative numerical algorithms need both a meaningful convergence criterion and, in practice, a defensive termination limit.

## Problem 23 — Matrix Traversal

### Problem

Given

```python
matrix = [
    [2, 4, 6],
    [1, 3, 5],
    [7, 8, 9],
]
```

use nested loops to calculate the sum of all matrix elements.

Then calculate the sum of each row separately.

Do not use `sum()`.

### Solution

For the total, one accumulator can be updated for every element:

```python
total = 0

for row in matrix:
    for value in row:
        total += value
```

For the row sums, a new accumulator must be initialised for every row:

```python
row_sums = []

for row in matrix:
    row_total = 0

    for value in row:
        row_total += value

    row_sums.append(row_total)
```

The total is `45`, and the row sums are `[12, 9, 24]`.

#### Verification

````python
matrix = [
    [2, 4, 6],
    [1, 3, 5],
    [7, 8, 9],
]

total = 0
row_sums = []

for row in matrix:
    row_total = 0

    for value in row:
        total += value
        row_total += value

    row_sums.append(row_total)

total, row_sums
````

### Key Takeaway

Nested iteration naturally represents traversal of nested data structures; accumulator placement determines whether state is global or reset for each group.

## Problem 24 — Pairwise Comparisons

### Problem

Given

```python
values = [3, 8, 2, 7]
```

use nested loops to print every ordered pair $(x_i,x_j)$.

How many pairs are produced?

Then modify the loops so that each unordered pair of distinct elements is considered exactly once.

For a collection containing $n$ elements, how many such pairs should be produced?

### Solution

For every ordered pair, each of the $n$ elements can be paired with every one of the $n$ elements:

```python
for x in values:
    for y in values:
        print(x, y)
```

For $n=4$, this produces

$$
n^2 = 16
$$

ordered pairs, including self-pairs.

To generate each unordered pair of distinct positions exactly once, the second index can begin immediately after the first:

```python
for i in range(len(values)):
    for j in range(i + 1, len(values)):
        print(values[i], values[j])
```

The number of such pairs is

$$
\binom{n}{2}
=
\frac{n(n-1)}{2}.
$$

For $n=4$, this gives `6`.

#### Verification

````python
values = [3, 8, 2, 7]

ordered = []
for x in values:
    for y in values:
        ordered.append((x, y))

unordered = []
for i in range(len(values)):
    for j in range(i + 1, len(values)):
        unordered.append((values[i], values[j]))

print("ordered:", len(ordered))
print("unordered:", len(unordered))
print(unordered)
````

### Key Takeaway

Pair-generation logic depends on whether order matters and whether self-pairs or duplicate pairings are allowed.

## Problem 26 — `zip()`

### Problem

Given

```python
observed = [1, 0, 1, 1, 0, 1]
predicted = [1, 1, 1, 0, 0, 1]
```

use `zip()` to count

- correct predictions;
- incorrect predictions.

Then calculate the accuracy

$$
\text{accuracy}
=
\frac{\text{correct predictions}}
     {\text{total predictions}}.
$$

Do not use a machine-learning library.

### Solution

`zip()` allows corresponding observed and predicted labels to be processed as pairs:

```python
correct = 0
incorrect = 0

for y_true, y_pred in zip(observed, predicted):
    if y_true == y_pred:
        correct += 1
    else:
        incorrect += 1
```

The pairs are

```text
(1,1) correct
(0,1) incorrect
(1,1) correct
(1,0) incorrect
(0,0) correct
(1,1) correct
```

Thus:

```text
correct   = 4
incorrect = 2
```

and

$$
\text{accuracy}
=
\frac{4}{6}
=
\frac{2}{3}
\approx 0.6667.
$$

#### Verification

````python
observed = [1, 0, 1, 1, 0, 1]
predicted = [1, 1, 1, 0, 0, 1]

correct = 0
incorrect = 0

for y_true, y_pred in zip(observed, predicted):
    if y_true == y_pred:
        correct += 1
    else:
        incorrect += 1

accuracy = correct / len(observed)

correct, incorrect, accuracy
````

### Key Takeaway

`zip()` directly expresses parallel iteration over corresponding observations and predictions.

## Problem 27 — Different Lengths with `zip()`

### Problem

Predict what happens:

```python
x = [1, 2, 3, 4]
y = [10, 20]

for a, b in zip(x, y):
    print(a, b)
```

How many iterations occur?

Why can this behaviour be dangerous when the two collections are expected to represent corresponding observations?

Investigate how `strict=True` changes the behaviour.

### Solution

By default, `zip()` stops when the shortest iterable is exhausted. The loop therefore performs only two iterations and prints

```text
1 10
2 20
```

The remaining values `3` and `4` from `x` are silently ignored.

This can be dangerous when equal lengths are an invariant of the data, because silent truncation can hide a preceding alignment or preprocessing error.

With

```python
zip(x, y, strict=True)
```

Python raises a `ValueError` when it detects that the iterables have different lengths. This is preferable when unequal lengths should be treated as invalid.

#### Verification

````python
x = [1, 2, 3, 4]
y = [10, 20]

print("default zip:")
for a, b in zip(x, y):
    print(a, b)

print("strict zip:")
try:
    for a, b in zip(x, y, strict=True):
        print(a, b)
except ValueError as exc:
    print(type(exc).__name__ + ":", exc)
````

### Key Takeaway

Default `zip()` truncates to the shortest iterable; use `strict=True` when mismatched lengths should be considered an error.

## Problem 29 — Threshold Classification

### Problem

Given a sequence of probabilities and a threshold $t$, convert each probability into a binary prediction according to

$$
\hat y =
\begin{cases}
1, & p \geq t,\\
0, & p < t.

For

```python
probabilities = [0.12, 0.51, 0.49, 0.87, 0.73]
```

construct a list called `predictions` using an explicit `for` loop.

### Solution

Initialise an empty output list and append one prediction for each probability:

```python
predictions = []
```

```python
[0, 1, 0, 1, 1]
```

Threshold classification is a direct example of combining iteration with a two-way conditional decision for every observation.

## Problem 30 — Early Stopping

### Problem

Suppose the validation losses from successive training epochs are

```python
losses = [0.52, 0.44, 0.39, 0.36, 0.355, 0.354, 0.3545, 0.355]
```

Traverse the values in order.

Stop when the current loss is greater than the previous loss and report the epoch at which this first occurs.

Use `enumerate()` and keep track of the previous value explicitly.

Explain why this is a simplified form of an **early-stopping** rule rather than a realistic complete implementation.

### Solution

Assuming human-readable epochs start at 1, one implementation is:

```python
previous_loss = losses[0]

for epoch, current_loss in enumerate(losses[1:], start=2):
    if current_loss > previous_loss:
        print(epoch)
        break

    previous_loss = current_loss
```

The sequence decreases through epoch 6:

```text
0.52
0.44
0.39
0.36
0.355
0.354
```

At epoch 7, the loss becomes `0.3545`, which is greater than `0.354`. The first increase therefore occurs at **epoch 7**.

This is only a simplified early-stopping rule. Real training procedures often use a patience parameter, a minimum meaningful improvement, checkpointing of the best model, and potentially noisy validation metrics. Stopping after a single small increase can be too sensitive to random variation.

#### Verification

````python
losses = [0.52, 0.44, 0.39, 0.36, 0.355, 0.354, 0.3545, 0.355]

previous_loss = losses[0]

for epoch, current_loss in enumerate(losses[1:], start=2):
    if current_loss > previous_loss:
        print("First increase at epoch:", epoch)
        print("previous:", previous_loss)
        print("current:", current_loss)
        break

    previous_loss = current_loss
````

### Key Takeaway

Early stopping is fundamentally a control-flow decision, but realistic implementations need robustness against noisy validation behaviour.

## Problem 32 — Reasoning about Control Flow

### Problem

Without executing the code, determine the exact output:

```python
for i in range(1, 5):
    if i == 2:
        continue

    for j in range(1, 4):
        if j == 3:
            break

        print(i, j)
else:
    print("done")
```

Explain separately the effect of

- `continue`;
- the inner `break`;
- the outer `else`.

In particular, explain why the inner `break` does or does not prevent the final `else` from executing.

### Solution

The outer loop produces `i = 1, 2, 3, 4`.

For `i = 1`, the inner loop prints `(1, 1)` and `(1, 2)`. When `j == 3`, the inner `break` terminates only the inner loop.

For `i = 2`, `continue` skips the remainder of the outer-loop body, so no inner loop runs.

For `i = 3` and `i = 4`, the same inner-loop pattern as for `i = 1` occurs.

The output is therefore:

```text
1 1
1 2
3 1
3 2
4 1
4 2
done
```

The final `else` belongs to the **outer `for` loop**. That outer loop is never terminated by `break`; it completes normally after processing `i = 4`. The `break` statements occur inside the inner loop and therefore do not suppress the outer loop's `else`.

#### Verification

````python
for i in range(1, 5):
    if i == 2:
        continue

    for j in range(1, 4):
        if j == 3:
            break

        print(i, j)
else:
    print("done")
````

### Key Takeaway

When control structures are nested, `break`, `continue`, and loop `else` must be interpreted relative to the specific loop to which each statement belongs.

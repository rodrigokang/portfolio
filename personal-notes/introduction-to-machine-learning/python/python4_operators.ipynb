# 6. Operators in Python

An operator in programming is a symbol or keyword that performs specific actions on one or more operands.

## 6.1. Decision Operators

Decision operators evaluate conditions (such as equality or comparison) to control program flow.

### 6.1.1. `if`

When we want to check if a certain condition is met, we use the `if` decision operator. The syntax we must follow is as follows:

```python
# if condition:
#     consequence
```


**Note**: The syntax of the colon after the condition and the indentation (equivalent to a tabulation, a total of 4 spaces) preceding the consequence is crucial. In fact, if either of these is omitted or if there is an indentation error, it will result in an error.

- To indent, press the tab key once.
- You can indent blocks by selecting the lines of code you want to indent and then pressing the tab key.
- You can unindent by pressing Shift + Tab.
- You can unindent blocks by selecting the lines of code you want to unindent and then pressing Shift + Tab.

Continuing with our example, if the user is older than 16 but less than 40, then they can get the job interview.

```python
age = 23
if (age >= 16 and age <= 40):
    print("You can have the job interview")
```

### 6.1.2. `else`

Now, we might wonder what we could tell the user in the case where the condition is not satisfied. This is where the `else` decision operator comes into play. This time, the syntax to follow is as follows:

```python
# if condición:
#     consecuencia_si_es_verdad
# else:
#     consecuencia_si_es_falsa
```

Following the previous example, if the user is 16 years or older but less than 40, then they can have the job interview. If not, we will inform them that they do not meet a basic requirement to have it.

```python
age = 13
if (age >= 16 and age <= 40):
    print("You can have the job interview")
else:
    print("You do not meet a basic requirement to have the job interview")
```

### 6.1.3. `elif`

Now, instead of checking whether a condition is met or not, we could ask ourselves how we would check for more than one condition. We could do it brute force by nesting `if` statements, that is, putting one `if` inside another; or we could do it using the `elif` decision operator.

The `elif` operator works as follows: it starts with an `if` statement; if the condition of this `if` is not met, we move on to the next possible condition preceded by an `elif`; if this condition is also not met, we move on to the next `elif`; we continue this way until either a condition is satisfied and its consequence is executed, or we reach the `else`, which implies that none of the previous conditions have been satisfied.

The syntax of the `elif` decision operator is as follows:

```python
# if condición_1:
#     consecuence
# elif condición_2:
#     consecuence
# elif condición_3:
#     consecuence
# . 
# . 
# . 
# else:
#     consecuence
```

Let's revisit the example where we need to be 16 years or older but less than 40 to have a job interview. We will enhance what we achieved with `if` and `else` by adding the `elif` operator.

Additionally, we will see what it tells us with an age of 20 years.

```python
age = 20

if age > 40:
    print("You cannot have the job interview; you exceed the age limit set by the company.")
elif age >= 16:
    print("You can apply for the job interview. You still need to pass it.")
else:
    print("You are still too young for the job.")
```

The operation of the above code is as follows:

- The `if` statement checks if the entered age is greater than 40.
- The `elif` statement checks if the age is in the range [16, 40].
- The `else` statement implies that the entered age is less than 16.

### 6.1.4. Ternary Operator

If we want to create a simple `if`/`else` statement in a single line of code, we can use the ternary operator, which has the following structure:

```python
# consecuence_true if condición else consecuence_false
```

For example, let's consider the case of being of legal age. If the age is greater than or equal to 18, then the person is of legal age in Argentina. Otherwise, they are a minor in Argentina.

```python
age = 20
text_major = "You are of legal age in Argentina"
text_minor = "You are a minor in Argentina"

print(text_major) if age >= 18 else print(text_minor)
```

### 6.1.5. Nested Operators

```python
age = 20
name = "Martin"

if age >= 18:
    if name.startswith("M") or name.startswith("m"):
        print("You are of legal age, as you are {} years old and your name, which is {}, starts with M.".format(age, name))
    else:
        print("You are of legal age, as you are {} years old.".format(age))
else:
    print("You are too young.")
```

## 6.2. Iteration Operators

Iteration operators execute a set of instructions repeatedly until a condition is met or a certain number of iterations are completed, facilitating repetitive tasks and enabling dynamic control over program execution.

### 6.2.1. "The `while` loop".

The idea of the `while` loop is: as long as the condition is true, we continue executing the lines inside the loop. Once the condition becomes false, we exit the loop.

Its structure is as follows:

```python
# Initialization of the condition variable
# while true condition:
    # statement 1
    # statement 2
    # .
    # .
    # .
    # statement n
```

Both the colons after the condition and the indentation before the instructions inside the loop reappear.

It should be noted that one of the instructions inside the `while` loop must modify the condition variable. Otherwise, if the condition variable is never modified, the condition will never become false, and the loop will never end, resulting in what is called an infinite loop.

One of the utilities of a `while` loop is to avoid the copy-pasting of certain functions. For example, if we want to print the first 10 natural numbers, instead of copying the `print()` function 10 times, once for each number, we can achieve this with a `while` loop:

```python
i = 1  # Initialize the variable
while i <= 10:  # We want i to be at most 10
    print(i)  # Print the numbers
    i += 1  # Increment by one in each iteration
```

### 6.2.2. `break` Statement

`break` is very useful if, given a condition, we want to exit a `while` loop immediately. Let's see it with an example:

**Example**

The Fibonacci sequence is an infinite sequence characterized by the fact that each term is the sum of the two preceding ones. Some of its terms are 1, 1, 2, 3, 5, 8, 13...

Let's say we want to print the first 20 terms of this series. Therefore, we will need both the terms of the series and the indices they occupy.

```python
fibo_ant = 1  # Previous term
fibo = 1  # Current term
idx = 3  # Since we already have the first two terms, we start with index 3

print("Term {} is at position {}".format(fibo_ant, 1))
print("Term {} is at position {}".format(fibo, 2))

while fibo <= 500000:  # Set a limit to prevent an infinite loop
    temp = fibo  # Temporarily store the current Fibonacci term
    fibo = fibo + fibo_ant  # Calculate the new term in the sequence
    fibo_ant = temp  # Modify the value of the previous term

    print("Term {} is at position {}".format(fibo, idx))

    if idx == 20:  # If we reach the twentieth index,
        break      # exit the loop

    idx += 1  # Increment the index value
```

The previous example could have been done perfectly without the need to use the `break` statement. Think about how you would modify the previous code to achieve the same result using only a `while` loop.

```python
fibo_ant = 1  # Previous term
fibo = 1  # Current term
idx = 1  # Start with index 1

while idx <= 20:  # Execute the loop for the first 20 indices
    print("Term {} is at position {}".format(fibo, idx))

    temp = fibo  # Temporarily store the current Fibonacci term
    fibo = fibo + fibo_ant  # Calculate the new term in the sequence
    fibo_ant = temp  # Modify the value of the previous term

    idx += 1  # Increment the index value
```

### 6.2.3. `while` ... `else` Combination

We can combine a `while` loop with the `else` operator to execute a block of code once the `while` condition has become false.

```python
i = 10
print("Ready for takeoff. Countdown begins.")
while i >= 0:
    print(i)
    i -= 1
else:
    print("Countdown has finished.")
```

### 6.2.4. `for` Loop

The idea of the for loop is: for all elements of the key, we continue executing the lines of the loop. Once we run out of elements, we exit the loop.

Its structure is as follows:

```python
# for key:
#    statement 1
#    statement 2
#    .
#    .
#    .
#    statement n
```

Both the colons after the key and the indentation before the instructions inside the loop reappear.

An example of using a for loop is iterating over all the characters of a string:

```python
s = "I like mathematics"

for c in s:
    print(c)
```

What the previous code chunk does is print each and every character, identified by `c`, that is present in the string `s`.

### 6.2.5. `range()` Function

The `range()` function has 3 possible arguments:

- `start`
- `stop`
- `step`
Let's see the usage of the `range()` function with an example. 

Let's revisit the example where we wanted to print the first 10 natural numbers:

```python
for i in range(1, 11, 1):
    print(i)
```

```python
for x in range(5):
    print(x)
```

Things to consider when using the `range()` function:

- The element specified in the `stop` argument is never included.
- If we don't specify any element in the `start` argument, it defaults to 0.
- The default value for the `step` argument is 1.

Therefore, we would get the same result as in the previous example by executing the following lines of code:

```python
for i in range(1, 11):
    print(i)
```

And what if we wanted to print the first 10 natural numbers in reverse order? Well, with a for loop, we would do it as follows:

```python
for i in range(10, 0, -1):
    print(i)
```

### 6.2.6. `continue` Statement

The `continue` statement is similar to `break`, but instead of exiting the loop, it interrupts the current iteration and starts the next one.

```python
for i in range(101):
    if i % 2 == 0 or i % 5 == 0:
        continue
    print(i)
```

### 6.2.7. Nested Loops

It involves loops within loops.

```python
for i in range(1, 11):
    print("\nMultiplication table for {}".format(i))
    for j in range(1, 21):
        print("{} x {} = {}".format(i, j, i * j))
```

Develop a program to check whether a year is a leap year or not. A leap year is divisible by four but not a multiple of one hundred unless it is divisible by four hundred.

```python
year = int(input("Year: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print("The year {} is a leap year".format(year))
        else:
            print("The year {} is not a leap year".format(year))
    else:
        print("The year {} is a leap year".format(year))
else:
    print("The year {} is not a leap year".format(year))
```

### 6.3.10. Problem 10

Imagine the letters of the alphabet arranged in a circle. That is, to the right of A is B, then C, and so on until Z. To the right of Z, the letter A is again found.

Have the user enter an integer value `n`, which corresponds to the rotation that will bring a certain letter `n` positions to its right. For example, if the rotation is 4, then A will become E, B to F, ... , Y to C, and Z to D.

HINT: Investigate the `chr()` and `ord()` functions to convert from the ASCII value of a character to the character itself and vice versa.

```python
n = int(input("Enter a rotation: "))
i = 65

while i <= 90:
    if i + n <= 90:
        print(chr(i) + ": " + chr(i + n))
    else:
        print(chr(i) + ": " + chr((i - 26) + n))
    i += 1
```

### 6.3.12. Problem 12

```python
s = "We think too much and feel too little"

letter = input("Enter a letter: ")
n = int(input("Specify the maximum number of times you want the previous letter to appear: "))
count = 0
letter = letter.lower()
s = s.lower()

for c in s:
    if count >= n:
        print("\nThe number of appearances has been exceeded")
        break
    if c == letter:
        count += 1
    elif c == "a" or c == "e" or c == "i" or c == "o" or c == "u":
        continue
  
    print(c, end="")
```

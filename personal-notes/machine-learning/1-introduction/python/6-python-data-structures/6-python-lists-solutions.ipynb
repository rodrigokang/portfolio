## Problem 1 — Indexing

### Problem

Given

```python
values = [4, 7, 1, 9, 3]
```

determine the result of

```python
values[0]
values[3]
values[-1]
values[-3]
```

without executing the code.

For each expression, state the corresponding non-negative index.

### Solution

The list has length 5, so its non-negative indices are `0` through `4`.

- `values[0]` is `4`.
- `values[3]` is `9`.
- `values[-1]` refers to the final element, which is non-negative index `4`, so the value is `3`.
- `values[-3]` refers to the third element from the end, which is non-negative index `2`, so the value is `1`.

Thus the corresponding index pairs are

```text
0  <-> -5
1  <-> -4
2  <-> -3
3  <-> -2
4  <-> -1
```

#### Verification

````python
values = [4, 7, 1, 9, 3]
values[0], values[3], values[-1], values[-3]
````

### Key Takeaway

Negative indices are alternative positional references counted from the end of the sequence.

## Problem 2 — Index Bounds

### Problem

Given

```python
values = [10, 20, 30]
```

what happens when evaluating

```python
values[3]
```

and

```python
values[-4]
```

?

Explain why both accesses are invalid.

### Solution

The list has length 3. Its valid non-negative indices are `0`, `1`, and `2`, while its valid negative indices are `-1`, `-2`, and `-3`.

Therefore, both `values[3]` and `values[-4]` are outside the valid index range and raise

```text
IndexError: list index out of range
```

The valid index intervals are

$$
0 \leq i < 3
$$

for non-negative indexing and

$$
-3 \leq i \leq -1
$$

for negative indexing.

#### Verification

````python
values = [10, 20, 30]

for i in (3, -4):
    try:
        print(values[i])
    except IndexError as exc:
        print(i, type(exc).__name__, exc)
````

### Key Takeaway

Direct indexing requires an existing position; an index outside either valid range raises `IndexError`.

## Problem 3 — Basic Slicing

### Problem

Given

```python
values = [10, 20, 30, 40, 50, 60]
```

predict:

```python
values[1:4]
values[:3]
values[3:]
values[:]
```

Explain why the element at the stopping index is excluded.

### Solution

The results are

```text
values[1:4] -> [20, 30, 40]
values[:3]  -> [10, 20, 30]
values[3:]  -> [40, 50, 60]
values[:]   -> [10, 20, 30, 40, 50, 60]
```

Python slices follow a half-open interval convention: the start index is included and the stop index is excluded. Therefore `values[1:4]` selects indices `1`, `2`, and `3`, but not `4`.

The expression `values[:]` returns a new outer list containing references to the same elements.

#### Verification

````python
values = [10, 20, 30, 40, 50, 60]
values[1:4], values[:3], values[3:], values[:]
````

### Key Takeaway

Slicing uses the half-open interval `[start, stop)` and normally returns a new list.

## Problem 4 — Slice Steps

### Problem

Predict the result of:

```python
values = list(range(10))

values[::2]
values[1::2]
values[2:8:2]
values[::-1]
```

### Key Takeaway

The third slice component controls spacing and direction; the stop boundary remains excluded.

## Problem 5 — Indexing versus Slicing

### Problem

Given

```python
values = [1, 2, 3, 4]
```

compare the behaviour of

```python
values[20]
```

with

```python
values[1:20]
```

Why does one operation raise an exception while the other does not?

### Solution

`values[20]` requests one specific element at a position that does not exist, so Python raises `IndexError`.

By contrast, `values[1:20]` requests a slice. Slice boundaries are clipped to the available sequence. Python therefore returns all available elements starting at index `1`:

```python
[2, 3, 4]
```

Slicing is deliberately tolerant of out-of-range boundaries, while direct indexing requires an actual position.

#### Verification

````python
values = [1, 2, 3, 4]

try:
    print(values[20])
except IndexError as exc:
    print(type(exc).__name__, exc)

print(values[1:20])
````

### Key Takeaway

Indexing requires an existing element; slicing safely clips boundaries to the sequence.

## Problem 8 — `append()` versus `extend()`

### Problem

Without executing the code, determine the final values of `a` and `b`:

```python
a = [1, 2]
b = [1, 2]

a.append([3, 4])
b.extend([3, 4])
```

Explain the structural difference between the two resulting lists.

### Solution

`append()` adds its argument as one element:

```python
a == [1, 2, [3, 4]]
```

`extend()` iterates over its argument and adds each produced element separately:

```python
b == [1, 2, 3, 4]
```

Thus `a` contains a nested list as its final element, whereas `b` remains a flat list of four integers.

#### Verification

````python
a = [1, 2]
b = [1, 2]

a.append([3, 4])
b.extend([3, 4])

a, b
````

### Key Takeaway

`append()` adds one object; `extend()` adds the elements produced by an iterable.

## Problem 14 — `sort()` versus `sorted()`

### Problem

Predict the final state:

```python
values = [4, 1, 3, 2]

result = values.sort()
```

What are the values of `values` and `result`?

Then rewrite the operation so that the original list remains unchanged and a sorted list is stored in `result`.

### Solution

`list.sort()` modifies the list in place and returns `None`:

```python
values == [1, 2, 3, 4]
result is None
```

To preserve the original list, use

```python
result = sorted(values)
```

on the original unsorted input. `sorted()` returns a new list and does not modify its argument.

#### Verification

````python
values = [4, 1, 3, 2]
result = values.sort()
print(values, result)

values = [4, 1, 3, 2]
result = sorted(values)
print(values, result)
````

### Key Takeaway

`sort()` mutates and returns `None`; `sorted()` returns a new sorted list.

## Problem 15 — Sorting by a Derived Criterion

### Problem

Given

```python

    ["linear", 0.82],
    ["forest", 0.91],
    ["tree", 0.78],
]
```

sort the records from highest score to lowest score using the second element of each inner list as the sorting key.

Do not manually rearrange the elements.

### Solution

The score is at index `1`, so it can be used as the sorting key:

```python
ordered = sorted(

    reverse=True,
)
```

The result is

```python
[
    ["forest", 0.91],
    ["linear", 0.82],
    ["tree", 0.78],
]
```

#### Verification

### Key Takeaway

The `key` argument separates the object being sorted from the criterion used to order it.

## Problem 16 — Equality and Identity

### Problem

Predict:

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

a == b
a is b
a == c
a is c
```

Explain every result.

### Solution

`a` and `b` are distinct list objects with equal contents. `c = a` creates another reference to the same object as `a`.

Therefore:

```text
a == b -> True
a is b -> False
a == c -> True
a is c -> True
```

`==` compares values, while `is` compares object identity.

#### Verification

````python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

a == b, a is b, a == c, a is c
````

### Key Takeaway

Equality concerns contents; identity concerns whether two names denote the same object.

## Problem 17 — Aliasing

### Problem

Without executing the code, determine the final values of `a` and `b`:

```python
a = [1, 2, 3]
b = a

b.append(4)

```

How many list objects are involved?

```python
[1, 2, 3, 4]
```

Then `a[0] = 100` changes the same object to

```python
[100, 2, 3, 4]
```

Both `a` and `b` therefore display the same final list.

Assignment between list variables copies a reference, not the list object.

## Problem 18 — Rebinding versus Mutation

### Problem

Consider:

```python
a = [1, 2, 3]
b = a

b = [4, 5, 6]
```

What are the final values of `a` and `b`?

Compare this with:

```python
a = [1, 2, 3]
b = a

b[:] = [4, 5, 6]
```

Explain why the results differ.

### Solution

In the first case, `b = [4, 5, 6]` **rebinds** `b` to a new list. `a` still refers to the original:

```text
a -> [1, 2, 3]
b -> [4, 5, 6]
```

In the second case, `b[:] = ...` performs slice assignment on the shared list object. Since `a` and `b` are aliases, both observe

```python
[4, 5, 6]
```

The distinction is between changing a name's binding and mutating the object to which the name refers.

#### Verification

````python
a = [1, 2, 3]
b = a
b = [4, 5, 6]
print(a, b, a is b)

a = [1, 2, 3]
b = a
b[:] = [4, 5, 6]
print(a, b, a is b)
````

### Key Takeaway

Rebinding changes which object a name references; mutation changes the shared object itself.

## Problem 19 — Shallow Copy

### Problem

Predict:

What are `a` and `b`?

What are the values of

```text
a == [1, 2, 3]
b == [1, 2, 3, 4]
```

After the modification:

```text
a == b -> False
```

### Key Takeaway

A shallow copy creates an independent outer list, which is sufficient when the contained elements do not require independent mutation.

## Problem 20 — Shallow Copy with Nested Lists

### Problem

Consider:

```python
a = [[1, 2], [3, 4]]
b = a.copy()

b[0].append(5)
```

Predict both lists.

Then predict the values of

```python
a is b
a[0] is b[0]
```

Explain the apparent contradiction between the two identity tests.

### Solution

The outer lists are distinct, but a shallow copy copies references to the same inner lists.

Therefore `b[0].append(5)` modifies an inner list shared by both outer lists:

```text
a == [[1, 2, 5], [3, 4]]
b == [[1, 2, 5], [3, 4]]
```

The identity tests are

```text
a is b       -> False
a[0] is b[0] -> True
```

There is no contradiction: identity is being tested at two different levels of the object structure.

#### Verification

````python
a = [[1, 2], [3, 4]]
b = a.copy()

b[0].append(5)

a, b, a is b, a[0] is b[0]
````

### Key Takeaway

A shallow copy duplicates the outer container but preserves references to nested objects.

## Problem 21 — Replacement versus Inner Mutation

### Problem

Starting independently from

```python
a = [[1, 2], [3, 4]]
b = a.copy()
```

compare the effects of

```python
b[0].append(99)
```

and

```python
b[0] = [99]
```

on `a`.

Explain the difference in terms of references.

### Solution

With `b[0].append(99)`, the inner list at index `0` is shared, so it is mutated and `a` changes too.

With `b[0] = [99]`, the outer list `b` is modified so that its index `0` now refers to a new list. The inner list still referenced by `a[0]` is untouched.

Thus mutation follows the shared reference; replacement changes only the reference stored in one outer list.

#### Verification

````python
a = [[1, 2], [3, 4]]
b = a.copy()
b[0].append(99)
print("inner mutation:", a, b)

a = [[1, 2], [3, 4]]
b = a.copy()
b[0] = [99]
print("replacement:", a, b)
````

### Key Takeaway

Mutating a shared nested object is visible through every reference; replacing one outer reference is local to that outer list.

## Problem 22 — Deep Copy

### Problem

Use `copy.deepcopy()` to create an independent copy of

```python
original = [[1, 2], [3, 4]]
```

Modify one of the inner lists of the copy and verify that the original remains unchanged.

Then compare

```python
original[0] is copied[0]
```

with the corresponding result from a shallow copy.

### Solution

A deep copy recursively duplicates the nested mutable structure:

```python
import copy

copied = copy.deepcopy(original)
copied[0].append(99)
```

The original remains

```python
[[1, 2], [3, 4]]
```

while the copy becomes

```python
[[1, 2, 99], [3, 4]]
```

For the deep copy,

```python
original[0] is copied[0]
```

is `False`.

For a shallow copy, the corresponding identity test is `True` because the inner list is shared.

#### Verification

````python
import copy

original = [[1, 2], [3, 4]]
copied = copy.deepcopy(original)
copied[0].append(99)

print(original)
print(copied)
print(original[0] is copied[0])

shallow = original.copy()
print(original[0] is shallow[0])
````

### Key Takeaway

Deep copying recursively separates nested mutable objects; shallow copying separates only the outer container.

## Problem 23 — Repeated Nested Lists

Predict the output:


### Key Takeaway

Repeating a container repeats references; use repeated construction when independent nested mutable objects are required.

## Problem 25 — `+=` and Aliasing

### Key Takeaway

For mutable lists, augmented assignment may preserve object identity, while ordinary concatenation followed by assignment rebinds the variable.

## Problem 27 — Constructing a List

### Problem

Given

```python
values = [3, -1, 5, -4, 7, 0]
```

construct a new list containing only the non-negative values using

1. an explicit `for` loop;
2. a list comprehension.

The original list must remain unchanged.

### Solution

Explicit loop:

```python
filtered = []

for value in values:
    if value >= 0:
        filtered.append(value)
```

List comprehension:

```python
filtered = [value for value in values if value >= 0]
```

Both produce

```python
[3, 5, 7, 0]
```

without modifying `values`.

#### Verification

````python
values = [3, -1, 5, -4, 7, 0]

explicit = []
for value in values:
    if value >= 0:
        explicit.append(value)

comprehension = [value for value in values if value >= 0]

explicit, comprehension, values
````

### Key Takeaway

A filtering comprehension is the concise form of a common explicit loop that conditionally appends elements.

## Problem 32 — Nested Traversal

### Problem

## Problem 34 — Structural Mutation during Iteration

### Problem

Predict the result:

```python
values = [1, 2, 2, 2, 3]

for value in values:
    if value == 2:
        values.remove(value)
```

Why might the final list still contain a `2`?

Rewrite the operation so that every `2` is removed without structurally modifying the list being traversed.

### Solution

The final list is

```python
[1, 2, 3]
```

The reason is that removing an element shifts later elements left while the iterator continues advancing. Some elements are therefore skipped.

A safe solution is to construct a new list:

```python
values = [value for value in values if value != 2]
```

or equivalently use an explicit loop that appends only retained values.

#### Verification

````python
values = [1, 2, 2, 2, 3]

for value in values:
    if value == 2:
        values.remove(value)

print("mutated during iteration:", values)

values = [1, 2, 2, 2, 3]
values = [value for value in values if value != 2]
print("filtered safely:", values)
````

### Key Takeaway

Structural mutation can change the positions the iterator has yet to visit; filtering into a new list avoids this class of bug.

## Problem 37 — Selecting Valid Measurements

### Problem

Given

```python
values = [0.42, None, -0.1, 0.71, None, 0.88, -0.3]
```

construct a new list containing only non-`None`, non-negative values.

First solve the problem with an explicit loop and then with a list comprehension.

Be careful about the order of the conditions.

### Solution

With an explicit loop:

```python
valid = []

for value in values:
    if value is None:
        continue

    if value >= 0:
        valid.append(value)
```

A comprehension can be written as

```python
valid = [
    value
    for value in values
    if value is not None and value >= 0
]
```

The `None` test must occur before the numerical comparison because evaluating `None >= 0` would raise `TypeError`.

The result is

```python

```

````python
values = [0.42, None, -0.1, 0.71, None, 0.88, -0.3]

explicit = []
for value in values:
    if value is None:
        continue
    if value >= 0:
        explicit.append(value)

comprehension = [
    value
    for value in values
    if value is not None and value >= 0
]

explicit, comprehension
````

### Key Takeaway

## Problem 39 — Copying a Configuration

### Problem

Suppose

```python
config = [
    ["learning_rate", 0.01],
    ["epochs", 100],
]
```

and you want to create a second configuration that can be modified independently.

Explain why

```python
config_b = config
```

is not sufficient.

Then explain why

```python
config_b = config.copy()
```

may still be insufficient if the inner lists are going to be modified.

Create an appropriately independent copy and demonstrate the difference.

### Solution

`config_b = config` creates only another reference to the same outer list, so any mutation of the list is shared.

`config.copy()` creates a new outer list but leaves the inner lists shared. Therefore an operation such as

```python
config_b[0][1] = 0.02
```

would also modify `config[0]`.

```python
import copy
config_b = copy.deepcopy(config)
```

Then nested mutation of `config_b` does not affect `config`.

#### Verification

### Key Takeaway

The required copying strategy depends on which levels of a nested mutable object must be independent.

## Problem 40 — Reasoning about Objects

### Problem

Without executing the code, determine the exact output:

```python
a = [[1, 2], [3, 4]]
b = a
c = a.copy()
d = a[:]

b[0].append(5)
c.append([6, 7])
d[1][0] = 99

print(a)
print(b)
print(c)
print(d)

print(a is b)
print(a is c)
print(a[0] is c[0])
print(a[1] is d[1])
```

Trace the object relationships carefully before predicting the result.

### Solution

Initially, `b` aliases `a`, while `c` and `d` are shallow copies. Therefore all four outer lists share the original inner lists.

1. `b[0].append(5)` mutates the first shared inner list to `[1, 2, 5]`.
2. `c.append([6, 7])` modifies only the outer list `c`, adding a new third element.
3. `d[1][0] = 99` mutates the second shared inner list from `[3, 4]` to `[99, 4]`.

The exact printed lists are therefore:

```text
[[1, 2, 5], [99, 4]]
[[1, 2, 5], [99, 4]]
[[1, 2, 5], [99, 4], [6, 7]]
[[1, 2, 5], [99, 4]]
```

The identity tests are:

```text
True
False
True
True
```

because `a` and `b` are the same outer list, `c` is a different outer list, and the relevant inner lists remain shared by the shallow copies.

#### Verification

````python
a = [[1, 2], [3, 4]]
b = a
c = a.copy()
d = a[:]

b[0].append(5)
c.append([6, 7])
d[1][0] = 99

print(a)
print(b)
print(c)
print(d)

print(a is b)
print(a is c)
print(a[0] is c[0])
print(a[1] is d[1])
````

### Key Takeaway

When reasoning about nested mutable structures, track identity separately at each level: outer-container identity and contained-object identity are independent questions.

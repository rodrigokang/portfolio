## Problem 1 — Basic Function Definition

### Problem

Write a function `rectangle_area()` that receives the width and height of a rectangle and returns its area.

Use the function to calculate the area of a rectangle with width `7.5` and height `4`.

Then explain the roles of the parameters, arguments, and return value in the function call.

### Solution

`width` and `height` are parameters because they appear in the function definition. In `rectangle_area(7.5, 4)`, the values `7.5` and `4` are arguments supplied to those parameters.

The `return` statement sends the computed area back to the caller. Returning the value, rather than merely printing it, makes the result available for assignment, testing, or further computation.

#### Implementation

````python
def rectangle_area(width, height):
    return width * height

area = rectangle_area(7.5, 4)
area
````

### Key Takeaway

Parameters describe a function's inputs, arguments provide concrete values during a call, and `return` exposes the result to the caller.

## Problem 2 — Positional and Keyword Arguments

### Problem

Consider:

```python
def calculate_total(price, quantity, discount=0):
    return price * quantity * (1 - discount)
```

Determine whether each call is valid and identify the parameter bindings and returned value:

```python
calculate_total(10, 5)
calculate_total(10, 5, 0.2)
calculate_total(price=10, quantity=5, discount=0.2)
calculate_total(quantity=5, price=10)
calculate_total(10, quantity=5)
```

### Solution

All five calls are valid.

- `calculate_total(10, 5)` gives `price=10`, `quantity=5`, `discount=0` and returns `50`.
- `calculate_total(10, 5, 0.2)` gives `discount=0.2` and returns `40`.
- The fully keyword-based call also returns `40`.
- `calculate_total(quantity=5, price=10)` uses the default discount and returns `50`.
- `calculate_total(10, quantity=5)` mixes a positional argument with a keyword argument and also returns `50`.

A parameter may not receive more than one value in the same call.

#### Verification

````python
def calculate_total(price, quantity, discount=0):
    return price * quantity * (1 - discount)

[
    calculate_total(10, 5),
    calculate_total(10, 5, 0.2),
    calculate_total(price=10, quantity=5, discount=0.2),
    calculate_total(quantity=5, price=10),
    calculate_total(10, quantity=5),
]
````

### Key Takeaway

Python binds arguments either by position or by parameter name; defaults are used only when the corresponding argument is omitted.

## Problem 3 — Default Arguments

### Problem

Write `calculate_power()` so that it receives a numeric value and an optional exponent that defaults to `2`.

These calls should be valid:

```python
calculate_power(5)
calculate_power(5, 3)
calculate_power(5, exponent=4)
```

### Solution

The default is declared in the signature as `exponent=2`. When the caller omits `exponent`, Python uses `2`. A positional or keyword argument explicitly supplied for `exponent` overrides that default for the current call.

#### Implementation

````python
def calculate_power(value, exponent=2):
    return value ** exponent

calculate_power(5), calculate_power(5, 3), calculate_power(5, exponent=4)
````

### Key Takeaway

A default parameter establishes standard behaviour while preserving configurability.

## Problem 4 — Mutable Default Arguments

### Problem

Consider:

```python
def add_observation(value, observations=[]):
    observations.append(value)
    return observations
```

Predict the results of three successive calls with `10`, `20`, and `30`, explain the behaviour, and rewrite the function safely.

### Solution

The successive results are `[10]`, `[10, 20]`, and `[10, 20, 30]`.

The default list is created once when the function definition executes. Every omitted `observations` argument therefore reuses the same mutable list. The usual solution is to use `None` as a sentinel and create a fresh list inside the function.

#### Corrected Implementation

````python
def add_observation(value, observations=None):
    if observations is None:
        observations = []
    observations.append(value)
    return observations

add_observation(10), add_observation(20), add_observation(30)
````

### Key Takeaway

Mutable defaults can unintentionally preserve state across calls because default expressions are evaluated at function definition time.

## Problem 5 — Variable-Length Positional Arguments

### Problem

Write `calculate_mean()` using `*args` so that it accepts any number of positional numeric arguments. Explain the type used internally and decide how to handle an empty call.

### Solution

Values collected by `*args` are stored in a tuple. Because the arithmetic mean is undefined for zero observations, an empty call should be handled explicitly. Raising `ValueError` gives the caller a clear failure mode.

#### Implementation

````python
def calculate_mean(*values):
    if not values:
        raise ValueError("At least one value is required.")
    return sum(values) / len(values)

calculate_mean(10, 20, 30), calculate_mean(1, 2, 3, 4, 5)
````

### Key Takeaway

`*args` collects additional positional arguments into a tuple; undefined edge cases should be validated explicitly.

## Problem 6 — Variable-Length Keyword Arguments

### Problem

Write `model_configuration()` using `**kwargs` so that arbitrary keyword arguments are returned as a dictionary. Explain how call-site keywords relate to dictionary entries.

### Solution

`**kwargs` collects additional keyword arguments into a dictionary. Each keyword name becomes a key and the associated argument becomes its value. The name `kwargs` is conventional; the `**` syntax is the relevant mechanism.

#### Implementation

````python
def model_configuration(**kwargs):
    return kwargs

model_configuration(
    model="random_forest",
    n_estimators=200,
    max_depth=10
)
````

### Key Takeaway

`**kwargs` provides a flexible keyword interface by collecting names and values into a dictionary.

## Problem 7 — Return Values

### Problem

Write `descriptive_range()` so that it returns the minimum and maximum of a numeric sequence. Explain why Python is often said to return two values even though one object is technically returned.

### Solution

`return min(values), max(values)` creates a tuple. Tuple unpacking allows the caller to assign its elements to two names, which creates the convenient appearance of two separate returned values.

#### Implementation

````python
def descriptive_range(values):
    return min(values), max(values)

minimum, maximum = descriptive_range([4, 8, 1, 9, 3])
result = descriptive_range([4, 8, 1, 9, 3])

minimum, maximum, result, type(result)
````

### Key Takeaway

Multiple related outputs are commonly represented by a single returned tuple and then unpacked by the caller.

## Problem 8 — Implicit Return Value

### Problem

Given:

```python
def display_result(value):
    print(f"Result: {value}")
```

What is assigned by `result = display_result(10)`? Explain the difference between printed output and returned data.

### Solution

The function prints `Result: 10` but implicitly returns `None`.

Printing is a side effect. Returning is part of the function's data interface. If another function expects the computed value, a printed representation cannot substitute for an explicit return value.

#### Verification

````python
def display_result(value):
    print(f"Result: {value}")

result = display_result(10)
print("Returned value:", result)
print("Returned type:", type(result))
````

### Key Takeaway

Printed output and returned values are conceptually different: one produces an effect, the other communicates data to the caller.

## Problem 9 — Scope and Name Resolution

### Problem

Predict the output and explain it using LEGB:

```python
x = 10

def outer():
    x = 20
    def inner():
        x = 30
        print(x)
    inner()
    print(x)

outer()
print(x)
```

### Solution

The output is `30`, then `20`, then `10`. Each assignment creates a binding in a different scope. Inside `inner()`, the local `x` wins. Back in `outer()`, its own local `x` is used. At module level, the global `x` remains `10`.

LEGB stands for Local, Enclosing, Global, Built-in.

#### Verification

````python
x = 10

def outer():
    x = 20
    def inner():
        x = 30
        print(x)
    inner()
    print(x)

outer()
print(x)
````

### Key Takeaway

Identically spelled names can coexist in different scopes; Python resolves each use according to LEGB.

## Problem 10 — Enclosing Scope

### Problem

Given a global `x="global"` and an `outer()` function containing `x="enclosing"`, explain which value an inner function prints when it contains only `print(x)`.

### Solution

The inner function prints `"enclosing"`. No local `x` exists inside it, so Python searches the enclosing function scope and finds `x` there before reaching the global scope.

#### Verification

````python
x = "global"

def outer():
    x = "enclosing"
    def inner():
        print(x)
    inner()

outer()
````

### Key Takeaway

Nested functions can read names from an enclosing function scope, and lookup stops at the first matching binding.

## Problem 11 — Rebinding Versus Mutation

### Problem

Predict the output:

```python
def modify(x, values):
    x = x + 1
    values.append(x)

number = 10
numbers = [1, 2, 3]

modify(number, numbers)

print(number)
print(numbers)
```

Explain the difference using rebinding, mutation, and mutability.

### Solution

The output is `10` and `[1, 2, 3, 11]`.

`x = x + 1` creates a new integer and rebinds the local parameter `x`; the caller's name `number` remains unchanged. `values.append(x)` mutates the shared list object, so the change is visible through `numbers`.

Integers are immutable and lists are mutable, but the deeper distinction is between rebinding a local name and mutating a shared object.

#### Verification

````python
def modify(x, values):
    x = x + 1
    values.append(x)

number = 10
numbers = [1, 2, 3]
modify(number, numbers)

print(number)
print(numbers)
````

### Key Takeaway

Parameters are local names bound to objects; rebinding is local, whereas mutating a shared mutable object can affect the caller.

## Problem 12 — Object Identity

### Problem

Compare the values printed by `id(numbers)` and `id(values)` when a list is passed into a function. Use the result to discuss Python's argument-passing model.

### Solution

The two IDs are equal during the call because the caller's name and the parameter are bound to the same list object.

This illustrates why simple labels such as “pass by value” and “pass by reference” can mislead. A useful model is **call by sharing**: the function receives another reference to the same object, while the parameter name itself remains local.

#### Verification

````python
def inspect(values):
    print("Inside:", id(values))

numbers = [1, 2, 3]
print("Outside:", id(numbers))
inspect(numbers)
````

### Key Takeaway

Python calls share references to objects, even though parameter names belong to the callee's local scope.

## Problem 13 — Avoiding Unintended Mutation

### Problem

Rewrite a function that removes negative values from a list so that it returns a new list and leaves the original unchanged. Compare the two designs and identify cases where mutation may still be appropriate.

### Solution

A list comprehension naturally constructs a new result. This design makes the transformation explicit and preserves the input.

Mutation can still be intentional when an API clearly documents in-place behaviour, when object identity matters, or when avoiding a large copy is materially important.

#### Implementation

````python
def remove_negative_values(values):
    return [value for value in values if value >= 0]

original = [3, -1, 5, -4, 0]
cleaned = remove_negative_values(original)

print("Original:", original)
print("Result:", cleaned)
````

### Key Takeaway

Returning a new object often improves predictability; mutation should be deliberate and visible in the function contract.

## Problem 14 — Functions as First-Class Objects

### Problem

Explain the difference between referring to `operation` and calling `operation(5)` after `operation = square`. Also distinguish `operation = square` from `operation = square(5)`.

### Solution

`operation = square` binds a second name to the function object. `operation(5)` calls that function and returns `25`.

By contrast, `operation = square(5)` calls `square` immediately and binds `operation` to the returned integer. The first form works because Python functions are first-class objects.

#### Verification

````python
def square(x):
    return x ** 2

operation = square

print(operation)
print(operation(5))
print(operation is square)
````

### Key Takeaway

Referring to a function object and invoking that function are distinct operations.

## Problem 15 — Passing Behaviour as an Argument

### Problem

Define `square()`, `cube()`, and a higher-order `apply_operation(operation, values)` that applies the supplied function to every value and returns a list.

### Solution

`apply_operation()` accepts a callable as data and invokes it for each element. It is higher-order because one of its arguments is itself a function. The iteration structure stays fixed while the behaviour can vary.

#### Implementation

````python
def square(x):
    return x ** 2

def cube(x):
    return x ** 3

def apply_operation(operation, values):
    return [operation(value) for value in values]

apply_operation(square, [1, 2, 3, 4]), apply_operation(cube, [1, 2, 3, 4])
````

### Key Takeaway

Higher-order functions separate a stable processing structure from configurable behaviour.

## Problem 16 — Lambda Expressions

### Problem

Rewrite `squared_difference(x, y)` as a lambda, then sort records of `(name, score)` first ascending and then descending by score using a lambda key.

### Solution

The equivalent lambda is `lambda x, y: (x - y) ** 2`. For sorting, the key function should return the tuple's second element. The same key works in both directions; `reverse=True` controls descending order.

#### Implementation

````python
squared_difference = lambda x, y: (x - y) ** 2

records = [
    ("Alice", 82),
    ("Bob", 91),
    ("Carol", 76),
    ("David", 88),
]

ascending = sorted(records, key=lambda record: record[1])
descending = sorted(records, key=lambda record: record[1], reverse=True)

squared_difference(7, 3), ascending, descending
````

### Key Takeaway

Lambdas are best suited to short, local pieces of behaviour such as simple key functions.

## Problem 17 — `map()`, `filter()`, and Comprehensions

### Problem

From `[-3, -2, -1, 0, 1, 2, 3, 4]`, produce the squares of only the positive values using both `filter()`/`map()` and a list comprehension. Compare readability.

### Solution

The functional version first filters positive values and then maps the squaring operation over them. The comprehension expresses both steps in one compact construct and is generally easier to read in this case.

#### Comparison

````python
values = [-3, -2, -1, 0, 1, 2, 3, 4]

functional_result = list(
    map(lambda x: x ** 2, filter(lambda x: x > 0, values))
)

comprehension_result = [x ** 2 for x in values if x > 0]

functional_result, comprehension_result
````

### Key Takeaway

`map()` and `filter()` are valid higher-order tools, but comprehensions are often more direct for simple Python transformations.

## Problem 18 — Functions as Sorting Criteria

### Problem

Given model dictionaries containing `name` and `rmse`, use `min(..., key=...)` to select the lowest RMSE model and `sorted(..., key=...)` to order the models. Explain why whole dictionaries are returned.

### Solution

The `key` function computes a comparison value but does not replace the original element. `min()` therefore returns the original dictionary associated with the smallest RMSE, and `sorted()` returns the original dictionaries reordered by their keys.

#### Implementation

````python
models = [
    {"name": "model_a", "rmse": 3.4},
    {"name": "model_b", "rmse": 2.8},
    {"name": "model_c", "rmse": 3.1},
]

best_model = min(models, key=lambda model: model["rmse"])
ordered_models = sorted(models, key=lambda model: model["rmse"])

best_model, ordered_models
````

### Key Takeaway

A key function supplies a comparison criterion while preserving the original objects as the returned elements.

## Problem 19 — Docstrings and Type Hints

### Problem

Implement:

```python
def accuracy(predictions: list[int], targets: list[int]) -> float:
    ...
```

Add a suitable docstring and identify appropriate validation before calculation.

### Solution

The function should ensure that both sequences have equal length and are not empty. It can count element-wise matches with `zip()` and divide by the number of targets.

Type hints describe the intended interface; the docstring and validation define behavioural details that annotations alone do not express.

#### Implementation

````python
def accuracy(
    predictions: list[int],
    targets: list[int],
) -> float:
    """
    Return the proportion of predictions equal to their targets.

    Parameters
    ----------
    predictions : list of int
        Predicted class labels.
    targets : list of int
        True class labels.

    Returns
    -------
    float
        Proportion of correctly predicted labels.

    Raises
    ------
    ValueError
        If the inputs have different lengths or are empty.
    """
    if len(predictions) != len(targets):
        raise ValueError("Predictions and targets must have the same length.")
    if not targets:
        raise ValueError("Predictions and targets must not be empty.")

    correct = sum(
        prediction == target
        for prediction, target in zip(predictions, targets)
    )
    return correct / len(targets)

accuracy([1, 0, 1, 1], [1, 1, 1, 1])
````

### Key Takeaway

Type hints document intended types, while validation and docstrings define the broader function contract.

## Problem 20 — Pure Functions and Side Effects

### Problem

Compare a function that appends `x ** 2` to a global list with a function that simply returns `x ** 2`. Identify the pure function and the side effect.

### Solution

The function that appends to the global list changes external state and implicitly returns `None`; it has a side effect. The function that returns `x ** 2` depends only on its input and does not modify external state, so it is pure for ordinary numeric inputs.

The pure version is easier to test and compose because its result is explicit.

#### Comparison

````python
results = []

def square_a(x):
    results.append(x ** 2)

def square_b(x):
    return x ** 2

returned_a = square_a(5)
returned_b = square_b(5)

print(results)
print(returned_a)
print(returned_b)
````

### Key Takeaway

Pure functions make dependencies and outputs explicit; side effects introduce state that callers must track.

## Problem 21 — Recursion

### Problem

Write recursive and iterative versions of `sum_to_n()` for a positive integer `n`. Identify the base case, recursive case, and progression toward termination.

### Solution

For positive integers, `n == 1` is a natural base case. The recursive case returns `n + sum_to_n(n - 1)`, and every call decreases `n` by one.

The iterative version avoids recursive call overhead and stack growth. For this simple accumulation, iteration is generally more practical in Python.

#### Recursive and Iterative Implementations

````python
def sum_to_n(n):
    if n < 1:
        raise ValueError("n must be a positive integer.")
    if n == 1:
        return 1
    return n + sum_to_n(n - 1)

def sum_to_n_iterative(n):
    if n < 1:
        raise ValueError("n must be a positive integer.")
    total = 0
    for value in range(1, n + 1):
        total += value
    return total

sum_to_n(10), sum_to_n_iterative(10)
````

### Key Takeaway

Recursion needs a base case and a step toward it; in Python, iteration is often preferable for straightforward repeated accumulation.

## Problem 22 — Designing a Data Transformation Function

### Problem

Implement `standardize(values: list[float]) -> list[float]` using the sample mean and sample standard deviation. Preserve the input list, return a new list, use type hints and a docstring, and define behaviour for zero variance.

### Solution

`statistics.stdev()` computes the sample standard deviation required by the problem. A comprehension can construct a new standardized list without mutating the input.

Standardization is undefined when the standard deviation is zero. A sample standard deviation also requires at least two observations, so both conditions should be validated.

#### Implementation

````python
from statistics import mean, stdev

def standardize(values: list[float]) -> list[float]:
    """Return sample-standardized values without modifying the input."""
    if len(values) < 2:
        raise ValueError("At least two observations are required.")

    sample_mean = mean(values)
    sample_std = stdev(values)

    if sample_std == 0:
        raise ValueError("Cannot standardize values with zero standard deviation.")

    return [(value - sample_mean) / sample_std for value in values]

values = [12.0, 15.0, 18.0, 21.0, 24.0]
standardized = standardize(values)

print("Original:", values)
print("Standardized:", standardized)
print("Mean:", mean(standardized))
print("Sample standard deviation:", stdev(standardized))
````

### Key Takeaway

Good transformation functions expose clear inputs and outputs, preserve inputs unless mutation is intended, and define edge-case behaviour explicitly.

## Problem 23 — Refactoring a Machine Learning Workflow

### Problem

Critique a single `run_model()` function that loads data, cleans it, creates features, fits a model, predicts, evaluates, prints metrics, and saves the model. Propose a better decomposition and justify the boundaries.

### Solution

The original function mixes file I/O, data preparation, model computation, evaluation, display, and persistence. That hides intermediate data flow and makes stages harder to test or reuse independently.

A useful decomposition gives each stage one coherent responsibility. The orchestration function can then connect these stages explicitly. Computation-oriented functions should preferably return values, while I/O functions make side effects visible.

#### One Possible Decomposition

````python
def load_data(file_path):
    ...

def clean_data(data):
    ...

def create_features(data):
    ...

def fit_model(features, targets):
    ...

def predict(model, features):
    ...

def evaluate(targets, predictions):
    ...

def display_metrics(metrics):
    ...

def save_model(model, output_path):
    ...

def run_model(file_path, output_path):
    data = load_data(file_path)
    clean = clean_data(data)
    features, targets = create_features(clean)
    model = fit_model(features, targets)
    predictions = predict(model, features)
    metrics = evaluate(targets, predictions)

    display_metrics(metrics)
    save_model(model, output_path)

    return model, metrics
````

### Key Takeaway

Function decomposition should expose meaningful stages and data flow rather than merely divide code into shorter blocks.

## Problem 24 — Reasoning About an Unfamiliar Function

### Problem

Analyse:

```python
def process(values, transform=lambda x: x, condition=lambda x: True):
    return [
        transform(value)
        for value in values
        if condition(value)
    ]
```

Predict the results of the default call, a squaring transformation, and a squaring transformation restricted to even values. Explain which parameters represent data and behaviour and why this is a higher-order function.

### Solution

The three results are `[1, 2, 3, 4]`, `[1, 4, 9, 16]`, and `[4, 16]`.

`values` supplies the data. `transform` and `condition` supply behaviour. The default transform is the identity function, while the default condition accepts every value.

`process()` is higher-order because it accepts functions as arguments. This pattern is useful when a pipeline structure is fixed but filtering and transformation rules should remain configurable.

#### Verification

````python
def process(values, transform=lambda x: x, condition=lambda x: True):
    return [
        transform(value)
        for value in values
        if condition(value)
    ]

result_1 = process([1, 2, 3, 4])
result_2 = process([1, 2, 3, 4], transform=lambda x: x ** 2)
result_3 = process(
    [1, 2, 3, 4],
    transform=lambda x: x ** 2,
    condition=lambda x: x % 2 == 0,
)

result_1, result_2, result_3
````

### Key Takeaway

Higher-order functions can keep the processing structure fixed while receiving configurable behaviour as function arguments.

## Final Review

### Core Concepts

The problems in this notebook reinforce the core ideas developed in the technical notes:

- **Function interfaces:** parameters define expected inputs, arguments supply values, and return values expose results.
- **Argument binding:** positional arguments, keyword arguments, defaults, `*args`, and `**kwargs` provide different interface patterns.
- **Default evaluation:** mutable default objects can persist across calls.
- **Scope:** names are resolved according to the LEGB hierarchy.
- **Object references:** parameters are local names bound to caller-supplied objects; rebinding and mutation are different operations.
- **Mutability:** changes to shared mutable objects can be visible outside the function.
- **First-class functions:** functions can be assigned, passed, returned, and stored like other objects.
- **Higher-order functions:** behaviour itself can be supplied as input to an algorithm.
- **Lambda expressions:** anonymous functions are useful for short, local behaviour.
- **Comprehensions and functional tools:** `map()`, `filter()`, and comprehensions provide related transformation patterns.
- **Function contracts:** names, docstrings, type hints, validation, and predictable side effects clarify intended behaviour.
- **Pure functions and side effects:** explicit input-output transformations are generally easier to test and compose.
- **Recursion:** recursive solutions require a base case and progress toward it.
- **Function design:** useful decomposition follows coherent responsibilities and meaningful abstractions.

For technical interviews, the main objective is not merely to recall syntax. It is to explain precisely how arguments are bound, what names refer to, whether an operation rebinds or mutates an object, where side effects occur, and why a particular function interface is appropriate.

## Problem 1 — String literals

### Problem

Without running the code, determine the value and type of each expression:

```python
a = "42"
b = '42'
c = """42"""

a == b == c
type(a)
```

Explain whether the different quotation styles affect the resulting object.

### Solution

All three assignments create the same Python string, `"42"`. Single quotes, double quotes, and triple quotes are different **literal syntaxes**, but they do not create different string types.

Therefore:

- `a == b == c` evaluates to `True`;
- `type(a)` is `str`.

Triple quotes become especially useful when a literal spans multiple lines, but a single-line triple-quoted literal is still an ordinary `str`.

#### Verification

````python
a = "42"
b = '42'
c = """42"""

a == b == c, type(a)
````

### Key Takeaway

Quotation style affects how a string literal is written, not the type of object created.

## Problem 2 — Escape sequences

### Problem

Without running the code, determine the exact output:

```python

print(text)
```

Explain the role of each escape sequence.

### Solution

The escape sequence `\t` inserts a horizontal tab and `\n` inserts a newline. The output therefore contains two rows and tab-separated fields:

```text

```

The exact visual spacing of a tab depends on the display environment, but the logical characters are fixed: a tab between each pair of fields and a newline between the two records.

#### Verification

````python
text = "model\taccuracy\nlinear\t0.91"
print(text)
````

### Key Takeaway

Escape sequences are interpreted when the string literal is created; they represent characters inside the resulting string.

## Problem 3 — Raw strings

### Problem

Consider:

```python
path_1 = "C:\new_folder\test"
path_2 = r"C:\new_folder\test"
```

Without running the code:

1. Explain why the two strings do not represent the same sequence of characters.
2. Predict what happens when each is printed.
3. Explain what the `r` prefix changes and what it does not change.

### Solution

In `path_1`, the sequences `\n` and `\t` are interpreted as newline and tab characters. The value therefore does not contain the literal backslash followed by `n` or `t` at those positions.

In `path_2`, the raw-string prefix suppresses the normal interpretation of most backslash escape sequences, so the backslashes remain literal characters.

The prefix changes the **parsing of the literal**, not the resulting data type. Both values are instances of `str`.

#### Verification

````python
path_1 = "C:\new_folder\test"
path_2 = r"C:\new_folder\test"

print(path_1)
print("---")
print(path_2)
print(type(path_1), type(path_2))
````

### Key Takeaway

Raw strings are literal syntax for writing backslash-heavy text conveniently; they do not define a separate string type.

## Problem 4 — Unicode code points

### Problem

Consider:

```python
text = "café"
```

Answer the following:

1. What does `len(text)` return?
2. What does `ord("é")` represent?
3. Is a Unicode code point the same thing as a byte?
4. Why might `len(text)` differ from `len(text.encode("utf-8"))`?

### Solution

`len(text)` returns `4` because the string contains four Unicode characters.

`ord("é")` returns the Unicode code point assigned to the character `é`, namely the integer `233` (`U+00E9`).

A Unicode code point is an abstract numerical identifier for a character; it is not a byte. An encoding such as UTF-8 maps code points to one or more bytes.

ASCII characters require one byte in UTF-8, while `é` requires two. Consequently, the number of characters and the number of encoded bytes can differ.

#### Verification

````python
text = "café"

len(text), ord("é"), len(text.encode("utf-8")), text.encode("utf-8")
````

### Key Takeaway

Characters, Unicode code points, and encoded bytes are related concepts but are not interchangeable.

## Problem 6 — Indexing versus slicing

### Problem

Given:

```python

```

predict the result of:

```python

```

Why does one expression raise an exception while the other does not?

### Solution

`text[10]` raises an `IndexError` because direct indexing requests one specific element that does not exist.

## Problem 8 — Half-open intervals

### Problem

Python slices use the convention:

```text
[start:stop]
```

where `start` is included and `stop` is excluded.

Explain why this convention is useful. In particular, explain the relationship between:

```python
text[:k]
text[k:]
```

and the original string.

### Solution

Half-open intervals make adjacent slices fit together without overlap or gaps. For any valid split position `k`:

```python
text[:k] + text[k:] == text
```

The first slice contains exactly the first `k` characters, and the second starts at the same boundary.

This convention also makes the length of a unit-step slice easy to reason about: when bounds are in range, `text[a:b]` has length `b - a`.

#### Verification

````python
text = "machine_learning"
k = 7

text[:k], text[k:], text[:k] + text[k:] == text
````

### Key Takeaway

Half-open intervals simplify both length calculations and the composition of adjacent slices.

## Problem 9 — String immutability

### Key Takeaway

Immutability means that string transformations create new objects rather than changing characters in place.

## Problem 14 — `find()` versus `index()`



What is the behavioural difference between these methods, and when might one be preferable to the other?

### Solution

Both methods search for the first occurrence of a substring.

If the substring is absent:

- `.find()` returns `-1`;

Use `.find()` when absence is an expected condition that will be handled as a returned value. Use `.index()` when absence should be treated as exceptional.

`find()` encodes absence as `-1`, whereas `index()` signals it with an exception.

## Problem 15 — `strip()` is not substring removal

### Problem

Without running the code, reason about the result of:

```python

text.strip("w.com")
```

Why should `.strip()` not be interpreted as "remove this exact substring"?

Which type of string operation should be used when the intention is to remove an exact prefix or suffix?

### Solution

The argument to `.strip(chars)` is treated as a **set of removable characters**, not as a literal substring. Python repeatedly removes matching characters from both ends until it reaches a character not contained in that set.

Therefore the result can remove more than an expected exact prefix or suffix.

For exact removal, use `.removeprefix()` or `.removesuffix()`.

#### Verification

````python
text = "www.example.com"

text.strip("w.com"), text.removeprefix("www."), text.removesuffix(".com")
````

### Key Takeaway

`strip()` removes characters from the ends; `removeprefix()` and `removesuffix()` remove exact substrings.

### Key Takeaway

## Problem 24 — Why `isdigit()` is not a numeric parser

### Problem

Consider:

```python
values = [
    "123",
    "-123",
    "12.5",
    "1e3"
]
```

Explain why `.isdigit()` is insufficient for deciding whether every value can be interpreted numerically.

What general approach would be more reliable when the goal is to parse floating-point values?

### Solution

`.isdigit()` tests a Unicode character property. It does not implement the syntax of Python numeric literals.

Therefore signs, decimal points, and exponent markers cause it to return `False` even when `float()` can parse the string successfully.

A more reliable approach is to attempt the intended conversion and handle `ValueError`.

#### Verification

````python
values = ["123", "-123", "12.5", "1e3"]

def parse_float(value):
    try:
        return float(value)
    except ValueError:
        return None

[(value, value.isdigit(), parse_float(value)) for value in values]
````

### Key Takeaway

Validation should match the grammar of the value you intend to parse; character predicates are not general numeric parsers.

## Problem 30 — Incorrect decoding

### Problem

Suppose UTF-8 encoded bytes are decoded using an incompatible encoding.

What kinds of outcomes are possible?

Explain why an encoding problem is fundamentally different from an ordinary string-manipulation problem.

### Solution

Two broad outcomes are possible:

1. decoding may fail with `UnicodeDecodeError` because the byte sequence is invalid under the chosen encoding;
2. decoding may succeed but produce incorrect characters (often called *mojibake*) if the wrong encoding accepts those byte values.

The problem occurs at the boundary between bytes and text. Ordinary string methods operate **after** bytes have already been interpreted as Unicode characters, so they cannot reliably reconstruct the intended text when that interpretation was wrong.

#### Demonstration

````python
text = "café"
encoded = text.encode("utf-8")

utf8_decoded = encoded.decode("utf-8")
latin1_decoded = encoded.decode("latin-1")

utf8_decoded, latin1_decoded
````

### Key Takeaway

Encoding errors concern how bytes are interpreted as characters, not merely how existing characters are rearranged or transformed.

## Problem 40 — First non-repeated character

### Problem

Implement:

```python
def first_unique_character(text):
    ...
```

that returns the first character appearing exactly once in the string.

For example:

```python

```

should return the first character whose total frequency in the string is one.

If no such character exists, return `None`.

Discuss the time complexity of your solution.

### Solution

A good interview solution performs two passes:

1. count every character once;
2. scan the original string again and return the first character with frequency `1`.

Using a dictionary makes both passes linear on average, so the overall time complexity is **O(n)** and the auxiliary space complexity is **O(k)**, where `k` is the number of distinct characters.

#### Implementation

````python
def first_unique_character(text):
    frequencies = {}

    for char in text:
        frequencies[char] = frequencies.get(char, 0) + 1

    for char in text:
        if frequencies[char] == 1:
            return char

    return None


first_unique_character("statistics")
````

### Key Takeaway

A frequency table avoids repeatedly scanning the string and reduces the problem from a naive quadratic approach to linear time on average.

## Problem 42 — Anagram detection

### Problem

Implement:

```python
def are_anagrams(a, b):
    ...
```

that determines whether two strings are anagrams.

For this problem:

- ignore letter case;
- ignore whitespace.

For example:

```python

```

should return `True`.

Discuss at least two possible implementation strategies and compare their time complexity.

### Solution

First normalize each string by removing whitespace and applying `.casefold()`.

Two common strategies are:

1. **Sort the characters** and compare the sorted sequences. This is simple but costs O(n log n) time.
2. **Count character frequencies** and compare the mappings. This costs O(n) average time with O(k) extra space.

A frequency-table solution is asymptotically better for long strings.

#### Implementation

````python
def normalize_for_anagram(text):
    return "".join(text.split()).casefold()


def frequency_table(text):
    counts = {}
    for char in text:
        counts[char] = counts.get(char, 0) + 1
    return counts


def are_anagrams(a, b):
    a_normalized = normalize_for_anagram(a)
    b_normalized = normalize_for_anagram(b)

    return frequency_table(a_normalized) == frequency_table(b_normalized)


are_anagrams("Data", "A Tad")
````

## Problem 43 — Longest common prefix

### Problem

Implement:

```python
def longest_common_prefix(strings):
    ...
```

that returns the longest prefix shared by every string in a list.

For example:

```python
strings = [
    "machine",
    "matching",
    "material"
]
```

should return:

```text
ma
```

Define sensible behaviour for an empty input list.

### Solution

For an empty list, returning `""` is a sensible convention.

One straightforward strategy uses the first string as the candidate reference and checks each character position against every other string. Stop as soon as a string ends or a character differs.

If `m` is the length of the shortest relevant prefix and `n` is the number of strings, the worst-case time is O(nm).

#### Implementation

````python
def longest_common_prefix(strings):
    if not strings:
        return ""

    first = strings[0]

    for index, char in enumerate(first):
        for string in strings[1:]:
            if index >= len(string) or string[index] != char:
                return first[:index]

    return first


strings = ["machine", "matching", "material"]

longest_common_prefix(strings)
````

### Key Takeaway

Prefix problems are naturally solved by stopping at the first position where the shared invariant fails.

## Problem 44 — Run-length encoding

### Problem

Implement a simple run-length encoder:

```python
def run_length_encode(text):
    ...
```

For example:

```python
run_length_encode("aaabbccccdaa")
```

should produce a representation equivalent to:

```text
a3b2c4d1a2
```

Discuss the time and space complexity of your implementation.

### Solution

Scan the input once while tracking the current character and run length. When the character changes, append the completed run to an output list.

The algorithm is O(n) time. The output itself may require O(n) space in the worst case, for example when every character differs from its neighbours.

#### Implementation

````python
def run_length_encode(text):
    if not text:
        return ""

    encoded = []
    current = text[0]
    count = 1

    for char in text[1:]:
        if char == current:
            count += 1
        else:
            encoded.append(f"{current}{count}")
            current = char
            count = 1

    encoded.append(f"{current}{count}")

    return "".join(encoded)


run_length_encode("aaabbccccdaa")
````

### Key Takeaway

Track contiguous runs in one pass and accumulate pieces in a list before joining them.

## Problem 48 — Data leakage through text preprocessing

### Problem

Suppose you are building a machine-learning model and create a text preprocessing pipeline using the entire dataset before splitting it into training and test sets.

Which string-cleaning operations are inherently independent of the dataset, and which kinds of learned text preprocessing could introduce data leakage?

Explain the distinction.

### Solution

Deterministic transformations that operate independently on each individual value—such as `.strip()`, fixed case normalization, or replacing a known literal delimiter—do not learn information from the dataset distribution.

Leakage becomes a concern when preprocessing **learns state from the full dataset**. Examples include:

- building a vocabulary from all documents;
- selecting tokens by global frequency;
- learning TF-IDF document frequencies on all observations;
- choosing categories based on global frequency;
- learning imputation values from all rows;
- fitting embeddings or feature selectors using information that includes the test set.

The correct pattern is to split first, fit stateful transformations on the training data only, and then apply the fitted transformation to validation/test data.

### Key Takeaway

Leakage depends on whether preprocessing estimates information from data that should have remained unseen, not merely on whether text is being transformed.

## Problem 50 — Designing a robust parser

### Problem

You receive values such as:

```python
values = [
    " 1,250.50 ",
    "$750.00",
    "N/A",
    "",
    "  92.5% ",
    "-42.7"
]
```

You need to prepare the data for a machine-learning pipeline.

Describe how you would design a robust parsing strategy.

Your answer should address:

- whitespace;
- formatting characters;
- percentages;
- missing values;
- invalid values;
- type conversion;
- error handling;
- whether all values should necessarily be interpreted using the same rule.

Do not write the complete implementation. Focus on the reasoning and design decisions.

### Solution

A robust parser should begin with an explicit **data contract** rather than a sequence of ad-hoc replacements.

A sensible design would be:

1. **Normalize superficial whitespace** using `.strip()`.
2. **Identify missing-value markers** such as `""` and `"N/A"` before numerical conversion.
3. **Classify known formats** rather than applying every transformation to every value:
   - ordinary formatted numbers may allow thousands separators;
   - currency values may allow an expected currency symbol;
   - percentages require removal of `%` followed by division by `100`;
   - signed decimal values should be passed directly to the numeric parser.
4. **Remove only formatting characters that the schema permits**. For example, commas should not be removed blindly if they might be decimal separators in another locale.
5. **Attempt explicit conversion** with `float()` (or another type dictated by the schema) and catch `ValueError`.
6. **Distinguish missing from invalid data**. A known missing marker may become a missing value such as `None`/`NaN`; malformed text should normally be logged, counted, rejected, or handled according to a documented policy rather than silently coerced.
7. **Preserve provenance where useful**, for example retaining the raw value or recording parse failures for auditability.
8. **Fit downstream stateful preprocessing only on training data** if this parser feeds a machine-learning pipeline.

The values should not necessarily all follow one rule. A percentage and a currency amount represent different semantics even though both can ultimately become floating-point numbers.

A production implementation would ideally be schema-driven and tested against edge cases, rather than relying on a single permissive cleaning expression.

### Key Takeaway

Robust parsing is a data-contract problem: identify semantics first, normalize only known representations, convert explicitly, and make failures visible.

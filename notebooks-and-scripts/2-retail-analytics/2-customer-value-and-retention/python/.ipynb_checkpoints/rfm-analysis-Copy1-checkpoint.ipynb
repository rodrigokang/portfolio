## Overview

This notebook accompanies the project **Customer Value and Retention** and contains the computational implementation of the customer segmentation workflow developed using transactional data from the AdventureWorks database.

The objective is to identify meaningful customer groups based on purchasing behavior and to establish a quantitative foundation for the churn prediction and customer lifetime value (CLV) models developed later in the project.

The analysis begins with the construction of Recency, Frequency, and Monetary Value (RFM) features, followed by exploratory analysis, customer scoring, and customer segmentation using K-Means clustering. The number of segments is not fixed in advance; instead, it is evaluated using clustering diagnostics such as inertia, silhouette score, Calinski-Harabasz score, and Davies-Bouldin score.

In addition to customer segmentation, the notebook introduces a continuous intra-cluster scoring framework designed to rank customers within each segment according to their relative behavioral importance. This additional layer of analysis provides a practical mechanism for prioritization once customer groups have been identified.

The notebook emphasizes reproducibility, interpretability, and practical decision-making. The focus is not only on obtaining statistically sound segments, but also on understanding how those segments can support customer retention, marketing actions, and long-term value creation.

### Extract Transactional Data

The query below extracts order-level and product-level transaction data.

Each row represents one product line within a customer order. This level of detail is useful because it allows the analysis to aggregate customer behaviour in different ways, including total spend, number of orders, number of purchased items, and product category diversity.

```python
query = """
SELECT
    c.customerid AS customer_id,
    soh.salesorderid AS sales_order_id,
    soh.orderdate AS order_date,
    soh.duedate AS due_date,
    soh.shipdate AS ship_date,
    soh.status AS order_status,
    soh.territoryid AS territory_id,
    st.name AS territory_name,
    st.countryregioncode AS country_region_code,
    sod.salesorderdetailid AS sales_order_detail_id,
    sod.productid AS product_id,
    p.name AS product_name,
    pc.name AS product_category,
    ps.name AS product_subcategory,
    sod.orderqty AS order_quantity,
    sod.unitprice AS unit_price,
    sod.unitpricediscount AS unit_price_discount,
    (
        sod.orderqty
        * sod.unitprice
        * (1 - sod.unitpricediscount)
    ) AS line_total
FROM sales.customer AS c
INNER JOIN sales.salesorderheader AS soh
    ON c.customerid = soh.customerid
INNER JOIN sales.salesorderdetail AS sod
    ON soh.salesorderid = sod.salesorderid
INNER JOIN production.product AS p
    ON sod.productid = p.productid
LEFT JOIN production.productsubcategory AS ps
    ON p.productsubcategoryid = ps.productsubcategoryid
LEFT JOIN production.productcategory AS pc
    ON ps.productcategoryid = pc.productcategoryid
LEFT JOIN sales.salesterritory AS st
    ON soh.territoryid = st.territoryid;
"""

transactions = pd.read_sql(query, engine)
```

## RFM Feature Engineering

Customer segmentation begins with the construction of Recency, Frequency, and Monetary Value (RFM) variables.

These variables summarize customer purchasing behaviour from three complementary perspectives:

- Recency measures how recently a customer has placed an order.
- Frequency measures how often a customer purchases.
- Monetary Value measures the total amount spent by the customer.

Together, these variables provide a compact representation of customer engagement and value, making them a natural starting point for segmentation analysis.

### Reference Date

Recency is calculated relative to a fixed reference date corresponding to the most recent transaction observed in the dataset.

Using a common reference date ensures that recency values are comparable across all customers.

```python
reference_date = (
    transactions["order_date"].max()
    + pd.Timedelta(days=1)
)

reference_date
```

### Customer-Level Aggregation

The transactional dataset is aggregated at the customer level to construct the Recency, Frequency, and Monetary Value variables.

The resulting dataset contains one row per customer and forms the basis for all subsequent segmentation analyses.

```python
rfm = (
    transactions
    .groupby("customer_id")
    .agg(
        recency=(
            "order_date",
            lambda x: (
                reference_date - x.max()
            ).days
        ),
        frequency=(
            "sales_order_id",
            "nunique"
        ),
        monetary=(
            "line_total",
            "sum"
        )
    )
    .reset_index()
)

rfm.head()
```

```python
rfm.shape
```

### Distribution of RFM Variables

Before applying any transformation, it is useful to inspect the empirical distributions of the RFM variables.

Customer transaction data often exhibits strong asymmetry, particularly for Frequency and Monetary Value. A small proportion of customers tend to account for a large share of transactions and revenue, producing highly skewed distributions.

Understanding these patterns helps motivate the transformations applied later in the workflow.

### Distributional Characteristics

The empirical distributions of the RFM variables reveal substantial positive skewness, particularly for Frequency and Monetary Value.

This behaviour is common in customer transaction data. Most customers place relatively few orders and generate modest revenue, while a much smaller group accounts for a disproportionately large share of transactions and spending.

Such asymmetry can distort distance-based clustering methods because extreme observations may dominate the geometry of the feature space.

To mitigate this effect, a Box-Cox transformation is applied to each RFM variable before standardization and clustering.

```python
rfm_bc = rfm.copy()

rfm_bc["recency_bc"], lambda_recency = boxcox(
    rfm_bc["recency"] + 1
)

rfm_bc["frequency_bc"], lambda_frequency = boxcox(
    rfm_bc["frequency"] + 1
)

rfm_bc["monetary_bc"], lambda_monetary = boxcox(
    rfm_bc["monetary"] + 1
)
```

```python
boxcox_lambdas = pd.DataFrame({
    "Variable": [
        "Recency",
        "Frequency",
        "Monetary"
    ],
    "Lambda": [
        lambda_recency,
        lambda_frequency,
        lambda_monetary
    ]
})

boxcox_lambdas
```

```python
rfm_bc[
    [
        "recency_bc",
        "frequency_bc",
        "monetary_bc"
    ]
].skew()
```

```python
fig, ax = plt.subplots()

sns.histplot(
    rfm_bc["recency_bc"],
    bins=30,
    kde=True,
    ax=ax
)

ax.set_title("Recency Distribution After Box-Cox")

plt.show()
```

```python
fig, ax = plt.subplots()

sns.histplot(
    rfm_bc["frequency_bc"],
    bins=30,
    kde=True,
    ax=ax
)

ax.set_title("Frequency Distribution After Box-Cox")

plt.show()
```

```python
fig, ax = plt.subplots()

sns.histplot(
    rfm_bc["monetary_bc"],
    bins=30,
    kde=True,
    ax=ax
)

ax.set_title("Monetary Distribution After Box-Cox")

plt.show()
```

### Effect of the Box-Cox Transformation

The Box-Cox transformation substantially reduces the skewness of all three RFM variables.

The largest improvement is observed for Monetary Value, which exhibits extreme positive skewness in its original form. Recency also becomes approximately symmetric after transformation, while Frequency retains a modest degree of asymmetry due to its discrete nature.

Overall, the transformed variables provide a more balanced representation of customer behaviour and are better suited for distance-based clustering methods such as K-Means.

### Standardization

Although the Box-Cox transformation reduces skewness, the variables remain measured on different scales.

To ensure that Recency, Frequency, and Monetary Value contribute comparably to the clustering process, the transformed variables are standardized using z-score normalization.

```python
scaler = StandardScaler()

rfm_scaled = scaler.fit_transform(
    rfm_bc[
        [
            "recency_bc",
            "frequency_bc",
            "monetary_bc"
        ]
    ]
)
```

```python
rfm_scaled = pd.DataFrame(
    rfm_scaled,
    columns=[
        "recency_scaled",
        "frequency_scaled",
        "monetary_scaled"
    ]
)

rfm_scaled.head()
```

## Selection of the Number of Clusters

Unlike traditional RFM approaches that define customer groups using predefined rules, the objective here is to allow the data to determine the segmentation structure.

Several clustering validation metrics are evaluated across different values of $K$ in order to identify a suitable number of customer segments.

The analysis considers:

- Within-cluster inertia.
- Silhouette Score.
- Calinski-Harabasz Index.
- Davies-Bouldin Index.

Together, these metrics provide complementary perspectives on cluster quality.

```python
k_values = range(2, 11)

inertia_scores = []
silhouette_scores = []
calinski_scores = []
davies_scores = []
```

```python
for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels = model.fit_predict(rfm_scaled)

    inertia_scores.append(
        model.inertia_
    )

    silhouette_scores.append(
        silhouette_score(
            rfm_scaled,
            labels
        )
    )

    calinski_scores.append(
        calinski_harabasz_score(
            rfm_scaled,
            labels
        )
    )

    davies_scores.append(
        davies_bouldin_score(
            rfm_scaled,
            labels
        )
    )
```

```python
cluster_metrics = pd.DataFrame({
    "k": list(k_values),
    "inertia": inertia_scores,
    "silhouette": silhouette_scores,
    "calinski_harabasz": calinski_scores,
    "davies_bouldin": davies_scores
})

cluster_metrics
```

```python
fig, ax = plt.subplots()

ax.plot(
    cluster_metrics["k"],
    cluster_metrics["inertia"],
    marker="o"
)

ax.set_title("Elbow Method")
ax.set_xlabel("Number of Clusters")
ax.set_ylabel("Inertia")

plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(
    cluster_metrics["k"],
    cluster_metrics["silhouette"],
    marker="o"
)

ax.set_title("Silhouette Score")
ax.set_xlabel("Number of Clusters")
ax.set_ylabel("Score")

plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(
    cluster_metrics["k"],
    cluster_metrics["calinski_harabasz"],
    marker="o"
)

ax.set_title("Calinski-Harabasz Index")
ax.set_xlabel("Number of Clusters")
ax.set_ylabel("Score")

plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(
    cluster_metrics["k"],
    cluster_metrics["davies_bouldin"],
    marker="o"
)

ax.set_title("Davies-Bouldin Index")
ax.set_xlabel("Number of Clusters")
ax.set_ylabel("Score")

plt.show()
```

### Cluster Validation Summary

The final number of clusters is selected by jointly considering all validation metrics rather than relying on a single criterion.

This approach reduces the risk of over-interpreting local optima and provides a more robust segmentation structure.

### Candidate Cluster Solutions

The validation metrics suggest several plausible clustering solutions.

Rather than selecting a single value of $K$ immediately, two candidate solutions are explored in more detail.

This approach reflects a practical business-oriented workflow in which statistical metrics are combined with interpretability and operational usefulness before making a final decision.

```python
candidate_k = [7, 9]

candidate_models = {}

for k in candidate_k:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels = model.fit_predict(rfm_scaled)

    candidate_models[k] = {
        "model": model,
        "labels": labels
    }
```

### Final Cluster Selection

Both candidate solutions provide meaningful customer segmentation structures.

Although the K = 9 solution achieves slightly better clustering metrics, it also introduces smaller and more specialized customer groups.

The K = 7 solution offers a more balanced trade-off between statistical performance, interpretability, and operational usefulness.

For this reason, the remainder of the analysis is based on the seven-cluster solution.

```python
final_k = 7

kmeans = KMeans(
    n_clusters=final_k,
    random_state=42,
    n_init=20
)

rfm["cluster"] = kmeans.fit_predict(
    rfm_scaled
)

rfm.head()
```

## Cluster Profiling

The clustering algorithm identifies groups of customers with similar purchasing behaviours.

To interpret these segments from a business perspective, the clusters are profiled using the original RFM variables rather than the transformed and standardized values used during model training.

This step translates the mathematical segmentation into customer groups that can support marketing, retention, and customer value strategies.

```python
cluster_profile = (
    rfm
    .groupby("cluster")
    .agg(
        customers=(
            "customer_id",
            "count"
        ),
        recency_mean=(
            "recency",
            "mean"
        ),
        frequency_mean=(
            "frequency",
            "mean"
        ),
        monetary_mean=(
            "monetary",
            "mean"
        )
    )
    .round(2)
)

cluster_profile
```

```python
global_recency = rfm["recency"].mean()
global_frequency = rfm["frequency"].mean()
global_monetary = rfm["monetary"].mean()

global_recency, global_frequency, global_monetary
```

```python
cluster_profile["R"] = np.where(
    cluster_profile["recency_mean"] > global_recency,
    "↑",
    "↓"
)

cluster_profile["F"] = np.where(
    cluster_profile["frequency_mean"] > global_frequency,
    "↑",
    "↓"
)

cluster_profile["M"] = np.where(
    cluster_profile["monetary_mean"] > global_monetary,
    "↑",
    "↓"
)

cluster_profile
```

### Preliminary Cluster Interpretation

The relative position of each cluster across Recency, Frequency, and Monetary Value provides a simple framework for interpreting customer behaviour.

The arrows indicate whether the average value of each RFM metric within a cluster is above or below the corresponding global average.

For Recency, lower values indicate more recent purchases and therefore more favourable customer behaviour. Consequently:

- $R \downarrow$ indicates more recent customer activity.
- $R \uparrow$ indicates longer periods since the last purchase.
- $F \uparrow$ indicates higher purchasing frequency.
- $M \uparrow$ indicates higher customer spending.

The resulting profiles serve as a starting point for assigning business-oriented segment labels.

## Intra-Cluster Customer Ranking

Customer segmentation provides a high-level grouping of customers with similar purchasing behaviour.

However, customers belonging to the same segment may still exhibit substantial differences in recency, purchasing frequency, and monetary contribution.

To support prioritization within each segment, an intra-cluster ranking methodology is introduced.

The approach combines the standardized RFM variables into a continuous score computed separately for each cluster. This allows customers to be ranked relative to their peers while preserving the structure identified by the clustering algorithm.

The resulting score ranges between 0 and 1, where larger values indicate stronger customer performance within the corresponding segment.

### Weighting Strategy

The contribution of each RFM component is determined using its variance within the corresponding cluster.

Variables exhibiting little variability provide limited discriminatory power because most customers share similar values.

Conversely, variables with larger variability contribute more information for ranking customers within the cluster.

This approach automatically adapts the ranking model to the characteristics of each segment.

```python
rfm_rank = pd.concat(
    [
        rfm,
        rfm_scaled
    ],
    axis=1
)

rfm_rank.head()
```

```python
rfm_rank["score_recency"] = (
    -rfm_rank["recency_scaled"]
)

rfm_rank["score_frequency"] = (
    rfm_rank["frequency_scaled"]
)

rfm_rank["score_monetary"] = (
    rfm_rank["monetary_scaled"]
)
```

```python
def compute_cluster_score(df):

    variances = df[
        [
            "score_recency",
            "score_frequency",
            "score_monetary"
        ]
    ].var()

    weights = (
        variances /
        variances.sum()
    )

    score = (
        weights["score_recency"]
        * df["score_recency"]
        +
        weights["score_frequency"]
        * df["score_frequency"]
        +
        weights["score_monetary"]
        * df["score_monetary"]
    )

    scaler = MinMaxScaler()

    score = scaler.fit_transform(
        score.to_frame()
    ).flatten()

    return pd.Series(
        score,
        index=df.index
    )
```

```python
rfm_rank["rfm_score"] = (
    rfm_rank
    .groupby("cluster")
    .apply(compute_cluster_score)
    .reset_index(level=0, drop=True)
)
```

## Intra-Cluster Weight Analysis

The proposed ranking methodology assigns weights according to the variability of each RFM component within a cluster.

To better understand the behaviour of the scoring system, the resulting weights are analysed across customer segments.

This analysis reveals which dimensions contribute most to customer differentiation inside each segment.

```python
cluster_weights = []

for cluster_id, group in rfm_rank.groupby("cluster"):

    variances = group[
        [
            "score_recency",
            "score_frequency",
            "score_monetary"
        ]
    ].var()

    weights = variances / variances.sum()

    cluster_weights.append(
        {
            "cluster": cluster_id,
            "recency_weight": weights["score_recency"],
            "frequency_weight": weights["score_frequency"],
            "monetary_weight": weights["score_monetary"]
        }
    )

cluster_weights = pd.DataFrame(cluster_weights)

cluster_weights
```

```python
cluster_weights = cluster_weights.merge(
    cluster_profile[
        ["segment"]
    ],
    left_on="cluster",
    right_index=True
)

cluster_weights
```

```python
cluster_weights_plot = (
    cluster_weights
    .set_index("segment")
    [
        [
            "recency_weight",
            "frequency_weight",
            "monetary_weight"
        ]
    ]
)

cluster_weights_plot.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Intra-Cluster Variable Weights")
plt.ylabel("Weight")
plt.xlabel("Segment")

plt.xticks(rotation=45)

plt.show()
```

### Interpretation of Variable Weights

The proposed weighting scheme does not attempt to measure variable importance globally.

Instead, weights quantify the discriminatory power of each variable within a specific customer segment.

Variables exhibiting little variability inside a cluster contribute less to the final ranking because they provide limited information for differentiating customers belonging to that segment.

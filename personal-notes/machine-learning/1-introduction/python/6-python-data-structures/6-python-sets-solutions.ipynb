## Problem 1 — Creating a Set

### Problem

Given:

```python
values = [1, 2, 2, 3, 3, 3, 4]
```

construct a set from `values`.

What is the length of the resulting set?

Explain why it differs from `len(values)`.

### Solution

Converting the list to a set removes duplicate occurrences:

```python
unique = set(values)
```

The resulting set contains the distinct values

```python
{1, 2, 3, 4}
```

so its length is `4`.

The original list has length `7` because lists preserve repeated occurrences. Sets represent membership only, so each distinct element occurs at most once.

#### Verification

````python
values = [1, 2, 2, 3, 3, 3, 4]
unique = set(values)

print(unique)
print(len(unique))
print(len(values))
````

### Key Takeaway

Lists preserve occurrences; sets preserve only distinct membership.

## Problem 2 — Empty Set

### Problem

Predict:

```python
type({})
type(set())
```

Why does Python require `set()` to create an empty set?

### Solution

`{}` is reserved for the empty dictionary literal, so

```python
type({}) is dict
```

while

```python
type(set()) is set
```

An empty set therefore requires the constructor `set()` because the brace syntax is ambiguous at zero elements and Python assigns `{}` to dictionaries.

#### Verification

````python
print(type({}))
print(type(set()))
````

### Key Takeaway

Use `set()` for an empty set; `{}` creates an empty dictionary.

## Problem 3 — String Conversion

### Problem

Compare:

```python
{"python"}
```

with:

```python
set("python")
```

How many elements does each structure contain?

### Solution

`{"python"}` is a set containing one string, so its length is `1`.

`set("python")` iterates over the string and inserts its distinct characters. Since `"python"` contains the letters `p`, `y`, `t`, `h`, `o`, and `n`, its set has length `6`.

The difference is between storing one object and constructing a set from an iterable.

#### Verification

````python
a = {"python"}
b = set("python")

print(a, len(a))
print(b, len(b))
````

### Key Takeaway

A set literal stores its listed objects; `set(iterable)` consumes the iterable's elements.

## Problem 5 — Invalid Set Elements

### Problem

Which of the following can be inserted into a set?

```python
42
"model"
(1, 2)
[1, 2]
{1, 2}
frozenset({1, 2})
```

Explain every answer using hashability.

### Solution

The following are hashable and can be elements:

```text
42
"model"
(1, 2)
frozenset({1, 2})
```

The list and ordinary set are unhashable and cannot be elements:

```text
[1, 2]
{1, 2}
```

Set elements must be hashable because set membership is implemented through hash-based lookup.

#### Verification

````python
objects = [
    42,
    "model",
    (1, 2),
    [1, 2],
    {1, 2},
    frozenset({1, 2}),
]

for obj in objects:
    try:
        print(repr(obj), "-> hash:", hash(obj))
    except TypeError as exc:
        print(repr(obj), "->", type(exc).__name__)
````

### Key Takeaway

The precise requirement for set elements is hashability, not a memorised list of allowed types.

## Problem 7 — Subset versus Proper Subset

### Problem

Given:

```python
A = {1, 2}
B = {1, 2, 3}
C = {1, 2}
```

predict:

```python
A <= B
A < B
A <= C
A < C
```

Explain the role of equality.

### Solution

The results are:

```text
A <= B -> True
A < B  -> True
A <= C -> True
A < C  -> False
```

`<=` tests subset and allows equality. `<` tests proper subset and therefore requires strict containment.

#### Verification

````python
A = {1, 2}
B = {1, 2, 3}
C = {1, 2}

print(A <= B)
print(A < B)
print(A <= C)
print(A < C)
````

### Key Takeaway

Subset allows equality; proper subset excludes equality.

## Problem 9 — Disjointness

### Problem

Given:

```python
A = {1, 2, 3}
B = {4, 5, 6}
C = {3, 7, 8}
```

predict:

```python
A.isdisjoint(B)
A.isdisjoint(C)
```

Verify each result using intersection.

### Solution

`A` and `B` share no elements, so `A.isdisjoint(B)` is `True`.

`A` and `C` both contain `3`, so `A.isdisjoint(C)` is `False`.

This is equivalent to checking whether the intersection is empty.

#### Verification

````python
A = {1, 2, 3}
B = {4, 5, 6}
C = {3, 7, 8}

print(A.isdisjoint(B), A & B)
print(A.isdisjoint(C), A & C)
````

### Key Takeaway

Two sets are disjoint exactly when their intersection is empty.

## Problem 12 — Difference Is Directional

### Problem

```python

```

for:

```python

```

Explain why difference is not commutative.

### Solution

The results are:

```text
A - B -> {1, 2}
B - A -> {4, 5}
```

Difference keeps elements from the left operand that do not belong to the right operand. Reversing the operands changes the membership condition, so difference is not commutative.

#### Verification

````python
A = {1, 2, 3}
B = {3, 4, 5}

print(A - B)
print(B - A)
````

### Key Takeaway

Set difference is directional: left-only membership is different from right-only membership.

## Problem 13 — Symmetric Difference

### Problem

For the same sets, calculate:

```python
A ^ B
```

Then verify:

```python
(A ^ B) == ((A - B) | (B - A))
```

Explain the identity.

### Solution

The symmetric difference is

```python
{1, 2, 4, 5}
```

It contains elements belonging to exactly one of the two sets.

The identity

$$
A \triangle B = (A-B) \cup (B-A)
$$

holds because the two directional differences capture the elements unique to each side, and their union combines them.

#### Verification

````python
A = {1, 2, 3}
B = {3, 4, 5}

print(A ^ B)
print((A ^ B) == ((A - B) | (B - A)))
````

### Key Takeaway

Symmetric difference is the union of the two directional differences.

## Problem 20 — `remove()` versus `discard()`

### Problem

Starting independently from:

```python
values = {1, 2, 3}
```

compare:

```python
values.remove(4)
```

with:

```python
values.discard(4)
```

Explain the appropriate use case for each method.

### Solution

`remove(4)` raises `KeyError` because `4` is absent.

`discard(4)` leaves the set unchanged and raises no exception.

Use `remove()` when absence indicates an unexpected state. Use `discard()` when absence is acceptable and the goal is simply to ensure that the element is not present.

#### Verification

````python
values = {1, 2, 3}

try:
    values.remove(4)
except KeyError as exc:
    print("remove:", type(exc).__name__, exc)

values = {1, 2, 3}
values.discard(4)
print("discard:", values)
````

### Key Takeaway

`remove()` enforces presence; `discard()` tolerates absence.

## Problem 21 — `pop()`

### Problem

Given:

```python
values = {10, 20, 30}
```

execute:

```python
x = values.pop()
```

What can be guaranteed about `x`?

What cannot be guaranteed?

### Solution

It can be guaranteed that:

- `x` was an element of the set immediately before the call;
- `x` is removed from the set;
- the set becomes one element smaller.

It cannot be guaranteed which particular element is removed from the programmer's perspective. Set `pop()` should not be interpreted as removing a first, last, smallest, or largest element.

#### Verification

````python
values = {10, 20, 30}
before = values.copy()

x = values.pop()

print("removed:", x)
print("was member:", x in before)
print("remaining:", values)
````

### Key Takeaway

Set `pop()` removes an arbitrary member; code must not depend on a particular set iteration order.

## Problem 22 — `clear()` and Aliasing

### Problem

Predict:

```python
A = {1, 2, 3}
B = A

A.clear()

print(A)
print(B)
```

Then compare this with rebinding:

```python
A = {1, 2, 3}
B = A

A = set()
```

Explain the difference.

### Solution

With `clear()`, the shared set object is mutated, so both names observe `set()`.

With `A = set()`, only `A` is rebound to a new empty set. `B` still refers to the original `{1, 2, 3}`.

This is mutation versus rebinding.

#### Verification

````python
A = {1, 2, 3}
B = A
A.clear()
print(A, B, A is B)

A = {1, 2, 3}
B = A
A = set()
print(A, B, A is B)
````

### Key Takeaway

Aliases observe mutation of a shared object; rebinding affects only the name being rebound.

## Problem 23 — In-Place Intersection

### Problem



### Solution

`intersection_update()` mutates `A` so that

### Key Takeaway

In-place set algebra preserves object identity; ordinary set algebra returns a new set.

## Problem 28 — `frozenset`

## Problem 30 — Duplicate Results in a Set Comprehension

### Problem

Predict:

```python
{
    x ** 2
    for x in range(-3, 4)
}
```

Why does the result contain fewer than seven elements?

### Solution

The inputs are

```text
-3, -2, -1, 0, 1, 2, 3
```

but opposite values have equal squares:

```text
(-3)^2 = 3^2 = 9
(-2)^2 = 2^2 = 4
(-1)^2 = 1^2 = 1
```

The resulting set is therefore

```python
{0, 1, 4, 9}
```

with only four distinct results.

#### Verification

````python
result = {
    x ** 2
    for x in range(-3, 4)
}

print(result)
print(len(result))
````

### Key Takeaway

Set comprehensions evaluate every iteration but retain only distinct resulting values.

## Problem 36 — Maximum without `max()`

```python
iterator = iter(values)
```

## Problem 37 — Sieve of Eratosthenes

### Problem

Given an integer $n \geq 2$, construct the set of all prime numbers between `2` and `n` using the **Sieve of Eratosthenes**.

A possible strategy is:

1. initialise the candidate set with all integers from `2` through `n`;
2. repeatedly remove multiples of each prime candidate beginning at $p^2$;
3. stop eliminating multiples when

$$
p^2 > n.
$$

Explain why starting elimination at $p^2$ is sufficient.

### Solution

A set-based implementation can maintain the surviving prime candidates:

```python
candidates = set(range(2, n + 1))

p = 2

while p * p <= n:
    if p in candidates:
        candidates.difference_update(
            range(p * p, n + 1, p)
        )

    p += 1
```

At the end, `candidates` contains the primes.

Starting at $p^2$ is sufficient because every smaller multiple of $p$ has the form

$$
kp
$$

with $k < p$ and therefore already has a factor smaller than $p$. Such composites would have been eliminated when processing the smaller factor.

#### Verification

````python
def sieve(n):
    candidates = set(range(2, n + 1))
    p = 2

    while p * p <= n:
        if p in candidates:
            candidates.difference_update(
                range(p * p, n + 1, p)
            )
        p += 1

    return candidates

print(sorted(sieve(50)))
````

### Key Takeaway

The sieve combines set removal with the number-theoretic fact that composites below $p^2$ already have a smaller prime factor.

## Problem 38 — Removing Duplicates while Preserving Order

### Problem

Given:

```python
values = [
    "A",
    "B",
    "A",
    "C",
    "B",
    "D",
]
```

construct:

```python
["A", "B", "C", "D"]
```

using both

- a set for membership tracking;
- a list for preserving first-occurrence order.

Explain why:

```python
list(set(values))
```

does not express the same requirement.

### Solution

Use a set to track previously seen values and a list to preserve sequence order:

```python
seen = set()
unique = []

for value in values:
    if value not in seen:
        seen.add(value)
        unique.append(value)
```

`list(set(values))` removes duplicates, but it does not express the requirement to preserve first-occurrence order because set order is not sequence semantics.

#### Verification

````python
values = ["A", "B", "A", "C", "B", "D"]

seen = set()
unique = []

for value in values:
    if value not in seen:
        seen.add(value)
        unique.append(value)

unique
````

### Key Takeaway

Combine data structures when the problem requires both efficient membership tracking and stable sequence order.

## Problem 39 — Class-Set Comparison

### Problem

Given:

```python
y_true = [0, 0, 1, 1, 2, 2]
y_pred = [0, 1, 1, 2, 2, 3]
```

determine

1. the unique observed classes;
2. the unique predicted classes;
3. classes predicted but never observed;
4. classes observed but never predicted;
5. classes appearing in either collection but not both.

### Solution

Construct:

```python
true_classes = set(y_true)
predicted_classes = set(y_pred)
```

Then:

```text
observed classes               -> {0, 1, 2}
predicted classes              -> {0, 1, 2, 3}
predicted but never observed   -> predicted_classes - true_classes = {3}
observed but never predicted   -> true_classes - predicted_classes = set()
either but not both            -> true_classes ^ predicted_classes = {3}
```

#### Verification

````python
y_true = [0, 0, 1, 1, 2, 2]
y_pred = [0, 1, 1, 2, 2, 3]

true_classes = set(y_true)
predicted_classes = set(y_pred)

print(true_classes)
print(predicted_classes)
print(predicted_classes - true_classes)
print(true_classes - predicted_classes)
print(true_classes ^ predicted_classes)
````

### Key Takeaway

Set algebra provides compact diagnostics for comparing categorical membership across collections.

## Problem 40 — Reasoning about Mutation and References

### Problem

Without executing the code, determine the results:

```python
A = {1, 2, 3}
B = A
C = A.copy()

B.add(4)
C.remove(1)
A |= {5}

print(A)
print(B)
print(C)

print(A is B)
print(A is C)
print(A == B)
print(A == C)
```

Trace which operations mutate shared objects and which set is independent.

### Solution

Initially:

- `B` aliases `A`;
- `C` is an independent copy.

`B.add(4)` mutates the shared `A`/`B` object:

```python
{1, 2, 3, 4}
```

`C.remove(1)` affects only `C`:

```python
{2, 3}
```

`A |= {5}` mutates the shared `A`/`B` object in place:

```python
{1, 2, 3, 4, 5}
```

Therefore:

```text
A -> {1, 2, 3, 4, 5}
B -> {1, 2, 3, 4, 5}
C -> {2, 3}

A is B -> True
A is C -> False
A == B -> True
A == C -> False
```

#### Verification

````python
A = {1, 2, 3}
B = A
C = A.copy()

B.add(4)
C.remove(1)
A |= {5}

print(A)
print(B)
print(C)

print(A is B)
print(A is C)
print(A == B)
print(A == C)
````

### Key Takeaway

Track object identity separately from set equality: aliases share mutations, while copies evolve independently.

### 13.1.1. Scatter Plot

To create a scatter plot, we use the `.scatter()` method from the `plt` module.

Some of the parameters of this method are:

* `x`: scalar, array, or list indicating the first coordinate of the observations
* `y`: scalar, array, or list indicating the second coordinate of the observations
* `c`: to change the fill color
* `edgecolors`: to change the edge color
* `alpha`: to change the transparency
* `marker`: to change the [shape of the point](https://matplotlib.org/3.2.1/api/markers_api.html#module-matplotlib.markers)
* `s`: to change the size of the points (measured in points)
* `linewidths`: to change the edge thickness

```python
height = [174.3, 153.5, 162.1, 157.9, 174.8, 169.2, 172.0, 160.4, 152.8, 163.3] 
weight = [ 65.7,  59.2,  57.3,  61.5,  77.3,  90.7,  85.4,  63.1,  73.2, 105.5]

plt.title("Heights vs. Weights")
plt.xlabel("Heights (cm)")
plt.ylabel("Weights (kg)")
plt.scatter(x=height, y=weight, 
            c="black", edgecolors="red", alpha=1,
            marker="X", s=200, linewidths=2)
plt.show()
```

We could display 2 different sets of data on a scatter plot:

```python
height_boys = [174.3, 153.5, 162.1, 157.9, 174.8, 169.2, 172.0, 160.4, 152.8, 163.3] 
weight_boys = [65.7, 59.2, 57.3, 61.5, 77.3, 90.7, 85.4, 63.1, 73.2, 105.5]

height_girls = [154.2, 156.1, 160.3, 157.9, 162.6, 153.9, 170.1, 165.2, 157.6, 163.5] 
weight_girls = [55.3, 54.2, 47.3, 62.4, 77.5, 60.3, 52.4, 58.1, 50.2, 80.1]

plt.title("Heights vs. Weights")
plt.xlabel("Heights (cm)")
plt.ylabel("Weights (kg)")

plt.scatter(x=height_boys, y=weight_boys, c="red", marker="^", s=100)
plt.scatter(x=height_girls, y=weight_girls, c="blue", marker="v", s=100)

plt.show()
```

We can draw each point with a different size by passing a list to the `s` parameter:

```python
x = [1, 2, 3, 4, 5, 6, 7]
sizes = [100, 150, 200, 250, 300, 350, 400]

plt.scatter(x=x, y=x, s=sizes, c="m", alpha=0.5)
plt.title("Different Sizes")
plt.show()
```

We can paint each point with a different color by passing a list of colors to the `c` parameter:

```python
x = [1, 2, 3, 4, 5, 6, 7]
colors = ["red", "orange", "yellow", "green", "cyan", "blue", "purple"]
sizes = [100, 150, 200, 250, 300, 350, 400]

plt.scatter(x=x, y=x, c=colors, s=sizes)
plt.title("Rainbow")
plt.show()
```

We could have also used a colormap to color the points.

To do this, instead of a list of colors, we need to specify the colormap to use in the `cmap` parameter and, in the `c` parameter, indicate a 1D list or array of values between 0 and 100 to indicate the color of each point for the specified colormap.

```python
colors = np.array([0, 10, 20, 30, 40, 50, 60])

plt.scatter(x=x, y=x, c=colors, cmap="plasma", s=200)
plt.title("Plasma Color Map")
plt.show()
```

We can attach the color map used to the plot using the `.colorbar()` method.

```python
colors = np.array([0, 10, 20, 30, 40, 50, 60])

plt.scatter(x=x, y=x, c=colors, cmap="plasma", s=200)
plt.title("Plasma Color Map")
plt.colorbar()
plt.show()
```

```python
colors = np.array([0, 10, 20, 30, 40, 50, 60])

plt.scatter(x = x, y = x, c = colors, cmap = "plasma_r", s = 200)
plt.title("Mapa de color plasma")
plt.colorbar()
plt.show()
```

We can invert the color map by adding `_r` to the color map name.

### 13.1.2. Line Plot

To create a line plot, we use the `.plot()` method from the `plt` module.

Some of the parameters of this method are:

* `x`: scalar, array, or list indicating the first coordinate of the observations.
* `y`: scalar, array, or list indicating the second coordinate of the observations.
* `color`: to change the fill color. This parameter can also be referred to by its abbreviation `c`.
* `fmt`: to quickly set a basic format as a string. For example, `"or"` represents red circles. The recommended order to input the format in this way is `"[marker][linestyle][color]"`, although `"[color][marker][linestyle]"` is also accepted.
* `linewidth`: to change the line thickness. This parameter can also be referred to by its abbreviation `lw`.
* `linestyle`: to change the line style. This parameter can also be referred to by its abbreviation `ls`.
* `alpha`: to change the transparency.
* `marker`: to change the marker shape. If this parameter is not indicated, the points are not drawn.
* `markersize`: to change the size of the points. This parameter can also be referred to by its abbreviation `ms`.
* `markeredgecolor`: to change the color of the point outline. This parameter can also be referred to by its abbreviation `mec`.
* `markerfacecolor`: to change the fill color of the point. This parameter can also be referred to by its abbreviation `mfc`.

| `"v"` | down triangle | `":"` or `"dotted"` | dotted | `"c"` or `"cyan"` | cyan |
| `"^"` | up triangle | `"None"` | no line | `"m"` or `"magenta"` | magenta |
| `"<"` | left triangle | | | `"y"` or `"yellow"` | yellow | 
| `">"` | right triangle | | | `"k"` or `"black"` | black | 
| `"1"` | down triangle | | | `"w"` or `"white"` | white |
| `"2"` | up triangle | | | | |
| `"3"` | left triangle | | | | | 
| `"4"` | right triangle | | | | | 
| `"s"` | square | | | | | 
| `"p"` | pentagon | | | | |

| `"h"` | hexagon 1 | | | | |

| `"D"` | diamond | | | | |

```python
plt.title("Heights vs. Weights")
plt.xlabel("Heights (cm)")
plt.ylabel("Weights (kg)")
plt.plot(sorted(height), weight, 
            c = "blue", linestyle = "--", linewidth = 2,
            marker = "*", markersize = 20, markerfacecolor = "m")
plt.show()
```

We could achieve the same result using the `fmt` parameter.

```python
plt.title("Heights vs. Weights")
plt.xlabel("Heights (cm)")
plt.ylabel("Weights (kg)")
plt.plot(sorted(height), weight, "*--b", lw=2,
         markersize=20, markerfacecolor="m")
plt.show()
```

If we don't specify the `x` parameter, the first coordinates are considered as the integers 0, 1, 2, ..., $n-1$, where $n$ is the total number of observations:

```python
rainy_days = [20, 15, 19, 25, 13, 8, 3, 5, 10, 16, 20, 18]

plt.title("Rainy Days per Month")
# (0 = January, 1 = February, ..., 11 = December)
plt.xlabel("Month")
plt.ylabel("Total Rainy Days")
plt.plot(rainy_days, c = "cyan", marker = "o")
plt.show()
```

In the same plot, we can draw more than one line:

```python
l1 = np.array([3, 5, 1, 9])
l2 = np.array([8, 2, 6, 4])

plt.plot(l1, c = "blue")
plt.plot(l2, c = "magenta")
plt.title("2 lines in 1 plot")
plt.show()
```

### 13.1.3. Bar Plot

To make a bar plot, we use the `.bar()` method from the `plt` module.

Some of the parameters of this method are:

- `x`: float, array, or list indicating the categories or coordinates of the bars
- `height`: float, array, or list indicating the heights of the bars
- `width`: to change the width of the bars
- `linewidth`: to change the size of the edge of the bars
- `bottom`: to change the minimum vertical value of the bars
- `align`: position of the horizontal axis tick with respect to the base of the bars. By default, it is "center"
- `color`: to change the colors of the bars. If we pass a list, we can assign a different color to each bar
- `edgecolor`: to change the color of the edge of the bars

```python
grades = ["Fail", "Pass", "Good", "Excellent"]
count = [35, 55, 23, 7]

plt.bar(grades, count, color=["#ff7f7d", "#ffc07d", "#817dff", "#7dff86"])
plt.xlabel("Grades")
plt.ylabel("Number of Students")
plt.title("Grades of Students in a Class")
plt.show()
```

```python
plt.bar(grades, count, color=["#ff7f7d", "#ffc07d", "#817dff", "#7dff86"], width=0.5)
plt.xlabel("Grades")
plt.ylabel("Number of Students")
plt.title("Grades of Students in a Class")
plt.show()
```

```python
plt.barh(grades, count, color=["#ff7f7d", "#ffc07d", "#817dff", "#7dff86"])
plt.xlabel("Grades")
plt.ylabel("Number of Students")
plt.title("Grades of Students in a Class")
plt.show()
```

```python
plt.barh(grades, count, color=["#ff7f7d", "#ffc07d", "#817dff", "#7dff86"], height=0.5)
plt.xlabel("Grades")
plt.ylabel("Number of Students")
plt.title("Grades of Students in a Class")
plt.show()
```

### 13.1.4. Pie Chart

To create a pie chart, we use the `.pie()` method from the `plt` module.

Some of the parameters for this method are:

- `x`: 1D array or list
- `labels`: to specify the labels
- `colors`: to change the colors of the sectors. If we pass a list, we can assign a different color to each circular sector
- `autopct`: to edit the format in which the percentage is displayed
- `labeldistance`: to adjust the radial distance of the labels
- `radius`: to modify the radius of the chart
- `startangle`: to change the angle (in degrees) at which it starts (default is 0)
- `explode`: vector of distances to indicate how separated we want the corresponding sector to be from the rest.
- `shadow`: to add shadow to the sectors

```python
options = ["With", "Without"]
count = [20, 80] 

plt.pie(count, colors=["#ff85b0", "#85ffac"], labels=options, autopct="%0.3f%%")
plt.title("Pizza with or without pineapple?")
plt.show()
```

```python
plt.pie(count, colors=["#ff85b0", "#85ffac"], labels=options, 
        startangle=90, explode=[0.2, 0], shadow=True)
plt.title("Pizza with or without pineapple?")
plt.legend(title="Options")
plt.show()
```

### 13.2.1. Creating `datetime` Objects

To create `datetime` objects, we can use the `.datetime()` method.

This method takes parameters such as year (`year`), month (`month`), day (`day`), hour (`hour`), minute (`minute`), second (`second`), and microsecond (`microsecond`). The first three, related to the date, are mandatory parameters. The remaining parameters, related to time, are optional.

```python
d = dt.datetime(2020, 9, 22, 12, 30)
print(d)
```

**Observation:** The `.datetime()` method takes an additional parameter, `tzinfo`, which is also optional. This parameter is supplied with the time zone. By default, it takes the value `None`.

The `datetime` module offers the `.strftime()` method to change the format in which we display the date and time. In the `format` parameter of this method, we can specify one of the following options:

| Code | Description |
| :---: | :--- |
| `%Y` | Full year |
| `%y` | Abbreviated year (without century) |
| `%B` | Full month name (English language) |
| `%b` | Abbreviated month name (English language) |
| `%m` | Numeric month |
| `%d` | Day of the month |
| `%j` | Day of the year, 001-366 |
| `%A` | Full weekday name (English language) |
| `%a` | Abbreviated weekday name (English language)|
| `%w` | Numeric representation of the weekday (0 = Sunday, 1 = Monday, ..., 6 = Saturday) |
| `%W` | Week number of the year, with Monday as the first day |
| `%U` | Week number of the year, with Sunday as the first day |
| `%H` | Hour (00-23) |
| `%I` | Hour (00-12) |
| `%p` | AM / PM |
| `%M` | Minute (00-59) |
| `%S` | Second (00-59) |
| `%f` | Microsecond (000000-999999) |
| `%Z` | Time zone |
| `%z` | UTC offset |
| `%c` | Local version of date and time |
| `%x` | Local version of date |
| `%X` | Local version of time |
| `%%` | Percent sign `%` |
| `%G` | ISO 8601 year |
| `%u` | ISO 8601 weekday |
| `%V` | ISO 8601 week number |

**Observation:** When we refer to UTC offset, we mean the number of hours by which the time zone is ahead or behind Coordinated Universal Time (UTC).

```python
d = dt.datetime(2020, 9, 22, 12, 30)
```

```python
print(d.strftime("%B %d, %Y"))
```

```python
print(d.strftime("%d/%m/%Y"))
```

```python
print(d.strftime("%d %b %y"))
```

```python
print(d.strftime("%d-%m-%Y %H:%M:%S"))
```

```python
print(d.strftime("%d %b %Y, %I:%M%p"))
```

When talking about date formats, it's important to be familiar with the ISO 8601 concept, as listed in the table above. It is an international standard for the representation of dates and times.

Python provides a method that quickly generates a date formatted with ISO 8601: the `.isoformat()` method.

```python
d.isoformat()
```

It's time to talk about the opposite method to `.strftime()`. This is the `.strptime()` method, which allows us to create a `datetime` object from a string. However, since the string can have any format, we need to specify to `.strptime()` which format we are using through its parameter.

```python
dt.datetime.strptime("September 22, 2020", "%B %d, %Y")
```

```python
dt.datetime.strptime("22/09/2020", "%d/%m/%Y")
```

```python
dt.datetime.strptime("22 Sep 20", "%d %b %y")
```

```python
dt.datetime.strptime("22-09-2020 12:30:00", "%d-%m-%Y %H:%M:%S")
```

```python
dt.datetime.strptime("22 Sep 2020, 12:30PM", "%d %b %Y, %I:%M%p")
```

**Observation:** The pattern must match perfectly. This includes commas, colons, and whitespace.

### 13.2.2. Time Zones

[Time zones](https://www.timeanddate.com/time/map/) are a bit more complex than what we've seen so far about `datetime` objects.

By default, we've already seen that a `datetime` object doesn't contain information about the time zone.

```python
print(today)
```

To specifically access information about the time zone of a `datetime` object, we can use the `.tzinfo` attribute.

```python
print(today.tzinfo)
```

To specify the time zone or UTC offset for a `datetime` object, we can use the `pytz` module.

```python
import pytz
```

We can obtain all time zones with the `.all_timezones` attribute.

```python
pytz.all_timezones
```

The `pytz` module also has the `.country_names` attribute, which is a dictionary that contains ISO Alpha-2 country codes as keys, and their corresponding full names as values.

```python
for key, value in pytz.country_names.items():
    print(key, "=", value)
```

For example, USA corresponds to the pair of letters US:

```python
pytz.country_names["US"]
```

Then there is also the `.country_timezones` attribute, which is also a dictionary that has the countries in ISO Alpha-2 code as keys and the time zones for each country key as values.

```python
for key, value in pytz.country_timezones.items():
    print(key, "=", value)
```

For example, the available time zones for USA (US) are:

```python
print(pytz.country_timezones["US"])
```

To specify a time zone for a `datetime` object, we use the `.localize` method.

```python
d_without_tz = dt.datetime(2020, 9, 22, 12, 30)
tz_usa = pytz.timezone("America/New_York")
d_with_tz = tz_usa.localize(d_without_tz)
d_with_tz.strftime("%Z")
```

## 13.4.  Multiple Plots

With the `.subplot()` method we can draw multiple plots in the same figure.

This method takes 3 parameters that describe the layout of the figure. The layout is organized in rows and columns, which are represented by the first two parameters. The third parameter represents the index occupied by the current plot.

```python
# plot 1
fig, (ax1, ax2) = plt.subplots(2, 1)
l1 = np.array([3, 5, 1, 9])

ax1.plot(l1, c="magenta", ls=":", lw=2)

# plot 2
l2 = np.array([8, 2, 6, 4])

ax2.plot(l2, c="blue", ls="--", lw=2)

plt.show()
```

**Note:** In the previous example, `plt.subplot(1, 2, 1)` indicates that the layout has one row, two columns, and that the subsequent plot will occupy the first subplot. Meanwhile, `plt.subplot(1, 2, 2)` indicates again that the layout consists of one row and two columns, but in this case the subsequent plot will be drawn in the second subplot.

If, on the other hand, we want a layout of the figure consisting of 2 rows and 1 column, we would do the following:

```python
# plot 1
plt.subplot(2, 1, 1)
plt.plot(l1, c = "magenta", ls = ":", lw = 2)

# plot 2
plt.subplot(2, 1, 2)
plt.plot(l2, c = "blue", ls = "--", lw = 2)

plt.show()
```

We can draw as many subplots as we want within a figure. We always need to specify the number of rows, the number of columns, and the index that each plot occupies.

In this case, the indices of the plots range from 1 to \( n \), where \( n \) = rows \(\times\) columns of the layout. The first index corresponds to the cell of the first row and first column; the second, to the first row and second column; the third corresponds to the cell of the first row and third column; and so on until completing the first row. Then, we continue with the second row and so on until all rows are completed.

```python
# plot 1 (first row, first column)
plt.subplot(2, 4, 1)
plt.plot(l1, c = "blue", ls = "-", lw = 2)

# plot 2 (first row, second column)
plt.subplot(2, 4, 2)
plt.plot(l2, c = "magenta", ls = "--", lw = 2)

# plot 3 (first row, third column)
plt.subplot(2, 4, 3)
plt.plot(l1, c = "black", ls = ":", lw = 2)

# plot 4 (first row, fourth column)
plt.subplot(2, 4, 4)
plt.plot(l2, c = "cyan", ls = "-.", lw = 2)

# plot 5 (second row, first column)
plt.subplot(2, 4, 5)
plt.plot(l2, c = "black", ls = "--", lw = 2)

# plot 6 (second row, second column)
plt.subplot(2, 4, 6)
plt.plot(l1, c = "cyan", ls = "-.", lw = 2)

# plot 7 (second row, third column)
plt.subplot(2, 4, 7)
plt.plot(l2, c = "blue", ls = "-", lw = 2)

# plot 8 (second row, fourth column)
plt.subplot(2, 4, 8)
plt.plot(l1, c = "magenta", ls = ":", lw = 2)

plt.show()
```

We can add a title to each subplot using the `.title()` method.

```python
# plot 1
l1 = np.array([3, 5, 1, 9])

plt.subplot(1, 2, 1)
plt.title("Subplot 1")
plt.plot(l1, c = "magenta", ls = ":", lw = 2)

# plot 2
l2 = np.array([8, 2, 6, 4])

plt.subplot(1, 2, 2)
plt.title("Subplot 2")
plt.plot(l2, c = "blue", ls = "--", lw = 2)

plt.show()
```

We can add a title to the figure using the `.suptitle()` method.

```python
# plot 1
plt.subplot(1, 2, 1)
plt.title("Subplot 1")
plt.xlabel("X-axis 1")
plt.ylabel("Y-axis 1")
plt.plot(l1, c = "magenta", ls = ":", lw = 2)

# plot 2
plt.subplot(1, 2, 2)
plt.title("Subplot 2")
plt.xlabel("X-axis 2")
plt.ylabel("Y-axis 2")
plt.plot(l2, c = "blue", ls = "--", lw = 2)

plt.suptitle("2 Subplots in one figure")

plt.show()
```

With the `add_subplot()` method, we can also draw multiple plots in the same figure.

To do this, we have to follow these steps:

* Create the `figure` object with the `.figure()` method
* Indicate the number of rows and columns we want with the `.add_gridspec()` method
* Create as many subplots as we want to display with the `.add_subplot()` method and indicate the positions each of them will take

```python
# We create the figure object
fig = plt.figure(constrained_layout = True)

# We set up with 3 rows and 4 columns
gs = fig.add_gridspec(3, 4)

# We create the subplots. In this case 3
# * It occupies all rows and all columns except the last one
fig_ax1 = fig.add_subplot(gs[:, :-1])
fig_ax1.set_title("gs[:, :-1]")
# * It occupies all rows except the last one and the last column
fig_ax2 = fig.add_subplot(gs[:-1, -1])
fig_ax2.set_title("gs[:-1, -1]")
# * It occupies the last row and the last column
fig_ax3 = fig.add_subplot(gs[-1, -1])
fig_ax3.set_title("gs[-1, -1]")

plt.show()
```

## 13.5. Graphs with `networkx`

The `networkx` module allows us to represent graphs.

**Graph:** A collection of objects called nodes connected by links called edges, which allow us to represent binary relationships between elements of a set.

```python
import networkx as nx
```

To create an empty `Graph` object, we use the `.Graph()` method.

```python
G = nx.Graph()
print(type(G))
```

### 13.5.1. Adding Nodes

We can add nodes to the graph G using the `.add_node()` method.

```python
G.add_node(1)
```

With the previous instruction, we have added node 1 to the graph G.

There are other ways to add nodes. For example, from an iterable object, such as a list, using the `.add_nodes_from()` method.

```python
G.add_nodes_from([2, 3])
```

With the previous instruction, we have added nodes 2 and 3 to the graph G. Therefore, our graph G now has 3 nodes: 1, 2, and 3.

If instead of associating nodes with numbers, we want to associate them with letters, we can use both `.add_node()` and `.add_nodes_from()` and provide strings. However, we must be careful because `.add_node("abcd")` adds a single node identified by `abcd`, while `.add_nodes_from("abcd")` will add 4 different nodes: `a`, `b`, `c`, and `d`.

Both the `.add_node()` and `.add_nodes_from()` methods can be used to add nodes from another existing graph, H.

If we use `.add_node(H)`, we will be adding the graph H as a node to G, creating a graph of graphs. Conversely, if we use `.add_nodes_from(H)`, we would be adding the nodes of H to G.

We can view the nodes that make up our graph G using the `.nodes` attribute.

```python
print(G.nodes)
```

We can calculate the total number of nodes in our graph G using the `.number_of_nodes()` method.

```python
G.number_of_nodes()
```

### 13.5.2. Adding Edges

We can add edges between nodes using the `.add_edge()` method. Remember that $G$ now consists of 3 nodes: 1, 2, and 3. If we wanted an edge between nodes 1 and 2, we would do

```python
G.add_edge(1, 2)
```

Alternatively, we could first create the edge as a tuple and then supply it to the `.add_edge()` method as follows

```python
e = (1, 2)
G.add_edge(*e) # We use the asterisk to unpack the tuple
```

As much as we want to add the same node or edge repeatedly, `networkx` ignores it and adds each node or edge only once.

There is also the possibility of adding edges from a list of tuples with the `.add_edges_from()` method

```python
e = [(1, 2), (2, 3)]
G.add_edges_from(e)
```

We can see the edges that make up our graph $G$ with the `.edges()` method

```python
print(G.edges)
```

If we supply the `.edges()` method by parameter with a node or a list of nodes present in the graph, it will return a list of edges in which the nodes by parameters are present.

```python
print(G.edges([1,3]))
```

We can calculate the total number of edges of our graph $G$ with the `.number_of_edges()` method

```python
G.number_of_edges()
```

The `.adj[]` method is used to calculate the neighbors of a node. Two nodes are said to be neighbors when they are joined by an edge.

```python
list(G.adj[1])
```

```python
list(G.adj[2])
```

```python
list(G.adj[3])
```

The `.degree()` method returns the degree of a node or a list of nodes. A node has degree $n$ if it has $n$ incident edges.

```python
list(G.degree([1, 2, 3]))
```

In the previous result, the first entry of each tuple is the node and the second is its degree.

Finally, we could add weight to the edges with the `.add_edge()` methods next to the `weight` or `.add_weighted_edges_from()` attribute

```python
G.add_edge(1, 3, weight = 5)
```

```python
G.add_weighted_edges_from([(1, 2, 10), (2, 3, 7.5)])
```

With the last code cells, we have created a new edge between nodes 1 and 3, whose weight is 5 and then we have added weight to the two existing edges: edge (1, 2) now has a weight of 10 and edge (2, 3) now has a weight of 7.5.

We can access the weight of an edge with the syntax

```python
G.edges[2, 3]["weight"]
```

### 13.5.3. Removing nodes and edges

We can remove nodes from a graph $G$ with the `.remove_node()` or `.remove_nodes_from()` methods.

```python
G.remove_node(2)
print(G.nodes)
```

```python
G.remove_nodes_from([1, 3])
print(G.nodes)
```

With the two previous cells, we have left the graph without nodes and, therefore, without edges.

```python
print(G.edges)
```

Let's build the graph $G$ again. This time it will have 5 nodes (`a`, `b`, `c`, `d` and `e`) and 7 edges.

```python
G.add_nodes_from("abcde")
G.add_edges_from([("a", "b"), ("a", "c"), ("a", "d"), ("a", "e"),
                  ("b", "e"), ("c", "d"), ("d", "e")])
print(G.nodes)
print(G.edges)
```

We can remove edges from a graph $G$ with the `.remove_edge()` or `.remove_edges_from()` methods.

```python
G.remove_edge("b", "e")
print(G.edges)
```

```python
G.remove_edges_from([("c", "d"), ("d", "e")])
print(G.edges)
```

We could remove all the edges with the `.clear_edges()` method

```python
G.clear_edges()
print(G.nodes)
print(G.edges)
```

The `.clear()` method removes all nodes and edges from a graph

```python
G.clear()
print(G.nodes)
print(G.edges)
```

Make a graph of the surface $f(x,y) = \sin(x) \cdot \cos(y)$ for $-5 \leq x \leq 5$, $-5 \leq y \leq 5$ using the `plot_surface` function. Be careful to use a fine enough discretization on `x` and `y` so that the graph looks smooth.

```python
fig = plt.figure(figsize = (12,10))
ax = plt.axes(projection="3d")

x = np.arange(-5, 5.1, 0.2)
y = np.arange(-5, 5.1, 0.2)

X, Y = np.meshgrid(x, y)
Z = np.sin(X)*np.cos(Y)

surf = ax.plot_surface(X, Y, Z, cmap = plt.cm.cividis)

# Set axes label
ax.set_xlabel("x", labelpad=20)
ax.set_ylabel("y", labelpad=20)
ax.set_zlabel("z", labelpad=20)

fig.colorbar(surf, shrink=0.5, aspect=8)

plt.show()
```

### 13.6.11. Problem 11

Create an animation of a red circle following a blue sine wave.

```python
from matplotlib.animation import FuncAnimation

# Number of points
n = 1000

# Generate x values from 0 to 6*pi
x = np.linspace(0, 6*np.pi, n)

# Calculate y values as sine of x
y = np.sin(x)

# Initialize the plot
fig, ax = plt.subplots()
ax.set_xlim(0, 6*np.pi)
ax.set_ylim(-1.1, 1.1)

# Plot the sine wave line
sine_line, = ax.plot(x, y, 'b')

# Plot the red circle (initially at the starting point)
red_circle, = ax.plot(x[0], y[0], 'ro', markersize=10)

# Function to update the frame of the animation
def update(frame):
    red_circle.set_data(x[frame], y[frame])  # Update the position of the red circle
    return red_circle,

# Create the animation
ani = FuncAnimation(fig, update, frames=n, interval=50)

# Display the animation
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.title("Red Circle Following a Blue Sine Wave")
plt.show()
```

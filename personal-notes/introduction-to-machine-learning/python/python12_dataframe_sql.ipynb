## 14.8. `Python` & `SQL` integration

### 14.8.1. Database and Relational Database definition

1. **Database**: A database is an organized collection of data that is stored and can be electronically accessed from a computer system. It is designed to allow for efficient retrieval, updating, and management of that data. Databases can vary in size and complexity, from simple datasets to complex systems used in large enterprises and organizations.

2. **Relational Database**: A relational database is a specific type of database that organizes data into related tables. These tables are comprised of rows and columns, where each row represents an individual instance of data and each column represents a specific attribute of that data. The relationship between tables is established through primary and foreign keys, enabling efficient and consistent querying and manipulation of the data. Relational databases follow the principles of the relational model proposed by Edgar F. Codd in the 1970s and are widely used in business and software applications due to their flexibility and scalability.

### 14.8.2. Database Management System (DBMS): 

A database management system is software that allows users to interact with a database, providing functions to create, modify, organize, and query data efficiently. These systems handle tasks such as data storage, retrieval, and security, as well as transaction management to ensure the integrity and consistency of the data. DBMS can be of different types, including relational systems, NoSQL database systems, object-oriented systems, among others.

1. [**SQLite**](https://www.sqlite.org/): A lightweight, open-source relational database management system that is implemented as a library within applications. SQLite is known for its ease of use, portability, and ability to be embedded into applications without requiring a separate database server. It is commonly used in mobile applications, web browsers, and other environments where a simple and efficient local database is required.

2. **MySQL**: An open-source relational database management system widely used for its performance, scalability, and ease of use.

3. **PostgreSQL**: Another open-source relational database management system known for its robustness, standards compliance, and advanced features such as support for custom data types and extensions.

4. **Oracle Database**: A relational database management system developed by Oracle Corporation. It is known for its scalability, reliability, and extensive capabilities for enterprises.

5. **Microsoft SQL Server**: A relational database management system developed by Microsoft. It is widely used in enterprise environments and tightly integrates with other Microsoft products.

6. **MongoDB**: A document-oriented NoSQL database management system. It is known for its schema flexibility and ability to handle unstructured or semi-structured data.

7. **Redis**: A NoSQL database management system commonly used as an in-memory database or as a data structure store in memory to accelerate application performance.

These are some examples of database management systems, each with its own features and specific use cases.

### 14.8.3. Primary and Foreign keys

**Primary Keys**:
- Unique: Each value in the primary key must be unique within the table.
- Immutable: Primary key values should not change once they are set.
- Never null: Primary key values cannot be null.

**Foreign Keys**:
- Do not have to be unique: Foreign key values can repeat within a table.
- Multiple foreign keys in an entity: A table can contain multiple foreign keys referencing different tables.
- Never null: Foreign key values must reference an existing value in the referenced table.

**Composite Primary Keys**:
- Used when two or more attributes are required to identify a record in a table.
- Used when it is not possible to take a single attribute as a primary key.
- Some rules for creating a composite primary key include:
  1. Use the minimum number of attributes possible.
  2. Avoid using attributes that are prone to change.

These definitions and considerations are important for proper design and implementation of relational databases.

### 14.8.5. Most common queries in `SQL`

| Command   | Description                                                                                           |
|-----------|-------------------------------------------------------------------------------------------------------|
| `CREATE`  | Used to create new database objects such as tables, indexes, or views in SQLite.                      |
| `ALTER`   | Used to modify existing database objects such as tables by adding, modifying, or dropping columns.     |
| `DROP`    | Used to delete existing database objects such as tables, indexes, or views from the database.          |
| `TRUNCATE`| Used to delete all rows from a table in SQLite, without removing the table structure.                 |

```python
# Connect to SQLite database (create if not exists)
conn = sqlite3.connect('example.db')
cur = conn.cursor()

# CREATE
cur.execute("""CREATE TABLE IF NOT EXISTS Students (
                    id INTEGER PRIMARY KEY,
                    name TEXT,
                    age INTEGER
                );""")

# INSERT
cur.execute("INSERT INTO Students (name, age) VALUES (?, ?);", ('Alice', 20))
cur.execute("INSERT INTO Students (name, age) VALUES (?, ?);", ('Bob', 22))
conn.commit()

cur.execute("SELECT * FROM Students;")
print("SELECT:", cur.fetchall())

# UPDATE
cur.execute("UPDATE Students SET age=? WHERE name=?;", (21, 'Alice'))
conn.commit()

cur.execute("SELECT * FROM Students;")
print("SELECT:", cur.fetchall())

# DELETE
cur.execute("DELETE FROM Students WHERE name=?;", ('Bob',))
conn.commit()

# DROP
cur.execute("DROP TABLE IF EXISTS Students;")
conn.commit()

# Close connection
conn.close()
```

### 14.8.6. Temporary Tables and Views in `SQL`

- Temporary tables in SQLite are tables that exist for the duration of a session or a transaction. They are useful for storing intermediate results or temporary data that is needed within a specific scope but does not need to persist beyond that scope. Temporary tables are automatically dropped when the session ends or the transaction completes.

- Views in SQLite are virtual tables that represent the result set of a SELECT statement. They provide a way to simplify complex queries and encapsulate logic by storing them as named objects in the database. Views can be queried like regular tables, but they do not store any data themselves. Instead, they dynamically generate the result set based on the underlying query.

```python
# Connect to the Northwind database
conn = sqlite3.connect('Northwind.db')

# Create a temporary table with the top 5 orders
conn.execute("""
CREATE TEMPORARY TABLE temp_orders AS
SELECT * FROM Orders
ORDER BY OrderDate DESC
LIMIT 5;
""")

# Query the temporary table
df_temp_orders = pd.read_sql_query("""
SELECT * FROM temp_orders;
""", conn)

# Close the connection
conn.close()

# Display the temporary table
print("Temporary Table (Top 5 Orders):")
df_temp_orders
```

```python
# Connect to the Northwind database
conn = sqlite3.connect('northwind.db')

# Create a cursor
cursor = conn.cursor()

# Drop the view if it exists
cursor.execute("DROP VIEW IF EXISTS expensive_products_view")

# Commit the changes
conn.commit()

# Close the cursor
cursor.close()

# Create a new cursor
cursor = conn.cursor()

# Create the view
cursor.execute("""
CREATE VIEW expensive_products_view AS
SELECT *
FROM Products
WHERE Price > 50
""")

# Commit the changes
conn.commit()

# Query the view
df_expensive_products = pd.read_sql_query("""
SELECT * FROM expensive_products_view
""", conn)

# Close the connection
conn.close()

# Display the view
print("\nView (Expensive Products):")
df_expensive_products
```

### 14.8.7. Databases from `.csv` files

It is possible to create databases from `.csv` files

```python
import os
import csv

# Example data for CSV files
data1 = [
    {'ID': 1, 'Name': 'John', 'Age': 30},
    {'ID': 2, 'Name': 'Alice', 'Age': 25},
    {'ID': 3, 'Name': 'Bob', 'Age': 35}
]

data2 = [
    {'ProductID': 101, 'ProductName': 'Phone', 'Price': 500},
    {'ProductID': 102, 'ProductName': 'Laptop', 'Price': 1000},
    {'ProductID': 103, 'ProductName': 'Tablet', 'Price': 300}
]

# Write data to CSV files
with open('data1.csv', 'w', newline='') as csvfile:
    fieldnames = ['ID', 'Name', 'Age']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(data1)

with open('data2.csv', 'w', newline='') as csvfile:
    fieldnames = ['ProductID', 'ProductName', 'Price']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(data2)

# Connect to SQLite database
conn = sqlite3.connect('example_csv.db')
conn.execute("PRAGMA foreign_keys = ON")
conn.execute("BEGIN")

cursor = conn.cursor()

# Check if tables exist, drop them if they do
cursor.execute('''DROP TABLE IF EXISTS users''')
cursor.execute('''DROP TABLE IF EXISTS products''')

# Commit changes and close connection
conn.commit()
conn.close()

# Connect to SQLite database again
conn = sqlite3.connect('example_csv.db')
cursor = conn.cursor()

# Create tables from CSV files
cursor.execute('''CREATE TABLE IF NOT EXISTS users
                  (ID INTEGER PRIMARY KEY, Name TEXT, Age INTEGER)''')

cursor.execute('''CREATE TABLE IF NOT EXISTS products
                  (ProductID INTEGER PRIMARY KEY, ProductName TEXT, Price INTEGER)''')

# Import data from CSV files
with open('data1.csv', 'r') as csvfile:
    reader = csv.DictReader(csvfile)
    for row in reader:
        cursor.execute('INSERT INTO users (ID, Name, Age) VALUES (?, ?, ?)', (row['ID'], row['Name'], row['Age']))

with open('data2.csv', 'r') as csvfile:
    reader = csv.DictReader(csvfile)
    for row in reader:
        cursor.execute('INSERT INTO products (ProductID, ProductName, Price) VALUES (?, ?, ?)', 
                       (row['ProductID'], row['ProductName'], row['Price']))

# Commit changes and close connection
conn.commit()
conn.close()
```

```python
# Connect to the SQLite database
conn = sqlite3.connect('example_csv.db')
cursor = conn.cursor()

# Execute a query to fetch data from the users table
cursor.execute("SELECT * FROM users")
users_data = cursor.fetchall()

# Execute a query to fetch data from the products table
cursor.execute("SELECT * FROM products")
products_data = cursor.fetchall()

# Print the retrieved data
print("Users Data:")
for row in users_data:
    print(row)

print("\nProducts Data:")
for row in products_data:
    print(row)

# Close the connection
conn.close()
```

# 5. Strings in `Python`

A **string** is an organized sequence of characters.

A variable of string type in `Python` (`str`) is used to store strings. When declaring a variable as a string, the content must be enclosed either in double quotes (`" "`) or single quotes (`' '`).

```python
s1 = "This is a string within double quotes"
type(s1)
```

```python
s2 = 'This is a string within single quotes'
type(s2)
```

`Python` doesn't handle accents well. Therefore, even if it's not perfectly written, it's better to avoid them to save ourselves many problems.

## 5.1. String literals

The fact that the content of variables of type `str` is enclosed in quotes, whether single or double, leads to the need for some characters to be treated in a special way.

This is where string literals come into play. Some of the most commonly used ones are shown in the following table:

| Code | Meaning                 |
|------|-------------------------|
| `\\` | Backslash, `\`          |
| `\'` | Single quote, `'`       |
| `\"` | Double quote, `"`       |
| `\n` | Newline                 |
| `\t` | Horizontal tab          |


For more information about `string literals`, refer to the [documentación](https://docs.python.org/3.7/reference/lexical_analysis.html#string-and-bytes-literals).

```python
# Backslash, \
backslash_example = "~/home/username/Documents\\file.txt"
print("Backslash Example:", backslash_example)

# Single quote, '
single_quote_example = 'It\'s raining'
print("Single Quote Example:", single_quote_example)

# Double quote, "
double_quote_example = "She said, \"Hello!\""
print("Double Quote Example:", double_quote_example)

# Newline
newline_example = "First line\nSecond line\nThird line"
print("Newline Example:")
print(newline_example)

# Horizontal tab
tab_example = "Name:\tJohn\tAge:\t25"
print("Tab Example:", tab_example)
```

## 5.2. String Concatenation

Concatenation is an operation that joins two or more strings into a single string.

In `Python`, to concatenate two string variables, we use the `+` operator.

```python
# Concatenating using the + operator
first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name
print("Full Name:", full_name)

# Concatenating with variables and literals
greeting = "Hello, "
name = "Alice"
message = greeting + name + "!"
print("Message:", message)
```

## 5.3. String Repetition

Repetition is an operation that repeats a string variable as many times as specified.

In `Python`, to repeat a string variable, we use the `*` operator. The order of the factors does not alter the result. In other words, it doesn't matter whether you use the syntax `num_repetitions * string_variable` or `string_variable * num_repetitions`.

```python
# Using the * operator
repeated1 = "Hello " * 3
print("Repeated 3 times:", repeated1)

# Using the * operator with variables
word = "Python"
repeated2 = word * 4
print("Repeated 4 times:", repeated2)
```

## 5.4. Substrings

To access a character in a string variable, we use the `[]` syntax.

```python
s = "I am a fan of video games"
```

```python
s[0] # First character
```

```python
s[5] # Sixth character
```

In `Python`, indices always start at 0, unlike some other programming languages, such as `R`.

If we precede the index with a `-`, then we start counting from the end.

```python
s[-1] # Last element
```

```python
s[-7] # Seventh element starting from the end
```

If we want to access several consecutive characters, we can use the `:` function.

```python
s[4:7] # From the fifth to the seventh
```

```python
s[:7] # From the first to the seventh
```

```python
s[8:] # From ninth to last
```

In `Python`, whenever we use the `:` function, the index on the right side is never included, as we have seen in the previous examples.

If we precede the index to the left of `:` with `-` and don't put any index to its right, we obtain the last elements.

```python
s[-10:] #Ten last elements
```

On the contrary, if we precede the index on the right with `-` without putting any index to the left of `:`, we will get all elements except the number of elements indicated by the index (remember that if we preceded with `-`, the indices counted from the end).

```python
s[:-10]
```

## 5.5. Strings Comparisons

Not only can we compare numerical data, but we can also compare strings based on alphabetical order.

```python
"Mallorca" < "Dubai"
```

The result we obtain is not because Mallorca is better than Dubai, but because the first letter of the first word, M, does not come before the first letter of the second word, D, in the alphabet.

In the event that the first letter of each of the compared words matches, the characters in the next position are compared. In the case of a tie, we would continue comparing the characters in the third position and so on.

```python
"Mallorca" >= "Madrid"
```

## 5.6. Methods for Working with Strings

In `Python`, a method is a function associated with an object that allows you to perform operations on that object or manipulate its data.

| Method        | Description                                                   |
|---------------|---------------------------------------------------------------|
| `.format()`   | Formats a string by replacing placeholders with values.       |
| `.lower()`    | Converts all characters in a string to lowercase.             |
| `.upper()`    | Converts all characters in a string to uppercase.             |
| `.count()`    | Counts the occurrences of a substring in a string.            |
| `.capitalize()` | Converts the first character of a string to uppercase.         |
| `.title()`    | Converts the first character of each word to uppercase.       |
| `.swapcase()` | Swaps the case of all characters in a string.                 |
| `.replace()`  | Replaces a substring with another substring in a string.      |
| `.split()`    | Splits a string into a list of substrings based on a delimiter.|
| `.strip()`    | Removes leading and trailing whitespace from a string.        |
| `.rstrip()`   | Removes trailing whitespace from a string.                    |
| `.lstrip()`   | Removes leading whitespace from a string.                     |
| `.find()`     | Finds the first occurrence of a substring in a string.        |
| `.index()`    | Returns the index of the first occurrence of a substring.     |
| `.rindex()`   | Returns the highest index of the substring in a string.       |
| `.startswith()` | Checks if a string starts with a specified substring.        |
| `.endswith()` | Checks if a string ends with a specified substring.           |
| `.isalnum()`  | Checks if all characters in a string are alphanumeric.        |
| `.isalpha()`  | Checks if all characters in a string are alphabetic.          |
| `.isdigit()`  | Checks if all characters in a string are digits.              |
| `.isspace()`  | Checks if all characters in a string are whitespace.          |
| `.islower()`  | Checks if all characters in a string are lowercase.           |
| `.isupper()`  | Checks if all characters in a string are uppercase.           |
| `.istitle()`  | Checks if a string is titlecased (first character uppercase, rest lowercase).|

```python
# .format()
name = "John"
age = 30
formatted_string = "My name is {}, and I am {} years old.".format(name, age)
print(formatted_string)

# .lower()
string_lower = "HELLO"
print(string_lower.lower())

# .upper()
string_upper = "hello"
print(string_upper.upper())

# .count()
sentence = "How much wood would a woodchuck chuck if a woodchuck could chuck wood?"
print(sentence.count("wood"))

# .capitalize()
text = "hello world"
print(text.capitalize())

# .title()
text = "hello world"
print(text.title())

# .swapcase()
text = "Hello World"
print(text.swapcase())

# .replace()
string = "I like apples"
print(string.replace("apples", "bananas"))

# .split()
sentence = "This is a sentence."
print(sentence.split())

# .strip()
string_with_whitespace = "  hello  "
print(string_with_whitespace.strip())

# .rstrip()
string_with_trailing_whitespace = "hello  "
print(string_with_trailing_whitespace.rstrip())

# .lstrip()
string_with_leading_whitespace = "  hello"
print(string_with_leading_whitespace.lstrip())

# .find()
sentence = "This is a sentence."
print(sentence.find("is"))

# .index()
sentence = "This is a sentence."
print(sentence.index("a"))

# .rindex()
sentence = "This is a sentence."
print(sentence.rindex("is"))

# .startswith()
sentence = "This is a sentence."
print(sentence.startswith("This"))

# .endswith()
sentence = "This is a sentence."
print(sentence.endswith("."))

# .isalnum()
string1 = "abc123"
string2 = "abc 123"
print(string1.isalnum())  # True
print(string2.isalnum())  # False

# .isalpha()
string1 = "abc"
string2 = "abc123"
print(string1.isalpha())  # True
print(string2.isalpha())  # False

# .isdigit()
string1 = "123"
string2 = "abc123"
print(string1.isdigit())  # True
print(string2.isdigit())  # False

# .isspace()
string1 = "   "
string2 = "  abc  "
print(string1.isspace())  # True
print(string2.isspace())  # False

# .islower()
string1 = "hello"
string2 = "Hello"
print(string1.islower())  # True
print(string2.islower())  # False

# .isupper()
string1 = "HELLO"
string2 = "Hello"
print(string1.isupper())  # True
print(string2.isupper())  # False

# .istitle()
string1 = "Hello World"
string2 = "hello world"
print(string1.istitle())  # True
print(string2.istitle())  # False
```

### 5.8.2 Problem 2

Given a string ("The road is closed but surely we can find an alternative"), prompt the user to enter a word belonging to the string to obtain the substring without the word indicated by the user using the `.find()` method and the `len() function.

```python
# Given string
original_string = "The road is closed but surely we can find an alternative"

# Ask the user to input a word from the string
user_input = input("Enter a word from the given string: ")

# Find the index of the user-input word in the original string
index = original_string.find(user_input)

# Check if the word is found in the string
if index != -1:
    # Calculate the length of the user-input word
    word_length = len(user_input)
    
    # Obtain the substring without the user-input word
    substring_without_word = original_string[:index] + original_string[index + word_length:]
    
    # Print the result
    print("Original String:", original_string)
    print("Substring without the word:", substring_without_word)
else:
    print("The entered word is not found in the original string.")
```

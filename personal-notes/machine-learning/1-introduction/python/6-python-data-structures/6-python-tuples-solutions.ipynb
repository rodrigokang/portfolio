## Problem 1 — Basic Tuple Creation

### Problem

Given:

```python
values = (
    10,
    20,
    30,
)
```

determine:

```python
type(values)
len(values)
values[0]
values[-1]
```

Explain the positional relationship between the positive and negative indices.

### Solution

`values` is a tuple of length `3`.

The results are:

```text
type(values) -> tuple
len(values)  -> 3
values[0]    -> 10
values[-1]   -> 30
```

For a tuple of length $n$, positive indices run from `0` through `n - 1`, while negative indices run from `-n` through `-1`. Thus, for this tuple:

```text
0  <-> -3
1  <-> -2
2  <-> -1
```

#### Verification

````python
values = (10, 20, 30)

print(type(values))
print(len(values))
print(values[0])
print(values[-1])
````

### Key Takeaway

Tuples are ordered sequences and support the same positional indexing model as lists.

## Problem 2 — The Single-Element Tuple

### Problem

Predict:

```python
a = ("python")
b = ("python",)

type(a)
type(b)
len(b)
```

Explain why the comma is required.

### Solution

`a` is a string because parentheses alone only group the expression.

`b` is a one-element tuple because the trailing comma creates the tuple structure.

Therefore:

```text
type(a) -> str
type(b) -> tuple
len(b)  -> 1
```

#### Verification

````python
a = ("python")
b = ("python",)

print(type(a))
print(type(b))
print(len(b))
````

### Key Takeaway

For a one-element tuple, the comma is essential; parentheses alone do not create the tuple.

## Problem 4 — Converting an Iterable

### Problem

Predict:

```python
tuple("data")
```

and:

```python
("data",)
```

Explain the structural difference.

### Solution

`tuple("data")` iterates over the string and produces:

```python
("d", "a", "t", "a")
```

`("data",)` is a one-element tuple containing the complete string:

```python
("data",)
```

The first construction consumes an iterable; the second stores one object.

#### Verification

````python
print(tuple("data"))
print(("data",))
````

### Key Takeaway

`tuple(iterable)` consumes elements from the iterable, whereas tuple literal syntax stores the explicitly listed objects.

## Problem 7 — Membership

### Problem

Given:

```python
features = (
    "age",
    "income",
    "balance",
)
```

predict:

```python
"income" in features
"country" in features
"country" not in features
```

Why does repeated tuple membership testing have different computational characteristics from set membership?

### Solution

The results are:

```text
"income" in features      -> True
"country" in features     -> False
"country" not in features -> True
```

Tuple membership generally performs a sequential search, giving $O(n)$ behaviour in the general case.

Set membership uses hash-based lookup and typically has average-case $O(1)$ behaviour.

#### Verification

````python
features = ("age", "income", "balance")

print("income" in features)
print("country" in features)
print("country" not in features)
````

### Key Takeaway

Tuples are sequences, not hash-based membership containers.

## Problem 9 — Immutability

### Problem

Predict what happens:

```python
values = (
    1,
    2,
    3,
)

values[1] = 10
```

What property of tuples causes the result?

### Solution

Python raises a `TypeError` because tuples do not support item assignment.

The tuple's sequence of references is fixed after creation, so an existing position cannot be replaced.

#### Verification

````python
values = (1, 2, 3)

try:
    values[1] = 10
except TypeError as exc:
    print(type(exc).__name__, exc)
````

### Key Takeaway

Tuple immutability prevents replacing, adding, or removing positions in the outer sequence.

## Problem 10 — Rebinding Is Not Mutation

### Problem

Consider:

```python
values = (
    1,
    2,
    3,
)

values = (
    4,
    5,
    6,
)
```

Why is this valid even though tuples are immutable?

### Solution

The statement does not modify the original tuple.

It creates a new tuple and rebinds the variable name `values` to that new object.

Immutability applies to the tuple object, not to the permanence of variable bindings.

#### Verification

````python
values = (1, 2, 3)
old = values

values = (4, 5, 6)

print(old)
print(values)
print(old is values)
````

### Key Takeaway

Rebinding changes which object a name refers to; it does not mutate the original object.

## Problem 11 — Mutable Object inside a Tuple

### Problem

Predict:

```python
data = (
    "model",
    [0.81, 0.84],
)

data[1].append(0.89)

print(data)
```

Why does this not violate tuple immutability?

### Solution

The result is:

```python
("model", [0.81, 0.84, 0.89])
```

The tuple still contains the same two references. The list at position `1` is mutable and its internal state changes.

Tuple immutability does not imply recursive immutability of every contained object.

#### Verification

````python
data = (
    "model",
    [0.81, 0.84],
)

data[1].append(0.89)
print(data)
````

### Key Takeaway

A tuple fixes its sequence of references, not the internal state of mutable objects referenced by those positions.

## Problem 12 — Replacing a Mutable Element

### Problem

Using the same initial tuple:

```python

    "model",
    [0.81, 0.84],
)
```

predict what happens with:

```python

```

Explain the distinction between mutating the contained list and replacing the tuple element.

### Solution

The assignment raises `TypeError`.

`data[1].append(...)` mutates the existing list object without changing the tuple reference.

`data[1] = ...` attempts to replace the tuple's reference at position `1`, which tuple immutability forbids.

#### Verification

````python
data = (
    "model",
    [0.81, 0.84],
)

try:
    data[1] = [0.90]
except TypeError as exc:
    print(type(exc).__name__, exc)
````

### Key Takeaway

Mutating a referenced mutable object is different from replacing the tuple slot that refers to it.

## Problem 13 — Tuple Hashability

### Problem

Determine whether each tuple can be hashed:

```python
(1, 2, 3)
("model", 42)
((1, 2), (3, 4))
(1, [2, 3])
```

Explain each result.

### Solution

The first three tuples are hashable because all of their relevant components are hashable.

The final tuple is not hashable because it contains a list, which is unhashable.

Tuple immutability alone does not guarantee hashability; its contents must also support hashing.

#### Verification

````python
objects = [
    (1, 2, 3),
    ("model", 42),
    ((1, 2), (3, 4)),
    (1, [2, 3]),
]

for obj in objects:
    try:
        print(repr(obj), "->", hash(obj))
    except TypeError as exc:
        print(repr(obj), "->", type(exc).__name__)
````

### Key Takeaway

A tuple is hashable only when its contents allow a stable tuple hash to be computed.

## Problem 18 — Unpacking Error

### Problem

Predict what happens:

```python
record = (
    "C001",
    42,
    0.87,
)

customer_id, probability = record
```

Why does Python raise an exception rather than silently discarding one value?

### Solution

Python raises:

```text
ValueError: too many values to unpack
```

Ordinary unpacking requires structural agreement between the number of produced values and the number of targets.

Silently dropping a value would hide a structural mismatch. If extra values should be collected, starred unpacking must be requested explicitly.

#### Verification

````python
record = ("C001", 42, 0.87)

try:
    customer_id, probability = record
except ValueError as exc:
    print(type(exc).__name__, exc)
````

### Key Takeaway

Ordinary unpacking is strict about structure; use a starred target when variable-length remainder handling is intended.

## Problem 31 — `zip()` Iterator Consumption

### Key Takeaway

## Problem 32 — Unequal Lengths with `zip()`

### Problem

Predict:

```python
a = [1, 2, 3]
b = ["A", "B"]

list(zip(a, b))
```

Which value is discarded?

Then explain how:

```python
zip(a, b, strict=True)
```

changes the behaviour.

### Solution

Default `zip()` stops when the shortest iterable is exhausted:

```python
[(1, "A"), (2, "B")]
```

The value `3` is not represented.

With `strict=True`, Python raises `ValueError` when the iterable lengths differ, making alignment assumptions explicit.

#### Verification

````python
a = [1, 2, 3]
b = ["A", "B"]

print(list(zip(a, b)))

try:
    print(list(zip(a, b, strict=True)))
except ValueError as exc:
    print(type(exc).__name__, exc)
````

### Key Takeaway

Use `strict=True` when unequal lengths indicate malformed positional data rather than intentional truncation.

## Problem 34 — Classification by Sign

### Problem

Given a sequence of integers, construct a list of tuples of the form:

```text
(number, classification)
```

where the classification is:

```text
"positive"
"negative"
"zero"
```

according to the sign of each integer.

For example:

```python
values = [
    -3,
    0,
    5,
]
```

should produce:

```python
[
    (-3, "negative"),
    (0, "zero"),
    (5, "positive"),
]
```

Solve the problem first using an explicit loop.

### Solution

Use a conditional branch for each value and append a tuple record:

```python
classified = []

for value in values:
    if value > 0:
        label = "positive"
    elif value < 0:
        label = "negative"
    else:
        label = "zero"

    classified.append((value, label))
```

#### Verification

````python
values = [-3, 0, 5]

classified = []

for value in values:
    if value > 0:
        label = "positive"
    elif value < 0:
        label = "negative"
    else:
        label = "zero"

    classified.append((value, label))

classified
````

### Key Takeaway

A list of tuples is useful when the outer collection is dynamic but every generated record has a fixed two-field structure.

## Problem 35 — Grade Classification

### Problem

Given integer grades from `1` through `10`, classify each grade using:

$$
[1,5)
\rightarrow
\text{fail},
$$

$$
[5,7)
\rightarrow
\text{pass},
$$

$$
[7,9)
\rightarrow
\text{notable},
$$

and:

$$
[9,10]
\rightarrow
\text{excellent}.
$$

Create tuples of the form:

```text
(grade, classification)
```

for a supplied sequence of grades.

### Solution

Order the threshold tests so that each grade falls into exactly one interval:

```python
classified = []

for grade in grades:
    if 1 <= grade < 5:
        label = "fail"
    elif grade < 7:
        label = "pass"
    elif grade < 9:
        label = "notable"
    elif grade <= 10:
        label = "excellent"
    else:
        label = "invalid"

    classified.append((grade, label))
```

#### Verification

````python
grades = [2, 5, 6, 7, 8, 9, 10]

classified = []

for grade in grades:
    if 1 <= grade < 5:
        label = "fail"
    elif grade < 7:
        label = "pass"
    elif grade < 9:
        label = "notable"
    elif grade <= 10:
        label = "excellent"
    else:
        label = "invalid"

    classified.append((grade, label))

classified
````

### Key Takeaway

Ordered interval tests plus tuple records provide a clear mapping from scalar inputs to fixed structured outputs.

## Problem 36 — Character Counts as Tuple Records

### Problem

Given a phrase, calculate:

- number of vowels;
- number of consonants;
- number of spaces.

Store the results as:

```python
[
    ("Vowels", ...),
    ("Consonants", ...),
    ("Spaces", ...),
]
```

Treat uppercase and lowercase letters equivalently.

### Solution

Iterate through the normalised text and maintain three counters:

```python
vowels = 0
consonants = 0
spaces = 0

for character in phrase.lower():
    if character in "aeiou":
        vowels += 1
    elif character.isalpha():
        consonants += 1
    elif character.isspace():
        spaces += 1
```

Then represent the result as fixed two-field tuple records.

#### Verification

````python
phrase = "Machine Learning"

vowels = 0
consonants = 0
spaces = 0

for character in phrase.lower():
    if character in "aeiou":
        vowels += 1
    elif character.isalpha():
        consonants += 1
    elif character.isspace():
        spaces += 1

result = [
    ("Vowels", vowels),
    ("Consonants", consonants),
    ("Spaces", spaces),
]

result
````

### Key Takeaway

Tuple records can provide compact labelled summaries inside a larger mutable result collection.

## Problem 37 — Average by Subject

### Problem

Given records of the form:

```python
(subject, grade)
```

for subjects such as:

```text
Math
Language
History
Computer Science
Music
```

calculate the average grade for every subject that has at least one observation.

Use tuples to represent the input observations.

Choose an appropriate additional data structure for grouping the grades and explain why that structure is useful.

### Solution

A dictionary is appropriate for grouping because each subject can serve as a key and map to a list of grades.

For example:

```python
groups = {}

for subject, grade in records:
    groups.setdefault(subject, []).append(grade)

averages = {
    subject: sum(grades) / len(grades)
    for subject, grades in groups.items()
}
```

The input remains a sequence of fixed `(subject, grade)` tuples, while the dictionary supports dynamic grouping by subject.

#### Verification

````python
records = [
    ("Math", 8),
    ("History", 7),
    ("Math", 9),
    ("Music", 10),
    ("History", 8),
    ("Computer Science", 9),
]

groups = {}

for subject, grade in records:
    groups.setdefault(subject, []).append(grade)

averages = {
    subject: sum(grades) / len(grades)
    for subject, grades in groups.items()
}

print(groups)
print(averages)
````

### Key Takeaway

Different data structures can collaborate: tuples encode fixed records, while dictionaries provide key-based grouping.

## Problem 38 — Points and Quadrants

### Problem

Given a collection of two-dimensional points represented by:

```python
(x, y)
```

construct tuples of the form:

```python
(x, y, region)
```

where `region` identifies the quadrant.

Treat points lying on an axis separately rather than assigning them arbitrarily to two possible quadrants.

For example, distinguish at least:

```text
quadrant I
quadrant II
quadrant III
quadrant IV
x-axis
y-axis
origin
```

### Solution

The origin and axes should be handled before quadrant conditions:

```python
classified = []

for x, y in points:
    if x == 0 and y == 0:
        region = "origin"
    elif y == 0:
        region = "x-axis"
    elif x == 0:
        region = "y-axis"
    elif x > 0 and y > 0:
        region = "quadrant I"
    elif x < 0 and y > 0:
        region = "quadrant II"
    elif x < 0 and y < 0:
        region = "quadrant III"
    else:
        region = "quadrant IV"

    classified.append((x, y, region))
```

This avoids overlapping conditions for points lying on boundaries.

#### Verification

````python
points = [
    (2, 3),
    (-2, 4),
    (-3, -1),
    (4, -2),
    (0, 5),
    (3, 0),
    (0, 0),
]

classified = []

for x, y in points:
    if x == 0 and y == 0:
        region = "origin"
    elif y == 0:
        region = "x-axis"
    elif x == 0:
        region = "y-axis"
    elif x > 0 and y > 0:
        region = "quadrant I"
    elif x < 0 and y > 0:
        region = "quadrant II"
    elif x < 0 and y < 0:
        region = "quadrant III"
    else:
        region = "quadrant IV"

    classified.append((x, y, region))

classified
````

### Key Takeaway

Boundary cases should be represented explicitly when the geometry distinguishes axes and origin from the open quadrants.

## Problem 39 — Words and Initial Letters

### Problem

Given:

```python
words = [
    "wave",
    "snail",
    "pineapple",
    "beach",
    "barbecue",
    "window",
    "fly",
]
```

create an iterable of tuples:

```text
(word, first_letter)
```

using `zip()`.

Then convert the result to a tuple of tuples.

### Solution

First construct the initial letters:

```python
initials = [word[0] for word in words]
```

Then establish positional correspondence with `zip()` and materialise as a tuple:

```python
result = tuple(zip(words, initials))
```

#### Verification

````python
words = [
    "wave",
    "snail",
    "pineapple",
    "beach",
    "barbecue",
    "window",
    "fly",
]

initials = [word[0] for word in words]
result = tuple(zip(words, initials))

result
````

### Key Takeaway

`zip()` creates positional records, and `tuple()` can materialise those records into an immutable outer collection.

## Problem 40 — Reasoning about Tuples and References

### Problem

Without executing the code, determine the exact results:

```python
A = (
    [1, 2],
    "model",
)

B = A
C = A[:]

A[0].append(3)

print(A)
print(B)
print(C)

print(A is B)
print(A is C)
print(A == C)
print(A[0] is C[0])
```

Explain carefully:

1. why mutating the inner list is allowed;
2. whether slicing creates a distinct tuple object in this case;
3. why object identity and tuple equality must be considered separately.

### Solution

`B = A` creates an alias.

For a full slice of a tuple, Python may return the original tuple object because tuples are immutable. In CPython, `A[:] is A` is `True`. Therefore `C` may be the same tuple object as `A`; code should not depend on full slicing as a tuple-copy operation.

The inner list is mutable and shared through the tuple reference. After:

```python
A[0].append(3)
```

the visible tuple value is:

```python
([1, 2, 3], "model")
```

through `A`, `B`, and `C`.

In the standard CPython behaviour used here:

```text
A is B      -> True
A is C      -> True
A == C      -> True
A[0] is C[0] -> True
```

The mutation affects the contained list, not the tuple structure itself.

#### Verification

````python
A = (
    [1, 2],
    "model",
)

B = A
C = A[:]

A[0].append(3)

print(A)
print(B)
print(C)

print(A is B)
print(A is C)
print(A == C)
print(A[0] is C[0])
````

### Key Takeaway

Tuple immutability permits implementation-level object reuse; do not assume that full slicing creates an independent tuple the way list slicing creates a new list.

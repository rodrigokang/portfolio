## Overview

This notebook accompanies the project **Customer Value and Retention** and contains the computational implementation of the customer segmentation workflow developed using transactional data from the AdventureWorks database.

The objective is to identify customer groups with distinct purchasing patterns and to create a practical foundation for the churn prediction and customer lifetime value (CLV) models developed later in the project.

The analysis begins with the construction of Recency, Frequency, and Monetary Value (RFM) features, followed by exploratory analysis and customer segmentation using K-Means clustering. K-Means was selected because it provides a simple and interpretable framework for grouping customers with similar purchasing behaviour, which is often more valuable in practice than pursuing incremental improvements in clustering complexity.

The number of segments is not fixed in advance. Instead, several clustering diagnostics are evaluated, including inertia, silhouette score, Calinski-Harabasz score, and Davies-Bouldin score.

In addition to customer segmentation, the notebook introduces a continuous intra-cluster scoring framework designed to rank customers within each segment according to their relative commercial value and purchasing behaviour. This additional layer of analysis provides a practical mechanism for prioritising customers once the broader customer groups have been identified.

The notebook emphasizes reproducibility, interpretability, and practical decision-making. The focus is not only on obtaining statistically coherent segments, but also on understanding how those segments can support customer retention initiatives, customer prioritisation, and future revenue opportunities.

Although this notebook uses the full transaction history to produce a current view of the customer base, the same RFM framework will later be applied within rolling observation windows when developing the churn prediction and CLV models. This distinction is important because customer segmentation is treated here as a descriptive task, whereas the subsequent models are predictive and therefore require a strict separation between observation data and future outcomes.

### Extract Transactional Data

The query below extracts transaction data at the order-line level.

Each row represents one product purchased within a customer order. This level of detail is useful because it keeps the dataset flexible: customer behaviour can later be aggregated by total spend, number of orders, quantity purchased, product category diversity, territory, or time period.

The extraction query follows the simplified data model introduced above, joining customer, order, product, category, and territory information into a single analytical dataset.

```python
query = """
SELECT
    c.customerid AS customer_id,
    soh.salesorderid AS sales_order_id,
    soh.orderdate AS order_date,
    soh.duedate AS due_date,
    soh.shipdate AS ship_date,
    soh.status AS order_status,
    soh.territoryid AS territory_id,
    st.name AS territory_name,
    st.countryregioncode AS country_region_code,
    sod.salesorderdetailid AS sales_order_detail_id,
    sod.productid AS product_id,
    p.name AS product_name,
    pc.name AS product_category,
    ps.name AS product_subcategory,
    sod.orderqty AS order_quantity,
    sod.unitprice AS unit_price,
    sod.unitpricediscount AS unit_price_discount,
    (
        sod.orderqty
        * sod.unitprice
        * (1 - sod.unitpricediscount)
    ) AS line_total
FROM sales.customer AS c
INNER JOIN sales.salesorderheader AS soh
    ON c.customerid = soh.customerid
INNER JOIN sales.salesorderdetail AS sod
    ON soh.salesorderid = sod.salesorderid
INNER JOIN production.product AS p
    ON sod.productid = p.productid
LEFT JOIN production.productsubcategory AS ps
    ON p.productsubcategoryid = ps.productsubcategoryid
LEFT JOIN production.productcategory AS pc
    ON ps.productcategoryid = pc.productcategoryid
LEFT JOIN sales.salesterritory AS st
    ON soh.territoryid = st.territoryid;
"""

transactions = pd.read_sql(query, engine)
```

## RFM Feature Engineering

Customer segmentation begins with the construction of Recency, Frequency, and Monetary Value (RFM) variables.

RFM analysis is one of the most widely used approaches for customer segmentation because it captures three fundamental dimensions of purchasing behaviour:

- Recency measures how recently a customer has placed an order.
- Frequency measures how often a customer purchases.
- Monetary Value measures the total revenue generated by the customer.

Together, these variables provide a compact and interpretable representation of customer activity, making them a practical starting point for segmentation analysis.

### Reference Date

Recency is calculated relative to a fixed reference date corresponding to the most recent transaction observed in the dataset.

Using a common reference date ensures that recency values are comparable across all customers.

```python
reference_date = (
    transactions["order_date"].max()
    + pd.Timedelta(days=1)
)

reference_date
```

### Customer-Level Aggregation

The transactional dataset is aggregated at the customer level to construct the Recency, Frequency, and Monetary Value variables.

The resulting dataset contains one row per customer and forms the basis for all subsequent segmentation analyses.

```python
rfm = (
    transactions
    .groupby("customer_id")
    .agg(
        recency=(
            "order_date",
            lambda x: (
                reference_date - x.max()
            ).days
        ),
        frequency=(
            "sales_order_id",
            "nunique"
        ),
        monetary=(
            "line_total",
            "sum"
        )
    )
    .reset_index()
)

rfm.head()
```

```python
rfm.shape
```

### Exploratory Analysis of RFM Variables

Before applying transformations and clustering algorithms, it is useful to examine the statistical properties of the RFM variables.

The objective of this analysis is to understand the distribution of recency, frequency, and monetary value, identify potential outliers, evaluate relationships between variables, and assess the degree of skewness present in the data.

These diagnostics help determine whether additional preprocessing steps are required before constructing customer segments.

```python
# Sturges' Rule for determining the number of bins in histograms
number_bins = 1 + int(np.log2(len(rfm['recency'])))

# Columns to visualize
columns_to_visualize = ['recency', 'frequency', 'monetary']

# Histograms
plt.figure(figsize=(15, 4))
for i, column in enumerate(columns_to_visualize, 1):
    plt.subplot(1, len(columns_to_visualize), i)
    plt.hist(rfm[column], bins=number_bins, edgecolor='black')
    plt.title('Distribution of ' + column.capitalize())
    plt.xlabel(column)
    plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

The distributions exhibit substantial positive skewness, particularly for Frequency and Monetary Value.

Most customers place only a small number of orders and generate relatively modest revenue, while a small subset of customers contributes disproportionately large purchasing volumes. This behaviour is typical of transactional datasets and is often observed in retail and e-commerce environments.

```python
# Scatter Plots
plt.figure(figsize=(15, 5))

# Scatter plot of Recency vs. Frequency
plt.subplot(1, 3, 1)
sns.scatterplot(data=rfm, x='recency', y='frequency')
plt.title('Recency vs. Frequency')

# Scatter plot of Recency vs. Monetary Value
plt.subplot(1, 3, 2)
sns.scatterplot(data=rfm, x='recency', y='monetary')
plt.title('Recency vs. Monetary Value')

# Scatter plot of Frequency vs. Monetary Value
plt.subplot(1, 3, 3)
sns.scatterplot(data=rfm, x='frequency', y='monetary')
plt.title('Frequency vs. Monetary Value')

plt.tight_layout()
plt.show()
```

The pairwise relationships reveal several characteristics of the customer base.

Customers with higher purchase frequency generally generate higher revenue, although considerable variability remains within each frequency level. In contrast, recency appears to be less strongly associated with monetary value, suggesting that both dimensions capture different aspects of customer behaviour.

These patterns support the use of all three RFM variables in the segmentation process, as each contributes distinct information.

A three-dimensional representation of the RFM feature space is also included. Although the structure is difficult to interpret directly due to the extreme scale differences between variables, the visualization provides a useful baseline for comparison after the transformation and scaling steps applied later in the workflow.

```python
from scipy.stats import skew

rfm_summary = pd.DataFrame({
    "Variable": ["Recency", "Frequency", "Monetary"],
    "Mean": [
        rfm["recency"].mean(),
        rfm["frequency"].mean(),
        rfm["monetary"].mean()
    ],
    "Standard Deviation": [
        rfm["recency"].std(),
        rfm["frequency"].std(),
        rfm["monetary"].std()
    ],
    "Skewness": [
        skew(rfm["recency"]),
        skew(rfm["frequency"]),
        skew(rfm["monetary"])
    ]
})

rfm_summary.round(4)
```

The summary statistics confirm the visual impression obtained from the histograms and scatter plots. All three variables exhibit positive skewness, with Frequency and Monetary Value showing particularly extreme asymmetry.

Such distributions can adversely affect distance-based clustering algorithms because a small number of high-value observations may dominate the clustering structure. To mitigate this effect and improve cluster stability, the RFM variables will be transformed before applying K-Means.

### RFM Transformation and Scaling

The raw RFM variables are strongly skewed and measured on very different scales. This matters because K-Means is a distance-based algorithm: variables with larger ranges or extreme values can dominate the clustering result.

To reduce this effect, the RFM variables are first transformed using the Box-Cox transformation and then standardized using `StandardScaler`.

Box-Cox is applied after adding one to each variable, ensuring that all values are strictly positive. The transformed variables are then scaled to have approximately zero mean and unit variance.

The goal is not to force the data to be perfectly normal, but to create a more balanced feature space for clustering.

```python
rfm_transformed = rfm.copy()

rfm_features = ["recency", "frequency", "monetary"]

boxcox_lambdas = {}

for feature in rfm_features:
    transformed_values, fitted_lambda = boxcox(rfm_transformed[feature] + 1)

    rfm_transformed[f"{feature}_boxcox"] = transformed_values
    boxcox_lambdas[feature] = fitted_lambda

boxcox_lambdas
```

```python
scaler = StandardScaler()

score_columns = [
    "score_recency",
    "score_frequency",
    "score_monetary"
]

rfm_transformed[score_columns] = scaler.fit_transform(
    rfm_transformed[
        [
            "recency_boxcox",
            "frequency_boxcox",
            "monetary_boxcox"
        ]
    ]
)

rfm_transformed.head()
```

```python
number_bins = 1 + int(np.log2(len(rfm_transformed)))

score_columns = [
    "score_recency",
    "score_frequency",
    "score_monetary"
]

score_titles = [
    "Distribution of Recency Score",
    "Distribution of Frequency Score",
    "Distribution of Monetary Score"
]

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, column, title in zip(axes, score_columns, score_titles):
    ax.hist(
        rfm_transformed[column],
        bins=number_bins,
        edgecolor="black"
    )

    ax.set_title(title)
    ax.set_xlabel(column)
    ax.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
scatter_pairs = [
    ("score_recency", "score_frequency", "Recency Score vs. Frequency Score"),
    ("score_recency", "score_monetary", "Recency Score vs. Monetary Score"),
    ("score_frequency", "score_monetary", "Frequency Score vs. Monetary Score")
]

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for ax, (x_col, y_col, title) in zip(axes, scatter_pairs):
    sns.scatterplot(
        data=rfm_transformed,
        x=x_col,
        y=y_col,
        alpha=0.6,
        ax=ax
    )

    ax.set_title(title)

plt.tight_layout()
plt.show()
```

```python
rfm_score_summary = pd.DataFrame({
    "Variable": ["Recency Score", "Frequency Score", "Monetary Score"],
    "Mean": [
        rfm_transformed["score_recency"].mean(),
        rfm_transformed["score_frequency"].mean(),
        rfm_transformed["score_monetary"].mean()
    ],
    "Standard Deviation": [
        rfm_transformed["score_recency"].std(),
        rfm_transformed["score_frequency"].std(),
        rfm_transformed["score_monetary"].std()
    ],
    "Skewness": [
        skew(rfm_transformed["score_recency"]),
        skew(rfm_transformed["score_frequency"]),
        skew(rfm_transformed["score_monetary"])
    ]
})

rfm_score_summary.round(4)
```

The transformed and scaled variables are now centred around zero and have comparable variance. The skewness values are also substantially lower than in the raw RFM variables, especially for Recency and Monetary Value.

Frequency remains discrete because it is based on the number of distinct orders. Even after transformation, this variable retains visible bands in the plots, which is expected for low-count purchase data.

## Selection of the Number of Clusters

Unlike traditional RFM approaches that define customer groups using predefined business rules, the objective here is to allow the clustering structure to emerge from the data.

Because no single metric provides a complete assessment of clustering quality, several complementary diagnostics are evaluated across a range of candidate values for $K$.

The analysis considers:

- Within-cluster inertia (Elbow Method).
- Silhouette Score.
- Calinski-Harabasz Index.
- Davies-Bouldin Index.

Each metric emphasizes a different aspect of cluster quality. Rather than relying on a single criterion, the final choice of $K$ is based on the combined evidence provided by all four diagnostics.

```python
# Features used for clustering

clustering_features = [
    "score_recency",
    "score_frequency",
    "score_monetary"
]

# Candidate values for K

k_values = range(2, 11)

# Containers for evaluation metrics

inertia_scores = []
silhouette_scores = []
calinski_scores = []
davies_scores = []

# Evaluate clustering performance

for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels = model.fit_predict(
        rfm_transformed[clustering_features]
    )

    inertia_scores.append(
        model.inertia_
    )

    silhouette_scores.append(
        silhouette_score(
            rfm_transformed[clustering_features],
            labels
        )
    )

    calinski_scores.append(
        calinski_harabasz_score(
            rfm_transformed[clustering_features],
            labels
        )
    )

    davies_scores.append(
        davies_bouldin_score(
            rfm_transformed[clustering_features],
            labels
        )
    )

# Summary table

cluster_metrics = pd.DataFrame({
    "k": list(k_values),
    "inertia": inertia_scores,
    "silhouette": silhouette_scores,
    "calinski_harabasz": calinski_scores,
    "davies_bouldin": davies_scores
})

cluster_metrics.round(4)
```

```python
# Clustering diagnostics

fig, axes = plt.subplots(
    nrows=2,
    ncols=2,
    figsize=(12, 8)
)

plots = [
    (
        "inertia",
        "Elbow Method",
        "Inertia",
        axes[0, 0]
    ),
    (
        "silhouette",
        "Silhouette Score",
        "Score",
        axes[0, 1]
    ),
    (
        "calinski_harabasz",
        "Calinski-Harabasz Index",
        "Score",
        axes[1, 0]
    ),
    (
        "davies_bouldin",
        "Davies-Bouldin Index",
        "Score",
        axes[1, 1]
    )
]

for metric, title, ylabel, ax in plots:

    ax.plot(
        cluster_metrics["k"],
        cluster_metrics[metric],
        marker="o",
        linewidth=2
    )

    ax.set_title(title)
    ax.set_xlabel("Number of Clusters")
    ax.set_ylabel(ylabel)

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```

The clustering diagnostics indicate that customer structure continues to improve as additional clusters are introduced. In particular, the Silhouette Score reaches its highest value at $K = 9$, while the Davies-Bouldin Index reaches its minimum value at the same point.

Although the statistical evidence slightly favours nine clusters, customer segmentation also requires practical interpretability. For this reason, the final decision will consider both clustering quality and the business meaning of the resulting customer groups.

The next step is therefore to examine the cluster profiles and determine whether the additional complexity introduced by larger values of $K$ produces genuinely distinct customer segments.

### Cluster Profile Analysis

Clustering diagnostics provide useful statistical guidance, but the final choice of $K$ should also be evaluated from a business perspective.

In customer segmentation, the objective is not only to maximize clustering metrics, but also to identify groups that are sufficiently distinct, interpretable, and actionable.

To support the final selection, the candidate solutions are examined through their cluster profiles, including cluster size and average RFM characteristics. This analysis helps determine whether additional clusters reveal genuinely different customer behaviours or simply subdivide existing groups.

#### Candidate Solution: K = 8

The first candidate solution uses eight clusters. The distribution of customers across clusters is examined first, followed by the average RFM characteristics of each segment.

```python
k_candidate = 8

kmeans_8 = KMeans(
    n_clusters=k_candidate,
    random_state=42,
    n_init=20
)

rfm["cluster_8"] = kmeans_8.fit_predict(
    rfm_transformed[
        [
            "score_recency",
            "score_frequency",
            "score_monetary"
        ]
    ]
)
```

The first step is to evaluate the number of customers assigned to each cluster. Extremely small clusters may indicate outliers or excessive fragmentation.

```python
cluster_sizes_8 = (
    rfm
    .groupby("cluster_8")
    .size()
    .reset_index(name="customers")
    .sort_values(
        "customers",
        ascending=False
    )
)

cluster_sizes_8
```

The behavioural characteristics of each segment can be examined using the original RFM variables.

```python
cluster_profile_8 = (
    rfm
    .groupby("cluster_8")
    .agg(
        customers=("customer_id", "count"),
        recency=("recency", "mean"),
        frequency=("frequency", "mean"),
        monetary=("monetary", "mean")
    )
    .round(2)
    .sort_values(
        "monetary",
        ascending=False
    )
)

cluster_profile_8
```

#### Candidate Solution: K = 9

The same analysis is repeated using nine clusters in order to determine whether the additional segment captures genuinely different customer behaviour.

```python
k_candidate = 9

kmeans_9 = KMeans(
    n_clusters=k_candidate,
    random_state=42,
    n_init=20
)

rfm["cluster_9"] = kmeans_9.fit_predict(
    rfm_transformed[
        [
            "score_recency",
            "score_frequency",
            "score_monetary"
        ]
    ]
)
```

```python
cluster_sizes_9 = (
    rfm
    .groupby("cluster_9")
    .size()
    .reset_index(name="customers")
    .sort_values(
        "customers",
        ascending=False
    )
)

cluster_sizes_9
```

```python
cluster_profile_9 = (
    rfm
    .groupby("cluster_9")
    .agg(
        customers=("customer_id", "count"),
        recency=("recency", "mean"),
        frequency=("frequency", "mean"),
        monetary=("monetary", "mean")
    )
    .round(2)
    .sort_values(
        "monetary",
        ascending=False
    )
)

cluster_profile_9
```

The comparison between the eight-cluster and nine-cluster solutions suggests that the additional cluster introduced at $K = 9$ captures a meaningful behavioural distinction rather than simply fragmenting an existing segment.

In the eight-cluster solution, a group of low-spending but relatively frequent customers appears as a single segment. When the number of clusters is increased to nine, this group separates into two distinct profiles: one characterised by recent purchasing activity and higher order frequency, and another characterised by older purchasing activity and lower engagement.

This additional separation is consistent with the underlying business behaviour and improves the interpretability of the segmentation. Combined with the favourable clustering diagnostics observed earlier, these results support the selection of $K = 9$ as the final segmentation solution.

## Customer Prioritisation Within Clusters

The cluster assignment identifies groups of customers with similar purchasing behaviour. However, customers belonging to the same segment may still differ substantially in their relative importance.

To support prioritisation within each cluster, an additional RFM score is calculated for every customer. The score combines the transformed and standardized Recency, Frequency, and Monetary variables into a single continuous measure.

Since lower recency indicates more recent activity, the Recency component enters the score with a negative sign, while Frequency and Monetary contribute positively.

Rather than replacing the cluster labels, the score provides a ranking mechanism that helps identify the strongest and weakest customers within each segment.

### Cluster-Specific Weighting

The contribution of each RFM component is allowed to vary across clusters.

Weights are derived from the within-cluster variance of the transformed RFM variables. Variables exhibiting greater variation receive larger weights because they contribute more strongly to differentiating customers within that particular segment.

This approach produces a cluster-specific prioritisation score while preserving the overall segmentation structure.

```python
def calculate_cluster_weights(
    data,
    cluster_column,
    score_columns
):
    """
    Calculate cluster-specific weights from within-cluster variance.

    Input:
        data (pd.DataFrame): Dataset containing cluster labels and score variables.
        cluster_column (str): Name of the column containing cluster labels.
        score_columns (list): List of standardized score columns used for weighting.

    Output:
        pd.DataFrame: Cluster-level weights for each score variable.

    Author:
        Rodrigo Kang
    """

    variance_by_cluster = (
        data
        .groupby(cluster_column)[score_columns]
        .var()
    )

    cluster_weights = variance_by_cluster.div(
        variance_by_cluster.sum(axis=1),
        axis=0
    )

    return cluster_weights
```

```python
def assign_rfm_score(
    data,
    cluster_column,
    score_columns,
    cluster_weights
):
    """
    Assign a weighted RFM score to each customer.

    Input:
        data (pd.DataFrame): Dataset containing customer-level RFM scores and cluster labels.
        cluster_column (str): Name of the column containing cluster labels.
        score_columns (list): List containing recency, frequency, and monetary score columns.
        cluster_weights (pd.DataFrame): Cluster-specific weights.

    Output:
        pd.DataFrame: Dataset with cluster-specific weights and RFM scores.

    Author:
        Rodrigo Kang
    """

    output = data.copy()

    recency_col, frequency_col, monetary_col = score_columns

    output["omega_recency"] = output[cluster_column].map(
        cluster_weights[recency_col]
    )

    output["omega_frequency"] = output[cluster_column].map(
        cluster_weights[frequency_col]
    )

    output["omega_monetary"] = output[cluster_column].map(
        cluster_weights[monetary_col]
    )

    output["rfm_score"] = (
        - output[recency_col] * output["omega_recency"]
        + output[frequency_col] * output["omega_frequency"]
        + output[monetary_col] * output["omega_monetary"]
    )

    scaler = MinMaxScaler()

    output["rfm_normalized_score"] = scaler.fit_transform(
        output[["rfm_score"]]
    )

    return output
```

```python
score_columns = [
    "score_recency",
    "score_frequency",
    "score_monetary"
]

scoring_data = (
    rfm
    .merge(
        rfm_transformed[
            [
                "customer_id",
                "score_recency",
                "score_frequency",
                "score_monetary"
            ]
        ],
        on="customer_id",
        how="left"
    )
)
```

```python
cluster_weights = calculate_cluster_weights(
    data=scoring_data,
    cluster_column="cluster",
    score_columns=score_columns
)

cluster_weights.round(4)
```

The resulting weights vary across clusters, reflecting the different sources of behavioural variation present within each customer segment.

For example, some clusters are primarily differentiated by recency, while others are more strongly influenced by spending behaviour or purchase frequency. As a result, the prioritisation score adapts to the characteristics of each segment rather than applying a single weighting scheme to the entire customer base.

```python
rfm_scored = assign_rfm_score(
    data=scoring_data,
    cluster_column="cluster",
    score_columns=score_columns,
    cluster_weights=cluster_weights
)
```

```python
rfm_scored[
    [
        "customer_id",
        "recency",
        "frequency",
        "monetary",
        "cluster",
        "omega_recency",
        "omega_frequency",
        "omega_monetary",
        "score_recency",
        "score_frequency",
        "score_monetary",
        "rfm_score",
        "rfm_normalized_score"
    ]
].head()
```

The normalized RFM score ranges from 0 to 1 and can be interpreted as a relative ranking within the customer base.

Customers with higher scores generally exhibit more favourable combinations of recency, frequency, and monetary value, while lower scores indicate weaker engagement and commercial contribution.

This score complements the clustering results by introducing a continuous prioritisation measure that can support marketing, retention, and customer management initiatives.

## Intra-Cluster Customer Ranking

Customer segmentation provides a high-level grouping of customers with similar purchasing behaviour.

However, customers belonging to the same segment may still exhibit substantial differences in recency, purchasing frequency, and monetary contribution.

To support prioritization within each segment, an intra-cluster ranking methodology is introduced.

The approach combines the standardized RFM variables into a continuous score computed separately for each cluster. This allows customers to be ranked relative to their peers while preserving the structure identified by the clustering algorithm.

The resulting score ranges between 0 and 1, where larger values indicate stronger customer performance within the corresponding segment.

### Weighting Strategy

The contribution of each RFM component is determined using its variance within the corresponding cluster.

Variables exhibiting little variability provide limited discriminatory power because most customers share similar values.

Conversely, variables with larger variability contribute more information for ranking customers within the cluster.

This approach automatically adapts the ranking model to the characteristics of each segment.

```python
rfm_rank = pd.concat(
    [
        rfm,
        rfm_scaled
    ],
    axis=1
)

rfm_rank.head()
```

```python
rfm_rank["score_recency"] = (
    -rfm_rank["recency_scaled"]
)

rfm_rank["score_frequency"] = (
    rfm_rank["frequency_scaled"]
)

rfm_rank["score_monetary"] = (
    rfm_rank["monetary_scaled"]
)
```

```python
def compute_cluster_score(df):

    variances = df[
        [
            "score_recency",
            "score_frequency",
            "score_monetary"
        ]
    ].var()

    weights = (
        variances /
        variances.sum()
    )

    score = (
        weights["score_recency"]
        * df["score_recency"]
        +
        weights["score_frequency"]
        * df["score_frequency"]
        +
        weights["score_monetary"]
        * df["score_monetary"]
    )

    scaler = MinMaxScaler()

    score = scaler.fit_transform(
        score.to_frame()
    ).flatten()

    return pd.Series(
        score,
        index=df.index
    )
```

```python
rfm_rank["rfm_score"] = (
    rfm_rank
    .groupby("cluster")
    .apply(compute_cluster_score)
    .reset_index(level=0, drop=True)
)
```

## Intra-Cluster Weight Analysis

The proposed ranking methodology assigns weights according to the variability of each RFM component within a cluster.

To better understand the behaviour of the scoring system, the resulting weights are analysed across customer segments.

This analysis reveals which dimensions contribute most to customer differentiation inside each segment.

```python
cluster_weights = []

for cluster_id, group in rfm_rank.groupby("cluster"):

    variances = group[
        [
            "score_recency",
            "score_frequency",
            "score_monetary"
        ]
    ].var()

    weights = variances / variances.sum()

    cluster_weights.append(
        {
            "cluster": cluster_id,
            "recency_weight": weights["score_recency"],
            "frequency_weight": weights["score_frequency"],
            "monetary_weight": weights["score_monetary"]
        }
    )

cluster_weights = pd.DataFrame(cluster_weights)

cluster_weights
```

```python
cluster_weights = cluster_weights.merge(
    cluster_profile[
        ["segment"]
    ],
    left_on="cluster",
    right_index=True
)

cluster_weights
```

```python
cluster_weights_plot = (
    cluster_weights
    .set_index("segment")
    [
        [
            "recency_weight",
            "frequency_weight",
            "monetary_weight"
        ]
    ]
)

cluster_weights_plot.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Intra-Cluster Variable Weights")
plt.ylabel("Weight")
plt.xlabel("Segment")

plt.xticks(rotation=45)

plt.show()
```

### Interpretation of Variable Weights

The proposed weighting scheme does not attempt to measure variable importance globally.

Instead, weights quantify the discriminatory power of each variable within a specific customer segment.

Variables exhibiting little variability inside a cluster contribute less to the final ranking because they provide limited information for differentiating customers belonging to that segment.

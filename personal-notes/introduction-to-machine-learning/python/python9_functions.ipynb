# 11. `Python` Functions

**Function.** A function in `Python` is a reusable piece of code that only runs when it is called.

It is defined using the `def` keyword and the general structure is as follows:

```python
# def function_name(input1, input2, ..., inputn):
    # Function body
    # return output
```

The `return` statement terminates the execution of the function and returns the result specified after it. If nothing is specified, the function would terminate but return nothing.

As we have seen, in general, functions consist of three parts:

- **Inputs (parameters or arguments).** These are the values we pass as input to the function.
- **Body.** These are all the operations that the function performs.
- **Output.** This is the result that the function returns.

Parameters are local variables of the function. If we try to execute one of these variables in the global environment, it will raise an error. We will discuss this topic in more detail later.

With what we have seen before, when constructing a function, we must ask ourselves the following questions:

- What data does the function need to know? (inputs)
- What does the function do? (body)
- What does it return? (output)

Inputs and output are optional: we can define a function without providing inputs and without it returning anything.

Once a function is defined, to call it, we use its name followed by parentheses:

Let's create a function that calculates the integer division of two numbers and returns the quotient and the remainder:

```python
def euclidean_division(x, y):
    q = x // y
    r = x % y
    return q, r
```

This function, which we've named `euclidean_division`, calculates the quotient and remainder of any two numbers and returns these calculated numbers as a result.

Now, let's use our function to calculate the quotient and remainder of the division $41 \div 7$.

```python
euclidean_division(x = 41, y = 7)
```

```python
euclidean_division(41, 7)
```

```python
euclidean_division(y = 7, x = 41)
```

When calling the function and providing the parameters `x = 41` and `y = 7`, we obtained the tuple `(5, 6)` as a result. The meaning of this result is that the integer quotient of $41 \div 7$ is 5, while the remainder is 6. Indeed,

$$41 = 7 \cdot 5 + 6$$

We could also save the results returned by our function in different variables, so we can work with them in the global environment.

```python
quotient, remainder = euclidean_division(x = 41, y = 7)
print(quotient)
print(remainder)
print(41 == 7 * quotient + remainder)
```

### 11.1.1. Arbitrary Number of Arguments

If we don't know the number of parameters that will be introduced, we add an asterisk `*` before the parameter name in the function definition. The entered values will be stored in a tuple.

```python
def sum_numbers(*numbers):
    sum = 0
    for n in numbers:
        sum += n

    return sum
```

```python
sum_numbers(1, 2, 3)
```

```python
sum_numbers(2, 4, 6, 8, 10)
```

## 11.3. Function Variables

Within a function in Python, there are two types of variables:

- **Local variable:** This is created and only exists within the function.
- **Global variable:** This is created in the global environment.

Given the following function:

```python
def arithmetic_operations(x, y):
    sum = x + y
    diff = x - y
    prod = x * y
    div = x / y
    return {"sum": sum,
           "difference": diff,
           "product": prod,
           "division": div}
```

If we try to print, for example, the value of the variable `prod` in the global environment, we will encounter an error. ### 11.3.1. Pass by Value vs. Pass by Reference

Depending on the type of data we pass as a parameter to the function, we can differentiate between:

- **Pass by value:** A local copy of the variable is created within the function.
- **Pass by reference:** The variable is directly manipulated, and changes made within the function also affect the global level.

In general, basic data types such as integers, floats, strings, or booleans are passed by value, while data structures like lists, dictionaries, sets, or tuples, as well as other objects, are passed by reference.

An example of pass by value would be:

```python
def double_value(n):
    n = n*2
    return n
```

```python
num = 5
print(double_value(num))
print(num)
```

An example of a pass by reference would be:

```python
def double_values(ns):
    for i, n in enumerate(ns):
        ns[i] *= 2

    return ns
```

```python
nums = [1, 2, 3, 4, 5]
print(double_values(nums))
print(nums)
```

To avoid modifying the original list, we can do so by passing a copy of that list as a parameter.

```python
nums = [1, 2, 3, 4, 5]
print(double_values(nums[:]))
print(nums)
```

Now let's see a function that contains a `for` loop and, given an integer, prints its multiplication table with its first 10 multiples and returns a list with all those multiples:

```python
def multiplication_table10(num):
    """
    Given an integer, we print its multiplication table with
    the first 10 multiples and return a list of the multiples.

    Args:
        num (int): value for which we will calculate its multiplication table

    Returns:
        multiples (list): list with the first 10 multiples of num
    """
    multiples = []
    print("Multiplication table of {}:".format(num))

    for i in range(1, 11):
        multiple = num * i
        print("{} x {} = {}".format(num, i, multiple))
        multiples.append(multiple)

    return multiples
```

Let's now improve the `multiplication_table10()` function so that if the user decides to enter a non-integer number, our function will alert them and explain the error they are making:

```python
def multiplication_table10(num):
    """
    Given an integer, first we check if it is an integer.
    If it's not, we return nothing.
    If it is, we print its multiplication table with the first
    10 multiples and return a list of the multiples.

    Args:
        num (int): value for which we will calculate its first 10 multiples

    Returns:
        multiples (list): list with the first 10 multiples of num
    """

    if type(num) != type(1):
        print("The entered number is not an integer")
        return

    multiples = []
    print("Multiplication table of {}:".format(num))

    for i in range(1, 11):
        multiple = num * i
        print("{} x {} = {}".format(num, i, multiple))
        multiples.append(multiple)

    return multiples
```

Let's now create a function that, given a sentence ending in a period, returns whether or not it contains the letter "a" using a `while` loop.

```python
def contains_a(sentence):
    i = 0
    while sentence[i] != ".":
        if sentence[i] == "a":
            return True
        i += 1
    return False
```

## 11.5. Recursive Functions

**Recursive Function:** It is a function that calls itself.

Be very careful with this type of functions because we can fall into an infinite loop. That is, the function will never finish executing.

An example of a recursive function that would enter into an infinite loop would be the following:


```python
def powers(x, n):
    print(x ** n)
    powers(x, n + 1)
```

Why do we say it enters an infinite loop? Because it will only stop if we interrupt the execution.

This is because we have not specified a stopping condition for the function, called a base case.

**Base case:** It is the case that indicates when the recursion should break. It must always be specified to avoid falling into an infinite loop.

In the case of the `powers()` function, we can specify as a base case when the resulting value exceeds 1000000. We specify this with an `if` statement.

```python
def powers(x, n):
    if x ** n > 1000000:
        return x ** n

    print(x ** n)
    powers(x, n + 1)
```

```python
powers(2, 1)
```

Now let's see a classic example of a recursive function that works correctly.

We want a function that prints the $i$-th term of the Fibonacci sequence. That is, we specify the index of the term and the function returns the value of that term.

The Fibonacci sequence is

$$1, 1, 2, 3, 5, 8, 13, \dots$$

That is, each term is obtained by summing the two previous terms.

$$F_0 = F_1 = 1$$
$$F_n = F_{n-1} + F_{n-2}, n\geq 2$$

So, the function we want, which we have named `Fibonacci()`, is:

```python
def Fibonacci(index):
    if index == 0 or index == 1:
        return 1

    return Fibonacci(index - 1) + Fibonacci(index - 2) 
```

As you can see, we have specified when the function should stop. That is, the base case is when the index is 0, as negative indices do not exist.

```python
Fibonacci(index = 7)
```

```python
Fibonacci(8)
```

```python
Fibonacci(30)
```

## 11.6. Funciones helper

Just like functions can call themselves, they can also call other functions.

**Helper Function:** It is a function whose purpose is to avoid code repetition.

If we are given the following function:

```python
def sign_sum(x, y):
    if x + y > 0:
        print("The result of adding {} and {} is positive".format(x, y))
    elif x + y == 0:
        print("The result of adding {} and {} is zero".format(x, y))
    else:
        print("The result of adding {} and {} is negative".format(x, y))
```

```python
sign_sum(5, 4)
sign_sum(3, -3)
sign_sum(1, -8)
```

We see that the `print` statement is repeated except for the last word.

We could think of creating the following helper function:

```python
def helper_print(x, y, sign):
    print("The result of adding {} and {} is {}.".format(x, y, sign))
```

If we use the helper function, the `sign_sum()` function would be modified as follows:

```python
def sign_sum(x, y):
    if x + y > 0:
        helper_print(x, y, "positive")
    elif x + y == 0:
        helper_print(x, y, "zero")
    else:
        helper_print(x, y, "negative")
```

```python
sign_sum(5, 4)
sign_sum(3, -3)
sign_sum(1, -8)
```

Thus, there is no longer any repeated code.

And as you can see, the original function works correctly.

### 11.7.1. The `filter()` Function

- It applies a function to all elements of an iterable object.
- It returns a generator object, hence we use the `list()` function to convert it to a list.
- As output, it returns the elements for which applying the function has returned `True`.

Using lambda functions, let's apply `filter()` to keep only the multiples of 7 from the following list called `nums`:

### 11.7.2. The `reduce()` Function

- It continuously applies the same function to the elements of an iterable object.

    1. It applies the function to the first two elements.
    2. It applies the function to the result of the previous step and the third element.
    3. It applies the function to the result of the previous step and the fourth element.
    4. It continues like this until only one element remains.
    
- It returns the resulting value.

Using lambda functions, let's apply `reduce()` to calculate the product of all elements in a list.

```python
from functools import reduce
```

```python
nums = [1, 2, 3, 4, 5, 6]
reduce(lambda x, y: x * y, nums)
```

Again, the function provided to `reduce()` doesn't have to be a lambda function; it can be an existing one or one created by ourselves.

With the following lines of code, we will obtain the maximum of a given list using `reduce()` and the created function `bigger_than()`:

```python
def bigger_than(a, b):
    if a > b:
        return a
    return b
```

```python
bigger_than(14, 7)
```

```python
nums = [-10, 5, 7, -3, 16, -30, 2, 33]
reduce(bigger_than, nums)
```

### 11.8.2. Problem 2

Create a function that displays the [Pascal's triangle](https://en.wikipedia.org/wiki/Pascal%27s_triangle). The number of rows to display will be indicated by the parameter. By default, `n` will be 1. For more information, see [binomial coefficient](https://en.wikipedia.org/wiki/Binomial_coefficient).

```python
def choose(n, k):
    """
    Calculates the binomial coefficient n choose k
    """
    import math
    if (n >= k and k >= 0):
        return math.factorial(n) / (math.factorial(k) * math.factorial(n - k))
    else:
        return "Cannot compute the factorial number indicated"

def pascal_triangle(n = 1):
    print(1)
    for i in range(1, n + 1):
        for k in range(i + 1):
            print(choose(i, k), end = " ")
        print("")

pascal_triangle(5)
```

### 11.8.14. Problem 14

Given a list of words, keep the word with the most "a"s.

```python
from functools import reduce

def total_a(w):
    """
    Returns the total occurrences of the letter 'a'
    Args:
        w: Word in string format
    Returns:
        a: Integer number
    """
    a = 0
    for c in w.lower():
        if c == "a":
            a += 1
    return a

def more_a(w1, w2):
    """
    Returns the word with the highest number of 'a'
    Args:
        w1: Word in string format
        w2: Word in string format
    Returns:
        Word in string format
    """
    if total_a(w1) >= total_a(w2):
        return w1
    return w2

words = ["shoe", "friend", "thing", "bitter", "grass"]
reduce(lambda w1, w2: more_a(w1,w2), words)
```

### 11.8.16. Problem 16

Sort a list of words by the number of occurrences of the letter indicated by the user. The order will be descending.

```python
def total_letter(w, l):
    """
    Returns the total occurrences of the letter indicated by the parameter in the word w
    Args:
        w: Word in string format
        l: Letter in string format
    Returns:
        letter: Integer
    """
    letter = 0
    for c in w.lower():
        if c == l.lower():
            letter += 1
    return letter

l = input("Enter a letter: ")
words = ["tutu", "toyshop", "cello", "boastful", "piano", "bat"]
sorted(words, key = lambda w: total_letter(w, l), reverse = True)
```

### 11.8.22. Problem 22

Write the factorial function using recursion. Use the function to calculate the factorial of 3.

```python
def factorial(n):
    """Computes and returns the factorial of n,
    a positive integer.
    """
    if n == 1: # Base case!
        return 1
    else: # Recursive step
        return n * factorial(n - 1) # Recursive call
    
factorial(3)
```

### 11.8.23. Problem 23

Write a recursive function to calculate the n-th Fibonacci number. Use the function to calculate the first five Fibonacci numbers.

```python
def fibonacci(n):
    """Computes and returns the Fibonacci of n,
    a positive integer.
    """
    if n == 1: # first base case
        return 1
    elif n == 2: # second base case
        return 1
    else: # Recursive step
        return fibonacci(n-1) + fibonacci(n-2)
    
print(fibonacci(1))
print(fibonacci(2))
print(fibonacci(3))
print(fibonacci(4))
print(fibonacci(5))
```

### 11.8.24. Problem 24

Write a function `fib_disp(5)` based on the modification of the `fibonacci` function from problem 113. Can we determine the order in which the Fibonacci numbers will appear on the screen for `fib_disp(5)`?

```python
def fib_disp(n):
    """Computes and returns the Fibonacci of n,
    a positive integer.
    """
    if n == 1: # first base case
        out = 1
        print(out)
        return out
    elif n == 2: # second base case
        out = 1
        print(out)
        return out
    else: # Recursive step
        out = fib_disp(n-1) + fib_disp(n-2)
        print(out)
        return out # Recursive call

fib_disp(5)
```

### 11.8.25. Problem 25

Make an iterative implementation to calculate Fibonacci numbers. Calculate the 25th Fibonacci number using this implementation and the `fibonacci` function from problem 119. Use the `timeit` magic command to measure the execution time of each one. Note the significant difference in execution times.

```python
import numpy as np

def iter_fib(n):
    fib = np.ones(n)
    
    for i in range(2, n):
        fib[i] = fib[i - 1] + fib[i - 2]
        
    return fib

iter_fib(25)
```

### 11.8.26. Problem 26

Build a program to solve the [Tower of Hanoi problem](https://en.wikipedia.org/wiki/Tower_of_Hanoi) for `N = 3`. Verify that the solution is correct by inspection.

```python
def my_towers(N, from_tower, to_tower, alt_tower):
    """
    Displays the moves required to move a tower of size
    N from the "from_tower" to the "to_tower".
    "from_tower", "to_tower" and "alt_tower" are
    either 1, 2, 3 referring to tower 1, 2, and 3.
    """
    if N != 0:
        # recursive call that moves N-1 stack from
        # starting tower to alternate tower
        my_towers(N-1, from_tower, alt_tower, to_tower)
        # display to screen movement of bottom disk from
        # starting tower to final tower
        print("Move disk %d from tower %d to tower %d."\
              %(N, from_tower, to_tower))
        # recursive call that moves N-1 stack from
        # alternate tower to final tower
        my_towers(N-1, alt_tower, to_tower, from_tower)
        
my_towers(3, 1, 3, 2)
```

### 11.8.27. Problem 27

Perform an implementation of the [quicksort algorithm](https://en.wikipedia.org/wiki/Quicksort).

```python
def my_quicksort(lst):
    
    if len(lst) <= 1:
        # list of length 1 is easiest to sort
        # because it is already sorted
        
        sorted_list = lst
    else:
        
        # select pivot as the first element of the list
        pivot = lst[0]
        
        # initialize lists for bigger and smaller
        # elements as well those equal to the pivot
        bigger = []
        smaller = []
        same = []
        
        # put elements into appropriate array
        
        for item in lst:
            if item > pivot:
                bigger.append(item)
            elif item < pivot:
                smaller.append(item)
            else:
                same.append(item)
                
        sorted_list = my_quicksort(smaller) + same + my_quicksort(bigger)
        
    return sorted_list

my_quicksort([2, 1, 3, 5, 6, 3, 8, 10])
```

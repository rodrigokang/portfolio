# 4. Numbers in `Python`

## 4.1. Types of Numbers

In `Python`, numerical variables can be of different types, including:

1. **Integer (`int`)**: Integers are whole numbers without any decimal point. For example: `5`, `-3`, `1000`.

2. **Floating point numbers (`float`)**: Floats represent real numbers and can have a decimal point. For example: `3.14`, `-0.001`, `2.0`.

3. **Complex numbers (`complex`)**: A complex number is an ordered pair of real numbers $z = (a, b)$, where $a, b \in \mathbb{R}$. In `Python` we can represent a complex number as `a + bj`, where `a` is the real part, `b` is the imaginary part, and `j` represents the imaginary unit. For example: `3 + 4j`, `-2 - 6j`.

4. **Boolean (`bool`)**: While not strictly numerical, boolean variables represent either `True` or `False`, which can be interpreted as numeric values of `1` and `0` respectively in certain contexts.

To determine the data type of a number, you can use the `type()` function.

```python
type(5)
```

```python
type(5.0)
```

```python
type(3 + 4j)
```

```python
type(True)
```

In `Python`, when referring to `float` numbers with all 0's in the decimal part, such as `3.0`, it is sufficient to write `3.`. That is, `Python` understands that the numbers `3.0` and `3.` are the same, including being of the same type: `float`.

```python
type(3.0)
```

```python
type(3.)
```

We can specify the type of number we want to use using the `int()` and `float()` functions.

```python
type(int(7.0))
```

```python
type(int(9.))
```

```python
type(float(3))
```

It's easy to convert from integers to floating-point numbers since it's always possible, but we can't always convert from floating-point numbers to integers as it results in the loss of the decimal part.

```python
int(3.5)
```

## 4.2. Arithmetic Operations

Below is a description of common arithmetic operations in Python:

| Operation   | Description                                            |
|-------------|--------------------------------------------------------|
| `+`         | Addition: Adds two operands together.                  |
| `-`         | Subtraction: Subtracts the right operand from the left.|
| `*`         | Multiplication: Multiplies two operands.               |
| `/`         | Division: Divides the left operand by the right.       |
| `//`        | Floor Division: Divides and rounds down to the nearest integer (or the quotient in integer division). |
| `%`           | Modulus: Returns the remainder of the division.        |
| `**`        | Exponentiation: Raises the left operand to the power of the right. |
| `pow()`     | Exponentiation (alternative): Raises the first argument to the power of the second. |

```python
# Addition (+)
result_addition = 3 + 5
print("Addition:", result_addition)

# Subtraction (-)
result_subtraction = 7 - 4
print("Subtraction:", result_subtraction)

# Multiplication (*)
result_multiplication = 2 * 6
print("Multiplication:", result_multiplication)

# Division (/)
result_division = 8 / 2
print("Division:", result_division)

# Floor Division (//)
result_floor_division = 7 // 2
print("Floor Division:", result_floor_division)

# Modulus (%)
result_modulus = 7 % 2
print("Modulus:", result_modulus)

# Exponentiation (**)
result_exponentiation = 2 ** 3
print("Exponentiation:", result_exponentiation)

# Exponentiation with pow()
result_pow_function = pow(2, 3)
print("Exponentiation (pow()):", result_pow_function)
```

## 4.4. Complex Numbers

### 4.4.1. Definition

In `Python`, complex numbers are defined in binomial form, and instead of using `i`, the letter `j` is used to represent the imaginary unit. If you want to indicate that the imaginary part is 1 or -1, simply using `j` or `-j` is not sufficient. Instead, you need to write `1j` or `-1j` when defining the complex number in its binomial form.

```python
z = 2 + 5j
z
```

We can also define complex numbers in `Python` with the `complex()` function.

```python
z = complex(1, -7)
z
```

To obtain the real (imaginary) part, we use the `.real` (`.imag`) method.

```python
print("Re(z) = ", z.real)
print("Im(z) = ", z.imag)
```

### 4.4.2. Operations with complex numbers

| Operation      | Description                                      |
|----------------|--------------------------------------------------|
| `+`            | Addition: Adds two complex numbers together.     |
| `-`            | Subtraction: Subtracts one complex number from another. |
| `*`            | Multiplication: Multiplies two complex numbers.  |
| `/`            | Division: Divides one complex number by another. |
| `.conjugate()` | Conjugate: Returns the complex conjugate of a complex number.|
| `abs()`        | Absolute Value: Returns the absolute value (magnitude) of a complex number.|

```python
# Define two complex numbers
z1 = 3 + 4j
z2 = 2 - 5j

# Addition (+)
result_addition = z1 + z2
print("Addition:", result_addition)

# Subtraction (-)
result_subtraction = z1 - z2
print("Subtraction:", result_subtraction)

# Multiplication (*)
result_multiplication = z1 * z2
print("Multiplication:", result_multiplication)

# Division (/)
result_division = z1 / z2
print("Division:", result_division)

# Conjugate (.conjugate())
conjugate_z1 = z1.conjugate()
conjugate_z2 = z2.conjugate()
print("Conjugate of z1:", conjugate_z1)
print("Conjugate of z2:", conjugate_z2)

# Absolute Value (abs())
abs_z1 = abs(z1)
abs_z2 = abs(z2)
print("Absolute value of z1:", abs_z1)
print("Absolute value of z2:", abs_z2)
```

## 4.6. Boolean

**Boolean Data**. It is a type of data that can only take on two values: `True` (true) or `False` (false).

**Logical Variable**. A variable that stores boolean data.

```python
is_adult = True
type(is_adult)
```

### 4.6.1. Truth Tables


Given two logical variables, $A$ and $B$, we can define basic operators using truth tables, where the true value is represented by the letter $V$ or by $1$, while the false value is represented by the letter $F$ or by $0$.

The truth table for variable $A$ would be

| $A$ |
| :---: |
| $V$ |
| $F$ |

The truth table for variable $B$ would be

| $B$ |
| :---: |
| $V$ |
| $F$ |

### 4.6.2. Logical operators in Python

**Negation**

The negation operator applied to a variable is represented by $\neg$ and returns the opposite value.

| $A$ | $\neg A$ |
| :---: | :---: |
| $V$ | $F$ |
| $F$ | $V$ |

In `Python`, to perform negation, we use the `not` operator.

```python
A = True
not A
```

```python
B = False
not B
```

**Conjunction**

The conjunction between two variables is represented by $\wedge$ and returns true only when both variables are true.

| $A$ | $B$ | $A\wedge B$ |
| :---: | :---: | :---: |
| $V$ | $V$ | $V$ |
| $V$ | $F$ | $F$ |
| $F$ | $V$ | $F$ |
| $F$ | $F$ | $F$ |

To perform the conjunction between two logical variables, we use the `and` operator.

```python
A, B = True, True
A and B
```

```python
A and (not B)
```

**Disjunction**

The disjunction between two variables is represented by $\vee$ and returns true when at least one of the logical variables is true.

| $A$ | $B$ | $A\vee B$ |
| :---: | :---: | :---: |
| $V$ | $V$ | $V$ |
| $V$ | $F$ | $V$ |
| $F$ | $V$ | $V$ |
| $F$ | $F$ | $F$ |

To perform the conjunction between two logical variables, we use the `or` operator.

```python
A, B = False, False
A or B
```

```python
(not A) or B
```

### 4.6.3. Logical operators in Python

In `Python`, we can compare data and obtain a boolean result. The available comparison operators are:

| Operator | Meaning                |
| :---:    | :---                   |
| $>$      | Strictly greater than  |
| $\ge$    | Greater than or equal  |
| $<$      | Strictly less than     |
| $\le$    | Less than or equal     |
| $==$     | Equal                  |
| $!=$     | Not equal              |

```python
print(7 == 7.0)
print(3.14 > 9)
print(7 != "7")
print(0.01 <= 1)
```

Note that when we compared the number 7 in integer and float formats, we found that they were equal, while comparing the number 7 in integer format with the same number in string format returned that they are different. This tells us that numerically, `Python` considers integer and float numbers equal. However, it will never consider two data points equal if one is in numeric format and the other is in string format.

We can make multiple comparisons at once:

```python
age = 17
(age >= 16) and (age <= 40)
```

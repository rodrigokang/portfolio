# Microbial Population Dynamics (Lotka–Volterra Model)

## Caso de una sola especie

Antes de proceder a la integración de la ecuación de Lotka-Volterra para una sola especie, es conveniente verificar si la ecuación diferencial tiene **solución** y si dicha solución es **única** para una condición inicial dada. Esta verificación es fundamental, ya que garantiza que los resultados que obtengamos al integrar la ecuación sean **bien definidos y consistentes**.  

En particular, revisaremos si la función que define la ecuación cumple con la **condición de Lipschitz** en un intervalo relevante, lo cual, por el teorema de Picard–Lindelöf, nos asegura existencia y unicidad de la solución local. Además, conocer la constante de Lipschitz nos permite acotar el comportamiento de la solución dentro del intervalo físico de interés.

### Existencia y Unicidad de la ecuación de Lotka-Volterra para una especie

Vamos a analizar la ecuación

$$
\frac{dx}{dt} = \mu x \left( 1 - \frac{x}{K} \right),
$$

definiendo

$$
f(t,x) = \mu x \left( 1 - \frac{x}{K} \right).
$$

**Derivada parcial respecto a $x$**

Calculamos la derivada parcial:

$$
\frac{\partial f}{\partial x}(t,x) = \mu - \frac{2\mu}{K} x.
$$

Tomando el valor absoluto, podemos estimar:

$$
\left| \frac{\partial f}{\partial x}(t,x) \right|
= \left| \mu - \frac{2\mu}{K} x \right|
\le |\mu| + \frac{2|\mu|}{K} |x|.
$$

Esta desigualdad muestra que $\left|\frac{\partial f}{\partial x}\right|$ crece linealmente con $|x|$.

**Acotación global y Lipschitz local**

Observamos que cuando $|x|\to\infty$, el valor absoluto de la derivada tiende a infinito. Por lo tanto **no existe una cota global**, y $f$ **no es Lipschitz global**.  

Sin embargo, si restringimos $x$ a un intervalo acotado $|x| \le M$, podemos definir una constante de Lipschitz local:

$$
\left| \frac{\partial f}{\partial x}(t,x) \right| \le |\mu| + \frac{2|\mu|}{K} M := L(M).
$$

Por tanto, $f$ es **localmente Lipschitz**, ya que es $C^1$ en $x$.

**Caso físico: $x \in [0,K]$**

En el intervalo físico $[0,K]$, la derivada parcial varía entre $\mu$ (en $x=0$) y $-\mu$ (en $x=K$). Entonces:

$$
\sup_{x \in [0,K]} \left| \mu - \frac{2\mu}{K} x \right| = |\mu|.
$$

Podemos elegir como constante de Lipschitz:

$$
L = |\mu|.
$$

**Consecuencia para existencia y unicidad**

Como $f$ es $C^1$ en $(t,x)$ y localmente Lipschitz en $x$. Por el teorema de Picard–Lindelöf, para cualquier condición inicial $x(t_0)=x_0$ existe y es única la solución local.  

Si además sabemos que la solución permanece en un conjunto acotado (por ejemplo $[0,K]$), obtenemos existencia y unicidad de la solución en ese intervalo con la constante de Lipschitz correspondiente.

### Solución analítica de la ecuación de Lotka-Volterra para una especie

Para $N = 1$, las ecuaciones competitivas de **Lotka-Volterra** se reducen a la siguiente ecuación diferencial **[6]**:

$$
\frac{dx}{dt} = \mu x\left( 1 - \frac{x}{K} \right)
$$

Separando las variables $x$ y $t$ e integrando, se tiene que:

$$
\int_{x_0}^x \frac{K}{x'(K - x')} \, dx' = \int_0^t \mu \, dt'
$$

La integral del lado izquierdo se puede resolver por fracciones simples. Para ello, se considera:

$$
\begin{split}
\frac{1}{x(K - x)} &= \frac{a}{x} + \frac{b}{K - x} \\
&= \frac{a(K - x) + bx}{x(K - x)}
\end{split}
$$

Igualando los numeradores:

$$
a(K - x) + bx = 1
$$

Evaluando en puntos convenientes:

$$
\left\{
\begin{array}{ll}
x = 0 &\Rightarrow a = \frac{1}{K} \\
x = K &\Rightarrow b = \frac{1}{K}
\end{array}
\right.
$$

Entonces:

$$
\frac{K}{x(K - x)} = \frac{1}{x} + \frac{1}{K - x}
$$

Por lo tanto, la integral puede reescribirse como:

$$
\int_{x_0}^x \left( \frac{1}{x'} + \frac{1}{K - x'} \right) dx' = \int_0^t \mu \, dt'
$$

Resolviendo ambas integrales:

$$
\left[ \ln|x'| - \ln|K - x'| \right]_{x_0}^x = \left[ \mu t' \right]_0^t
$$

Lo que implica:

$$
\ln \left| \frac{x}{K - x} \right| - \ln \left| \frac{x_0}{K - x_0} \right| = \mu t
$$

Agrupando términos:

$$
\frac{x}{K - x} \cdot \frac{K - x_0}{x_0} = e^{\mu t}
$$

Definiendo una constante $C = \frac{K - x_0}{x_0}$:

$$
\frac{x}{K - x} \cdot C = e^{\mu t}
$$

Despejando $x(t)$:

$$
\begin{split}
x &= \frac{1}{C} e^{\mu t} (K - x) \\
x + \frac{1}{C} e^{\mu t} x &= \frac{1}{C} K e^{\mu t} \\
x \left( 1 + \frac{1}{C} e^{\mu t} \right) &= \frac{1}{C} K e^{\mu t} \\
x(t) &= \frac{\frac{1}{C} K e^{\mu t}}{1 + \frac{1}{C} e^{\mu t}} \\
x(t) &= \frac{\frac{x_0}{K - x_0} K e^{\mu t}}{1 + \frac{x_0}{K - x_0} e^{\mu t}} \\
x(t) &= \frac{x_0 K e^{\mu t}}{(K - x_0) + x_0 e^{\mu t}}
\end{split}
$$

O bien, expresado de manera más general:

$$
\boxed{
x(t) = \frac{K}{1 + C e^{-\mu t}}
}
$$

Esta es la conocida **función logística**, también llamada **sigmoide** o **curva S**.

### Comparación de la solución analítica con la solución numérica obtenida mediante RK4

La **ecuación logística** es un modelo clásico en dinámica de poblaciones que describe cómo una población crece inicialmente de manera exponencial y luego se estabiliza al aproximarse a una **capacidad de carga** impuesta por limitaciones ambientales. Su forma diferencial es:

$$
\frac{dx}{dt} = \mu \, x \left(1 - \frac{x}{K}\right)
$$

donde:
- $x(t)$ representa el tamaño de la población en el tiempo $t$,
- $\mu$ es la **tasa de crecimiento**, y  
- $K$ es la **capacidad de carga** del medio.

En esta sección, se compara la **solución exacta analítica** de la ecuación logística con la **solución numérica** obtenida mediante el método de Runge–Kutta de cuarto orden (RK4), implementado a través de la función `solve_ivp` de `SciPy`.

El objetivo es verificar la concordancia entre ambas soluciones y analizar el comportamiento de la población a lo largo del tiempo.

```python
# Importación de librerías
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Parámetros
K = 100 # Capacidad de carga
mu = 0.15 # Tasa de crecimiento
x0 = 0.05 # Condición inicial

t = np.linspace(0, 100, 500)

# Solución exacta
def logistic_exact(t, K, mu, x0):
    """
    Calcula la solución exacta de la ecuación logística para una sola especie.

    La ecuación logística es:
        dx/dt = mu * x * (1 - x / K)

    Inputs
    ------
    t : array_like
        Vector de tiempos donde se desea evaluar la solución.
    K : float
        Capacidad de carga máxima.
    mu : float
        Tasa de crecimiento.
    x0 : float
        Condición inicial de la población.

    Outputs
    -------
    x : np.ndarray
        Valor de la población en cada instante de tiempo del vector `t`.
    """
    C = (K - x0) / x0
    return K / (1 + C * np.exp(-mu * t))

x_exact = logistic_exact(t, K, mu, x0)

# Ecuación logística
def logistic(t, x):
    """
    Función que devuelve la derivada de la población según la ecuación logística.

    dx/dt = mu * x * (1 - x / K)

    Nota: `mu` y `K` se toman como variables globales en este ejemplo.

    Inputs
    ------
    t : float
        Tiempo (no se usa explícitamente, requerido por solve_ivp).
    x : float
        Valor actual de la población.

    Outputs
    -------
    dxdt : float
        Derivada de la población en el tiempo t.
    """
    return mu * x * (1 - x / K)

# Integración numérica RK4
sol = solve_ivp(logistic, [t[0], t[-1]], [x0], t_eval=t, method='RK45')
x_rk4 = sol.y[0]

# Graficar resultados
plt.figure(figsize=(10, 6))
plt.plot(t, x_exact, label='Solución Exacta', color='yellow', linewidth=2)
plt.plot(t, x_rk4, '--', label='Solución Numérica (RK4)', color='red',
         linewidth=2, markevery=20)
plt.axhline(K, color='gray', linestyle='--', label='Capacidad de carga K')
plt.title('Comparación de la solución exacta y la numérica', fontsize=16)
plt.xlabel('Tiempo t', fontsize=14)
plt.ylabel('Población x(t)', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(fontsize=12)
plt.tight_layout()
plt.show()
```

### Análisis del Error y la Estabilidad en la Ecuación Logística

A continuación analizamos dos propiedades clave del modelo logístico y del método numérico usado (Runge–Kutta RK4):

1. **Accuracy (Precisión):**  
   Evalúa qué tan cerca está la solución numérica de la solución exacta del modelo.

2. **Stability (Estabilidad):**  
   Analiza cómo responde el sistema (y el método numérico) ante pequeñas perturbaciones en el tiempo o en las condiciones iniciales.

#### Análisis de Exactitud

En esta sección se evalúa la precisión del método numérico RK4 al resolver la ecuación logística, comparando su solución con la solución exacta.

1. **Errores absoluto y relativo**  
Se calculan los errores en cada punto temporal:

$$
e_{\text{abs}}(t) = |x_{\text{exact}}(t) - x_{\text{RK4}}(t)|, \quad
e_{\text{rel}}(t) = \frac{e_{\text{abs}}(t)}{|x_{\text{exact}}(t)|}
$$

2. **Indicadores globales**  
Se muestran los errores máximo y promedio (absoluto y relativo) para cuantificar la precisión general del método.

3. **Visualización**  
Se muestran **dos gráficos** que permiten comparar magnitudes y tendencias:
- (A) Errores absoluto y relativo.
- (B) Errores máximo vs número de pasos.

De esta forma, se pueden apreciar las magnitudes reales de los errores y cómo se comportan al refinar la discretización temporal.

```python
# <><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><>
# Análisis de Exactitud
# <><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><>

print("\n" + "<>" * 36)
print("ANÁLISIS DE ERROR")
print("=" * 72)

# Errores absoluto y relativo
error_absoluto = np.abs(x_exact - x_rk4)
error_relativo = error_absoluto / np.abs(x_exact)

# Métricas globales
print(f"Error absoluto máximo: {np.max(error_absoluto):.2e}")
print(f"Error relativo máximo: {np.max(error_relativo):.2e}")
print(f"Error absoluto promedio: {np.mean(error_absoluto):.2e}")
print(f"Error relativo promedio: {np.mean(error_relativo):.2e}")

print("<>" * 36 + "\n")

# Estudio de convergencia
pasos = [50, 100, 200, 500, 1000]
errores_maximos = []

for n_pasos in pasos:
    t_local = np.linspace(0, 100, n_pasos)
    x_exact_local = logistic_exact(t_local, K, mu, x0)
    sol_local = solve_ivp(logistic, [t_local[0], t_local[-1]], [x0],
                         t_eval=t_local, method='RK45', rtol=1e-8, atol=1e-10)
    x_rk4_local = sol_local.y[0]
    error_max = np.max(np.abs(x_exact_local - x_rk4_local))
    errores_maximos.append(error_max)

# Visualización

fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# (A) Error absoluto y relativo en escala común
ax[0].plot(t, error_absoluto, label='Error absoluto', color='red')
ax[0].plot(t, error_relativo, label='Error relativo', color='blue')
ax[0].set_title('Error absoluto y relativo (escala común)', fontsize=13)
ax[0].set_xlabel('Tiempo t')
ax[0].set_ylabel('Error')
ax[0].grid(True, linestyle='--', alpha=0.7)
ax[0].legend()

# (B) Error máximo vs número de pasos en escala común
ax[1].plot(pasos, errores_maximos, 'o-', color='green', linewidth=2)
ax[1].set_title('Error máximo vs Número de Pasos', fontsize=13)
ax[1].set_xlabel('Número de pasos')
ax[1].set_ylabel('Error máximo')
ax[1].grid(True, linestyle='--', alpha=0.7)

# Marcar el paso con menor error máximo con línea punteada vertical
indice_min = np.argmin(errores_maximos)
mejor_paso = pasos[indice_min]
mejor_error = errores_maximos[indice_min]

ax[1].axvline(mejor_paso, color='red', linestyle='--', linewidth=1.5, label=f'Mínimo error: {mejor_paso} pasos')
ax[1].legend()

plt.tight_layout()
plt.show()
```

#### Estabilidad del Sistema

El objetivo del análisis de estabilidad es comprender cómo responde el sistema ante **pequeñas perturbaciones** alrededor de un punto de equilibrio. En otras palabras, buscamos saber si el sistema **retorna a su estado de equilibrio** o si **se aleja de él** cuando se lo perturba ligeramente.

Este análisis tiene dos dimensiones complementarias:

1. **Estabilidad del modelo dinámico (estabilidad teórica)**

    La ecuación logística puede escribirse como:
    
    $$
    \frac{dx}{dt} = f(x) = \mu x \left(1 - \frac{x}{K}\right)
    $$
    
    Un **punto de equilibrio** ($x^*$) se define como aquel donde ($f(x^*) = 0$).
    En el modelo logístico existen dos puntos fijos:
    
    $$
    x_1 = 0, \quad x_2 = K
    $$
    
    Para determinar la **estabilidad local** de cada equilibrio, se linealiza la ecuación alrededor de ($x^*$).
    Esto se hace evaluando la derivada (Jacobiano) de ($f(x)$):
    
    $$
    \lambda = f'(x^*) = \frac{df}{dx}\bigg|_{x=x^*} = \mu \left(1 - \frac{2x^*}{K}\right)
    $$
    
    Dado que este sistema consta de una sola especie, el Jacobiano se reduce a esta derivada escalar.
    El **signo de ($\lambda$)** determina la estabilidad:
    
    * Si ( $\lambda < 0$ ), el equilibrio es **estable** (atractor):
      pequeñas perturbaciones tienden a desaparecer con el tiempo.
    * Si ( $\lambda > 0$ ), el equilibrio es **inestable** (repulsor):
      las perturbaciones crecen y el sistema se aleja del equilibrio.
    
    En el caso logístico:
    
    $$
    f'(0) = \mu > 0 \quad \Rightarrow \text{el punto } x=0 \text{ es inestable}
    $$
    
    $$
    f'(K) = -\mu < 0 \quad \Rightarrow \text{el punto } x=K \text{ es estable}
    $$
    
    Esto significa que la población tenderá siempre hacia ($K$) (la capacidad de carga), independientemente del valor inicial ($x_0 > 0$).

2. **Estabilidad numérica**
    
    Además de la estabilidad del modelo continuo, es importante que el **método numérico** conserve el comportamiento cualitativo del sistema.
    Un integrador numérico estable debería reproducir correctamente la tendencia hacia el equilibrio estable ($x = K$).
    
    Para evaluar esto, se resuelve el sistema para diferentes condiciones iniciales ($x_0$) (por ejemplo, por debajo o por encima de $K$) y se analiza si todas las trayectorias numéricas **convergen hacia el mismo equilibrio**.
    Si el método es numéricamente inestable, podrían aparecer oscilaciones espurias o divergencias, incluso si el modelo teórico es estable.
    
    En resumen, el análisis de estabilidad permite verificar dos cosas fundamentales:   
    
    * Que el **modelo dinámico** posee el comportamiento esperado (atractores o repulsores).
    * Que el **método numérico** utilizado respeta esa estructura y no introduce inestabilidades artificiales.

```python
# <><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><>
# Análisis de estabilidad del sistema (teórico)
# <><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><>

# Importación de librerías
from scipy.optimize import fsolve

print("\n" + "<>" * 36)
print("ANÁLISIS DE ESTABILIDAD DEL SISTEMA")
print("=" * 72)

# 1. Puntos de equilibrio
def ecuacion_equilibrio(x):
    """
    Ecuación para determinar los puntos de equilibrio del sistema logístico.

    La ecuación logística es:
        dx/dt = mu * x * (1 - x / K)
    Un punto de equilibrio x* cumple dx/dt = 0.

    Inputs
    ------
    x : float
        Valor de la población.

    Outputs
    -------
    float
        Valor de dx/dt evaluado en x.
    """
    return mu * x * (1 - x / K)

# Encontrar puntos fijos
puntos_equilibrio = []
for x_guess in [0, 50, 150]:
    punto = fsolve(ecuacion_equilibrio, x_guess)[0]
    if not any(np.isclose(punto, p, atol=1e-6) for p in puntos_equilibrio):
        puntos_equilibrio.append(punto)

print("Puntos de equilibrio encontrados:")
print("-" * len("Puntos de equilibrio encontrados:"))
for i, punto in enumerate(puntos_equilibrio):
    print(f"  x_{i} = {punto:.6f}")

# 2. Análisis de estabilidad lineal
def jacobiano(x):
    """
    Jacobiano del sistema logístico para análisis de estabilidad lineal.

    El jacobiano en un punto de equilibrio x* se define como:
        λ = df/dx |_{x=x*} = mu * (1 - 2*x*/K)

    Inputs
    ------
    x : float
        Valor de la población (punto de equilibrio).

    Outputs
    -------
    float
        Autovalor del jacobiano en x.  
        Si λ < 0, el punto es estable; si λ > 0, es inestable.
    """
    return mu * (1 - 2*x/K)

print("\nAnálisis de estabilidad lineal:")
print("-" * len("Análisis de estabilidad lineal:"))
for punto in puntos_equilibrio:
    autovalor = jacobiano(punto)
    estabilidad = "Estable" if autovalor < 0 else "Inestable"
    print(f"  x = {punto:.2f}: λ = {autovalor:.4f} ({estabilidad})")

print("<>" * 36 + "\n")

# 3. Simulación con distintas condiciones iniciales
condiciones_iniciales = [0.01, 50, 150, 200]
colores = ['blue', 'green', 'orange', 'purple']

plt.figure(figsize=(10, 6))
for x0_test, color in zip(condiciones_iniciales, colores):
    x_exact_test = logistic_exact(t, K, mu, x0_test)
    plt.plot(t, x_exact_test, label=f'x₀ = {x0_test}', color=color, linewidth=2)

# Marcar puntos de equilibrio
for punto in puntos_equilibrio:
    plt.axhline(punto, color='red', linestyle=':', alpha=0.7, label=f'Equilibrio: {punto:.1f}')

plt.title('Estabilidad: Diferentes Condiciones Iniciales', fontsize=16)
plt.xlabel('Tiempo t')
plt.ylabel('Población x(t)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.ylim(-10, 220)
plt.tight_layout()
plt.show()
```

```python
# <><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><>
# Estabilidad Numérica usando RK45
# <><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><><>

# Condiciones iniciales de prueba
condiciones_iniciales = [0.01, 50, 150, 200]
colores = ['blue', 'green', 'orange', 'purple']

plt.figure(figsize=(10, 6))

# Integrar con RK45 para cada condición inicial
for x0_test, color in zip(condiciones_iniciales, colores):
    sol_rk45 = solve_ivp(
        logistic, [t[0], t[-1]], [x0_test],
        t_eval=t, method='RK45', rtol=1e-8, atol=1e-10
    )
    plt.plot(sol_rk45.t, sol_rk45.y[0], label=f'x₀ = {x0_test}', color=color, linewidth=2)

# Marcar puntos de equilibrio
for punto in puntos_equilibrio:
    plt.axhline(punto, color='red', linestyle=':', alpha=0.7, label=f'Equilibrio: {punto:.1f}')

plt.title('Estabilidad Numérica: Diferentes Condiciones Iniciales (RK45)', fontsize=16)
plt.xlabel('Tiempo t')
plt.ylabel('Población x(t)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.ylim(-10, 220)
plt.tight_layout()
plt.show()
```

## Caso multiespecie ($N = 2$)

Este análisis traduce los parámetros experimentales reportados en el paper de [Dedrick et al.](https://journals.asm.org/doi/full/10.1128/msystems.00757-22) al formato de la ecuación logística tipo Lotka–Volterra utilizada en este proyecto:

$$
\frac{dx_i}{dt}=\mu_i x_i\left(1-\frac{\sum_j \alpha_{ij}x_j}{K_i}\right)
$$

y a su forma equivalente generalizada (GLV):

$$
\frac{dx_i}{dt}=x_i\left(\mu_i+\sum_j \beta_{ij}x_j\right)
$$

**Datos reportados (paper)**

- $\mu_1=1.85\ \text{h}^{-1}$, $K_1=0.61\ \text{OD}_{600}$
- $\mu_2=1.56\ \text{h}^{-1}$, $K_2=0.35\ \text{OD}_{600}$
- Capacidad de carga en spent medium:
  - $K_{12}=0.17$ (efecto de 2 sobre 1)
  - $K_{21}=0.03$ (efecto de 1 sobre 2)
- Coeficientes reportados:
  - $c_{12}=-1.26$
  - $c_{21}=-0.52$

El paper define:

$$
c_{ij} = \frac{K_{ij}-K_i}{K_j}
$$

**Verificación:**

$$
c_{12}=\frac{0.17-0.61}{0.35}=-1.26, \qquad
c_{21}=\frac{0.03-0.35}{0.61}=-0.52
$$

**Relación entre notaciones**

Partiendo de la forma generalizada (GLV):

$$
\frac{dx_i}{dt} = x_i \left( \mu_i + \sum_j \beta_{ij} x_j \right)
$$

y comparándola con la forma logística:

$$
\frac{dx_i}{dt} = \mu_i x_i \left(1 - \frac{\sum_j \alpha_{ij} x_j}{K_i}\right)
$$

se obtiene la relación:

$$
\beta_{ij} = -\frac{\mu_i}{K_i}\alpha_{ij} \quad \Longleftrightarrow \quad \alpha_{ij} = -\frac{K_i}{\mu_i}\beta_{ij}
$$

El paper además vincula sus $c_{ij}$ con los coeficientes GLV mediante:

$$
\beta_{ij} = c_{ij} \frac{\mu_i}{K_i}
$$

Combinando ambas expresiones:

$$
\boxed{\alpha_{ij} = -c_{ij}} \quad \text{(para } i \ne j \text{)}
$$

**Valores numéricos**

Dado que el modelo logístico normaliza la autointeracción, se asume:

$$
\alpha_{11} = \alpha_{22} = 1
$$

y a partir de los $c_{ij}$ del paper:

$$
\alpha_{12} = -c_{12} = 1.26, \qquad
\alpha_{21} = -c_{21} = 0.52
$$

La matriz de interacciones queda:

$$
\alpha =
\begin{pmatrix}
1.00 & 1.26 \\
0.52 & 1.00
\end{pmatrix}
$$

**Forma GLV equivalente**

Usando la relación $\beta_{ij} = -\dfrac{\mu_i}{K_i}\alpha_{ij}$:

$$
\beta =
\begin{pmatrix}
-\dfrac{1.85}{0.61}(1.00) & -\dfrac{1.85}{0.61}(1.26) \\
-\dfrac{1.56}{0.35}(0.52) & -\dfrac{1.56}{0.35}(1.00)
\end{pmatrix}
\approx
\begin{pmatrix}
-3.03 & -3.82 \\
-2.32 & -4.46
\end{pmatrix}
$$

**Ecuaciones finales**

**Forma logística (modelo de Lotka–Volterra)**

$$
\begin{aligned}
\frac{dx_1}{dt} &= 1.85\,x_1\left(1 - \frac{1.00\,x_1 + 1.26\,x_2}{0.61}\right), \\
\frac{dx_2}{dt} &= 1.56\,x_2\left(1 - \frac{0.52\,x_1 + 1.00\,x_2}{0.35}\right)
\end{aligned}
$$

**Forma GLV equivalente**

$$
\begin{aligned}
\frac{dx_1}{dt} &= x_1\left(1.85 - 3.03\,x_1 - 3.82\,x_2\right), \\
\frac{dx_2}{dt} &= x_2\left(1.56 - 2.32\,x_1 - 4.46\,x_2\right)
\end{aligned}
$$

**Interpretación biológica**

- $\alpha_{ii}=1$ se adopta por **convención**, para conservar la interpretación de $K_i$ como la capacidad de carga intrínseca.  
- $\alpha_{12}=1.26$: la especie 2 ejerce un efecto inhibitorio mayor que la autolimitación de la especie 1.  
- $\alpha_{21}=0.52$: la especie 1 inhibe menos a la especie 2.  
- Signos negativos en $\beta_{ij}$ indican competencia.  
- Las unidades: $\mu_i$ en $\text{h}^{-1}$, $K_i$ en $\text{OD}_{600}$ (densidad óptica).

### Extracción y preprocesamiento de datos experimentales

Los datos experimentales provienen de ensayos de crecimiento en microplaca, donde se registran simultáneamente:

- **Densidad óptica total** (OD600), utilizada como proxy de biomasa total.
- **Fluorescencia GFP**, empleada para estimar la abundancia de una de las especies en cocultivo.

El objetivo de esta etapa es transformar las lecturas crudas del archivo experimental en **DataFrames estructurados**, adecuados para análisis estadístico, visualización y ajuste de modelos dinámicos poblacionales.

**Eje temporal**

Las mediciones se realizan a intervalos regulares de tiempo. El eje temporal se define como:

$$
t_n = n \cdot \Delta t, \qquad n = 1, 2, \dots, N
$$

donde:

- $\Delta t$ es el intervalo entre lecturas (en horas),
- $N$ es el número total de puntos temporales.

Cada lectura experimental queda asociada explícitamente a su correspondiente valor de $t_n$.

**Estructuración de los datos de microplaca**

Las lecturas de la microplaca se reorganizan en formato **long/tidy**, donde cada fila representa una medición individual caracterizada por:

- tiempo ($t$),
- fila de la placa,
- columna de la placa,
- identificador del pozo (*well*),
- tipo de medición (OD600 o GFP),
- valor de la señal.

Este formato permite un manejo flexible de los datos, facilitando el filtrado por condiciones experimentales, la agregación de réplicas y la integración con rutinas de análisis y visualización.

Conceptualmente, cada señal se representa como:

$$
y = y(\text{pozo}, t)
$$

**Selección del cocultivo 1:1**

Para el análisis de cocultivo se selecciona un subconjunto específico de pozos correspondientes a una condición 1:1 entre especies.  
Cada uno de estos pozos se considera una **réplica experimental independiente**.

A partir de esta selección se construye un DataFrame que contiene, para cada réplica y cada punto temporal:

- OD total,
- señal GFP,
- identificador de réplica,
- tiempo de medición.

**Corrección de línea base (*baseline correction*)**

Las señales experimentales pueden presentar un offset inicial asociado al medio de cultivo o al sistema de medición.  
Para cada réplica se estima una línea base utilizando los primeros puntos temporales:

$$
\bar{y}_{\text{baseline}} = \frac{1}{N_b} \sum_{n=1}^{N_b} y(t_n)
$$

donde $N_b$ es el número de puntos iniciales utilizados para el cálculo del baseline.

La señal corregida se define como:

$$
y_{\text{corr}}(t) = y(t) - \bar{y}_{\text{baseline}}
$$

Este procedimiento se aplica de manera independiente tanto a la OD total como a la señal GFP.

**Conversión de GFP a densidad óptica**

La fluorescencia GFP se utiliza como proxy de la abundancia de una de las especies del cocultivo.  
Para expresarla en las mismas unidades que la OD total, se aplica una conversión lineal:

$$
\text{OD}_{\text{Sa}}(t) = \gamma \cdot \text{GFP}_{\text{corr}}(t)
$$

donde $\gamma$ es un factor de calibración determinado experimentalmente.

Esta conversión permite trabajar con ambas especies en una escala común de densidad óptica.

**Reconstrucción de la segunda especie**

La abundancia de la segunda especie en el cocultivo se estima por diferencia, asumiendo que la densidad óptica total es aditiva:

$$
\text{OD}_{\text{Sna}}(t) = \text{OD}_{\text{total}}(t) - \text{OD}_{\text{Sa}}(t)
$$

De este modo, para cada réplica y cada tiempo se obtienen estimaciones consistentes de ambas poblaciones.

**DataFrame final del cocultivo**

El resultado de la etapa de extracción es un DataFrame que contiene, para cada réplica y punto temporal:

- tiempo ($t$),
- identificador de réplica,
- OD total corregida,
- OD estimada de la especie marcada,
- OD estimada de la especie complementaria.

Este DataFrame constituye la **base de datos principal** para el análisis posterior.

**Agregación de réplicas**

Para análisis a nivel poblacional, las réplicas se agregan punto a punto en el tiempo, calculando estadísticos descriptivos:

$$
\mu(t) = \frac{1}{R} \sum_{r=1}^{R} x_r(t),
\qquad
\sigma(t) = \sqrt{\frac{1}{R-1} \sum_{r=1}^{R} \left[x_r(t) - \mu(t)\right]^2}
$$

donde $R$ es el número de réplicas y $x_r(t)$ la señal correspondiente a la réplica $r$.

El DataFrame agregado se utiliza para la visualización de curvas promedio y la comparación directa con modelos dinámicos.

```python
# <><><><><><><><><><><><><><><><><><><><><><><><><><><><><><>
# Extracción de datos
# <><><><><><><><><><><><><><><><><><><><><><><><><><><><><><>

from __future__ import annotations

from pathlib import Path
import numpy as np
import pandas as pd


# ============================================================
# Parsing token-based del archivo de microplaca
# ============================================================
def find_token_index(tokens: list[str], target: str, start: int = 0) -> int:
    """
    Busca la posición de un token específico dentro de una lista tokenizada.

    Inputs:
        - tokens: Lista de strings (contenido del archivo separado por espacios).
        - target: Token exacto a buscar.
        - start: Índice desde el cual comenzar la búsqueda.

    Outputs:
        - idx: Índice entero donde aparece `target` por primera vez a partir de `start`.

    Autor:
        Tecnologías Digitales
    """
    for idx in range(start, len(tokens)):
        if tokens[idx] == target:
            return idx
    raise ValueError(f"Token '{target}' no encontrado en el archivo.")


def parse_plate_block(
    tokens: list[str],
    start_idx: int,
    n_rows: int,
    n_cols: int,
    n_timepoints: int,
) -> tuple[np.ndarray, int]:
    """
    Parsea un bloque de lecturas de microplaca (p. ej. OD600 o GFP) y lo devuelve como
    un array 3D con forma (rows, cols, time).

    Nota:
        El bloque se lee secuencialmente mediante tokens, respetando el patrón de
        encabezados y separadores del archivo.

    Inputs:
        - tokens: Lista de strings tokenizados.
        - start_idx: Índice a partir del cual comienza el bloque numérico (justo después del header "Time").
        - n_rows: Cantidad de filas de la placa.
        - n_cols: Cantidad de columnas de la placa.
        - n_timepoints: Número de lecturas temporales.

    Outputs:
        - values_3d: np.ndarray de forma (n_rows, n_cols, n_timepoints) con los valores numéricos.
        - next_idx: Índice del token donde termina el bloque (útil para continuar el parseo).

    Autor:
        Tecnologías Digitales
    """
    values_3d = np.zeros((n_rows, n_cols, n_timepoints), dtype=float)
    idx = start_idx

    for t in range(n_timepoints):
        _read_id = int(tokens[idx])   # identificador de lectura (no se usa aquí)
        idx += 1

        _time_str = tokens[idx]       # string de tiempo (no se usa aquí)
        idx += 1

        # saltos fijos del archivo
        idx += 3
        idx += 13

        for i in range(n_rows):
            for j in range(n_cols):
                values_3d[i, j, t] = float(tokens[idx])
                idx += 1
            idx += 4  # separadores de fin de fila

        idx += 3      # separadores de fin de bloque temporal

    return values_3d, idx


def plate_3d_to_long_dataframe(
    values_3d: np.ndarray,
    time_hours: np.ndarray,
    measurement: str,
) -> pd.DataFrame:
    """
    Convierte un array 3D (rows, cols, time) a un DataFrame long/tidy.

    Inputs:
        - values_3d: Array con forma (n_rows, n_cols, n_timepoints).
        - time_hours: Array con forma (n_timepoints,) con el tiempo en horas.
        - measurement: Nombre de la medición (por ejemplo "OD600" o "GFP").

    Outputs:
        - df_long: DataFrame con columnas:
            ["time_hours", "read_index", "plate_row", "plate_col", "well_id",
             "measurement", "signal_value"]

    Autor:
        Tecnologías Digitales
    """
    n_rows, n_cols, n_t = values_3d.shape

    # Mantener índices de placa en 1-based (más representativo en microplacas)
    plate_rows = np.arange(1, n_rows + 1)
    plate_cols = np.arange(1, n_cols + 1)

    rr, cc, tt = np.meshgrid(plate_rows, plate_cols, np.arange(n_t), indexing="ij")

    df_long = pd.DataFrame({
        "time_hours": time_hours[tt.ravel()],
        "read_index": (tt.ravel() + 1).astype(int),
        "plate_row": rr.ravel().astype(int),
        "plate_col": cc.ravel().astype(int),
        "measurement": measurement,
        "signal_value": values_3d.reshape(-1),
    })

    df_long["well_id"] = (
        "R" + df_long["plate_row"].astype(str).str.zfill(2)
        + "C" + df_long["plate_col"].astype(str).str.zfill(2)
    )

    df_long = df_long[
        ["time_hours", "read_index", "plate_row", "plate_col", "well_id", "measurement", "signal_value"]
    ].sort_values(["plate_row", "plate_col", "read_index"], ignore_index=True)

    return df_long


def load_microplate_txt_to_dataframes(
    filepath: str | Path,
    n_rows: int = 8,
    n_cols: int = 12,
    n_timepoints: int = 96,
    dt_hours: float = 10 / 60,
) -> tuple[pd.DataFrame, pd.DataFrame]:
    """
    Carga un archivo .txt de lecturas de microplaca y devuelve dos DataFrames long/tidy:
    uno para OD600 y otro para GFP (si ambos bloques están presentes).

    Inputs:
        - filepath: Ruta al archivo .txt.
        - n_rows: Filas de la placa (default 8).
        - n_cols: Columnas de la placa (default 12).
        - n_timepoints: Lecturas temporales (default 96).
        - dt_hours: Intervalo temporal entre lecturas en horas (default 10/60).

    Outputs:
        - df_od600: DataFrame long de OD600.
        - df_gfp: DataFrame long de GFP.

    Autor:
        Tecnologías Digitales
    """
    filepath = Path(filepath)
    raw_text = filepath.read_text(encoding="utf-8", errors="ignore")
    tokens = raw_text.split()

    time_hours = dt_hours * np.arange(1, n_timepoints + 1)

    # Bloque OD600
    idx_temp = find_token_index(tokens, "Temperature", start=0)
    idx_time = find_token_index(tokens, "Time", start=idx_temp)
    od600_3d, idx_after_od = parse_plate_block(tokens, idx_time + 1, n_rows, n_cols, n_timepoints)

    # Bloque GFP
    idx_temp2 = find_token_index(tokens, "Temperature:", start=idx_after_od)
    idx_time2 = find_token_index(tokens, "Time", start=idx_temp2)
    gfp_3d, _ = parse_plate_block(tokens, idx_time2 + 1, n_rows, n_cols, n_timepoints)

    df_od600 = plate_3d_to_long_dataframe(od600_3d, time_hours, measurement="OD600")
    df_gfp = plate_3d_to_long_dataframe(gfp_3d, time_hours, measurement="GFP")

    return df_od600, df_gfp


# ============================================================
# Curación: extracción de cocultivo 1:1 en DataFrame
# ============================================================
def extract_coculture_timeseries_dataframe(
    df_od600: pd.DataFrame,
    df_gfp: pd.DataFrame,
    plate_rows_1based: range = range(2, 8),  # replicates: rows 2..7
    plate_col_1based: int = 5,
    baseline_points: int = 5,
    gfp_to_od_factor: float = 1.0 / 192970,
) -> pd.DataFrame:
    """
    Construye un DataFrame con series temporales del cocultivo 1:1 a partir de OD600 y GFP.

    Procedimiento:
        1) Filtra un subconjunto de wells (rango de filas y una columna).
        2) Estima y sustrae baseline por replicate (primeros `baseline_points`).
        3) Convierte GFP→OD para estimar la contribución de la especie marcada (Sa).
        4) Reconstruye la otra especie como: other = total_od - sa_od_est.

    Inputs:
        - df_od600: DataFrame long OD600.
        - df_gfp: DataFrame long GFP.
        - plate_rows_1based: Filas (1-based) a considerar como replicates.
        - plate_col_1based: Columna (1-based) del well de interés.
        - baseline_points: Cantidad de puntos iniciales para baseline-correction.
        - gfp_to_od_factor: Factor de conversión de GFP a OD (calibración).

    Outputs:
        - df_coculture: DataFrame con columnas:
            ["time_hours", "read_index", "replicate_id",
             "sa_od_est", "total_od", "sna_od_est",
             "sa_baseline_gfp", "total_baseline_od"]

    Autor:
        Tecnologías Digitales
    """
    target_rows = list(plate_rows_1based)
    target_col = plate_col_1based

    od_sub = df_od600[
        (df_od600["plate_row"].isin(target_rows)) &
        (df_od600["plate_col"] == target_col)
    ].copy()

    gfp_sub = df_gfp[
        (df_gfp["plate_row"].isin(target_rows)) &
        (df_gfp["plate_col"] == target_col)
    ].copy()

    # Un identificador semántico de replicate
    od_sub["replicate_id"] = od_sub["plate_row"].apply(lambda r: f"replicate_row_{r:02d}")
    gfp_sub["replicate_id"] = gfp_sub["plate_row"].apply(lambda r: f"replicate_row_{r:02d}")

    # Baseline por replicate (primeros puntos)
    total_baseline = (
        od_sub[od_sub["read_index"] <= baseline_points]
        .groupby("replicate_id", as_index=False)["signal_value"]
        .mean()
        .rename(columns={"signal_value": "total_baseline_od"})
    )

    sa_baseline = (
        gfp_sub[gfp_sub["read_index"] <= baseline_points]
        .groupby("replicate_id", as_index=False)["signal_value"]
        .mean()
        .rename(columns={"signal_value": "sa_baseline_gfp"})
    )

    # Merge de baselines
    od_sub = od_sub.merge(total_baseline, on="replicate_id", how="left")
    gfp_sub = gfp_sub.merge(sa_baseline, on="replicate_id", how="left")

    # Baseline-correction
    od_sub["total_od"] = od_sub["signal_value"] - od_sub["total_baseline_od"]
    gfp_sub["sa_gfp_corrected"] = gfp_sub["signal_value"] - gfp_sub["sa_baseline_gfp"]

    # GFP → OD (Sa)
    gfp_sub["sa_od_est"] = gfp_to_od_factor * gfp_sub["sa_gfp_corrected"]

    # Join por replicate + tiempo
    df_coculture = (
        od_sub[["time_hours", "read_index", "replicate_id", "total_od", "total_baseline_od"]]
        .merge(
            gfp_sub[["time_hours", "read_index", "replicate_id", "sa_od_est", "sa_baseline_gfp"]],
            on=["time_hours", "read_index", "replicate_id"],
            how="inner",
        )
        .sort_values(["replicate_id", "read_index"], ignore_index=True)
    )

    # Reconstrucción de Sna
    df_coculture["sna_od_est"] = df_coculture["total_od"] - df_coculture["sa_od_est"]

    # Orden sugerido de columnas
    df_coculture = df_coculture[
        ["time_hours", "read_index", "replicate_id",
         "sa_od_est", "total_od", "sna_od_est",
         "sa_baseline_gfp", "total_baseline_od"]
    ]

    return df_coculture


def summarize_coculture_replicates(df_coculture: pd.DataFrame) -> pd.DataFrame:
    """
    Resume replicates del cocultivo generando estadísticas por tiempo.

    Inputs:
        - df_coculture: DataFrame generado por extract_coculture_timeseries_dataframe.

    Outputs:
        - df_summary: DataFrame con columnas:
            ["time_hours", "read_index",
             "sa_od_mean", "sa_od_std",
             "sna_od_mean", "sna_od_std",
             "total_od_mean", "total_od_std"]

    Autor:
        Tecnologías Digitales
    """
    df_summary = (
        df_coculture
        .groupby(["time_hours", "read_index"], as_index=False)
        .agg(
            sa_od_mean=("sa_od_est", "mean"),
            sa_od_std=("sa_od_est", "std"),
            sna_od_mean=("sna_od_est", "mean"),
            sna_od_std=("sna_od_est", "std"),
            total_od_mean=("total_od", "mean"),
            total_od_std=("total_od", "std"),
        )
        .sort_values(["read_index"], ignore_index=True)
    )
    return df_summary


# ============================================================
# DataFrames
# ============================================================

file_path = Path("datasets/dedrick/Cocultures_GFPSa_1850_1821_MC_Day1_29JUL2019.txt")

# df_od600:
# DataFrame en formato long/tidy que contiene todas las lecturas de densidad óptica (OD600)
# de la microplaca completa. Cada fila representa una medición individual en el tiempo
# para un pozo específico (identificado por fila y columna de la placa).
# Este DataFrame constituye la señal total de biomasa y sirve como base para:
#   - seleccionar subconjuntos de wells,
#   - calcular baselines,
#   - reconstruir contribuciones de especies en cocultivo.
df_od600, df_gfp = load_microplate_txt_to_dataframes(file_path)

# df_gfp:
# DataFrame en formato long/tidy que contiene todas las lecturas de fluorescencia GFP
# de la microplaca completa. Cada fila representa una medición individual en el tiempo
# para un pozo específico.
# Este DataFrame se utiliza para estimar la abundancia de la especie marcada
# (Sa) mediante conversión GFP → OD, luego de la corrección de baseline.
# (Se devuelve junto con df_od600 en la misma llamada al loader).

# df_coculture:
# DataFrame con las series temporales del cocultivo 1:1, restringido a un subconjunto
# específico de wells (replicates).
# Incluye, para cada replicate y tiempo:
#   - la OD total corregida (total_od),
#   - la estimación de OD de Sa a partir de GFP (sa_od_est),
#   - la estimación de OD de Sna por diferencia (sna_od_est).
# Este DataFrame es la base directa para el ajuste de modelos dinámicos.
df_coculture = extract_coculture_timeseries_dataframe(df_od600, df_gfp)

# df_coculture_summary:
# DataFrame agregado que resume el comportamiento promedio del cocultivo
# a lo largo del tiempo.
# Para cada punto temporal se reportan:
#   - media y desviación estándar de Sa,
#   - media y desviación estándar de Sna,
#   - media y desviación estándar de la OD total.
# Este DataFrame es el que se utiliza típicamente para:
#   - visualización de curvas promedio,
#   - comparación con simulaciones,
#   - ajuste de parámetros a nivel poblacional.
df_coculture_summary = summarize_coculture_replicates(df_coculture)
```

### Metodología de simulación y ajuste del modelo dinámico en cocultivo

En esta etapa se implementa un **modelo dinámico determinista** para describir la evolución temporal de un cocultivo bacteriano de dos especies, combinando:

- un **modelo logístico tipo Lotka–Volterra** con interacción interespecífica,
- **integración numérica continua** mediante un esquema Runge–Kutta adaptativo de orden 4(5),
- y **estimación de parámetros** por mínimos cuadrados no lineales a partir de datos experimentales.

El objetivo es **comparar las trayectorias poblacionales observadas experimentalmente** con las predicciones del modelo, tanto utilizando parámetros derivados de experimentos independientes (*spent medium*) como parámetros ajustados directamente al cocultivo.

#### Implementación del modelo dinámico

**Ecuaciones diferenciales**

La dinámica poblacional de las dos especies se describe mediante el siguiente sistema de ecuaciones diferenciales ordinarias:

$$
\begin{aligned}
\frac{dx_1}{dt} &= \mu_1 \, x_1
\left(
1 - \frac{x_1 - c_{12} x_2}{K_1}
\right), \\
\frac{dx_2}{dt} &= \mu_2 \, x_2
\left(
1 - \frac{x_2 - c_{21} x_1}{K_2}
\right),
\end{aligned}
$$

donde:

- $x_1(t)$ y $x_2(t)$ representan las densidades ópticas (OD) de cada especie,
- $\mu_i$ son las tasas de crecimiento intrínsecas,
- $K_i$ son las capacidades de carga en monocultivo,
- $c_{ij}$ son los coeficientes de interacción interespecífica.

Un valor $c_{ij} < 0$ indica que la especie $j$ ejerce un efecto inhibitorio sobre la especie $i$.

**Modelado del tiempo de latencia (*lag time*)**

Para capturar el retardo observado experimentalmente en el inicio del crecimiento, se introduce un **lag time específico por especie**, implementado como una dinámica por partes:

$$
\frac{dx_i}{dt} =
\begin{cases}
0, & t \le t_{i,\text{lag}}, \\
\mu_i x_i \left(1 - \dfrac{x_i - c_{ij} x_j}{K_i}\right),
& t > t_{i,\text{lag}}.
\end{cases}
$$

Antes del tiempo $t_{i,\text{lag}}$, la población $x_i$ permanece constante.  
Este enfoque reproduce el tratamiento utilizado en la implementación original del artículo de Dedrick *et al.*

**Integración numérica - Método RK4(5)**

El sistema de ecuaciones diferenciales se integra numéricamente utilizando un esquema **Runge–Kutta adaptativo de orden 4(5)**, implementado mediante `solve_ivp` con el método `"RK45"`.

Este método:

- ajusta automáticamente el paso temporal para controlar el error local,
- produce una solución continua en el tiempo,
- permite evaluar la solución exactamente en los tiempos experimentales $t_n$.

Formalmente, el integrador aproxima la solución del problema:

$$
\frac{d\mathbf{x}}{dt} = \mathbf{f}(\mathbf{x}, t; \theta),
\qquad
\mathbf{x}(t_0) = \mathbf{x}_0,
$$

donde $\theta$ representa el conjunto de parámetros del modelo.

**Datos experimentales utilizados**

Las trayectorias experimentales utilizadas para la comparación corresponden a los **promedios de réplicas** del cocultivo, organizadas en un DataFrame agregado que contiene:

- $x_{1,\text{obs}}(t_n)$: OD promedio de la especie marcada,
- $x_{2,\text{obs}}(t_n)$: OD promedio de la segunda especie,
- $t_n$: tiempos de medición, definidos como
  $$
  t_n = n \cdot \Delta t.
  $$

Para mantener coherencia con la implementación original, el primer punto temporal puede forzarse a cero tras la corrección de baseline.

**Estimación de parámetros por mínimos cuadrados**

**Parámetros ajustados**

Se ajustan los coeficientes de interacción:

$$
\theta = (c_{12}, c_{21}),
$$

mientras que los parámetros $\mu_i$, $K_i$ y los tiempos de latencia se consideran fijos y conocidos a partir de experimentos independientes.

**Función de residuales**

Para un conjunto dado de parámetros $\theta$, se simula el modelo y se define el vector de residuales como:

$$
\mathbf{r}(\theta) =
\begin{bmatrix}
x_{1,\text{sim}}(t_1;\theta) - x_{1,\text{obs}}(t_1) \\
\vdots \\
x_{1,\text{sim}}(t_N;\theta) - x_{1,\text{obs}}(t_N) \\
x_{2,\text{sim}}(t_1;\theta) - x_{2,\text{obs}}(t_1) \\
\vdots \\
x_{2,\text{sim}}(t_N;\theta) - x_{2,\text{obs}}(t_N)
\end{bmatrix}.
$$

**Problema de optimización**

La estimación de parámetros se formula como un problema de **mínimos cuadrados no lineales**:

$$
\min_{\theta}
\;\; \|\mathbf{r}(\theta)\|_2^2.
$$

Este problema se resuelve mediante algoritmos estándar de optimización numérica para mínimos cuadrados, conceptualmente equivalentes a la función `lsqnonlin` utilizada en MATLAB.

**Comparación modelo–experimento**

Se comparan tres conjuntos de curvas:

1. **Curvas experimentales promedio** del cocultivo.
2. **Simulación del modelo** utilizando coeficientes $c_{ij}$ derivados de experimentos en *spent medium*.
3. **Simulación del modelo ajustado**, utilizando los valores óptimos $(c_{12}^\ast, c_{21}^\ast)$ obtenidos por mínimos cuadrados.

Esta comparación permite evaluar:

- la capacidad predictiva del modelo usando parámetros independientes,
- la mejora obtenida al ajustar directamente a datos de cocultivo,
- la coherencia del modelo dinámico con las trayectorias observadas.

**Comentario metodológico final**

El enfoque adoptado es **determinista y fenomenológico**:

- el ruido experimental no se modela explícitamente,
- no se asume una distribución probabilística de errores,
- el ajuste se realiza directamente sobre trayectorias temporales discretas.

Este esquema es coherente con el objetivo principal del estudio:  
**evaluar si un modelo simple de interacción poblacional puede capturar la dinámica observada en cocultivo**, y comparar distintas estrategias de inferencia de parámetros.

```python
# ============================================================
# RK4(5) + Ajuste por mínimos cuadrados + Comparación en plot
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.optimize import least_squares


def coculture_rhs(t: float, y: np.ndarray, params: dict) -> np.ndarray:
    """
    Lado derecho (RHS) del modelo logístico tipo LV usado por Dedrick, con lag "duro".

    Inputs:
        - t: Tiempo (horas).
        - y: Estado del sistema, array [x1, x2].
        - params: Diccionario con parámetros:
            mu1, mu2, K1, K2, c12, c21, t1_lag, t2_lag

    Outputs:
        - dydt: Derivadas [dx1/dt, dx2/dt] como np.ndarray.

    Autor:
        Tecnologías Digitales
    """
    x1, x2 = y

    mu1 = params["mu1"]
    mu2 = params["mu2"]
    K1 = params["K1"]
    K2 = params["K2"]
    c12 = params["c12"]
    c21 = params["c21"]
    t1_lag = params.get("t1_lag", 0.0)
    t2_lag = params.get("t2_lag", 0.0)

    # Comentario: lag "duro" como en MATLAB: antes del lag, la dinámica se congela
    dx1 = 0.0
    dx2 = 0.0

    if t > t1_lag:
        dx1 = mu1 * x1 * (1.0 - (x1 - c12 * x2) / K1)

    if t > t2_lag:
        dx2 = mu2 * x2 * (1.0 - (x2 - c21 * x1) / K2)

    return np.array([dx1, dx2], dtype=float)


def simulate_coculture_rk45(
    t_eval: np.ndarray,
    x10: float,
    x20: float,
    params: dict,
    rtol: float = 1e-6,
    atol: float = 1e-9,
) -> tuple[np.ndarray, np.ndarray]:
    """
    Simula el cocultivo con RK4(5) (solve_ivp con method='RK45') evaluando en t_eval.

    Inputs:
        - t_eval: Vector 1D de tiempos (horas) donde evaluar la solución.
        - x10: Condición inicial de la especie 1.
        - x20: Condición inicial de la especie 2.
        - params: Parámetros del modelo (ver coculture_rhs).
        - rtol: Tolerancia relativa del integrador.
        - atol: Tolerancia absoluta del integrador.

    Outputs:
        - x1: Trayectoria simulada de la especie 1 en los tiempos t_eval.
        - x2: Trayectoria simulada de la especie 2 en los tiempos t_eval.

    Autor:
        Tecnologías Digitales
    """
    t0 = float(np.min(t_eval))
    tf = float(np.max(t_eval))

    sol = solve_ivp(
        fun=lambda t, y: coculture_rhs(t, y, params),
        t_span=(t0, tf),
        y0=np.array([x10, x20], dtype=float),
        t_eval=t_eval,
        method="RK45",   # RK4(5) adaptativo
        rtol=rtol,
        atol=atol,
    )

    if not sol.success:
        raise RuntimeError(f"Integración falló: {sol.message}")

    return sol.y[0], sol.y[1]


def residuals_c12_c21(
    c: np.ndarray,
    t_eval: np.ndarray,
    x10: float,
    x20: float,
    params_base: dict,
    x1_obs: np.ndarray,
    x2_obs: np.ndarray,
) -> np.ndarray:
    """
    Residuales para ajustar c12 y c21 por mínimos cuadrados no lineales usando RK45.

    Inputs:
        - c: Array [c12, c21].
        - t_eval: Tiempos donde evaluar la solución.
        - x10, x20: Condiciones iniciales.
        - params_base: Parámetros fijos (mu1,mu2,K1,K2,t1_lag,t2_lag) sin c12/c21.
        - x1_obs: Serie observada de la especie 1 (Sa).
        - x2_obs: Serie observada de la especie 2 (Sna).

    Outputs:
        - r: Vector de residuales concatenado: [x1_sim-x1_obs, x2_sim-x2_obs].

    Autor:
        Tecnologías Digitales
    """
    params = dict(params_base)
    params["c12"] = float(c[0])
    params["c21"] = float(c[1])

    x1_sim, x2_sim = simulate_coculture_rk45(t_eval, x10, x20, params)

    r1 = x1_sim - x1_obs
    r2 = x2_sim - x2_obs
    return np.concatenate([r1, r2])


def fit_c12_c21_rk45(
    t_eval: np.ndarray,
    x1_obs: np.ndarray,
    x2_obs: np.ndarray,
    x10: float,
    x20: float,
    params_base: dict,
    c0: tuple[float, float] = (-1.25, -0.5),
) -> dict:
    """
    Ajusta c12 y c21 por mínimos cuadrados no lineales, simulando con RK45.

    Inputs:
        - t_eval: Vector de tiempos (horas).
        - x1_obs: Serie observada especie 1.
        - x2_obs: Serie observada especie 2.
        - x10, x20: Condiciones iniciales.
        - params_base: Parámetros fijos (sin c12/c21).
        - c0: Valor inicial para [c12, c21].

    Outputs:
        - result: Diccionario con:
            c12_opt, c21_opt, optimizer_result (objeto de SciPy)

    Autor:
        Tecnologías Digitales
    """
    res = least_squares(
        fun=residuals_c12_c21,
        x0=np.array(c0, dtype=float),
        args=(t_eval, x10, x20, params_base, x1_obs, x2_obs),
    )

    return {
        "c12_opt": float(res.x[0]),
        "c21_opt": float(res.x[1]),
        "optimizer_result": res,
    }


# ============================================================
# 1) Levantar curvas experimentales desde df_coculture_summary
# ============================================================

t_eval = df_coculture_summary["time_hours"].to_numpy()

# Curvas experimentales (promedios)
x1_obs = df_coculture_summary["sa_od_mean"].to_numpy()
x2_obs = df_coculture_summary["sna_od_mean"].to_numpy()

# (Opcional) para emular el MATLAB que fuerza el primer punto a 0:
x1_obs = x1_obs.copy()
x2_obs = x2_obs.copy()
x1_obs[0] = 0.0
x2_obs[0] = 0.0


# ============================================================
# 2) Parámetros base (paper) + condiciones iniciales (paper)
# ============================================================

params_base = {
    "mu1": 1.85,
    "mu2": 1.56,
    "K1": 0.61,
    "K2": 0.35,
    "t1_lag": 2.0,
    "t2_lag": 1.6,
}

# Condiciones iniciales (las del script MATLAB)
x10 = 0.0025
x20 = 0.0025


# ============================================================
# 3) Comparación: modelo con c_ij desde spent medium (paper)
# ============================================================

# Dedrick define c_ij = (K_ij - K_i) / K_j
K12 = 0.17
K21 = 0.03

c12_spent = (K12 - params_base["K1"]) / params_base["K2"]
c21_spent = (K21 - params_base["K2"]) / params_base["K1"]

params_spent = dict(params_base, c12=c12_spent, c21=c21_spent)
x1_spent, x2_spent = simulate_coculture_rk45(t_eval, x10, x20, params_spent)


# ============================================================
# 4) Ajuste c12/c21 por mínimos cuadrados (RK45)
# ============================================================

fit = fit_c12_c21_rk45(
    t_eval=t_eval,
    x1_obs=x1_obs,
    x2_obs=x2_obs,
    x10=x10,
    x20=x20,
    params_base=params_base,
    c0=(-1.25, -0.5),
)

c12_opt = fit["c12_opt"]
c21_opt = fit["c21_opt"]

params_fit = dict(params_base, c12=c12_opt, c21=c21_opt)
x1_fit, x2_fit = simulate_coculture_rk45(t_eval, x10, x20, params_fit)

print(f"c12_spent = {c12_spent:.3f}, c21_spent = {c21_spent:.3f}")
print(f"c12_opt   = {c12_opt:.3f}, c21_opt   = {c21_opt:.3f}")


# ============================================================
# 5) Plot comparativo: Experimento vs Modelo (spent vs fit)
# ============================================================

plt.figure(figsize=(10, 6))

# Datos experimentales (promedio)
plt.plot(t_eval, x1_obs, label="Sa (exp, media)", linewidth=2.5)
plt.plot(t_eval, x2_obs, label="Sna (exp, media)", linewidth=2.5)

# Modelo con c desde spent medium
plt.plot(t_eval, x1_spent, linestyle=":", label="Sa (modelo, spent)", linewidth=2.0)
plt.plot(t_eval, x2_spent, linestyle=":", label="Sna (modelo, spent)", linewidth=2.0)

# Modelo con c ajustado
plt.plot(t_eval, x1_fit, linestyle="--", label="Sa (modelo, fit RK45)", linewidth=2.0)
plt.plot(t_eval, x2_fit, linestyle="--", label="Sna (modelo, fit RK45)", linewidth=2.0)

# (Opcional) marcar lags
plt.axvline(params_base["t1_lag"], linestyle=":", alpha=0.5)
plt.axvline(params_base["t2_lag"], linestyle=":", alpha=0.5)

plt.xlabel("Tiempo (horas)")
plt.ylabel("OD")
plt.title("Comparación de curvas: experimento vs modelo (RK4(5) / RK45)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

## Evaluación del rendimiento del modelo dinámico

### Objetivo de la evaluación

El objetivo de esta etapa es **cuantificar la concordancia entre las trayectorias poblacionales simuladas por el modelo dinámico y las curvas de crecimiento observadas experimentalmente**, así como diagnosticar posibles deficiencias estructurales o numéricas de la implementación.

La evaluación se realiza **a posteriori**, una vez definidas las trayectorias simuladas, y no interviene en el proceso de ajuste de parámetros.

---

### Datos utilizados para la evaluación

La comparación se efectúa utilizando:

- **Curvas experimentales promedio** del cocultivo:
  - $x_{1,\text{obs}}(t_n)$: densidad óptica promedio de la especie marcada (Sa).
  - $x_{2,\text{obs}}(t_n)$: densidad óptica promedio de la segunda especie (Sna).

- **Curvas simuladas por el modelo**:
  - $x_{1,\text{pred}}(t_n)$ y $x_{2,\text{pred}}(t_n)$, obtenidas mediante la integración numérica del modelo dinámico.

Todas las curvas se evalúan en el mismo conjunto de tiempos discretos $\{t_n\}_{n=1}^{N}$, definidos por el esquema experimental.

---

### Métricas cuantitativas de rendimiento

La calidad del ajuste se evalúa mediante métricas de error estándar, calculadas **por especie** y de forma **global**.

#### Error absoluto medio (MAE)

El error absoluto medio se define como:

$$
\text{MAE} = \frac{1}{N} \sum_{n=1}^{N} \left| y_{\text{pred}}(t_n) - y_{\text{obs}}(t_n) \right|.
$$

Esta métrica proporciona una medida directa del error promedio en unidades de densidad óptica.

---

#### Raíz del error cuadrático medio (RMSE)

La raíz del error cuadrático medio se define como:

$$
\text{RMSE} =
\sqrt{
\frac{1}{N}
\sum_{n=1}^{N}
\left( y_{\text{pred}}(t_n) - y_{\text{obs}}(t_n) \right)^2
}.
$$

La RMSE penaliza de manera más severa los errores grandes y es sensible a discrepancias estructurales entre modelo y datos.

---

#### Coeficiente de determinación ($R^2$)

El coeficiente de determinación se calcula como:

$$
R^2 = 1 -
\frac{
\sum_{n=1}^{N} \left( y_{\text{obs}}(t_n) - y_{\text{pred}}(t_n) \right)^2
}{
\sum_{n=1}^{N} \left( y_{\text{obs}}(t_n) - \bar{y}_{\text{obs}} \right)^2
}.
$$

Esta métrica cuantifica la fracción de la varianza experimental explicada por el modelo.

---

#### Evaluación global

Además de la evaluación por especie, se define una métrica **global** concatenando ambas trayectorias:

$$
\mathbf{y}_{\text{obs}} =
\begin{bmatrix}
x_{1,\text{obs}} \\
x_{2,\text{obs}}
\end{bmatrix},
\qquad
\mathbf{y}_{\text{pred}} =
\begin{bmatrix}
x_{1,\text{pred}} \\
x_{2,\text{pred}}
\end{bmatrix}.
$$

Las métricas MAE, RMSE y $R^2$ se calculan sobre estos vectores concatenados para obtener una medida integrada del desempeño del modelo.

---

### Diagnóstico de sanidad de la simulación

Además de las métricas cuantitativas, se realizan **chequeos de consistencia física y numérica** sobre las trayectorias simuladas:

- Verificación de ausencia de valores no numéricos (NaN o infinitos).
- Identificación de valores negativos en las poblaciones simuladas.
- Evaluación de los rangos máximos y mínimos alcanzados por las trayectorias.

Estos controles permiten detectar errores de implementación, inestabilidades numéricas o inconsistencias en los parámetros del modelo.

---

### Análisis de residuales

Como complemento a las métricas agregadas, se analizan los **residuales temporales** definidos como:

$$
r_i(t_n) = x_{i,\text{pred}}(t_n) - x_{i,\text{obs}}(t_n),
\qquad i \in \{1,2\}.
$$

El análisis de los residuales en función del tiempo permite:

- identificar sesgos sistemáticos del modelo,
- detectar regiones temporales mal representadas,
- distinguir entre ruido experimental y errores estructurales del modelo.

---

### Sensibilidad numérica

Finalmente, se evalúa la **robustez numérica** de la simulación analizando la sensibilidad de las trayectorias frente a variaciones en las tolerancias del integrador numérico.

Para distintos pares de tolerancias $(\text{rtol}, \text{atol})$, se compara la solución obtenida con una solución de referencia, cuantificando la diferencia máxima absoluta:

$$
\Delta_{\max} = \max_n \left| x(t_n) - x_{\text{ref}}(t_n) \right|.
$$

Una baja sensibilidad indica que los resultados del modelo no dependen de manera crítica de los parámetros numéricos del integrador.

---

### Comentario final

El esquema de evaluación adoptado es **determinista y descriptivo**:

- no se asume un modelo probabilístico del ruido experimental,
- no se realiza inferencia estadística formal,
- la comparación se basa en métricas de error y análisis de residuales.

Este enfoque es coherente con el objetivo principal del estudio:  
**verificar si un modelo dinámico sencillo de interacción poblacional reproduce razonablemente las trayectorias experimentales observadas**, y comparar el rendimiento entre distintas configuraciones del modelo.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.integrate import solve_ivp
from scipy.optimize import least_squares


# ============================================================
# 1) Modelo + simulación RK45
# ============================================================

def coculture_rhs(t: float, y: np.ndarray, params: dict) -> np.ndarray:
    """
    RHS del modelo logístico tipo LV (Dedrick) con lag "duro".

    Inputs:
        - t: tiempo (horas).
        - y: estado [x1, x2].
        - params: diccionario con mu1, mu2, K1, K2, c12, c21, t1_lag, t2_lag.

    Outputs:
        - dydt: derivadas [dx1/dt, dx2/dt].

    Autor:
        Tecnologías Digitales
    """
    x1, x2 = y

    mu1 = params["mu1"]
    mu2 = params["mu2"]
    K1 = params["K1"]
    K2 = params["K2"]
    c12 = params["c12"]
    c21 = params["c21"]
    t1_lag = params.get("t1_lag", 0.0)
    t2_lag = params.get("t2_lag", 0.0)

    dx1 = 0.0
    dx2 = 0.0

    # Comentario: lag duro como en MATLAB: hasta t_lag no hay crecimiento
    if t > t1_lag:
        dx1 = mu1 * x1 * (1.0 - (x1 - c12 * x2) / K1)

    if t > t2_lag:
        dx2 = mu2 * x2 * (1.0 - (x2 - c21 * x1) / K2)

    return np.array([dx1, dx2], dtype=float)


def simulate_coculture_rk45(
    t_eval: np.ndarray,
    x10: float,
    x20: float,
    params: dict,
    rtol: float = 1e-6,
    atol: float = 1e-9,
) -> tuple[np.ndarray, np.ndarray]:
    """
    Simula el cocultivo con RK4(5) (solve_ivp con method='RK45') y evalúa en t_eval.

    Inputs:
        - t_eval: tiempos (horas) donde evaluar.
        - x10, x20: condiciones iniciales.
        - params: parámetros del modelo.
        - rtol, atol: tolerancias del integrador.

    Outputs:
        - x1: trayectoria simulada de la especie 1.
        - x2: trayectoria simulada de la especie 2.

    Autor:
        Tecnologías Digitales
    """
    t0 = float(np.min(t_eval))
    tf = float(np.max(t_eval))

    sol = solve_ivp(
        fun=lambda t, y: coculture_rhs(t, y, params),
        t_span=(t0, tf),
        y0=np.array([x10, x20], dtype=float),
        t_eval=t_eval,
        method="RK45",
        rtol=rtol,
        atol=atol,
    )

    if not sol.success:
        raise RuntimeError(f"Integración falló: {sol.message}")

    return sol.y[0], sol.y[1]


# ============================================================
# 2) Ajuste por mínimos cuadrados (c12, c21)
# ============================================================

def residuals_c12_c21(
    c: np.ndarray,
    t_eval: np.ndarray,
    x10: float,
    x20: float,
    params_base: dict,
    x1_obs: np.ndarray,
    x2_obs: np.ndarray,
) -> np.ndarray:
    """
    Residuales para ajustar (c12, c21) por mínimos cuadrados.

    Inputs:
        - c: array [c12, c21].
        - t_eval: tiempos de evaluación.
        - x10, x20: condiciones iniciales.
        - params_base: parámetros fijos sin c12/c21.
        - x1_obs, x2_obs: series observadas (mismo largo que t_eval).

    Outputs:
        - r: residuales concatenados [x1_sim-x1_obs, x2_sim-x2_obs].

    Autor:
        Tecnologías Digitales
    """
    params = dict(params_base)
    params["c12"] = float(c[0])
    params["c21"] = float(c[1])

    x1_sim, x2_sim = simulate_coculture_rk45(t_eval, x10, x20, params)
    return np.concatenate([x1_sim - x1_obs, x2_sim - x2_obs])


def fit_c12_c21(
    t_eval: np.ndarray,
    x1_obs: np.ndarray,
    x2_obs: np.ndarray,
    x10: float,
    x20: float,
    params_base: dict,
    c0: tuple[float, float] = (-1.25, -0.5),
) -> dict:
    """
    Ajusta c12 y c21 por mínimos cuadrados no lineales.

    Inputs:
        - t_eval: tiempos (horas).
        - x1_obs, x2_obs: series observadas.
        - x10, x20: condiciones iniciales.
        - params_base: parámetros fijos.
        - c0: inicialización para (c12, c21).

    Outputs:
        - result: dict con c12_opt, c21_opt, optimizer_result.

    Autor:
        Tecnologías Digitales
    """
    res = least_squares(
        fun=residuals_c12_c21,
        x0=np.array(c0, dtype=float),
        args=(t_eval, x10, x20, params_base, x1_obs, x2_obs),
    )
    return {"c12_opt": float(res.x[0]), "c21_opt": float(res.x[1]), "optimizer_result": res}


# ============================================================
# 3) Métricas de evaluación (modelo vs datos)
# ============================================================

def regression_metrics(y_true: np.ndarray, y_pred: np.ndarray) -> dict:
    """
    Calcula métricas básicas de ajuste: MAE, RMSE, R2.

    Inputs:
        - y_true: valores observados.
        - y_pred: valores predichos.

    Outputs:
        - metrics: dict con mae, rmse, r2.

    Autor:
        Tecnologías Digitales
    """
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)

    mae = float(np.mean(np.abs(y_true - y_pred)))
    rmse = float(np.sqrt(np.mean((y_true - y_pred) ** 2)))

    ss_res = float(np.sum((y_true - y_pred) ** 2))
    ss_tot = float(np.sum((y_true - np.mean(y_true)) ** 2))
    r2 = float(1.0 - ss_res / ss_tot) if ss_tot > 0 else float("nan")

    return {"mae": mae, "rmse": rmse, "r2": r2}


def evaluate_fit(
    t_eval: np.ndarray,
    x1_obs: np.ndarray,
    x2_obs: np.ndarray,
    x1_pred: np.ndarray,
    x2_pred: np.ndarray,
) -> pd.DataFrame:
    """
    Resume métricas por especie y global (concatenando ambas).

    Inputs:
        - t_eval: tiempos.
        - x1_obs, x2_obs: observaciones.
        - x1_pred, x2_pred: predicciones.

    Outputs:
        - df_metrics: DataFrame con métricas.

    Autor:
        Tecnologías Digitales
    """
    m1 = regression_metrics(x1_obs, x1_pred)
    m2 = regression_metrics(x2_obs, x2_pred)

    m_all = regression_metrics(
        np.concatenate([x1_obs, x2_obs]),
        np.concatenate([x1_pred, x2_pred]),
    )

    return pd.DataFrame(
        [
            {"series": "x1 (Sa)", **m1},
            {"series": "x2 (Sna)", **m2},
            {"series": "global (x1+x2)", **m_all},
        ]
    )


# ============================================================
# 4) Sanity checks (signos, no negatividad, estabilidad)
# ============================================================

def sanity_checks(t_eval: np.ndarray, x1: np.ndarray, x2: np.ndarray, name: str = "model") -> dict:
    """
    Checks simples para detectar implementaciones con signos mal puestos o inestabilidades.

    Inputs:
        - t_eval: tiempos.
        - x1, x2: trayectorias.
        - name: etiqueta para reporting.

    Outputs:
        - checks: dict con flags y estadísticas.

    Autor:
        Tecnologías Digitales
    """
    x1 = np.asarray(x1, dtype=float)
    x2 = np.asarray(x2, dtype=float)

    checks = {
        "name": name,
        "x1_min": float(np.min(x1)),
        "x2_min": float(np.min(x2)),
        "x1_max": float(np.max(x1)),
        "x2_max": float(np.max(x2)),
        "any_nan": bool(np.isnan(x1).any() or np.isnan(x2).any()),
        "any_inf": bool(np.isinf(x1).any() or np.isinf(x2).any()),
        "negative_fraction_x1": float(np.mean(x1 < 0)),
        "negative_fraction_x2": float(np.mean(x2 < 0)),
    }
    return checks


def tolerance_sensitivity(
    t_eval: np.ndarray,
    x10: float,
    x20: float,
    params: dict,
    tol_list: list[tuple[float, float]] = [(1e-6, 1e-9), (1e-7, 1e-10), (1e-5, 1e-8)],
) -> pd.DataFrame:
    """
    Evalúa sensibilidad de la simulación a tolerancias (rtol, atol).

    Inputs:
        - t_eval: tiempos.
        - x10, x20: condiciones iniciales.
        - params: parámetros del modelo.
        - tol_list: lista de tuplas (rtol, atol).

    Outputs:
        - df: DataFrame con diferencias respecto a la primera tolerancia.

    Autor:
        Tecnologías Digitales
    """
    x1_ref, x2_ref = simulate_coculture_rk45(t_eval, x10, x20, params, rtol=tol_list[0][0], atol=tol_list[0][1])
    rows = []
    for rtol, atol in tol_list:
        x1, x2 = simulate_coculture_rk45(t_eval, x10, x20, params, rtol=rtol, atol=atol)
        rows.append({
            "rtol": rtol,
            "atol": atol,
            "max_abs_diff_x1": float(np.max(np.abs(x1 - x1_ref))),
            "max_abs_diff_x2": float(np.max(np.abs(x2 - x2_ref))),
        })
    return pd.DataFrame(rows)


# ============================================================
# 5) Pipeline de evaluación: spent vs fit + plots
# ============================================================

# Datos experimentales desde tus DataFrames (promedios)
t_eval = df_coculture_summary["time_hours"].to_numpy()
x1_obs = df_coculture_summary["sa_od_mean"].to_numpy().copy()
x2_obs = df_coculture_summary["sna_od_mean"].to_numpy().copy()

# Comentario: emula el MATLAB que fuerza el primer punto a 0
x1_obs[0] = 0.0
x2_obs[0] = 0.0

# Parámetros del paper
params_base = {
    "mu1": 1.85,
    "mu2": 1.56,
    "K1": 0.61,
    "K2": 0.35,
    "t1_lag": 2.0,
    "t2_lag": 1.6,
}

# Condiciones iniciales del script
x10, x20 = 0.0025, 0.0025

# Parámetros “spent medium”
K12, K21 = 0.17, 0.03
c12_spent = (K12 - params_base["K1"]) / params_base["K2"]
c21_spent = (K21 - params_base["K2"]) / params_base["K1"]

params_spent = dict(params_base, c12=c12_spent, c21=c21_spent)
x1_spent, x2_spent = simulate_coculture_rk45(t_eval, x10, x20, params_spent)

# Ajuste por mínimos cuadrados
fit = fit_c12_c21(t_eval, x1_obs, x2_obs, x10, x20, params_base, c0=(-1.25, -0.5))
c12_opt, c21_opt = fit["c12_opt"], fit["c21_opt"]

params_fit = dict(params_base, c12=c12_opt, c21=c21_opt)
x1_fit, x2_fit = simulate_coculture_rk45(t_eval, x10, x20, params_fit)

print(f"c12_spent={c12_spent:.3f}, c21_spent={c21_spent:.3f}")
print(f"c12_opt  ={c12_opt:.3f}, c21_opt  ={c21_opt:.3f}")

# Métricas
df_metrics_spent = evaluate_fit(t_eval, x1_obs, x2_obs, x1_spent, x2_spent)
df_metrics_fit = evaluate_fit(t_eval, x1_obs, x2_obs, x1_fit, x2_fit)

display(df_metrics_spent.assign(model="spent"))
display(df_metrics_fit.assign(model="fit"))

# Sanity checks
checks_spent = sanity_checks(t_eval, x1_spent, x2_spent, name="spent")
checks_fit = sanity_checks(t_eval, x1_fit, x2_fit, name="fit")
display(pd.DataFrame([checks_spent, checks_fit]))

# Sensibilidad a tolerancias (opcional)
df_tol = tolerance_sensitivity(t_eval, x10, x20, params_fit)
display(df_tol)

# ------------------------------------------------------------
# Plots: curvas + residuales
# ------------------------------------------------------------
plt.figure(figsize=(10, 6))
plt.plot(t_eval, x1_obs, linewidth=2.5, label="Sa exp (media)")
plt.plot(t_eval, x2_obs, linewidth=2.5, label="Sna exp (media)")

plt.plot(t_eval, x1_spent, linestyle=":", linewidth=2.0, label="Sa modelo (spent)")
plt.plot(t_eval, x2_spent, linestyle=":", linewidth=2.0, label="Sna modelo (spent)")

plt.plot(t_eval, x1_fit, linestyle="--", linewidth=2.0, label="Sa modelo (fit RK45)")
plt.plot(t_eval, x2_fit, linestyle="--", linewidth=2.0, label="Sna modelo (fit RK45)")

plt.axvline(params_base["t1_lag"], linestyle=":", alpha=0.5)
plt.axvline(params_base["t2_lag"], linestyle=":", alpha=0.5)

plt.xlabel("Tiempo (horas)")
plt.ylabel("OD")
plt.title("Curvas: experimento vs modelo (spent vs fit, RK4(5))")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()

# Residuales vs tiempo (fit)
res1 = x1_fit - x1_obs
res2 = x2_fit - x2_obs

plt.figure(figsize=(10, 4))
plt.plot(t_eval, res1, label="Residual Sa (fit)")
plt.plot(t_eval, res2, label="Residual Sna (fit)")
plt.axhline(0.0, linestyle="--", alpha=0.6)
plt.xlabel("Tiempo (horas)")
plt.ylabel("Residual (OD)")
plt.title("Residuales del ajuste (fit RK45)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# ============================================================
# Euler + Ajuste por mínimos cuadrados + Evaluación de rendimiento
# (Implementación completa autocontenida)
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import least_squares


# ============================================================
# 1) Modelo (RHS) + Simulación con Euler explícito
# ============================================================

def coculture_rhs(t: float, y: np.ndarray, params: dict) -> np.ndarray:
    """
    Lado derecho (RHS) del modelo logístico tipo LV usado por Dedrick, con lag "duro".

    Inputs:
        - t: Tiempo (horas).
        - y: Estado [x1, x2].
        - params: Diccionario con mu1, mu2, K1, K2, c12, c21, t1_lag, t2_lag.

    Outputs:
        - dydt: Derivadas [dx1/dt, dx2/dt] como np.ndarray.

    Autor:
        Tecnologías Digitales
    """
    x1, x2 = y

    mu1 = params["mu1"]
    mu2 = params["mu2"]
    K1 = params["K1"]
    K2 = params["K2"]
    c12 = params["c12"]
    c21 = params["c21"]
    t1_lag = params.get("t1_lag", 0.0)
    t2_lag = params.get("t2_lag", 0.0)

    # Comentario: lag "duro" como en MATLAB: antes del lag, dx/dt = 0
    dx1 = 0.0
    dx2 = 0.0

    if t > t1_lag:
        dx1 = mu1 * x1 * (1.0 - (x1 - c12 * x2) / K1)

    if t > t2_lag:
        dx2 = mu2 * x2 * (1.0 - (x2 - c21 * x1) / K2)

    return np.array([dx1, dx2], dtype=float)


def simulate_coculture_euler(
    t_eval: np.ndarray,
    x10: float,
    x20: float,
    params: dict,
    dt_hours: float,
) -> tuple[np.ndarray, np.ndarray]:
    """
    Simula el cocultivo usando Euler explícito y devuelve trayectorias evaluadas en t_eval.

    Requisito:
        - t_eval debe ser (aprox.) una grilla uniforme con paso dt_hours, como en el experimento.

    Inputs:
        - t_eval: Vector 1D de tiempos (horas) donde se requiere la solución.
        - x10, x20: Condiciones iniciales.
        - params: Parámetros del modelo (ver coculture_rhs).
        - dt_hours: Paso temporal de Euler (horas).

    Outputs:
        - x1: Trayectoria simulada de la especie 1 en t_eval.
        - x2: Trayectoria simulada de la especie 2 en t_eval.

    Autor:
        Tecnologías Digitales
    """
    t_eval = np.asarray(t_eval, dtype=float)
    n = t_eval.size

    x1 = np.zeros(n, dtype=float)
    x2 = np.zeros(n, dtype=float)
    x1[0] = float(x10)
    x2[0] = float(x20)

    for k in range(1, n):
        t_prev = t_eval[k - 1]
        y_prev = np.array([x1[k - 1], x2[k - 1]], dtype=float)

        dydt = coculture_rhs(t_prev, y_prev, params)

        y_next = y_prev + dt_hours * dydt

        x1[k] = y_next[0]
        x2[k] = y_next[1]

    return x1, x2


# ============================================================
# 2) Ajuste por mínimos cuadrados (c12, c21) con Euler
# ============================================================

def residuals_c12_c21_euler(
    c: np.ndarray,
    t_eval: np.ndarray,
    x10: float,
    x20: float,
    params_base: dict,
    x1_obs: np.ndarray,
    x2_obs: np.ndarray,
    dt_hours: float,
) -> np.ndarray:
    """
    Residuales para ajustar c12 y c21 por mínimos cuadrados no lineales usando Euler.

    Inputs:
        - c: Array [c12, c21].
        - t_eval: Tiempos donde evaluar la solución.
        - x10, x20: Condiciones iniciales.
        - params_base: Parámetros fijos sin c12/c21.
        - x1_obs, x2_obs: Series observadas.
        - dt_hours: Paso temporal de Euler (horas).

    Outputs:
        - r: Vector de residuales concatenado [x1_sim-x1_obs, x2_sim-x2_obs].

    Autor:
        Tecnologías Digitales
    """
    params = dict(params_base)
    params["c12"] = float(c[0])
    params["c21"] = float(c[1])

    x1_sim, x2_sim = simulate_coculture_euler(t_eval, x10, x20, params, dt_hours)

    r1 = x1_sim - x1_obs
    r2 = x2_sim - x2_obs
    return np.concatenate([r1, r2])


def fit_c12_c21_euler(
    t_eval: np.ndarray,
    x1_obs: np.ndarray,
    x2_obs: np.ndarray,
    x10: float,
    x20: float,
    params_base: dict,
    dt_hours: float,
    c0: tuple[float, float] = (-1.25, -0.5),
) -> dict:
    """
    Ajusta c12 y c21 por mínimos cuadrados no lineales, simulando con Euler.

    Inputs:
        - t_eval: Vector de tiempos (horas).
        - x1_obs, x2_obs: Series observadas.
        - x10, x20: Condiciones iniciales.
        - params_base: Parámetros fijos (sin c12/c21).
        - dt_hours: Paso temporal de Euler (horas).
        - c0: Inicialización para (c12, c21).

    Outputs:
        - result: Diccionario con:
            c12_opt, c21_opt, optimizer_result (objeto SciPy)

    Autor:
        Tecnologías Digitales
    """
    res = least_squares(
        fun=residuals_c12_c21_euler,
        x0=np.array(c0, dtype=float),
        args=(t_eval, x10, x20, params_base, x1_obs, x2_obs, dt_hours),
    )

    return {
        "c12_opt": float(res.x[0]),
        "c21_opt": float(res.x[1]),
        "optimizer_result": res,
    }


# ============================================================
# 3) Métricas + checks de sanidad + sensibilidad a dt
# ============================================================

def regression_metrics(y_true: np.ndarray, y_pred: np.ndarray) -> dict:
    """
    Calcula métricas de ajuste: MAE, RMSE y R2.

    Inputs:
        - y_true: Valores observados.
        - y_pred: Valores predichos.

    Outputs:
        - metrics: Diccionario con mae, rmse, r2.

    Autor:
        Tecnologías Digitales
    """
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)

    mae = float(np.mean(np.abs(y_true - y_pred)))
    rmse = float(np.sqrt(np.mean((y_true - y_pred) ** 2)))

    ss_res = float(np.sum((y_true - y_pred) ** 2))
    ss_tot = float(np.sum((y_true - np.mean(y_true)) ** 2))
    r2 = float(1.0 - ss_res / ss_tot) if ss_tot > 0 else float("nan")

    return {"mae": mae, "rmse": rmse, "r2": r2}


def evaluate_model_performance(
    x1_obs: np.ndarray,
    x2_obs: np.ndarray,
    x1_pred: np.ndarray,
    x2_pred: np.ndarray,
) -> pd.DataFrame:
    """
    Resume el rendimiento del modelo por especie y de forma global.

    Inputs:
        - x1_obs, x2_obs: Series observadas (Sa, Sna).
        - x1_pred, x2_pred: Series predichas por el modelo.

    Outputs:
        - df_metrics: DataFrame con MAE, RMSE, R2 por serie y global.

    Autor:
        Tecnologías Digitales
    """
    m1 = regression_metrics(x1_obs, x1_pred)
    m2 = regression_metrics(x2_obs, x2_pred)

    m_global = regression_metrics(
        np.concatenate([x1_obs, x2_obs]),
        np.concatenate([x1_pred, x2_pred]),
    )

    return pd.DataFrame([
        {"serie": "Sa", **m1},
        {"serie": "Sna", **m2},
        {"serie": "Global (Sa + Sna)", **m_global},
    ])


def sanity_checks(x1: np.ndarray, x2: np.ndarray, name: str = "model") -> dict:
    """
    Checks simples de sanidad para detectar problemas numéricos o físicos.

    Inputs:
        - x1, x2: Trayectorias simuladas.
        - name: Etiqueta del modelo.

    Outputs:
        - checks: Diccionario con flags y estadísticos básicos.

    Autor:
        Tecnologías Digitales
    """
    x1 = np.asarray(x1, dtype=float)
    x2 = np.asarray(x2, dtype=float)

    return {
        "name": name,
        "x1_min": float(np.min(x1)),
        "x2_min": float(np.min(x2)),
        "x1_max": float(np.max(x1)),
        "x2_max": float(np.max(x2)),
        "any_nan": bool(np.isnan(x1).any() or np.isnan(x2).any()),
        "any_inf": bool(np.isinf(x1).any() or np.isinf(x2).any()),
        "negative_fraction_x1": float(np.mean(x1 < 0)),
        "negative_fraction_x2": float(np.mean(x2 < 0)),
    }


def dt_sensitivity_euler(
    t_eval: np.ndarray,
    x10: float,
    x20: float,
    params: dict,
    dt_list: list[float],
) -> pd.DataFrame:
    """
    Sensibilidad del esquema de Euler al tamaño de paso dt (re-muestreando a t_eval).

    Estrategia:
        - Se integra con paso dt (posiblemente menor que el del experimento).
        - Se interpola linealmente la trayectoria a los tiempos t_eval.

    Inputs:
        - t_eval: Tiempos experimentales (horas).
        - x10, x20: Condiciones iniciales.
        - params: Parámetros del modelo.
        - dt_list: Lista de pasos dt (horas) a evaluar.

    Outputs:
        - df: DataFrame con max abs diff vs referencia (primer dt de la lista).

    Autor:
        Tecnologías Digitales
    """
    t_eval = np.asarray(t_eval, dtype=float)
    t0 = float(t_eval.min())
    tf = float(t_eval.max())

    def integrate_and_interpolate(dt: float) -> tuple[np.ndarray, np.ndarray]:
        n_steps = int(np.round((tf - t0) / dt))
        t_grid = t0 + dt * np.arange(n_steps + 1, dtype=float)

        x1_grid = np.zeros_like(t_grid)
        x2_grid = np.zeros_like(t_grid)
        x1_grid[0] = float(x10)
        x2_grid[0] = float(x20)

        for k in range(1, t_grid.size):
            t_prev = t_grid[k - 1]
            y_prev = np.array([x1_grid[k - 1], x2_grid[k - 1]], dtype=float)
            dydt = coculture_rhs(t_prev, y_prev, params)
            y_next = y_prev + dt * dydt
            x1_grid[k] = y_next[0]
            x2_grid[k] = y_next[1]

        x1_eval = np.interp(t_eval, t_grid, x1_grid)
        x2_eval = np.interp(t_eval, t_grid, x2_grid)
        return x1_eval, x2_eval

    x1_ref, x2_ref = integrate_and_interpolate(dt_list[0])

    rows = []
    for dt in dt_list:
        x1_eval, x2_eval = integrate_and_interpolate(dt)
        rows.append({
            "dt_hours": dt,
            "max_abs_diff_x1": float(np.max(np.abs(x1_eval - x1_ref))),
            "max_abs_diff_x2": float(np.max(np.abs(x2_eval - x2_ref))),
        })

    return pd.DataFrame(rows)


# ============================================================
# 4) Datos (desde df_coculture_summary) + Simulación + Fit + Evaluación
# ============================================================

# Datos experimentales
t_eval = df_coculture_summary["time_hours"].to_numpy()
x1_obs = df_coculture_summary["sa_od_mean"].to_numpy().copy()
x2_obs = df_coculture_summary["sna_od_mean"].to_numpy().copy()

# Comentario: emula el MATLAB que fuerza el primer punto a 0
x1_obs[0] = 0.0
x2_obs[0] = 0.0

# Paso temporal experimental (10 min)
dt_hours = float(np.median(np.diff(t_eval)))  # robusto si hay pequeñas variaciones

# Parámetros base (paper)
params_base = {
    "mu1": 1.85,
    "mu2": 1.56,
    "K1": 0.61,
    "K2": 0.35,
    "t1_lag": 2.0,
    "t2_lag": 1.6,
}

# Condiciones iniciales (script MATLAB)
x10, x20 = 0.0025, 0.0025

# Parámetros "spent medium" -> c_ij
K12, K21 = 0.17, 0.03
c12_spent = (K12 - params_base["K1"]) / params_base["K2"]
c21_spent = (K21 - params_base["K2"]) / params_base["K1"]

params_spent = dict(params_base, c12=c12_spent, c21=c21_spent)

# Simulación Euler con c_spent
x1_spent, x2_spent = simulate_coculture_euler(t_eval, x10, x20, params_spent, dt_hours)

# Ajuste c12, c21 (Euler)
fit = fit_c12_c21_euler(
    t_eval=t_eval,
    x1_obs=x1_obs,
    x2_obs=x2_obs,
    x10=x10,
    x20=x20,
    params_base=params_base,
    dt_hours=dt_hours,
    c0=(-1.25, -0.5),
)

c12_opt, c21_opt = fit["c12_opt"], fit["c21_opt"]
params_fit = dict(params_base, c12=c12_opt, c21=c21_opt)

# Simulación Euler con c_fit
x1_fit, x2_fit = simulate_coculture_euler(t_eval, x10, x20, params_fit, dt_hours)

print(f"dt_hours   = {dt_hours:.6f}")
print(f"c12_spent  = {c12_spent:.3f}, c21_spent  = {c21_spent:.3f}")
print(f"c12_opt    = {c12_opt:.3f}, c21_opt    = {c21_opt:.3f}")

# ============================================================
# 5) Evaluación de rendimiento (spent vs fit)
# ============================================================

df_metrics_spent = evaluate_model_performance(x1_obs, x2_obs, x1_spent, x2_spent).assign(model="spent")
df_metrics_fit = evaluate_model_performance(x1_obs, x2_obs, x1_fit, x2_fit).assign(model="fit")

display(pd.concat([df_metrics_spent, df_metrics_fit], ignore_index=True))

# Checks de sanidad
checks_spent = sanity_checks(x1_spent, x2_spent, name="spent")
checks_fit = sanity_checks(x1_fit, x2_fit, name="fit")
display(pd.DataFrame([checks_spent, checks_fit]))

# Sensibilidad a dt (opcional): comparar dt experimental vs refinamientos
dt_list = [dt_hours, dt_hours / 2, dt_hours / 4]
df_dt = dt_sensitivity_euler(t_eval, x10, x20, params_fit, dt_list=dt_list)
display(df_dt)

# ============================================================
# 6) Plots: curvas y residuales (spent vs fit)
# ============================================================

plt.figure(figsize=(10, 6))

# Experimento
plt.plot(t_eval, x1_obs, linewidth=2.5, label="Sa exp (media)")
plt.plot(t_eval, x2_obs, linewidth=2.5, label="Sna exp (media)")

# Modelo spent
plt.plot(t_eval, x1_spent, linestyle=":", linewidth=2.0, label="Sa modelo (spent, Euler)")
plt.plot(t_eval, x2_spent, linestyle=":", linewidth=2.0, label="Sna modelo (spent, Euler)")

# Modelo fit
plt.plot(t_eval, x1_fit, linestyle="--", linewidth=2.0, label="Sa modelo (fit, Euler)")
plt.plot(t_eval, x2_fit, linestyle="--", linewidth=2.0, label="Sna modelo (fit, Euler)")

# Marcar lag times
plt.axvline(params_base["t1_lag"], linestyle=":", alpha=0.5)
plt.axvline(params_base["t2_lag"], linestyle=":", alpha=0.5)

plt.xlabel("Tiempo (horas)")
plt.ylabel("OD")
plt.title("Comparación de curvas: experimento vs modelo (Euler)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()

# Residuales del fit
res_sa = x1_fit - x1_obs
res_sna = x2_fit - x2_obs

plt.figure(figsize=(10, 4))
plt.plot(t_eval, res_sa, label="Residual Sa (fit, Euler)")
plt.plot(t_eval, res_sna, label="Residual Sna (fit, Euler)")
plt.axhline(0.0, linestyle="--", alpha=0.6)
plt.xlabel("Tiempo (horas)")
plt.ylabel("Residual (OD)")
plt.title("Residuales del ajuste (Euler)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
from __future__ import annotations

from dataclasses import dataclass
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

from scipy.optimize import least_squares


# ============================================================
# Rutas a datos
# ============================================================
DATA_DIR = Path("datasets/dedrick")
INFILE = DATA_DIR / "Cocultures_GFPSa_1850_1821_MC_Day1_29JUL2019.txt"


# ============================================================
# Parámetros del modelo (paper / MATLAB)
# ============================================================
r1 = 1.85  # tasa de crecimiento especie 1 (Sa) [1/h]
r2 = 1.56  # tasa de crecimiento especie 2 (Sna1850) [1/h]

k1 = 0.61  # capacidad de carga Sa [OD600]
k2 = 0.35  # capacidad de carga Sna [OD600]

k12 = 0.17  # Sa en sobrenadante de Sna
k21 = 0.03  # Sna en sobrenadante de Sa

# Coeficientes de interacción (definición del MATLAB)
c12 = -(k1 - k12) / k2
c21 = -(k2 - k21) / k1

# Condiciones iniciales
s10 = 0.0025
s20 = 0.0025

# Tiempos de latencia (horas)
t1_lag = 2.0
t2_lag = 1.6

# Grilla temporal (MATLAB: dt = 10/60 h, 96 puntos ≈ 16 h)
dt = 10 / 60
nr = 96
t_grid = dt * np.arange(1, nr + 1)


# ============================================================
# Simulación Euler explícito + lag (replica exacta del MATLAB)
# ============================================================
def simulate_two_species_euler_with_lag(
    t: np.ndarray,
    dt: float,
    r1: float,
    r2: float,
    k1: float,
    k2: float,
    c12: float,
    c21: float,
    x1_0: float,
    x2_0: float,
    t1_lag: float,
    t2_lag: float,
) -> tuple[np.ndarray, np.ndarray]:
    """
    Simula el sistema logístico acoplado de dos especies usando
    Euler explícito con tiempo de latencia (lag), tal como en el código MATLAB.

    Actualizaciones:
        x1 += dt * r1 * x1 * (1 - (x1 - c12*x2)/k1)
        x2 += dt * r2 * x2 * (1 - (x2 - c21*x1)/k2)

    Durante el lag:
        si t <= t1_lag, x1 permanece constante
        si t <= t2_lag, x2 permanece constante
    """
    x1 = np.zeros_like(t, dtype=float)
    x2 = np.zeros_like(t, dtype=float)

    x1[0] = x1_0
    x2[0] = x2_0

    for i in range(1, len(t)):
        ti = t[i]

        x1_prev = x1[i - 1]
        x2_prev = x2[i - 1]

        # Especie 1
        if ti > t1_lag:
            x1[i] = x1_prev + dt * r1 * x1_prev * (
                1.0 - (x1_prev - c12 * x2_prev) / k1
            )
        else:
            x1[i] = x1_prev

        # Especie 2
        if ti > t2_lag:
            x2[i] = x2_prev + dt * r2 * x2_prev * (
                1.0 - (x2_prev - c21 * x1_prev) / k2
            )
        else:
            x2[i] = x2_prev

        # (Opcional) evitar valores negativos por artefactos numéricos
        # x1[i] = max(x1[i], 0.0)
        # x2[i] = max(x2[i], 0.0)

    return x1, x2


# ============================================================
# Parser del archivo .txt de microplaca (tipo fscanf)
# ============================================================
@dataclass
class PlateData:
    od6: np.ndarray  # matriz OD600, forma (r, c, nr)
    gfp: np.ndarray  # matriz GFP, forma (r, c, nr)


def _find_token_index(tokens: list[str], target: str, start: int = 0) -> int:
    """Busca la posición de un token específico en la lista."""
    for idx in range(start, len(tokens)):
        if tokens[idx] == target:
            return idx
    raise ValueError(f"No se encontró el token '{target}' en el archivo.")


def _scan_plate_block(tokens: list[str], start_idx: int, r: int, c: int, nr: int):
    """
    Replica el bucle interno del MATLAB basado en fscanf:
    lee un bloque de datos OD o GFP de tamaño (r x c x nr).
    """
    block = np.zeros((r, c, nr), dtype=float)
    idx = start_idx

    for n in range(nr):
        _ = int(tokens[idx])   # índice de fila (ignorado)
        idx += 1

        _ = tokens[idx]        # string de tiempo (ignorado)
        idx += 1

        idx += 3               # saltar tokens
        idx += 13

        for i in range(r):
            for j in range(c):
                block[i, j, n] = float(tokens[idx])
                idx += 1
            idx += 4           # saltar tokens de fin de fila

        idx += 3               # saltar tokens de fin de bloque

    return block, idx


def load_dedrick_plate_txt(filepath: Path, r: int = 8, c: int = 12, nr: int = 96) -> PlateData:
    """
    Carga el archivo .txt de cocultivo siguiendo la lógica del script MATLAB.
    """
    text = filepath.read_text(encoding="utf-8", errors="ignore")
    tokens = text.split()

    # Primer bloque: OD600
    idx_temp = _find_token_index(tokens, "Temperature", start=0)
    idx_time = _find_token_index(tokens, "Time", start=idx_temp)
    od6, idx_after_od = _scan_plate_block(tokens, idx_time + 1, r, c, nr)

    # Segundo bloque: GFP
    idx_temp2 = _find_token_index(tokens, "Temperature:", start=idx_after_od)
    idx_time2 = _find_token_index(tokens, "Time", start=idx_temp2)
    gfp, _ = _scan_plate_block(tokens, idx_time2 + 1, r, c, nr)

    return PlateData(od6=od6, gfp=gfp)


# ============================================================
# Extracción de cocultivo 1:1 (Sa–Sna)
# ============================================================
def extract_coculture_1to1_sa_sna(
    plate: PlateData,
    gfp_to_od: float,
    repl_rows_1based: range = range(2, 8),  # MATLAB 2:7
    col_1based: int = 5,
    baseline_points: int = 5,
):
    """
    Extrae las curvas experimentales de Sa y Sna a partir de GFP y OD600.
    """
    repl_idx = [i - 1 for i in repl_rows_1based]
    col_idx = col_1based - 1

    gfp_block = plate.gfp[repl_idx, col_idx, :]
    od_block = plate.od6[repl_idx, col_idx, :]

    # Sustracción de baseline
    gfp_baseline = np.nanmean(gfp_block[:, :baseline_points], axis=1, keepdims=True)
    od_baseline = np.nanmean(od_block[:, :baseline_points], axis=1, keepdims=True)

    gfp_corr = gfp_block - gfp_baseline
    od_corr = od_block - od_baseline

    sa_od = gfp_to_od * gfp_corr
    total_od = od_corr
    sna_od = total_od - sa_od

    sa_od_all = sa_od.T
    sna_od_all = sna_od.T

    s1e = np.nanmean(sa_od_all, axis=1)
    s2e = np.nanmean(sna_od_all, axis=1)

    s1e[0] = 0.0
    s2e[0] = 0.0

    return sa_od_all, sna_od_all, s1e, s2e


# ============================================================
# Ajuste de coeficientes de interacción (lsqnonlin → least_squares)
# ============================================================
def fit_interaction_coeffs(
    t: np.ndarray,
    dt: float,
    r1: float,
    r2: float,
    k1: float,
    k2: float,
    x1_0: float,
    x2_0: float,
    t1_lag: float,
    t2_lag: float,
    s1e: np.ndarray,
    s2e: np.ndarray,
    initial_c=(-1.25, -0.5),
    fit_window_hours=(3.0, 8.0),
):
    """
    Ajusta c12 y c21 minimizando el error cuadrático entre modelo y experimento
    en una ventana temporal dada.
    """
    t_min, t_max = fit_window_hours
    mask = (t >= t_min) & (t <= t_max)

    def residuals(params):
        c12_try, c21_try = params
        x1_sim, x2_sim = simulate_two_species_euler_with_lag(
            t, dt, r1, r2, k1, k2,
            c12_try, c21_try,
            x1_0, x2_0,
            t1_lag, t2_lag
        )
        return np.concatenate([
            x1_sim[mask] - s1e[mask],
            x2_sim[mask] - s2e[mask],
        ])

    result = least_squares(residuals, np.array(initial_c, dtype=float))
    c_opt = result.x

    x1_opt, x2_opt = simulate_two_species_euler_with_lag(
        t, dt, r1, r2, k1, k2,
        c_opt[0], c_opt[1],
        x1_0, x2_0,
        t1_lag, t2_lag
    )

    return c_opt, x1_opt, x2_opt


# ============================================================
# Ejecución principal
# ============================================================
# Simulación con coeficientes derivados del CFSM
s1_model, s2_model = simulate_two_species_euler_with_lag(
    t_grid, dt, r1, r2, k1, k2,
    c12, c21, s10, s20, t1_lag, t2_lag
)

print("Coeficientes de interacción (CFSM):")
print(f"  c12 = {c12:.4f}")
print(f"  c21 = {c21:.4f}")

# Cargar datos experimentales
plate = load_dedrick_plate_txt(INFILE)

# Factor de conversión GFP → OD (AJUSTAR según calibración real)
gfp_to_od = 1.0 / 192970

sa_od_all, sna_od_all, s1e, s2e = extract_coculture_1to1_sa_sna(
    plate, gfp_to_od
)

# Ajuste de coeficientes
c_opt, s1_fit, s2_fit = fit_interaction_coeffs(
    t_grid, dt, r1, r2, k1, k2,
    s10, s20, t1_lag, t2_lag,
    s1e, s2e
)

print("\nCoeficientes ajustados (least squares):")
print(f"  c12_opt = {c_opt[0]:.4f}")
print(f"  c21_opt = {c_opt[1]:.4f}")

# Gráficos
plt.figure(figsize=(10, 6))

plt.plot(t_grid, sa_od_all, color=(0.2, 0.4, 0.7), alpha=0.3)
plt.plot(t_grid, sna_od_all, color=(1.0, 0.4, 0.2), alpha=0.3)

plt.plot(t_grid, s1e, linewidth=2, label="Sa (exp)")
plt.plot(t_grid, s2e, linewidth=2, label="Sna (exp)")

plt.plot(t_grid, s1_model, "b:", linewidth=2, label="Sa (modelo CFSM)")
plt.plot(t_grid, s2_model, "r:", linewidth=2, label="Sna (modelo CFSM)")

plt.plot(t_grid, s1_fit, "b--", linewidth=2, label="Sa (modelo ajustado)")
plt.plot(t_grid, s2_fit, "r--", linewidth=2, label="Sna (modelo ajustado)")

plt.xlabel("Tiempo (horas)")
plt.ylabel("OD")
plt.xlim(2, 16)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
```

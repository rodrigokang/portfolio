## Overview

This notebook accompanies the project **Customer Value and Retention** and contains the computational implementation of the customer segmentation workflow developed using transactional data from the AdventureWorks database.

The objective is to identify customer groups with distinct purchasing patterns and to create a practical foundation for the churn prediction and customer lifetime value (CLV) models developed later in the project.

The analysis begins with the construction of Recency, Frequency, and Monetary Value (RFM) features, followed by exploratory analysis and customer segmentation using K-Means clustering. K-Means was selected because it provides a simple and interpretable framework for grouping customers with similar purchasing behaviour, which is often more valuable in practice than pursuing incremental improvements in clustering complexity.

The number of segments is not fixed in advance. Instead, several clustering diagnostics are evaluated, including inertia, silhouette score, Calinski-Harabasz score, and Davies-Bouldin score.

In addition to customer segmentation, the notebook introduces a continuous intra-cluster scoring framework designed to rank customers within each segment according to their relative commercial value and purchasing behaviour. This additional layer of analysis provides a practical mechanism for prioritising customers once the broader customer groups have been identified.

The notebook emphasizes reproducibility, interpretability, and practical decision-making. The focus is not only on obtaining statistically coherent segments, but also on understanding how those segments can support customer retention initiatives, customer prioritisation, and future revenue opportunities.

Although this notebook uses the full transaction history to produce a current view of the customer base, the same RFM framework will later be applied within rolling observation windows when developing the churn prediction and CLV models. This distinction is important because customer segmentation is treated here as a descriptive task, whereas the subsequent models are predictive and therefore require a strict separation between observation data and future outcomes.

```python
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)

pio.renderers.default = "notebook_connected"

seed = 42
np.random.seed(seed)

outputs_dir = Path("1-customer-value-and-retention-rfm-figures")
tables_dir = outputs_dir / "tables"

outputs_dir.mkdir(parents=True, exist_ok=True)
tables_dir.mkdir(parents=True, exist_ok=True)


def clean_filename(name):
    """
    Convert a table or figure name into a simple file-friendly stem.

    Inputs:
    -------
        name : str
            Original table or figure name.

    Outputs:
    --------
        str
            Sanitized filename.

    Author:
    -------
        Rodrigo Kang
    """

    return (
        str(name)
        .strip()
        .lower()
        .replace(" ", "-")
        .replace("_", "-")
    )


def save_table(data, filename, index=False):
    """
    Save a dataframe as CSV in the RFM output tables folder.

    Inputs:
    -------
        data : pandas.DataFrame
            Table to export.

        filename : str
            Output filename without extension.

        index : bool, default=False
            Whether to save the dataframe index.

    Outputs:
    --------
        pathlib.Path
            Path to the exported CSV file.

    Author:
    -------
        Rodrigo Kang
    """

    output_path = tables_dir / f"{clean_filename(filename)}.csv"

    index_label = "metric" if index else None
    data.to_csv(
        output_path,
        index=index,
        index_label=index_label
    )

    return output_path


def save_plotly_figure(fig, filename, width=1000, height=650, scale=2):
    """
    Save a Plotly figure as interactive HTML and static PNG.

    Notes:
    ------
        PNG export requires the kaleido package.

    Inputs:
    -------
        fig : plotly.graph_objects.Figure
            Figure to export.

        filename : str
            Output filename without extension.

        width : int, default=1000
            Figure width in pixels.

        height : int, default=650
            Figure height in pixels.

        scale : int, default=2
            Resolution scaling factor for PNG export.

    Outputs:
    --------
        tuple[pathlib.Path, pathlib.Path]
            Paths to the exported HTML and PNG files.

    Author:
    -------
        Rodrigo Kang
    """

    stem = clean_filename(filename)

    html_path = outputs_dir / f"{stem}.html"
    png_path = outputs_dir / f"{stem}.png"

    fig.write_html(html_path)

    try:
        fig.write_image(
            png_path,
            width=width,
            height=height,
            scale=scale
        )
    except Exception as error:
        print(
            "PNG export failed. Install or update kaleido if needed: "
            f"{error}"
        )

    return html_path, png_path


def apply_plotly_layout(fig, title, xaxis_title=None, yaxis_title=None):
    """
    Apply a consistent visual layout to portfolio figures.

    Inputs:
    -------
        fig : plotly.graph_objects.Figure
            Figure to format.

        title : str
            Figure title.

        xaxis_title : str, optional
            Label for the x-axis.

        yaxis_title : str, optional
            Label for the y-axis.

    Outputs:
    --------
        plotly.graph_objects.Figure
            Formatted Plotly figure.

    Author:
    -------
        Rodrigo Kang
    """

    fig.update_layout(
        title=title,
        template="plotly_white",
        width=1000,
        height=650,
        font=dict(
            family="Arial",
            size=14
        ),
        title_font=dict(size=20),
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=1.02,
            xanchor="right",
            x=1
        ),
        margin=dict(l=70, r=40, t=90, b=70)
    )

    if xaxis_title is not None:
        fig.update_xaxes(title_text=xaxis_title)

    if yaxis_title is not None:
        fig.update_yaxes(title_text=yaxis_title)

    return fig
```

### Extract Transactional Data

The query below extracts transaction data at the order-line level.

Each row represents one product purchased within a customer order. This level of detail is useful because it keeps the dataset flexible: customer behaviour can later be aggregated by total spend, number of orders, quantity purchased, product category diversity, territory, or time period.

The extraction query follows the simplified data model introduced above, joining customer, order, product, category, and territory information into a single analytical dataset.

```python
query = """
SELECT
    c.customerid AS customer_id,
    soh.salesorderid AS sales_order_id,
    soh.orderdate AS order_date,
    soh.duedate AS due_date,
    soh.shipdate AS ship_date,
    soh.status AS order_status,
    soh.territoryid AS territory_id,
    st.name AS territory_name,
    st.countryregioncode AS country_region_code,
    sod.salesorderdetailid AS sales_order_detail_id,
    sod.productid AS product_id,
    p.name AS product_name,
    pc.name AS product_category,
    ps.name AS product_subcategory,
    sod.orderqty AS order_quantity,
    sod.unitprice AS unit_price,
    sod.unitpricediscount AS unit_price_discount,
    (
        sod.orderqty
        * sod.unitprice
        * (1 - sod.unitpricediscount)
    ) AS line_total
FROM sales.customer AS c
INNER JOIN sales.salesorderheader AS soh
    ON c.customerid = soh.customerid
INNER JOIN sales.salesorderdetail AS sod
    ON soh.salesorderid = sod.salesorderid
INNER JOIN production.product AS p
    ON sod.productid = p.productid
LEFT JOIN production.productsubcategory AS ps
    ON p.productsubcategoryid = ps.productsubcategoryid
LEFT JOIN production.productcategory AS pc
    ON ps.productcategoryid = pc.productcategoryid
LEFT JOIN sales.salesterritory AS st
    ON soh.territoryid = st.territoryid;
"""

transactions = pd.read_sql(query, engine)
```

## RFM Feature Engineering

Customer segmentation begins with the construction of Recency, Frequency, and Monetary Value (RFM) variables.

RFM analysis is one of the most widely used approaches for customer segmentation because it captures three fundamental dimensions of purchasing behaviour:

- Recency measures how recently a customer has placed an order.
- Frequency measures how often a customer purchases.
- Monetary Value measures the total revenue generated by the customer.

Together, these variables provide a compact and interpretable representation of customer activity, making them a practical starting point for segmentation analysis.

### Reference Date

Recency is calculated relative to a fixed reference date corresponding to the most recent transaction observed in the dataset.

Using a common reference date ensures that recency values are comparable across all customers.

```python
reference_date = (
    transactions["order_date"].max()
    + pd.Timedelta(days=1)
)

reference_date
```

### Customer-Level Aggregation

The transactional dataset is aggregated at the customer level to construct the Recency, Frequency, and Monetary Value variables.

The resulting dataset contains one row per customer and forms the basis for all subsequent segmentation analyses.

```python
rfm = (
    transactions
    .groupby("customer_id")
    .agg(
        recency=(
            "order_date",
            lambda x: (
                reference_date - x.max()
            ).days
        ),
        frequency=(
            "sales_order_id",
            "nunique"
        ),
        monetary=(
            "line_total",
            "sum"
        )
    )
    .reset_index()
)

rfm_preview = rfm.head()

save_table(rfm_preview, "rfm-preview")

rfm_preview
```

Such distributions can adversely affect distance-based clustering algorithms because a small number of high-value observations may dominate the clustering structure. To mitigate this effect and improve cluster stability, the RFM variables will be transformed before applying K-Means.

### RFM Transformation and Scaling

The raw RFM variables are strongly skewed and measured on very different scales. This matters because K-Means is a distance-based algorithm: variables with larger ranges or extreme values can dominate the clustering result.

To reduce this effect, the RFM variables are first transformed using the Box-Cox transformation and then standardized using `StandardScaler`.

Box-Cox is applied after adding one to each variable, ensuring that all values are strictly positive. The transformed variables are then scaled to have approximately zero mean and unit variance.

The goal is not to force the data to be perfectly normal, but to create a more balanced feature space for clustering.

```python
rfm_transformed = rfm.copy()

rfm_features = ["recency", "frequency", "monetary"]

boxcox_lambdas = {}

for feature in rfm_features:
    transformed_values, fitted_lambda = boxcox(rfm_transformed[feature] + 1)

    rfm_transformed[f"{feature}_boxcox"] = transformed_values
    boxcox_lambdas[feature] = fitted_lambda

boxcox_lambdas
```

```python
scaler = StandardScaler()

score_columns = [
    "score_recency",
    "score_frequency",
    "score_monetary"
]

rfm_transformed[score_columns] = scaler.fit_transform(
    rfm_transformed[
        [
            "recency_boxcox",
            "frequency_boxcox",
            "monetary_boxcox"
        ]
    ]
)

rfm_transformed_preview = rfm_transformed.head()

save_table(rfm_transformed_preview, "rfm-transformed-preview")

rfm_transformed_preview
```

## Selection of the Number of Clusters

Unlike traditional RFM approaches that define customer groups using predefined business rules, the objective here is to allow the clustering structure to emerge from the data.

Because no single metric provides a complete assessment of clustering quality, several complementary diagnostics are evaluated across a range of candidate values for $K$.

The analysis considers:

- Within-cluster inertia (Elbow Method).
- Silhouette Score.
- Calinski-Harabasz Index.
- Davies-Bouldin Index.

Each metric emphasizes a different aspect of cluster quality. Rather than relying on a single criterion, the final choice of $K$ is based on the combined evidence provided by all four diagnostics.

```python
# Features used for clustering
clustering_features = [
    "score_recency",
    "score_frequency",
    "score_monetary"
]

# Candidate values for K
k_values = range(2, 11)

# Containers for evaluation metrics
inertia_scores = []
silhouette_scores = []
calinski_scores = []
davies_scores = []

# Evaluate clustering performance
for k in k_values:

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=20
    )

    labels = model.fit_predict(
        rfm_transformed[clustering_features]
    )

    inertia_scores.append(
        model.inertia_
    )

    silhouette_scores.append(
        silhouette_score(
            rfm_transformed[clustering_features],
            labels
        )
    )

    calinski_scores.append(
        calinski_harabasz_score(
            rfm_transformed[clustering_features],
            labels
        )
    )

    davies_scores.append(
        davies_bouldin_score(
            rfm_transformed[clustering_features],
            labels
        )
    )

# Summary table
cluster_metrics = pd.DataFrame({
    "k": list(k_values),
    "inertia": inertia_scores,
    "silhouette": silhouette_scores,
    "calinski_harabasz": calinski_scores,
    "davies_bouldin": davies_scores
})

cluster_metrics = cluster_metrics.round(4)

save_table(cluster_metrics, "cluster-metrics")

cluster_metrics
```

```python
metric_labels = {
    "inertia": "Inertia",
    "silhouette": "Silhouette Score",
    "calinski_harabasz": "Calinski-Harabasz Index",
    "davies_bouldin": "Davies-Bouldin Index"
}

cluster_metrics_long = cluster_metrics.melt(
    id_vars="k",
    value_vars=list(metric_labels.keys()),
    var_name="metric",
    value_name="value"
)

cluster_metrics_long["metric"] = (
    cluster_metrics_long["metric"]
    .map(metric_labels)
)

fig = px.line(
    cluster_metrics_long,
    x="k",
    y="value",
    markers=True,
    facet_col="metric",
    facet_col_wrap=2,
    labels={
        "k": "Number of Clusters",
        "value": "Score",
        "metric": ""
    },
    height=750
)

fig.update_traces(line_width=2)
fig.update_yaxes(matches=None, showticklabels=True)
fig.for_each_annotation(
    lambda annotation: annotation.update(
        text=annotation.text.split("=")[-1]
    )
)

fig = apply_plotly_layout(
    fig,
    title="Clustering Diagnostics",
    xaxis_title="Number of Clusters",
    yaxis_title="Score"
)

save_plotly_figure(
    fig,
    "clustering-diagnostics",
    width=1000,
    height=750
)

fig.show()
```

The clustering diagnostics indicate that customer structure continues to improve as additional clusters are introduced. In particular, the Silhouette Score reaches its highest value at $K = 9$, while the Davies-Bouldin Index reaches its minimum value at the same point.

Although the statistical evidence slightly favours nine clusters, customer segmentation also requires practical interpretability. For this reason, the final decision will consider both clustering quality and the business meaning of the resulting customer groups.

The next step is therefore to examine the cluster profiles and determine whether the additional complexity introduced by larger values of $K$ produces genuinely distinct customer segments.

## Customer Prioritisation Within Clusters

The cluster assignment identifies groups of customers with similar purchasing behaviour. However, customers belonging to the same segment may still differ substantially in their relative importance.

To support prioritisation within each cluster, an additional RFM score is calculated for every customer. The score combines the transformed and standardized Recency, Frequency, and Monetary variables into a single continuous measure.

Since lower recency indicates more recent activity, the Recency component enters the score with a negative sign, while Frequency and Monetary contribute positively.

Rather than replacing the cluster labels, the score provides a ranking mechanism that helps identify the strongest and weakest customers within each segment.

### Cluster-Specific Weighting

The contribution of each RFM component is allowed to vary across clusters.

Weights are derived from the within-cluster variance of the transformed RFM variables. Variables exhibiting greater variation receive larger weights because they contribute more strongly to differentiating customers within that particular segment.

This approach produces a cluster-specific prioritisation score while preserving the overall segmentation structure.

```python
def calculate_cluster_weights(
    data,
    cluster_column,
    score_columns
):
    """
    Calculate cluster-specific weights from within-cluster variance.

    Input:
        data (pd.DataFrame): Dataset containing cluster labels and score variables.
        cluster_column (str): Name of the column containing cluster labels.
        score_columns (list): List of standardized score columns used for weighting.

    Output:
        pd.DataFrame: Cluster-level weights for each score variable.

    Author:
        Rodrigo Kang
    """

    variance_by_cluster = (
        data
        .groupby(cluster_column)[score_columns]
        .var()
    )

    cluster_weights = variance_by_cluster.div(
        variance_by_cluster.sum(axis=1),
        axis=0
    )

    return cluster_weights
```

```python
def assign_rfm_score(
    data,
    cluster_column,
    score_columns,
    cluster_weights
):
    """
    Assign a weighted RFM score to each customer.

    Input:
        data (pd.DataFrame): Dataset containing customer-level RFM scores and cluster labels.
        cluster_column (str): Name of the column containing cluster labels.
        score_columns (list): List containing recency, frequency, and monetary score columns.
        cluster_weights (pd.DataFrame): Cluster-specific weights.

    Output:
        pd.DataFrame: Dataset with cluster-specific weights and RFM scores.

    Author:
        Rodrigo Kang
    """

    output = data.copy()

    recency_col, frequency_col, monetary_col = score_columns

    output["omega_recency"] = output[cluster_column].map(
        cluster_weights[recency_col]
    )

    output["omega_frequency"] = output[cluster_column].map(
        cluster_weights[frequency_col]
    )

    output["omega_monetary"] = output[cluster_column].map(
        cluster_weights[monetary_col]
    )

    output["rfm_score"] = (
        - output[recency_col] * output["omega_recency"]
        + output[frequency_col] * output["omega_frequency"]
        + output[monetary_col] * output["omega_monetary"]
    )

    scaler = MinMaxScaler()

    output["rfm_normalized_score"] = scaler.fit_transform(
        output[["rfm_score"]]
    )

    return output
```

```python
score_columns = [
    "score_recency",
    "score_frequency",
    "score_monetary"
]

scoring_data = (
    rfm
    .merge(
        rfm_transformed[
            [
                "customer_id",
                "score_recency",
                "score_frequency",
                "score_monetary"
            ]
        ],
        on="customer_id",
        how="left"
    )
)
```

```python
cluster_weights = calculate_cluster_weights(
    data=scoring_data,
    cluster_column="cluster",
    score_columns=score_columns
)

cluster_weights.round(4)
```

The resulting weights vary across clusters, reflecting the different sources of behavioural variation present within each customer segment.

For example, some clusters are primarily differentiated by recency, while others are more strongly influenced by spending behaviour or purchase frequency. As a result, the prioritisation score adapts to the characteristics of each segment rather than applying a single weighting scheme to the entire customer base.

```python
rfm_scored = assign_rfm_score(
    data=scoring_data,
    cluster_column="cluster",
    score_columns=score_columns,
    cluster_weights=cluster_weights
)
```

```python
rfm_scored_preview = rfm_scored[
    [
        "customer_id",
        "recency",
        "frequency",
        "monetary",
        "cluster",
        "omega_recency",
        "omega_frequency",
        "omega_monetary",
        "score_recency",
        "score_frequency",
        "score_monetary",
        "rfm_score",
        "rfm_normalized_score"
    ]
].head()

save_table(rfm_scored_preview, "rfm-scored-preview")

rfm_scored_preview
```

The normalized RFM score ranges from 0 to 1 and can be interpreted as a relative ranking within the customer base.

Customers with higher scores generally exhibit more favourable combinations of recency, frequency, and monetary value, while lower scores indicate weaker engagement and commercial contribution.

This score complements the clustering results by introducing a continuous prioritisation measure that can support marketing, retention, and customer management initiatives.

## Segment Interpretation

The clustering algorithm identifies groups of customers with similar purchasing behaviour, but the resulting cluster labels have no inherent business meaning.

To make the segmentation actionable, each cluster is interpreted using its average Recency, Frequency, and Monetary characteristics. The objective is to translate the statistical output of the clustering algorithm into customer segments that can support retention, marketing, and customer development initiatives.

The table below summarizes the behavioural profile of each segment.

```python
segment_summary = (
    rfm_scored
    .groupby("cluster")
    .agg(
        customers=("customer_id", "count"),
        recency=("recency", "mean"),
        frequency=("frequency", "mean"),
        monetary=("monetary", "mean"),
        avg_score=("rfm_normalized_score", "mean")
    )
    .round(2)
    .sort_values(
        "monetary",
        ascending=False
    )
)

save_table(segment_summary, "segment-summary-by-monetary", index=True)

segment_summary
```

```python
segment_names = {
    5: "Champions",
    1: "High Value but Cooling",
    7: "Lost Customers",
    8: "Recent Low-Value Customers",
    3: "Inactive Low-Value Customers",
    4: "Frequent Low-Spend Customers",
    6: "Declining Frequent Buyers",
    0: "Active Minimal Buyers",
    2: "Inactive Minimal Buyers"
}

segment_order = [
    "Champions",
    "High Value but Cooling",
    "Frequent Low-Spend Customers",
    "Recent Low-Value Customers",
    "Active Minimal Buyers",
    "Declining Frequent Buyers",
    "Inactive Low-Value Customers",
    "Inactive Minimal Buyers",
    "Lost Customers"
]

segment_summary = (
    rfm_scored
    .groupby("cluster")
    .agg(
        customers=("customer_id", "count"),
        recency=("recency", "mean"),
        frequency=("frequency", "mean"),
        monetary=("monetary", "mean"),
        avg_score=("rfm_normalized_score", "mean")
    )
    .round(2)
    .reset_index()
)

segment_summary["segment"] = segment_summary["cluster"].map(segment_names)

segment_summary["segment"] = pd.Categorical(
    segment_summary["segment"],
    categories=segment_order,
    ordered=True
)

segment_summary = (
    segment_summary
    .sort_values("segment")
    .reset_index(drop=True)
)

save_table(segment_summary, "segment-summary")

segment_summary
```

Several distinct customer behaviours emerge from the segmentation.

The most valuable segment consists of customers who purchase relatively frequently and generate substantially higher revenue than the rest of the customer base. These customers represent a natural target for retention and relationship management initiatives.

At the opposite end of the spectrum, several segments are characterised by low spending levels and limited purchasing activity. The distinction between active and inactive low-value customers is primarily driven by recency, suggesting different levels of future engagement.

The segmentation also reveals an intermediate group of customers who purchase frequently but generate relatively low revenue. These customers may represent opportunities for cross-selling, up-selling, or product mix optimisation.

The score distributions reveal meaningful differences across segments.

Champions and Frequent Low-Spend Customers exhibit the highest score levels, indicating the presence of customers that are particularly strong relative to others within their respective groups. Although both segments achieve similar score ranges, their business interpretation differs substantially because their average monetary values are very different.

Several intermediate segments, including High Value but Cooling and Declining Frequent Buyers, display moderate score distributions. These groups may contain customers with retention or development potential depending on the business objective.

At the opposite end of the spectrum, Lost Customers exhibit the lowest scores and the smallest prioritisation potential. Their long inactivity periods make them less attractive targets for immediate retention efforts.

Importantly, the normalized RFM score should be interpreted as a segment-specific prioritisation metric rather than a global measure of customer value. A customer with a high score in a low-value segment may still generate less revenue than a customer with a moderate score in a high-value segment.

The score should be interpreted primarily as a ranking mechanism within segments rather than as a global measure of customer value. Customers with higher scores occupy stronger positions relative to other customers in the same segment and may represent higher-priority targets for retention, development, or marketing initiatives.

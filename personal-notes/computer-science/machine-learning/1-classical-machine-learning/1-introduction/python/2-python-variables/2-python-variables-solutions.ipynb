## Problem 1 — From Algorithm to Execution

### Problem

Consider the following statements:

1. An algorithm is the same thing as a Python program.
2. A high-level programming language provides abstractions that reduce the need to work directly with machine-level operations.
3. Python source code is executed directly by the processor without any intermediate representation.
4. A programming language provides a formal way to express algorithms in a form that can ultimately be executed by a computer.

For each statement:

1. Determine whether it is correct.
2. If it is incorrect, identify the conceptual error.
3. Explain the relationship among an **algorithm**, a **programming language**, a **program**, and the underlying computer hardware.

### Solution

1. **Incorrect.** An algorithm is an abstract, finite procedure for solving a problem; a Python program is one concrete implementation expressed in Python.
2. **Correct.** High-level languages hide many processor- and memory-level details behind abstractions such as objects, functions, and data structures.
3. **Incorrect.** In CPython, source code is normally compiled to Python bytecode, which is then executed by the Python virtual machine rather than being executed directly by the processor as source text.
4. **Correct.** A programming language supplies formal syntax and semantics for expressing computations that an implementation can ultimately execute.

The relationship can be viewed as:

**algorithm → expression in a programming language → program → language implementation/runtime → machine operations on hardware**

The algorithm is the method; the language is the formal notation; the program is a particular expression of the method; and the implementation bridges that program to executable machine-level behaviour.

### Key Takeaway

Keep the abstraction layers distinct: an algorithm is not a program, and source code is not hardware execution.

## Problem 2 — Compiled or Interpreted?

### Problem

Python is often described as an interpreted language.

### Questions

1. Why is this description useful but incomplete?
2. Describe, at a high level, what happens to Python source code when it is executed by CPython.
3. What is Python bytecode?
4. What role does the Python virtual machine play?
5. Why should the distinction between compiled and interpreted languages not be treated as absolute?

### Solution

Calling Python “interpreted” is useful because normal Python development does not involve an explicit ahead-of-time native compilation step. It is incomplete because CPython does perform compilation internally.

At a high level, CPython parses source code and compiles it into **Python bytecode**. Bytecode is a lower-level, platform-independent instruction representation for the Python execution engine. The **Python virtual machine (PVM)** executes those bytecode instructions.

Therefore, “compiled” and “interpreted” describe implementation strategies rather than two perfectly exclusive classes of programming language. A language can involve both compilation and interpretation at different stages.

### Key Takeaway

Python execution can involve both compilation and interpretation; the distinction is implementation-dependent.

## Problem 6 — Modules and Namespaces

### Problem

Suppose a module named `statistics_tools.py` contains:

## Problem 7 — Import Styles

### Problem

Consider the following alternatives:

```python
import math
```

```python
import math as mt
```

```python
from math import sqrt, pi
```

```python
from math import *
```

### Questions

1. How does each form affect the current namespace?
2. Which forms preserve the module name as part of the reference?
3. What advantages can aliases provide?
4. Why are explicit imports generally preferable to wildcard imports?
5. Under what circumstances could two wildcard imports create a subtle error?
6. Which import style would you generally prefer in shared production code, and why?

### Solution

- `import math` binds the name `math` in the current namespace.
- `import math as mt` binds the module to the alias `mt`.
- `from math import sqrt, pi` binds `sqrt` and `pi` directly in the current namespace.
- `from math import *` introduces the module's exported public names directly into the current namespace.

The first two forms preserve a module qualifier (`math.sqrt` or `mt.sqrt`). Aliases are useful when they are conventional or when a module name is inconveniently long.

Explicit imports make dependencies and name origins visible. Wildcard imports can silently overwrite existing names; two wildcard imports exporting the same name can make the later binding replace the earlier one. In shared production code, explicit module imports or explicit `from ... import ...` forms are generally preferable because they make name provenance clear.

#### Import Examples

`````python
import math
math.sqrt(16)

import math as mt
mt.pi

from math import sqrt, pi
sqrt(25), pi
`````

### Key Takeaway

Prefer imports that make dependencies and name origins clear.

## Problem 9 — Module Caching

### Problem

Suppose a module contains:

```python
print("module loaded")
```

and is imported twice during the same Python process:

```python
import my_module
import my_module
```

### Questions

1. Would you normally expect `"module loaded"` to be printed once or twice?
2. Why?
3. What role does `sys.modules` play?
4. Suppose the source file changes after the first import. Why might importing it again not reflect the change?
5. What are three ways to obtain the updated module during interactive development?
6. What is the conceptual difference between restarting the kernel and reloading only one module?

### Solution

`"module loaded"` would normally be printed **once**. After the first successful import, Python records the loaded module in `sys.modules`. A subsequent ordinary import in the same process normally reuses that module object instead of executing the module from scratch.

If the source file changes afterwards, another ordinary `import my_module` can therefore continue using the already loaded version.

During interactive development, common approaches include:
1. `importlib.reload(my_module)`;
2. IPython's `autoreload` extension;
3. restarting the kernel/process and importing again.

Reloading targets a particular module while preserving most of the current process state. Restarting the kernel discards the whole in-memory Python state and creates a clean execution environment.

## Problem 16 — Assignment Is Not Equality

### Problem

Consider:

```python
x = 3
x = x + 1
```

### Questions

1. Why is the second statement valid in Python even though the corresponding mathematical equation

$$

$$

would be inconsistent over ordinary numbers?
2. Describe the order in which Python evaluates the second statement.
3. What object is associated with `x` before the second assignment?
4. What object is associated with `x` afterwards?
5. Why is the term **reassignment** appropriate?

### Solution

`=` is assignment, not mathematical equality. Python first evaluates the entire right-hand side `x + 1` using the current binding of `x`, obtains a result, and only then binds `x` to that result.

Initially, `x` is bound to the integer `3`. Evaluating `x + 1` produces the integer value `4`, after which `x` is rebound to `4`.

The term **reassignment** is appropriate because the same name receives a new binding. The mathematical expression \(x=x+1\) would assert equality between both sides simultaneously; the Python statement describes a state-changing operation performed in an evaluation order.

#### Reassignment

`````python
x = 3
x = x + 1
x
`````

### Key Takeaway

Assignment changes bindings; it does not assert mathematical equality.

## Problem 27 — Namespace Collision

### Problem

Suppose two modules contain functions with the same name:

```python
module_a.transform
module_b.transform
```

A programmer writes:

```python
from module_a import transform
from module_b import transform
```

### Questions

1. Which `transform` name will be available after both imports?
2. What problem has occurred conceptually?
3. How could module-qualified imports avoid the ambiguity?
4. Why is this an argument against indiscriminate wildcard imports?

### Solution

After both direct imports, the current name `transform` normally refers to the function imported from `module_b`, because the second binding replaces the first binding of that name.

The conceptual problem is a **namespace collision**: two different objects are being introduced under the same unqualified name.

Module-qualified imports avoid the ambiguity:

`import module_a`
`import module_b`

The functions can then remain explicitly distinguishable as `module_a.transform(...)` and `module_b.transform(...)`. Wildcard imports increase this risk because many names can enter the namespace without being individually visible in the import statement.

#### Avoiding the Collision

`````python
# Preferred conceptual pattern:
# import module_a
# import module_b
#
# module_a.transform(...)
# module_b.transform(...)
`````

### Key Takeaway

Namespace qualification prevents ambiguity when different modules expose the same names.

## Problem 28 — Reproducibility in Interactive Computing

### Problem

A data scientist performs the following actions in a notebook:

1. Imports a module.
2. Modifies the module's source file externally.
3. Runs a later notebook cell expecting the new behaviour.
4. Receives results corresponding to the old implementation.

### Questions

1. Explain why this can happen.
2. How could `importlib.reload()` help?
3. How could IPython's `autoreload` extension help?
4. Why might restarting the kernel still be the safest option in some situations?
5. What broader lesson does this illustrate about hidden execution state in interactive environments?

### Solution

This can happen because importing a module normally places the loaded module in `sys.modules`. Editing the source file does not automatically replace the already loaded module object.

`importlib.reload(module)` can re-execute a loaded module's code. IPython's `autoreload` extension can automate reloading during interactive development. Restarting the kernel is more disruptive but provides a clean process state, which can be safer when dependencies, imported names, references, or accumulated notebook state make selective reloading difficult to reason about.

The broader lesson is that an interactive environment has **execution state** that is not always obvious from the visible order of cells. Reproducible work should minimise reliance on hidden state and should be testable from a clean kernel in a defined order.

### Key Takeaway

### Solution

The scenario combines several foundational ideas:

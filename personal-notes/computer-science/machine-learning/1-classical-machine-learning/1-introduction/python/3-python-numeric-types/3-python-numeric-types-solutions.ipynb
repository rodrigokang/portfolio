## Problem 2 — Operator precedence

### Problem

Without executing the code, determine the result of each expression:

```python
True or False and False
(True or False) and False
not True or True
not (True or True)
```

Explain how operator precedence affects the results.

### Solution

Python's Boolean operators follow a precedence order: `not` is evaluated before `and`, and `and` before `or`, unless parentheses explicitly change the grouping.

The safest way to reason about a compound expression is therefore:
1. evaluate parenthesised subexpressions;
2. apply `not`;
3. apply `and`;
4. apply `or`.

Parentheses should be preferred whenever they make the intended logic clearer, even when operator precedence would already produce the desired result.

#### Precedence Check

````python
True or False and False
(True or False) and False
````

### Key Takeaway

Use operator precedence correctly, but prefer parentheses when they improve clarity.

## Problem 3 — Comparisons

### Problem

Determine whether each expression evaluates to `True` or `False`:

```python
7 == 7.0
7 == "7"
True == 1
False == 0
3 < 5 < 10
3 < 5 > 10
```

Explain any results that may initially appear surprising.

### Solution

`and` and `or` use **short-circuit evaluation**. Python evaluates only as much of the expression as is necessary to determine the result.

For `A and B`, if `A` is falsey, Python does not need to evaluate `B`. For `A or B`, if `A` is truthy, Python does not need to evaluate `B`.

This matters because the skipped expression may be expensive, may raise an exception, or may have side effects.

#### Short-circuit Example

````python
def trace():
    print("evaluated")
    return True

False and trace()
True or trace()
````

### Key Takeaway

Short-circuit evaluation can skip expressions that are unnecessary to determine the result.

## Problem 4 — Compound conditions

### Problem

Given:

```python

has_degree = True
```

Write a Boolean expression that evaluates to `True` only when a candidate:

- is at least 25 years old;
- has at least 3 years of experience; and
- either has a degree or has at least 7 years of experience.

### Solution

In Python, Boolean operators do not necessarily return literal `True` or `False`. They can return one of their operands.

`x and y` returns the first falsey operand encountered, or the last operand if all are truthy. `x or y` returns the first truthy operand encountered, or the last operand if all are falsey.

This behaviour follows directly from short-circuit evaluation and is widely used for compact conditional expressions.

#### Operand-returning Behaviour

````python
0 or 5
"data" and 42
````

### Key Takeaway

`and` and `or` can return operands rather than literal Boolean values.

Appropriate representations are:

The representation should be chosen from the semantics of the quantity, not merely from how its value happens to be written.

# 1. Statistical Learning

While some experts argue that Statistical Learning (**SL**) and Machine Learning (**ML**) are two distinct disciplines that complement each other, a more appropriate perspective would be that both are equivalent, and, in reality, it is more about **Statistical Learning** than **Machine Learning**. In other words:

$$\boxed{\mathbf{SL} \equiv \mathbf{ML}}$$

At its core, **SL** is based on principles of statistics (**S**), numerical analysis (**NA**), and computer science (**CS**):

$$\boxed{\mathbf{SL} \equiv \mathbf{S} + \mathbf{NA} + \mathbf{CS}}$$

## 1.1. Goals of SL

**SL** encompasses a set of methods primarily designed for two fundamental tasks:

1. Inference
2. Prediction

To comprehend this, consider a scenario where we have $M$ predictors, denoted as $X_1, X_2, \dots, X_M$, or represented compactly as:

$$X := (X_1, X_2, \dots, X_M)$$

Furthermore, suppose that $Y$ is a quantitative response and that there is a relationship between $Y$ and $X$. Without loss of generality, we can express this relationship as:

$$Y = f(X) + \varepsilon$$

where:

* $f$ represents some fixed but unknown function of the predictors.
* $\varepsilon$ denotes the random error term which satisfies:
    * $\varepsilon \neq \varepsilon(X)$ ($\varepsilon$ is independent of $X$)
    * $\bar{\varepsilon} = 0$

The primary objective of **SL** is to estimate the function $f$ to better understand the relationship between $Y$ and $X$ (inference) and to be able to predict the value of the response $Y$ knowing the predictors values $X$ (prediction).

### 1.1.1. Inference

In statistics, a key issue challenge lies in the inherent limitation of not having complete information,  rather, we are overall confronted with only have a part it.

![inference](inference.png)

To address this issue, we employ the concept of inference, which encompass a collection of methods and procedures that, through induction, allow us to deduce properties of a statistical population based on a subset of it.

* **Induction** involves drawing conclusions from the analysis of specific instances to generalize broader principles.

* **Population** refers to the set of elements or events under consideration to answer a question or conduct an experiment.

* **Sample** is a representative portion of the population.

### 1.1.2. Prediction

In order to predict the value of the response $Y$ we can use

$$\hat{Y} = \hat{f}(X)$$

where:

* $\hat{f}$ is the estimate of f
* $\hat{Y}$ is the resulting prediction of $Y$

#### Accuracy of $\hat{Y}$

$$
\begin{split}
\mathbb{E}[Y-\hat{Y}]^2 & = \mathbb{E}[f(X) + \varepsilon - \hat{f}]^2 \\
& = \mathbb{E}[f(X) - \hat{f}]^2 + 2\mathbb{E}\{ [f(X) - \hat{f}]\varepsilon \} + \mathbb{E}(\varepsilon^2)\\
& = [ f(X) - \hat{f}(X) ]^2 + 2 \; [ f(X) - \hat{f}(X) ] \; \mathbb{E}(\varepsilon) + \mathrm{Var}(\varepsilon) + \mathbb{E}^2(\varepsilon)
\end{split}
$$

\\
$$

$$
\begin{split}
& \Rightarrow \mathbb{E}[Y-\hat{Y}]^2 = [ f(X) - \hat{f}(X) ]^2 + \mathrm{Var}(\varepsilon) \\ \\
& \rightarrow \boxed{\mathrm{Error} = \mathrm{Reducible} + \mathrm{Irreducible}}
\end{split}
$$

**Remarks**

* The term $[ f(X) - \hat{f}(X) ]^2$ is reducible because we can enhance the estimation $\hat{f}$ by implementing a more suitable **SL** technique. On the other hand, the term $\mathrm{Var}(\varepsilon)$ is irreducible because $Y = Y(\varepsilon)$ and $\varepsilon \neq \varepsilon(X)$. No matter how good the estimation of $\hat{f}$ is, the term $\varepsilon \nrightarrow 0$.

* It may happen that $\varepsilon = \varepsilon(X')$, where $X'$ corresponds to other predictors that are useful in predicting $Y$ but are not measured. Therefore, $X'$ would not be included in $f$ to predict $Y$.

* $\varepsilon$ may encompass unmeasurable variations. For instance, in the case of an adverse reaction to a medication in a specific patient on a given day, it varies based on the manufacturing of the medication itself and the overall well-being sensation of the patient that day.

#### Methods for estimating $f$

Overall, the majority of statistical learning methods used to estimate $f$ can be classified as either parametric or non-parametric.

1. **Parametric Methods**:
    1. **Procedure**:
        * $f:=f(\theta_0,\theta_1, \dots)$
        * $\mathrm{TD} = \{(x_{ij},y_i); i=1,2, \dots, N; j=1,2, \dots, M\} \overset{\text{LS, ML, ...}}{\longrightarrow} \hat{\theta}_0, \hat{\theta}_1, \dots$
    2. **Advantages and Disadvantage**:
    
        * **Underfitting, Overfitting**:
          Underfitting occurs when a model is too simple to capture the underlying patterns in the data. It fails to learn the           relationships between the input features and the target variable, resulting in poor performance on both the training           and unseen data. In other words, the model lacks the complexity needed to represent the true underlying structure of           the data.
          On the other hand, overfitting indeed happens when a model is excessively complex and fits the training data too               closely, capturing not only the underlying patterns but also the noise or random fluctuations present in the data.             This excessive fitting to the training set can lead to poor generalization on new, unseen data, as the model has               essentially memorized the specific instances in the training data rather than learning the broader relationships and           trends.
          
        * **Non-flexible and "cheaps"**:
          Rigid models but they are not demanding in terms of computational resources (in general $M$ is small).
          
        * **Easy interpretation**:
          Straightforward interpretation of the relationship between $X$ and $Y$

2. **Non-Parametric Methods**:
    1. **Procedure**:
        * Instead of assuming an a priori functional form $f$, the objective is to seek the estimate $\hat{f}$ that best fits             the TD.
    2. **Advantages and Disadvantage**:
        * **Flexible and "expensives" (in general $N$ is large)**:
          Flexible models but they come with high computational demands, generally involving a large value for $N$.
        * **Hard interpretation**:
          Interpreting the relationship between $X$ and $Y$ poses a significant difficulty.

## 1.2. Algorithms in Statistical Learning

There is two types of algorithms in Statistical Learning, supervised and unsupervised

**Supervised**:
* **Description**: In supervised learning, the training data consists of observations $x_{ij}$, $i=1,2,\dots, N$, $j=1,2,\dots, M$ associated with a response $y_i$. The goal is to design a model that relates the response to the predictors, aiming to accurately predict the response for future observations (prediction) or better understand the relationship between the response and predictors (inference).
* **Examples of Applications**:
    * Classification: Predicting the category to which a new instance belongs.
    * Regression: Predicting a continuous numerical value.

**Unsupervised**:
* **Description**: In unsupervised learning, the training data consists of observations $x_{ij}$, $i=1,2,\dots, N$, $j=1,2,\dots, M$ that are not labeled with a response. The goal is to find patterns and understand the relationships and properties between the observations. These algorithms work well for tasks like classification.
* **Examples of Applications**:
    * Clustering: Grouping similar data points together.
    * Dimensionality Reduction: Reducing the number of variables in the data while preserving relevant information.
    * Association: Discovering patterns of association between variables.

There is an analogy quite useful in understanding the different ways humans can learn, which can be globally classified as assisted learning, i.e., through a teacher in an institution (supervised learning), or unassisted learning, i.e., self-taught (unsupervised learning).

## 2.1. Computer systems

Essentially, a computer system consists of two main components: **hardware** and **software**. Next, we will provide a brief description of each of these parts.

### 2.1.1.  Computer hardware concept

The term **hardware** refers to the physical and tangible component of a computer system. In broad terms, a computer comprises five main elements: **input devices**, **output devices**, **the processor** (also known as CPU, Central Processing Unit), **main memory**, and **secondary memory**. In essence, hardware represents the essential physical infrastructure that constitutes a computer system and enables its operation. The hardware infrastructure is illustrated in the following figure:

**Main Components of a Computer**

![hardware_components](hardware_components.png)

Below, we will concisely explain each of these parts:

- **Input Devices**: Any device that allows a person to transmit information to the computer, such as the keyboard or mouse.
- **Output Devices**: Any device that allows the computer to transmit information to us, such as the monitor or printer.
- **Memory**: In order to process information, computers have memory. A computer has two forms of memory, known as main memory and secondary memory.
    - **Main Memory**: Main memory consists of an extensive list of numbered locations known as memory addresses. The number of these addresses varies between computers, ranging from a few thousand to several million, and even billions in certain cases. Each memory address contains a sequence of **bits**, representing ones and zeros. In most computers, all memory addresses contain the same number of bits, with a **bit** being a **binary digit** that can be zero or one (a concept that will be explored later). Generally, memory addresses contain eight bits (or multiples of eight bits), referred to as a **byte**. Therefore, we can refer to these numbered locations in main memory as bytes. To simplify the description, we can conceive the computer's main memory as an extensive list of numbered locations called bytes. Each byte has a unique **address**. Data such as numbers or letters can be stored in these bytes, and the address is used to locate the data when needed. If data is too large to fit in a single byte, the computer uses adjacent bytes to store it, and the entire memory that houses that data is called a memory location. In practice, a computer's main memory resembles a list of memory locations of variable sizes. Each location is expressed in bytes, and the address of the first byte is used as the name or address of that memory location. Flexibility in memory location sizes allows dynamic changes when running a new program on the computer. In summary, main memory is organized into numbered locations called bytes, each with a unique address. To store data, such as numbers or letters, multiple consecutive bytes are used, and the address of the first byte becomes the address of the entire memory location. Commonly, main memory is known as **RAM** or **Random Access Memory**, as the computer can instantly access information at any memory address:
    
    ![main_memory](main_memory.png)
    
    - **Secondary Memory**: The computer also has another type of memory, known as secondary memory or secondary storage. In this context, the words memory and storage are synonymous. Secondary memory is used to maintain a permanent record of information before and after the computer is used. Common alternative terms for secondary memory include auxiliary memory, auxiliary storage, external memory, and external storage. Secondary memory often requires sequential access, implying that the computer must search through all, or at least many, memory locations until it finds the necessary data. Commonly, this type of memory is found on the hard disk or solid-state drive.
- **Processor**: The processor, also known as the central processing unit (CPU), acts as the "brain" of the computer. Despite its essential role, the processor is a relatively simple component. Its capacity is limited to executing a set of basic instructions provided by the programmer. Processor functions include simple mathematical operations such as addition, subtraction, multiplication, and division, as well as the ability to transfer data between different memory locations. Additionally, it can interpret strings of zeros and ones as letters and send them to output devices. Although the processor has some ability to rearrange the order of instructions, its operations are fundamental and limited.

### 2.1.2.  Data hierarchy

As mentioned earlier, the fundamental unit of information in computer systems is the **binary digit** or **bit**. A hierarchy of increasingly intricate data is built upon this unit. Data hierarchy refers to the methodical organization of data, often in a hierarchical structure. The organization of data encompasses characters, fields, records, files and database.

The components of the data hierarchy are outlined below:

- A **bit** corresponds to a digit in the binary numbering system and represents the smallest unit of information. The term is an acronym for a binary digit. While the decimal numbering system uses ten digits (ten symbols), binary employs only two digits, 0 and 1.

- A **byte** or **octet** (**B**) is the fundamental unit of information used in computing and telecommunications, composed of an ordered set of **eight** bits. There are two reasons why there are eight bits. Firstly, eight is a power of 2 (8 equals $2^3$). Since computers use bits, which have only two possible values, powers of two are more convenient than powers of 10. Secondly, due to the aforementioned reason, eight bits (one byte) are needed to encode a single character (such as a letter or any other keyboard symbol). The accepted standard for this encoding is known as [ASCII](https://en.wikipedia.org/wiki/ASCII).

- A **field** contains a single fact or attribute of an entity. Consider a date field, for example, "September 19, 2004." This can be treated as a single date field (e.g., date of birth) or three fields, i.e., day of the month, month, and year.

- A **record** is a collection of related fields. An Employee record may contain fields for name, address, date of birth, etc.

- A **file** is a collection of related records. If there are 100 employees, each employee would have a record (e.g., called 'Employee Personal Details' record), and the compilation of 100 such records would constitute a file (in this case, called 'Employee Personal Details' file).

- Files are integrated into a **database**. This is achieved using a database management system. If there are other aspects of employee data that we wish to capture, additional files could be created, such as the Employee Training History file and the Employee Work History file.

### 2.1.3.  Computer software concept

Software constitutes the logical and intangible part of a computer system, comprised of programs and data that instruct the hardware on how to perform specific tasks. Typically, direct communication with the computer is not established; instead, it occurs through the operating system. The operating system allocates the computer's resources to ensure it accomplishes assigned tasks. Essentially a program itself, the operating system oversees and coordinates the functionality of all other programs, handling user requests. When a program needs to be executed, the user communicates the filename containing it to the operating system, which then proceeds with the execution. Examples of operating systems include UNIX, DOS, Linux, Windows, and Mac OS.

It is crucial to appropriately define the concept of a **program**, and to do so, it becomes necessary to introduce certain definitions:

- **Algorithm**: It is a set of well-defined, non-ambiguous, ordered, and finite instructions or rules that typically allows solving a problem. The term "algorithm" has a long history, dating back to the 9th century with the name of the Persian mathematician and astronomer, al-Khowarizmi. He authored the renowned book on number manipulation and equations titled "Kitab al-jabr w’almugabala," which translates to "Rules of Reunion and Reduction" in Spanish. The word "algebra" was derived from the Arabic word "al-jabr," present in the book's title and often translated as reunion or reintegration. The meanings of the words "algebra" and "algorithm" used to be more closely related than they are today. In fact, until modern times, the term "algorithm" solely referred to algebraic rules for solving numerical equations.
- **Programming Languages**: As previously mentioned, the "language" that computers "understand" is binary. However, humans communicate in languages like English or Spanish. Therefore, a language incompatibility issue arises when giving instructions to the computer. To address this problem, instructions can be written either in the language machines understand, known as **low-level programming languages** (also called **machine code**), or in the language humans speak, referred to as **high-level programming languages** (or **source code**). However, writing instructions in a high-level language introduces the language incompatibility issue mentioned earlier. To solve this problem, two approaches exist. One is to use a language "translator" in the programming context, known as a **compiler**. A **compiler** is a program that translates a program written in a high-level language into machine language that the computer can directly understand and execute. The other way to address the "language" incompatibility problem is by using an "interpreter" that performs translation on-the-fly. **Interpreters** differ from compilers or assemblers in that, while the latter translate a program from its description in a programming language to the machine code of the system, interpreters only translate as needed, typically instruction by instruction, and usually do not save the result of such translation. High-level languages can be classified according to their **paradigm**, with the most important ones being **declarative**, **procedural**, or **multiparadigm**. In **declarative** programming languages, the programmer tells the computer what needs to be done, and through internal control mechanisms, the computer, via the programming language, finds the most optimal solution. On the other hand, in **procedural** programming languages, the user must instruct the machine on how to execute the procedure. As for **multiparadigm** languages, they are a combination of paradigms that can incorporate both **declarative** and **procedural** elements.

- **Classification of Programming Languages According to Abstraction Level**:
    - **Low-level languages (machine code)**
    - **High-level languages (source code)**
        - **Compiled high-level languages** - FORTRAN, C, C++, Java, COBOL, PASCAL, ...
        - **Interpreted high-level languages** - Python, JavaScript, SQL...

- **Classification of Programming Languages According to Paradigm**:
    - **Declarative languages** - SQL
    - **Procedural languages** - C
    - **Multiparadigm languages** - Python
        
- **Scripting Languages**:
  - Programs can be written in plain text files with an extension specific to the programming language used, referred to as scripts. Generally, it is necessary to install the programming language on the computer to be able to read and execute these scripts.
- **IDE (Integrated Development Environment)**:
  - An **Integrated Development Environment (IDE)**, is a computer application that provides comprehensive services to ease the software development process for developers or programmers.

## 2.4. Modules

In `Python`, a module is presented as a file that houses both Python definitions and statements. These definitions encompass functions, classes, and variables (concepts that we will see in the next lectures), and may also involve code execution upon module import. The central premise supporting the use of modules lies in separating code into files, providing better organization and efficient code reuse.

Highlighted features and advantages of modules in Python include:

- **Code Structuring**: Modules enable a more structured organization of code. By grouping related functions, classes, and variables in a single file, understanding and maintaining the code are simplified.

- **Code Reusability**: Modules allow code reuse in various parts of a program or even in different programs. By importing a module into any Python script or program, its functions and classes can be utilized without the need to rewrite the code.

- **Namespace**: Modules provide a separate namespace for the variables and functions they contain. This approach prevents naming conflicts in the code and contributes to maintaining coherence.

- **Encapsulation**: Modules facilitate the encapsulation of related logic in a single file, making it easier to hide unnecessary implementation details and promoting the creation of clean and user-friendly interfaces.
